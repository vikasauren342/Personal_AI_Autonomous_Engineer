````python
# KAGGLE-009
# Purpose: Inspect Qwen model assets and V3 model configuration.
# Expected result: Clear inventory of available Qwen assets and the model
# path/backend expected by the canonical V3 model manager.
#
# IMPORTANT:
# - No model download
# - No model modification
# - No training
# - Read-only diagnostic

from pathlib import Path
import os
import json

print("=== KAGGLE-009 : QWEN / MODEL ASSET AUDIT ===")

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

print("\nProject root:")
print(" ", PROJECT_ROOT)

# ------------------------------------------------------------
# 1. Inspect V3 model directory
# ------------------------------------------------------------

MODELS_DIR = PROJECT_ROOT / "models"

print("\n[1] V3 models directory")
print(" ", MODELS_DIR)
print("Exists:", MODELS_DIR.exists())

if MODELS_DIR.exists():
    for path in sorted(MODELS_DIR.rglob("*")):
        if path.is_file():
            size_mb = path.stat().st_size / (1024 * 1024)
            print(f"  FILE  {path.relative_to(MODELS_DIR)}  [{size_mb:.2f} MB]")
        elif path.is_dir():
            print(f"  DIR   {path.relative_to(MODELS_DIR)}")

# ------------------------------------------------------------
# 2. Inspect Qwen-specific assets
# ------------------------------------------------------------

QWEN_DIR = MODELS_DIR / "qwen_assets"

print("\n[2] Qwen assets")
print(" ", QWEN_DIR)
print("Exists:", QWEN_DIR.exists())

qwen_files = []

if QWEN_DIR.exists():
    for path in sorted(QWEN_DIR.rglob("*")):
        if path.is_file():
            size_mb = path.stat().st_size / (1024 * 1024)
            qwen_files.append(path)
            print(
                f"  {path.relative_to(QWEN_DIR)} "
                f"[{size_mb:.2f} MB]"
            )

print("\nQwen file count:", len(qwen_files))

# ------------------------------------------------------------
# 3. Look for likely model-weight files
# ------------------------------------------------------------

print("\n[3] Possible model-weight files")

weight_extensions = {
    ".safetensors",
    ".bin",
    ".pt",
    ".pth",
    ".gguf",
    ".ggml"
}

weight_files = []

for root in [
    PROJECT_ROOT / "models",
    Path("/kaggle/input"),
]:
    if root.exists():
        for path in root.rglob("*"):
            if path.is_file() and path.suffix.lower() in weight_extensions:
                weight_files.append(path)

if weight_files:
    for path in weight_files:
        size_gb = path.stat().st_size / (1024 ** 3)
        print(
            f"  WEIGHT: {path} "
            f"[{size_gb:.3f} GB]"
        )
else:
    print("  No model-weight files found.")

# ------------------------------------------------------------
# 4. Inspect V3 configuration files
# ------------------------------------------------------------

print("\n[4] Configuration files")

CONFIG_DIR = PROJECT_ROOT / "config"

config_files = []

if CONFIG_DIR.exists():
    for path in sorted(CONFIG_DIR.rglob("*")):
        if path.is_file():
            config_files.append(path)
            print(" ", path.relative_to(CONFIG_DIR))

# ------------------------------------------------------------
# 5. Print model-related configuration safely
# ------------------------------------------------------------

print("\n[5] Model-related configuration")

keywords = (
    "model",
    "qwen",
    "backend",
    "path",
    "checkpoint",
    "tokenizer",
    "device"
)

for path in config_files:
    try:
        text = path.read_text(errors="ignore")
    except Exception:
        continue

    matching_lines = []

    for line_number, line in enumerate(text.splitlines(), 1):
        lower = line.lower()

        if any(keyword in lower for keyword in keywords):
            matching_lines.append(
                (line_number, line.strip())
            )

    if matching_lines:
        print(f"\n--- {path.relative_to(CONFIG_DIR)} ---")

        for line_number, line in matching_lines[:80]:
            print(f"{line_number}: {line}")

# ------------------------------------------------------------
# 6. Environment model variables
# ------------------------------------------------------------

print("\n[6] Relevant environment variables")

env_keys = [
    key for key in os.environ
    if any(
        token in key.upper()
        for token in [
            "MODEL",
            "QWEN",
            "TOKENIZER",
            "CHECKPOINT",
            "CUDA"
        ]
    )
]

if env_keys:
    for key in sorted(env_keys):
        value = os.environ[key]

        # Avoid dumping potentially sensitive values.
        if any(secret in key.upper() for secret in ["TOKEN", "KEY", "SECRET"]):
            value = "<REDACTED>"

        print(f"  {key}={value}")
else:
    print("  No relevant model environment variables detected.")

# ------------------------------------------------------------
# 7. Final classification
# ------------------------------------------------------------

print("\n=== RESULT ===")

if weight_files:
    print("Model weights: FOUND")
else:
    print("Model weights: NOT FOUND")

if qwen_files:
    print("Qwen assets: FOUND")
else:
    print("Qwen assets: NOT FOUND")

print("V3 model audit: COMPLETE")

print("\n=== END KAGGLE-009 ===")
````

````python
# KAGGLE-011
# Purpose: Verify whether a complete Qwen Transformers checkpoint exists
# in Kaggle input, the V3 project, or another configured model directory.
# Expected result: A definitive classification of:
#   COMPLETE_CHECKPOINT
#   PARTIAL_CHECKPOINT
#   NO_CHECKPOINT
#
# IMPORTANT:
# - No model download
# - No model loading
# - No source modification
# - Read-only diagnostic

from pathlib import Path
import os
import json

print("=== KAGGLE-011 : QWEN CHECKPOINT VERIFICATION ===")

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

# ------------------------------------------------------------
# 1. Candidate model directories
# ------------------------------------------------------------

candidates = []

env_model_dir = os.environ.get("PERSONAL_AI_MODEL_DIR")

if env_model_dir:
    candidates.append(
        ("PERSONAL_AI_MODEL_DIR", Path(env_model_dir))
    )

candidates.extend([
    (
        "V3/models/qwen2.5-1.5b-instruct",
        PROJECT_ROOT / "models" / "qwen2.5-1.5b-instruct"
    ),
    (
        "V3/models/qwen",
        PROJECT_ROOT / "models" / "qwen"
    ),
    (
        "V3/models/qwen_assets",
        PROJECT_ROOT / "models" / "qwen_assets"
    ),
    (
        "Kaggle personal-ai-qwen",
        Path("/kaggle/input/personal-ai-qwen")
    ),
    (
        "Kaggle dataset personal-ai-qwen",
        Path("/kaggle/input/datasets/vikaschandoliya/personal-ai-qwen")
    ),
    (
        "Colab Qwen path",
        Path("/content/personal_ai/models/qwen2.5-1.5b-instruct")
    ),
])

# ------------------------------------------------------------
# 2. Check checkpoint signatures
# ------------------------------------------------------------

print("\n[1] Candidate directories")

weight_extensions = {
    ".safetensors",
    ".bin",
    ".pt",
    ".pth",
    ".gguf",
    ".ggml"
}

for name, path in candidates:

    print(f"\n--- {name} ---")
    print("Path:", path)
    print("Exists:", path.exists())

    if not path.exists():
        continue

    files = [
        p for p in path.rglob("*")
        if p.is_file()
    ]

    print("Files:", len(files))

    config = path / "config.json"
    tokenizer = path / "tokenizer.json"

    weights = [
        p for p in files
        if p.suffix.lower() in weight_extensions
    ]

    print("config.json:", config.exists())
    print("tokenizer.json:", tokenizer.exists())
    print("weight files:", len(weights))

    for weight in weights:
        size_gb = weight.stat().st_size / (1024 ** 3)
        print(
            f"  WEIGHT: {weight.name} "
            f"[{size_gb:.3f} GB]"
        )

# ------------------------------------------------------------
# 3. Look specifically for Transformers shard indexes
# ------------------------------------------------------------

print("\n[2] Transformers checkpoint indexes")

index_names = [
    "model.safetensors.index.json",
    "pytorch_model.bin.index.json",
]

found_indexes = []

search_roots = [
    PROJECT_ROOT / "models",
    Path("/kaggle/input"),
    Path("/kaggle/working"),
]

for root in search_roots:

    if not root.exists():
        continue

    for index_name in index_names:

        for index in root.rglob(index_name):

            if index.is_file():
                found_indexes.append(index)

                print("FOUND:", index)

# ------------------------------------------------------------
# 4. Validate any found shard index
# ------------------------------------------------------------

print("\n[3] Shard-index validation")

if found_indexes:

    for index in found_indexes:

        print(f"\n--- {index} ---")

        try:
            data = json.loads(
                index.read_text(errors="ignore")
            )

            weight_map = data.get("weight_map", {})

            print("Weight-map entries:", len(weight_map))

            missing = []

            for filename in set(weight_map.values()):

                shard = index.parent / filename

                if not shard.exists():
                    missing.append(filename)

            if missing:
                print(
                    "Missing shards:",
                    len(missing)
                )

                for filename in missing[:20]:
                    print("  MISSING:", filename)

            else:
                print("All referenced shards: FOUND")

        except Exception as exc:
            print("Could not parse index:", exc)

else:
    print("No Transformers shard indexes found.")

# ------------------------------------------------------------
# 5. Search for model-like directories under Kaggle input
# ------------------------------------------------------------

print("\n[4] Kaggle model directory discovery")

kaggle_root = Path("/kaggle/input")

model_like_dirs = []

if kaggle_root.exists():

    for path in kaggle_root.rglob("config.json"):

        parent = path.parent

        # Avoid listing the same directory repeatedly.
        if parent not in model_like_dirs:
            model_like_dirs.append(parent)

for path in sorted(model_like_dirs):

    files = [
        p for p in path.rglob("*")
        if p.is_file()
    ]

    weights = [
        p for p in files
        if p.suffix.lower() in weight_extensions
    ]

    if weights or "qwen" in str(path).lower():

        print(
            f"\n  {path}"
            f"\n    files={len(files)}"
            f"\n    weight_files={len(weights)}"
        )

# ------------------------------------------------------------
# 6. Final classification
# ------------------------------------------------------------

print("\n=== CLASSIFICATION ===")

all_weights = []

for name, path in candidates:

    if not path.exists():
        continue

    for p in path.rglob("*"):
        if p.is_file() and p.suffix.lower() in weight_extensions:
            all_weights.append(p)

# Deduplicate by resolved path.
all_weights = list({
    str(p.resolve()): p
    for p in all_weights
}.values())

if found_indexes:
    print("Transformers shard index: FOUND")

if all_weights:
    print("Model weight files: FOUND")
else:
    print("Model weight files: NOT FOUND")

if found_indexes and all_weights:
    classification = "COMPLETE_CHECKPOINT_CANDIDATE"
elif all_weights:
    classification = "WEIGHT_FILES_FOUND_NEEDS_VALIDATION"
else:
    classification = "NO_COMPLETE_CHECKPOINT"

print("\nRESULT:", classification)

print("\n=== END KAGGLE-011 ===")
````

````python
# KAGGLE-013
# Purpose: Download the Qwen 2.5 1.5B Instruct checkpoint into a
# dedicated writable model directory for the Personal AI runtime.
#
# Expected result:
#   /kaggle/working/personal_ai_models/qwen2.5-1.5b-instruct/
# contains the complete Hugging Face Transformers checkpoint.
#
# IMPORTANT:
# - Does NOT modify V3 source code.
# - Does NOT modify V3 configuration files.
# - Does NOT load the model.
# - Download only.
# - The model is kept OUTSIDE the canonical V3 source tree.

from pathlib import Path
import os
import shutil

print("=== KAGGLE-013 : ACQUIRE QWEN CHECKPOINT ===")

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"

MODEL_ROOT = Path(
    "/kaggle/working/personal_ai_models"
)

MODEL_DIR = (
    MODEL_ROOT /
    "qwen2.5-1.5b-instruct"
)

MODEL_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

print("\n[1] Model configuration")
print("Model ID :", MODEL_ID)
print("Target   :", MODEL_DIR)
print("Exists   :", MODEL_DIR.exists())

# ------------------------------------------------------------
# 2. Check whether a previous download exists
# ------------------------------------------------------------

print("\n[2] Existing checkpoint")

if MODEL_DIR.exists():

    existing_files = [
        p for p in MODEL_DIR.rglob("*")
        if p.is_file()
    ]

    print(
        "Existing files:",
        len(existing_files)
    )

    if existing_files:

        total_size = sum(
            p.stat().st_size
            for p in existing_files
        )

        print(
            "Existing size:",
            f"{total_size / (1024**3):.3f} GB"
        )

else:

    print("No previous checkpoint found.")

# ------------------------------------------------------------
# 3. Download using Hugging Face Hub
# ------------------------------------------------------------

print("\n[3] Download")

from huggingface_hub import snapshot_download

try:

    downloaded_path = snapshot_download(
        repo_id=MODEL_ID,
        local_dir=str(MODEL_DIR),
        local_dir_use_symlinks=False,
        resume_download=True,
    )

    print("\nDownload completed.")
    print("Resolved path:", downloaded_path)

except TypeError:

    # Compatibility fallback for newer huggingface_hub
    downloaded_path = snapshot_download(
        repo_id=MODEL_ID,
        local_dir=str(MODEL_DIR),
        resume_download=True,
    )

    print("\nDownload completed.")
    print("Resolved path:", downloaded_path)

except Exception as exc:

    print("\nDOWNLOAD FAILED")
    print("Error type:", type(exc).__name__)
    print("Error:", exc)

    raise

# ------------------------------------------------------------
# 4. Basic checkpoint inventory
# ------------------------------------------------------------

print("\n[4] Checkpoint inventory")

if not MODEL_DIR.exists():

    raise RuntimeError(
        "Model directory does not exist after download."
    )

files = [
    p for p in MODEL_DIR.rglob("*")
    if p.is_file()
]

total_size = sum(
    p.stat().st_size
    for p in files
)

print("File count :", len(files))
print(
    "Total size :",
    f"{total_size / (1024**3):.3f} GB"
)

for p in sorted(files):

    relative = p.relative_to(MODEL_DIR)

    size_mb = p.stat().st_size / (1024**2)

    print(
        f"  {relative} "
        f"[{size_mb:.2f} MB]"
    )

# ------------------------------------------------------------
# 5. Required Transformers files
# ------------------------------------------------------------

print("\n[5] Required files")

required = [
    "config.json",
    "tokenizer.json",
    "tokenizer_config.json",
]

for filename in required:

    path = MODEL_DIR / filename

    print(
        f"  [{'FOUND' if path.exists() else 'MISSING'}] "
        f"{filename}"
    )

# ------------------------------------------------------------
# 6. Weight-file discovery
# ------------------------------------------------------------

print("\n[6] Model weights")

weight_extensions = {
    ".safetensors",
    ".bin",
    ".pt",
    ".pth"
}

weight_files = [
    p for p in files
    if p.suffix.lower() in weight_extensions
]

print(
    "Weight files:",
    len(weight_files)
)

for p in weight_files:

    size_gb = p.stat().st_size / (1024**3)

    print(
        f"  {p.relative_to(MODEL_DIR)} "
        f"[{size_gb:.3f} GB]"
    )

# ------------------------------------------------------------
# 7. Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")

if weight_files:
    print("Qwen checkpoint acquisition: SUCCESS")
    print("Model weights: FOUND")
    print("Model loading: NOT PERFORMED")
    print("V3 source modification: NONE")
else:
    print("Qwen checkpoint acquisition: INCOMPLETE")
    print("Model weights: NOT FOUND")

print("\nModel directory:")
print(MODEL_DIR)

print("\n=== END KAGGLE-013 ===")
````

````python
# KAGGLE-014
# Purpose: Validate the downloaded Qwen checkpoint before GPU loading.
#
# Checks:
#   1. Required files
#   2. config.json validity
#   3. Model architecture/configuration
#   4. safetensors metadata
#   5. Tensor count / tensor metadata
#   6. Tokenizer loading
#   7. Transformers configuration loading
#
# Expected result:
#   CHECKPOINT VALIDATION: PASS
#
# IMPORTANT:
# - No model weights are loaded into GPU.
# - No model generation.
# - No V3 source modification.

from pathlib import Path
import os
import json

print("=== KAGGLE-014 : QWEN CHECKPOINT INTEGRITY VALIDATION ===")

MODEL_DIR = Path(
    "/kaggle/working/personal_ai_models/qwen2.5-1.5b-instruct"
).resolve()

print("\nModel directory:")
print(MODEL_DIR)

if not MODEL_DIR.exists():
    raise FileNotFoundError(
        f"Model directory not found: {MODEL_DIR}"
    )

# ------------------------------------------------------------
# 1. Required files
# ------------------------------------------------------------

print("\n[1] Required checkpoint files")

required_files = [
    "config.json",
    "generation_config.json",
    "tokenizer.json",
    "tokenizer_config.json",
    "vocab.json",
    "merges.txt",
    "model.safetensors",
]

all_required = True

for filename in required_files:

    path = MODEL_DIR / filename
    exists = path.exists()

    print(
        f"  [{'FOUND' if exists else 'MISSING'}] "
        f"{filename}"
    )

    if not exists:
        all_required = False

# ------------------------------------------------------------
# 2. Validate config.json
# ------------------------------------------------------------

print("\n[2] config.json validation")

config_path = MODEL_DIR / "config.json"

config = None

try:

    config = json.loads(
        config_path.read_text()
    )

    print("JSON parse: PASS")

    important_keys = [
        "model_type",
        "architectures",
        "hidden_size",
        "num_hidden_layers",
        "num_attention_heads",
        "vocab_size",
    ]

    for key in important_keys:

        print(
            f"  {key}:",
            config.get(key, "<not present>")
        )

except Exception as exc:

    print(
        "JSON parse: FAIL",
        type(exc).__name__,
        str(exc)
    )

    all_required = False

# ------------------------------------------------------------
# 3. Transformers configuration validation
# ------------------------------------------------------------

print("\n[3] Transformers configuration")

transformers_config_ok = False

try:

    from transformers import AutoConfig

    hf_config = AutoConfig.from_pretrained(
        MODEL_DIR,
        local_files_only=True
    )

    transformers_config_ok = True

    print("AutoConfig: PASS")
    print(
        "Model type:",
        getattr(hf_config, "model_type", "unknown")
    )

    print(
        "Architecture:",
        getattr(
            hf_config,
            "architectures",
            "unknown"
        )
    )

except Exception as exc:

    print(
        "AutoConfig: FAIL",
        type(exc).__name__,
        str(exc)
    )

# ------------------------------------------------------------
# 4. Safetensors metadata validation
# ------------------------------------------------------------

print("\n[4] safetensors validation")

safetensors_ok = False

weights_path = MODEL_DIR / "model.safetensors"

try:

    from safetensors import safe_open

    with safe_open(
        str(weights_path),
        framework="pt",
        device="cpu"
    ) as f:

        keys = list(f.keys())

        safetensors_ok = True

        print("File open: PASS")
        print("Tensor count:", len(keys))

        print("\nFirst tensor entries:")

        for key in keys[:15]:

            tensor = f.get_tensor(key)

            print(
                f"  {key}"
                f" | shape={tuple(tensor.shape)}"
                f" | dtype={tensor.dtype}"
            )

        if len(keys) > 15:
            print(
                "  ...",
                len(keys) - 15,
                "additional tensors"
            )

except Exception as exc:

    print(
        "Safetensors validation: FAIL",
        type(exc).__name__,
        str(exc)
    )

# ------------------------------------------------------------
# 5. Tokenizer validation
# ------------------------------------------------------------

print("\n[5] Tokenizer validation")

tokenizer_ok = False

try:

    from transformers import AutoTokenizer

    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_DIR,
        local_files_only=True
    )

    tokenizer_ok = True

    print("AutoTokenizer: PASS")
    print(
        "Tokenizer class:",
        tokenizer.__class__.__name__
    )

    print(
        "Vocabulary size:",
        len(tokenizer)
    )

    print(
        "Chat template:",
        "FOUND"
        if getattr(tokenizer, "chat_template", None)
        else "NOT FOUND"
    )

except Exception as exc:

    print(
        "AutoTokenizer: FAIL",
        type(exc).__name__,
        str(exc)
    )

# ------------------------------------------------------------
# 6. Model class resolution — WITHOUT loading weights
# ------------------------------------------------------------

print("\n[6] Model class resolution")

model_class_ok = False

try:

    from transformers import AutoModelForCausalLM

    model_class = AutoModelForCausalLM.from_config(
        hf_config
    )

    model_class_ok = True

    print(
        "Model class resolution: PASS"
    )

    print(
        "Resolved class:",
        model_class.__class__.__name__
    )

    # Delete the config-created model immediately.
    # This does not load checkpoint weights.
    del model_class

except Exception as exc:

    print(
        "Model class resolution: FAIL",
        type(exc).__name__,
        str(exc)
    )

# ------------------------------------------------------------
# 7. Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")

checks = {
    "required_files": all_required,
    "transformers_config": transformers_config_ok,
    "safetensors": safetensors_ok,
    "tokenizer": tokenizer_ok,
    "model_class": model_class_ok,
}

for name, passed in checks.items():

    print(
        f"  [{'PASS' if passed else 'FAIL'}] "
        f"{name}"
    )

if all(checks.values()):
    print("\nCHECKPOINT VALIDATION: PASS")
    print("Safe to proceed to GPU loading.")
else:
    print("\nCHECKPOINT VALIDATION: FAIL")
    print("Do NOT load the model yet.")

print("\n=== END KAGGLE-014 ===")
````

````python
# KAGGLE-016
# Purpose: Load the real Qwen 2.5-1.5B Instruct model on Kaggle GPU.
#
# Expected result:
#   CUDA available        = True
#   GPU count             = 2
#   Qwen model loaded     = SUCCESS
#   Model device mapping  = reported
#
# IMPORTANT:
# - This is the FIRST actual model load.
# - No V3 source modification.
# - No training.
# - No model fine-tuning.
# - No generation yet.

from pathlib import Path
import os
import sys
import torch

print("=== KAGGLE-016 : FIRST GPU MODEL LOAD ===")

# ------------------------------------------------------------
# 1. Runtime
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

MODEL_DIR = Path(
    os.environ["PERSONAL_AI_MODEL_DIR"]
).resolve()

print("\n[1] Runtime")

print("Project:", PROJECT_ROOT)
print("Model  :", MODEL_DIR)

print(
    "PyTorch:",
    torch.__version__
)

print(
    "CUDA available:",
    torch.cuda.is_available()
)

print(
    "GPU count:",
    torch.cuda.device_count()
)

if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA is not available."
    )

if torch.cuda.device_count() == 0:
    raise RuntimeError(
        "No GPU detected."
    )

# ------------------------------------------------------------
# 2. GPU information
# ------------------------------------------------------------

print("\n[2] GPUs")

for i in range(torch.cuda.device_count()):

    print(
        f"GPU {i}:",
        torch.cuda.get_device_name(i)
    )

# ------------------------------------------------------------
# 3. Load tokenizer
# ------------------------------------------------------------

print("\n[3] Loading tokenizer")

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_DIR,
    local_files_only=True
)

print(
    "Tokenizer:",
    tokenizer.__class__.__name__
)

print(
    "Vocabulary:",
    len(tokenizer)
)

# ------------------------------------------------------------
# 4. Load model
# ------------------------------------------------------------

print("\n[4] Loading Qwen model")

from transformers import AutoModelForCausalLM

model = AutoModelForCausalLM.from_pretrained(
    MODEL_DIR,
    device_map="auto",
    torch_dtype="auto",
    local_files_only=True,
)

print("\nModel load: SUCCESS")

# ------------------------------------------------------------
# 5. Model information
# ------------------------------------------------------------

print("\n[5] Model information")

print(
    "Model class:",
    model.__class__.__name__
)

print(
    "Parameter count:",
    f"{sum(p.numel() for p in model.parameters()):,}"
)

print(
    "Model dtype:",
    next(model.parameters()).dtype
)

print(
    "Input device:",
    model.device
)

# ------------------------------------------------------------
# 6. Device map
# ------------------------------------------------------------

print("\n[6] Device map")

device_map = getattr(
    model,
    "hf_device_map",
    None
)

if device_map:

    for device, count in sorted(
        (
            (str(v), list(device_map.values()).count(v))
            for v in set(device_map.values())
        ),
        key=lambda x: x[0]
    ):
        print(
            f"  {device}: {count} modules"
        )

else:

    print(
        "hf_device_map unavailable."
    )

# ------------------------------------------------------------
# 7. GPU memory
# ------------------------------------------------------------

print("\n[7] GPU memory")

for i in range(torch.cuda.device_count()):

    allocated = (
        torch.cuda.memory_allocated(i)
        / (1024 ** 3)
    )

    reserved = (
        torch.cuda.memory_reserved(i)
        / (1024 ** 3)
    )

    print(
        f"GPU {i}: "
        f"allocated={allocated:.3f} GB, "
        f"reserved={reserved:.3f} GB"
    )

# ------------------------------------------------------------
# Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")

print("Qwen tokenizer: LOADED")
print("Qwen model:     LOADED")
print("CUDA:           ACTIVE")
print("Generation:     NOT PERFORMED")
print("Training:       NOT PERFORMED")
print("V3 source:      UNMODIFIED")

print("\n=== END KAGGLE-016 ===")
````

````python
# KAGGLE-017
# Purpose: Generate the FIRST real response from the loaded
# Qwen 2.5-1.5B Instruct model.
#
# Expected result:
#   A real natural-language response generated by Qwen.
#
# IMPORTANT:
# - Uses the model already loaded by KAGGLE-016.
# - No training.
# - No fine-tuning.
# - No V3 source modification.

print("=== KAGGLE-017 : FIRST REAL AI RESPONSE ===")

# ------------------------------------------------------------
# 1. Verify loaded objects
# ------------------------------------------------------------

print("\n[1] Loaded model verification")

try:
    model
    tokenizer
except NameError:
    raise RuntimeError(
        "Model/tokenizer not found. Run KAGGLE-016 first."
    )

print("Model   :", model.__class__.__name__)
print("Tokenizer:", tokenizer.__class__.__name__)

# ------------------------------------------------------------
# 2. Create test conversation
# ------------------------------------------------------------

messages = [
    {
        "role": "system",
        "content": (
            "You are a helpful Personal AI assistant. "
            "Answer clearly and honestly."
        )
    },
    {
        "role": "user",
        "content": (
            "Hello. Introduce yourself briefly and explain "
            "what you can do."
        )
    }
]

# ------------------------------------------------------------
# 3. Apply Qwen chat template
# ------------------------------------------------------------

print("\n[2] Preparing prompt")

prompt = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

print("Prompt prepared.")

# ------------------------------------------------------------
# 4. Tokenize
# ------------------------------------------------------------

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

# Put inputs on the same device as the model's input embedding.
try:
    input_device = model.get_input_embeddings().weight.device
except Exception:
    input_device = model.device

inputs = {
    key: value.to(input_device)
    for key, value in inputs.items()
}

print(
    "Input device:",
    input_device
)

print(
    "Input tokens:",
    inputs["input_ids"].shape[1]
)

# ------------------------------------------------------------
# 5. Generate
# ------------------------------------------------------------

print("\n[3] Generating...")
print("-" * 60)

import torch

with torch.inference_mode():

    output = model.generate(
        **inputs,
        max_new_tokens=256,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
    )

# ------------------------------------------------------------
# 6. Decode only generated tokens
# ------------------------------------------------------------

input_length = inputs["input_ids"].shape[1]

generated_tokens = output[0][input_length:]

response = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
)

print("\n=== AI RESPONSE ===")
print(response)
print("===================")

# ------------------------------------------------------------
# 7. Basic inference statistics
# ------------------------------------------------------------

print("\n[4] Inference statistics")

print(
    "Generated tokens:",
    generated_tokens.shape[0]
)

print(
    "Output tensor shape:",
    tuple(output.shape)
)

# ------------------------------------------------------------
# 8. GPU memory
# ------------------------------------------------------------

print("\n[5] GPU memory after generation")

for i in range(torch.cuda.device_count()):

    allocated = (
        torch.cuda.memory_allocated(i)
        / (1024 ** 3)
    )

    reserved = (
        torch.cuda.memory_reserved(i)
        / (1024 ** 3)
    )

    print(
        f"GPU {i}: "
        f"allocated={allocated:.3f} GB, "
        f"reserved={reserved:.3f} GB"
    )

# ------------------------------------------------------------
# Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")
print("Qwen model:       RUNNING")
print("Inference:        SUCCESS")
print("Real AI response: GENERATED")
print("Training:         NOT PERFORMED")
print("V3 source:        UNMODIFIED")

print("\n=== END KAGGLE-017 ===")
````

````python
# KAGGLE-020
# Platform: Kaggle
# Purpose:
#   Perform the first complete end-to-end V3 Personal AI inference test.
#
#   Path being tested:
#       Bootstrap
#          -> Runtime
#          -> ModelManager
#          -> Router
#          -> AgentRegistry
#          -> Agent
#          -> Qwen model
#          -> Memory
#
# Expected result:
#   A real answer should be produced through Orchestrator.ask().
#   Health should show qwen_local + fallback and the V3 agents.
#   V3 source remains unmodified.
#
# Note:
#   QwenBackend is lazy-loaded, so the V3 runtime may load the checkpoint
#   again even though KAGGLE-016/017 already proved direct inference.

from pathlib import Path
import os
import sys
import json
import time

# ------------------------------------------------------------
# 1. Project
# ------------------------------------------------------------

PROJECT_ROOT = Path(os.environ["PERSONAL_AI_ROOT"]).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=== KAGGLE-020 : FIRST END-TO-END PERSONAL AI TEST ===\n")

print("[1] Project")
print("-" * 60)
print("PROJECT_ROOT:", PROJECT_ROOT)
print("Exists:", PROJECT_ROOT.exists())

# ------------------------------------------------------------
# 2. Environment
# ------------------------------------------------------------

print("\n[2] Environment")
print("-" * 60)

import torch

print("CUDA available:", torch.cuda.is_available())
print("GPU count:", torch.cuda.device_count())

if torch.cuda.is_available():
    for i in range(torch.cuda.device_count()):
        print(
            f"GPU {i}:",
            torch.cuda.get_device_name(i)
        )

# ------------------------------------------------------------
# 3. Exact V3 Bootstrap
# ------------------------------------------------------------

print("\n[3] Starting V3 Bootstrap")
print("-" * 60)

from personal_ai.bootstrap import Bootstrap

bootstrap = Bootstrap(PROJECT_ROOT)

print("Bootstrap class:", type(bootstrap).__name__)
print("Starting runtime...")

t0 = time.time()

runtime = bootstrap.start()

bootstrap_time = time.time() - t0

print("Runtime created:", runtime is not None)
print("Bootstrap time:", round(bootstrap_time, 2), "seconds")

# ------------------------------------------------------------
# 4. Runtime components
# ------------------------------------------------------------

print("\n[4] Runtime components")
print("-" * 60)

components = [
    "config",
    "env",
    "registry",
    "memory",
    "models",
    "tools",
    "agents",
    "router",
    "evaluator",
    "learning",
]

for name in components:
    obj = getattr(runtime, name, None)
    print(f"{name:12}: {type(obj).__name__ if obj is not None else 'MISSING'}")

# ------------------------------------------------------------
# 5. Model backend status
# ------------------------------------------------------------

print("\n[5] Model backends")
print("-" * 60)

for name, backend in runtime.models.backends.items():
    print(
        f"{name:15} "
        f"class={type(backend).__name__} "
        f"available={backend.available()}"
    )

# ------------------------------------------------------------
# 6. Agent status
# ------------------------------------------------------------

print("\n[6] Agents")
print("-" * 60)

print("Registered agents:")

try:
    for name in runtime.agents.agents:
        print(" ", name)
except Exception as e:
    print("Agent listing error:", repr(e))

# ------------------------------------------------------------
# 7. Orchestrator
# ------------------------------------------------------------

print("\n[7] Creating Orchestrator")
print("-" * 60)

from personal_ai.core.orchestrator import Orchestrator

orchestrator = Orchestrator(runtime)

print("Orchestrator:", type(orchestrator).__name__)
print("Planner:", type(orchestrator.planner).__name__)
print("Execution engine:", type(orchestrator.engine).__name__)

# ------------------------------------------------------------
# 8. Health
# ------------------------------------------------------------

print("\n[8] V3 Health")
print("-" * 60)

health = orchestrator.health()

print(json.dumps(health, indent=2, default=str))

# ------------------------------------------------------------
# 9. First real Personal AI query
# ------------------------------------------------------------

print("\n[9] FIRST REAL PERSONAL AI QUERY")
print("-" * 60)

query = (
    "Explain in simple terms what you are, what your main components are, "
    "and how the components work together."
)

print("User query:")
print(query)

print("\nGenerating...")
print("-" * 60)

t1 = time.time()

try:
    result = orchestrator.ask(query, autonomous=False)

    inference_time = time.time() - t1

    print("\n=== PERSONAL AI RESPONSE ===")
    print(result.get("answer", result))

    print("\n=== ROUTING ===")
    print("Route:", result.get("route"))

    print("\n=== MEMORY HITS ===")
    memories = result.get("memory_hits", [])
    print("Count:", len(memories))

    print("\n=== INFERENCE TIME ===")
    print(round(inference_time, 2), "seconds")

    # --------------------------------------------------------
    # 10. Final validation
    # --------------------------------------------------------

    print("\n[10] Validation")
    print("-" * 60)

    answer = result.get("answer")

    print("Result type:", type(result).__name__)
    print("Answer exists:", bool(answer))
    print("Route exists:", bool(result.get("route")))
    print("Memory returned:", "memory_hits" in result)

    print("\n=== RESULT ===")
    print("Bootstrap              : PASS")
    print("Runtime                : PASS")
    print("Model registration     : PASS")
    print("Orchestrator           : PASS")
    print("End-to-end inference   :", "PASS" if answer else "FAIL")
    print("V3 source modification : NONE")
    print("Training               : NOT PERFORMED")

except Exception as e:

    inference_time = time.time() - t1

    print("\n=== END-TO-END ERROR ===")
    print(type(e).__name__ + ":", str(e))

    import traceback
    traceback.print_exc()

    print("\n=== RESULT ===")
    print("Bootstrap              : PASS")
    print("Runtime                : PASS")
    print("Orchestrator           : CREATED")
    print("End-to-end inference   : FAIL")
    print("V3 source modification : NONE")

print("\n=== END KAGGLE-020 ===")
````

````python
# KAGGLE-021
# Platform: Kaggle
# Purpose:
#   Test the existing V3 autonomous execution path.
#
#   This verifies:
#       Orchestrator
#          -> Router
#          -> Agent
#          -> Planner
#          -> ExecutionEngine
#          -> Evaluation
#          -> LearningLedger
#
# Expected result:
#   autonomous=True returns an execution object containing the
#   execution/evaluation information produced by the existing V3 core.
#
# Important:
#   No V3 source modification.
#   No training.
#   We are testing existing architecture only.

from pathlib import Path
import os
import sys
import json
import time

PROJECT_ROOT = Path(os.environ["PERSONAL_AI_ROOT"]).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=== KAGGLE-021 : AUTONOMOUS EXECUTION TEST ===\n")

# ------------------------------------------------------------
# 1. Bootstrap
# ------------------------------------------------------------

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.orchestrator import Orchestrator

print("[1] Starting V3 Runtime")
print("-" * 60)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime:", type(runtime).__name__)

# ------------------------------------------------------------
# 2. Create orchestrator
# ------------------------------------------------------------

print("\n[2] Creating Orchestrator")
print("-" * 60)

orchestrator = Orchestrator(runtime)

print("Orchestrator:", type(orchestrator).__name__)
print("Planner:", type(orchestrator.planner).__name__)
print("ExecutionEngine:", type(orchestrator.engine).__name__)

# ------------------------------------------------------------
# 3. Autonomous query
# ------------------------------------------------------------

query = (
    "Analyze this task and determine the best way to answer it: "
    "Explain why modular architecture is useful for building a "
    "long-term personal AI system."
)

print("\n[3] Autonomous query")
print("-" * 60)
print(query)

# ------------------------------------------------------------
# 4. Execute
# ------------------------------------------------------------

print("\n[4] Running autonomous=True")
print("-" * 60)

t0 = time.time()

try:
    result = orchestrator.ask(
        query,
        autonomous=True
    )

    elapsed = time.time() - t0

    # --------------------------------------------------------
    # 5. Result
    # --------------------------------------------------------

    print("\n[5] Autonomous result")
    print("-" * 60)

    print("Result type:", type(result).__name__)

    if isinstance(result, dict):

        print("\nTop-level keys:")
        for key in result.keys():
            print(" ", key)

        print("\nRoute:")
        print(result.get("route"))

        print("\nAnswer:")
        print(result.get("answer"))

        print("\nMemory hits:")
        print(len(result.get("memory_hits", [])))

        execution = result.get("execution")

        print("\nExecution object:")
        print(type(execution).__name__)

        print("\nExecution details:")
        print(json.dumps(
            execution,
            indent=2,
            default=str
        ))

    else:
        print(result)

    # --------------------------------------------------------
    # 6. Validation
    # --------------------------------------------------------

    print("\n[6] Validation")
    print("-" * 60)

    print("Result exists:", bool(result))

    if isinstance(result, dict):
        print("Route present:", "route" in result)
        print("Answer present:", bool(result.get("answer")))
        print("Memory present:", "memory_hits" in result)
        print("Execution present:", "execution" in result)

    print("\nElapsed:", round(elapsed, 2), "seconds")

    print("\n=== RESULT ===")
    print("Runtime                 : PASS")
    print("Orchestrator            : PASS")
    print("Autonomous path         : REACHED")
    print(
        "Execution result        :",
        "PRESENT" if isinstance(result, dict) and "execution" in result
        else "MISSING"
    )
    print("V3 source modification  : NONE")
    print("Training                : NOT PERFORMED")

except Exception as e:

    print("\n=== AUTONOMOUS EXECUTION ERROR ===")
    print(type(e).__name__ + ":", str(e))

    import traceback
    traceback.print_exc()

    print("\n=== RESULT ===")
    print("Runtime                 : PASS")
    print("Orchestrator            : CREATED")
    print("Autonomous path         : FAILED")
    print("V3 source modification  : NONE")

print("\n=== END KAGGLE-021 ===")
````

````python
# KAGGLE-022
# Platform: Kaggle
#
# Purpose:
#   Inspect and test the EXISTING V3 tool layer before we modify anything.
#
#   We want to determine:
#       1. Which tools are registered?
#       2. Which tools are available?
#       3. What interfaces do they expose?
#       4. Can the runtime execute a simple deterministic tool directly?
#       5. Does the autonomous architecture currently have a usable
#          tool-execution foundation?
#
# Expected result:
#   A complete inventory of the existing V3 tools and a deterministic
#   execution test.
#
# Important:
#   - V3 source modification: NONE
#   - Model training: NONE
#   - Model reload: NONE
#   - We are inspecting/testing the existing architecture only.

from pathlib import Path
import os
import sys
import inspect
import json
import traceback

PROJECT_ROOT = Path(os.environ["PERSONAL_AI_ROOT"]).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=== KAGGLE-022 : TOOL LAYER AUDIT + EXECUTION TEST ===\n")

# ------------------------------------------------------------
# [1] Bootstrap exact V3 runtime
# ------------------------------------------------------------

from personal_ai.bootstrap import Bootstrap

print("[1] Starting exact V3 runtime")
print("-" * 60)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime:", type(runtime).__name__)
print("Tools object:", type(runtime.tools).__name__)

# ------------------------------------------------------------
# [2] Inspect ToolRegistry
# ------------------------------------------------------------

print("\n[2] ToolRegistry inspection")
print("-" * 60)

tools = runtime.tools

print("Class:", type(tools).__name__)
print("Module:", type(tools).__module__)

print("\nAttributes:")
for name in sorted(dir(tools)):
    if name.startswith("_"):
        continue
    try:
        value = getattr(tools, name)
        if callable(value):
            print(f"  {name}()")
        else:
            print(f"  {name} = {type(value).__name__}")
    except Exception:
        print(f"  {name} = <unreadable>")

# ------------------------------------------------------------
# [3] Inspect likely registry containers
# ------------------------------------------------------------

print("\n[3] Registered tool data")
print("-" * 60)

candidate_attributes = [
    "tools",
    "registry",
    "items",
    "data",
    "handlers",
    "_tools",
    "_registry"
]

found_registry_data = False

for attr in candidate_attributes:
    if hasattr(tools, attr):
        try:
            value = getattr(tools, attr)

            print(f"\nAttribute: {attr}")
            print("Type:", type(value).__name__)

            if isinstance(value, dict):
                found_registry_data = True

                print("Keys:")
                for key in value.keys():
                    print(" ", key)

            elif isinstance(value, (list, tuple, set)):
                found_registry_data = True

                print("Items:")
                for item in value:
                    print(" ", item)

            else:
                print("Value:", repr(value)[:1000])

        except Exception as e:
            print(f"Could not inspect {attr}: {e}")

if not found_registry_data:
    print("\nNo obvious registry container found.")

# ------------------------------------------------------------
# [4] Inspect public callable methods
# ------------------------------------------------------------

print("\n[4] Public callable interfaces")
print("-" * 60)

public_callables = []

for name in sorted(dir(tools)):
    if name.startswith("_"):
        continue

    try:
        value = getattr(tools, name)

        if callable(value):
            public_callables.append(name)

            try:
                signature = inspect.signature(value)
            except Exception:
                signature = "<signature unavailable>"

            print(f"{name}{signature}")

    except Exception:
        pass

# ------------------------------------------------------------
# [5] Inspect source code of ToolRegistry
# ------------------------------------------------------------

print("\n[5] ToolRegistry source")
print("-" * 60)

try:
    source = inspect.getsource(type(tools))
    print(source)
except Exception as e:
    print("Source inspection failed:", type(e).__name__, str(e))

# ------------------------------------------------------------
# [6] Attempt deterministic tool discovery
# ------------------------------------------------------------

print("\n[6] Tool discovery test")
print("-" * 60)

discovery_methods = [
    "list_tools",
    "list",
    "available",
    "names",
    "get_tools",
    "registry"
]

discovered = None
discovery_method = None

for method_name in discovery_methods:

    if not hasattr(tools, method_name):
        continue

    try:
        method = getattr(tools, method_name)

        if callable(method):

            result = method()

            print("Method:", method_name)
            print("Result type:", type(result).__name__)
            print("Result:")
            print(result)

            discovered = result
            discovery_method = method_name
            break

    except TypeError as e:
        print(f"{method_name} exists but requires arguments:", e)

    except Exception as e:
        print(
            f"{method_name} failed:",
            type(e).__name__,
            str(e)
        )

if discovery_method is None:
    print("No zero-argument tool discovery method found.")

# ------------------------------------------------------------
# [7] Attempt safe deterministic execution
# ------------------------------------------------------------

print("\n[7] Safe tool execution test")
print("-" * 60)

execution_candidates = [
    "execute",
    "run",
    "call",
    "invoke"
]

execution_attempted = False
execution_result = None

# We deliberately use a very simple candidate operation.
#
# The code below does NOT execute arbitrary shell commands.
# It only attempts to identify whether the registry exposes
# a safe execution interface.

for method_name in execution_candidates:

    if not hasattr(tools, method_name):
        continue

    method = getattr(tools, method_name)

    if not callable(method):
        continue

    execution_attempted = True

    print("Candidate execution method:", method_name)

    try:
        print(
            "Signature:",
            inspect.signature(method)
        )
    except Exception:
        pass

    # Do not guess an unsafe argument structure.
    # We inspect the interface instead.
    print(
        "Execution call intentionally not guessed; "
        "interface must be identified from the actual V3 source."
    )

    break

if not execution_attempted:
    print("No obvious execution method found.")

# ------------------------------------------------------------
# [8] Runtime health
# ------------------------------------------------------------

print("\n[8] Runtime health")
print("-" * 60)

try:
    health = runtime.models
    print("Models:", list(health.backends.keys()))
except Exception as e:
    print("Model inspection error:", e)

try:
    print("Agents:", list(runtime.agents.agents.keys()))
except Exception as e:
    print("Agent inspection error:", e)

try:
    print("Memory items:", len(runtime.memory.data.get("items", [])))
except Exception as e:
    print("Memory inspection error:", e)

# ------------------------------------------------------------
# [9] Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")

print("V3 Runtime              : PASS")
print("ToolRegistry loaded     : PASS")

if discovery_method:
    print("Tool discovery          : FOUND")
else:
    print("Tool discovery          : NEEDS INTERFACE REVIEW")

if execution_attempted:
    print("Tool execution interface: FOUND")
else:
    print("Tool execution interface: NOT FOUND")

print("V3 source modification  : NONE")
print("Model training          : NOT PERFORMED")
print("Model reload            : NOT PERFORMED")

print("\n=== END KAGGLE-022 ===")
````

````python
# KAGGLE-023
# Platform: Kaggle
#
# Purpose:
#   Test whether the EXISTING V3 Planner can generate tool-using actions.
#
#   We already know:
#       - Qwen works
#       - Orchestrator works
#       - Autonomous path works
#       - ToolRegistry exists
#
#   Now we test the missing connection:
#
#       Qwen Planner
#           ↓
#       structured plan
#           ↓
#       tool action
#
# Expected result:
#   Determine whether the planner can produce actions such as:
#
#       python
#       shell
#       filesystem_read
#
#   or whether it currently produces only model_response.
#
# Important:
#   - No V3 source modification
#   - No training
#   - No arbitrary tool execution
#   - No shell command generated by the model will be executed
#   - This is a planning/interface test only

from pathlib import Path
import os
import sys
import inspect
import json
import re
import traceback

PROJECT_ROOT = Path(os.environ["PERSONAL_AI_ROOT"]).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=== KAGGLE-023 : PLANNER → TOOL SELECTION TEST ===\n")

# ------------------------------------------------------------
# [1] Start exact V3 runtime
# ------------------------------------------------------------

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.planner import Planner

print("[1] Starting exact V3 runtime")
print("-" * 60)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime:", type(runtime).__name__)
print("Models:", list(runtime.models.backends.keys()))
print("Tools:", list(runtime.tools.tools.keys()))

# ------------------------------------------------------------
# [2] Create exact V3 Planner
# ------------------------------------------------------------

print("\n[2] Creating V3 Planner")
print("-" * 60)

planner = Planner(runtime.models)

print("Planner:", type(planner).__name__)
print("Planner module:", type(planner).__module__)

try:
    print("Planner source:")
    print(inspect.getsource(type(planner)))
except Exception as e:
    print("Planner source unavailable:", type(e).__name__, str(e))

# ------------------------------------------------------------
# [3] Inspect planner public interface
# ------------------------------------------------------------

print("\n[3] Planner public interface")
print("-" * 60)

for name in sorted(dir(planner)):
    if name.startswith("_"):
        continue

    try:
        value = getattr(planner, name)

        if callable(value):
            try:
                print(f"{name}{inspect.signature(value)}")
            except Exception:
                print(f"{name}()")

        else:
            print(f"{name} = {type(value).__name__}")

    except Exception:
        pass

# ------------------------------------------------------------
# [4] Inspect actual planner method
# ------------------------------------------------------------

print("\n[4] Identifying planning method")
print("-" * 60)

planning_method = None

for name in ["plan", "create_plan", "make_plan", "generate_plan"]:

    if hasattr(planner, name):

        candidate = getattr(planner, name)

        if callable(candidate):
            planning_method = candidate
            print("Planning method:", name)

            try:
                print("Signature:", inspect.signature(candidate))
            except Exception:
                pass

            break

if planning_method is None:
    print("No standard planning method found.")

# ------------------------------------------------------------
# [5] Tool-aware planning request
# ------------------------------------------------------------

query = """
Determine the current Python version available to this Personal AI runtime.

You should use an available tool rather than guessing the answer.

Create a structured execution plan showing:
1. Which tool should be used.
2. What action/arguments it should receive.
3. What observation should be collected.
4. How the result should be evaluated.

Do not execute the tool. Only produce the plan.
"""

print("\n[5] Tool-aware planning request")
print("-" * 60)
print(query)

# ------------------------------------------------------------
# [6] Generate plan
# ------------------------------------------------------------

print("\n[6] Generating plan")
print("-" * 60)

plan_result = None

if planning_method is not None:

    try:

        # Try the simplest likely interface first.
        plan_result = planning_method(query)

        print("Plan result type:", type(plan_result).__name__)
        print("\nPlan result:")
        print(plan_result)

    except Exception as e:

        print(
            "Planner invocation failed:",
            type(e).__name__,
            str(e)
        )

        traceback.print_exc()

else:

    print("Planning method unavailable.")

# ------------------------------------------------------------
# [7] Analyze generated plan
# ------------------------------------------------------------

print("\n[7] Plan analysis")
print("-" * 60)

plan_text = str(plan_result) if plan_result is not None else ""

tool_names = [
    "python",
    "shell",
    "filesystem_read"
]

detected_tools = []

for tool_name in tool_names:

    if re.search(
        r"\b" + re.escape(tool_name) + r"\b",
        plan_text,
        flags=re.IGNORECASE
    ):
        detected_tools.append(tool_name)

print("Detected tool references:")

if detected_tools:
    for tool_name in detected_tools:
        print(" ", tool_name)
else:
    print("  NONE")

# Detect model_response action.
model_response_detected = bool(
    re.search(
        r"model_response",
        plan_text,
        flags=re.IGNORECASE
    )
)

print("\nmodel_response action detected:", model_response_detected)

# ------------------------------------------------------------
# [8] Attempt structured JSON extraction
# ------------------------------------------------------------

print("\n[8] Structured-plan inspection")
print("-" * 60)

json_candidates = []

# Direct JSON
try:
    parsed = json.loads(plan_text)
    json_candidates.append(parsed)
except Exception:
    pass

# Markdown JSON blocks
for block in re.findall(
    r"```json\s*(.*?)\s*```",
    plan_text,
    flags=re.IGNORECASE | re.DOTALL
):

    try:
        parsed = json.loads(block)
        json_candidates.append(parsed)
    except Exception:
        pass

print("JSON structures detected:", len(json_candidates))

for i, parsed in enumerate(json_candidates, start=1):

    print(f"\nJSON structure {i}:")

    try:
        print(json.dumps(
            parsed,
            indent=2,
            ensure_ascii=False
        ))
    except Exception:
        print(parsed)

# ------------------------------------------------------------
# [9] Safety confirmation
# ------------------------------------------------------------

print("\n[9] Execution safety")
print("-" * 60)

print("Model-generated tool command executed: NO")
print("Python tool executed: NO")
print("Shell tool executed: NO")
print("Filesystem read executed: NO")

# ------------------------------------------------------------
# [10] Final classification
# ------------------------------------------------------------

print("\n=== RESULT ===")

print("V3 Runtime              : PASS")
print("Planner loaded          :", "PASS" if planning_method else "INTERFACE UNKNOWN")
print("Planning request        :", "COMPLETED" if plan_result is not None else "FAILED")
print(
    "Tool-aware planning     :",
    "DETECTED" if detected_tools else "NOT DETECTED"
)
print(
    "Structured plan         :",
    "FOUND" if json_candidates else "NOT FOUND"
)
print(
    "model_response action   :",
    "DETECTED" if model_response_detected else "NOT DETECTED"
)

print("Tool execution          : NOT PERFORMED")
print("V3 source modification  : NONE")
print("Training                : NOT PERFORMED")

print("\n=== END KAGGLE-023 ===")
````

````python
# KAGGLE-025
# Platform: Kaggle
#
# Purpose:
#   Test a structured Planner -> executable Plan contract.
#
# Goal:
#   Determine whether Qwen can produce:
#
#   {
#       "steps": [
#           {
#               "action": "tool",
#               "tool": "shell",
#               "args": ["python --version"],
#               "kwargs": {}
#           }
#       ],
#       "assumptions": []
#   }
#
# Expected result:
#   - Qwen generates structured JSON.
#   - JSON can be parsed.
#   - The resulting object matches the V3 Plan contract.
#   - No tool is executed.
#   - V3 source remains unchanged.

from pathlib import Path
import os
import sys
import json
import uuid

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task, Plan

print("=== KAGGLE-025 : STRUCTURED PLANNER CONTRACT TEST ===\n")

# ------------------------------------------------------------
# [1] Start runtime
# ------------------------------------------------------------

runtime = Bootstrap(PROJECT_ROOT).start()

print("[1] Runtime")
print("-" * 60)

print("Runtime :", type(runtime).__name__)
print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

# ------------------------------------------------------------
# [2] Create test task
# ------------------------------------------------------------

task = Task(
    id=str(uuid.uuid4()),
    goal="Check the installed Python version using an available system tool."
)

print("\n[2] Test task")
print("-" * 60)

print("Task ID :", task.id)
print("Goal    :", task.goal)

# ------------------------------------------------------------
# [3] Ask Qwen for STRICT structured JSON
# ------------------------------------------------------------

prompt = """
You are the planning component of a modular Personal AI system.

Create an EXECUTABLE plan for this task:

Check the installed Python version using an available system tool.

Available tools:

1. shell
   args:
     [command_string]

2. python
   args:
     [python_code]

3. filesystem_read
   args:
     [relative_path]

Return ONLY valid JSON.
Do not use Markdown.
Do not explain anything.

Required JSON schema:

{
  "steps": [
    {
      "action": "tool",
      "tool": "shell",
      "args": ["python --version"],
      "kwargs": {}
    }
  ],
  "assumptions": []
}

Rules:

- action MUST be "tool" for a tool execution.
- tool MUST be one of:
  "shell", "python", "filesystem_read"
- args MUST be an array.
- kwargs MUST be an object.
- Do not execute the tool.
- Do not invent tools.
- Keep the plan minimal.
"""

print("\n[3] Requesting structured plan from Qwen")
print("-" * 60)

raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": "You output strict JSON plans."
        },
        {
            "role": "user",
            "content": prompt
        }
    ],
    capability="planning",
    max_new_tokens=512,
    do_sample=False
)

print("Raw model output:")
print("-" * 60)
print(raw)

# ------------------------------------------------------------
# [4] Remove accidental Markdown fences if model adds them
# ------------------------------------------------------------

print("\n[4] JSON normalization")
print("-" * 60)

text = raw.strip()

if text.startswith("```"):
    lines = text.splitlines()

    if lines and lines[0].startswith("```"):
        lines = lines[1:]

    if lines and lines[-1].strip() == "```":
        lines = lines[:-1]

    text = "\n".join(lines).strip()

print("Normalized output:")
print(text)

# ------------------------------------------------------------
# [5] Parse JSON
# ------------------------------------------------------------

print("\n[5] JSON parsing")
print("-" * 60)

try:

    data = json.loads(text)

    print("JSON parse: PASS")
    print("Root type :", type(data).__name__)
    print("Keys      :", list(data.keys()))

except Exception as e:

    print("JSON parse: FAIL")
    print("Error:", type(e).__name__, str(e))

    raise

# ------------------------------------------------------------
# [6] Validate Plan schema
# ------------------------------------------------------------

print("\n[6] Executable Plan schema validation")
print("-" * 60)

assert isinstance(data, dict)
assert "steps" in data
assert isinstance(data["steps"], list)

steps = data["steps"]

assert len(steps) >= 1

for index, step in enumerate(steps):

    print(f"\nStep {index}:")
    print(step)

    assert isinstance(step, dict)

    assert step.get("action") == "tool", (
        f"Step {index}: action must be 'tool'"
    )

    assert step.get("tool") in {
        "shell",
        "python",
        "filesystem_read"
    }, (
        f"Step {index}: unknown tool"
    )

    assert isinstance(
        step.get("args", []),
        list
    )

    assert isinstance(
        step.get("kwargs", {}),
        dict
    )

print("\nSchema validation: PASS")

# ------------------------------------------------------------
# [7] Construct actual V3 Plan object
# ------------------------------------------------------------

print("\n[7] Constructing V3 Plan object")
print("-" * 60)

assumptions = data.get("assumptions", [])

plan = Plan(
    task_id=task.id,
    steps=steps,
    assumptions=assumptions
)

print("Plan object:", plan)
print("Plan type  :", type(plan).__name__)
print("Task ID    :", plan.task_id)
print("Step count :", len(plan.steps))

# ------------------------------------------------------------
# [8] Confirm tool execution did NOT happen
# ------------------------------------------------------------

print("\n[8] Safety verification")
print("-" * 60)

print("Tool executed     : NO")
print("Shell executed    : NO")
print("Python executed   : NO")
print("Filesystem read   : NO")
print("V3 source changed : NO")
print("Model trained     : NO")

# ------------------------------------------------------------
# [9] Final result
# ------------------------------------------------------------

print("\n=== RESULT ===")

print("Qwen structured planning : PASS")
print("JSON parsing             : PASS")
print("Plan schema              : PASS")
print("V3 Plan object           : PASS")
print("Tool execution           : NOT PERFORMED")
print("V3 source modification   : NONE")

print("\nNext:")
print("If this passes, KAGGLE-026 will test the")
print("structured Plan through the EXISTING")
print("ExecutionEngine and ToolRegistry.")

print("\n=== END KAGGLE-025 ===")
````

````python
# ============================================================
# KAGGLE-026
# Qwen -> Structured Plan -> ExecutionEngine -> ToolRegistry
# -> Safe Tool Execution -> Observation -> Evaluation -> Ledger
#
# PURPOSE:
# Test the FIRST REAL model-generated tool execution.
#
# SAFETY:
# Only the exact command "python --version" is permitted.
# No arbitrary model-generated shell command is executed.
# No V3 source modification.
# ============================================================

from pathlib import Path
import os
import sys
import json
import uuid
import traceback

# ------------------------------------------------------------
# [1] Locate exact V3 runtime
# ------------------------------------------------------------

PROJECT_ROOT = Path(os.environ["PERSONAL_AI_ROOT"]).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine

print("=== KAGGLE-026 : REAL SAFE TOOL EXECUTION TEST ===\n")

# ------------------------------------------------------------
# [2] Start existing V3 runtime
# ------------------------------------------------------------

runtime = Bootstrap(PROJECT_ROOT).start()

print("[1] Runtime")
print("-" * 60)
print("Runtime :", type(runtime).__name__)
print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

# ------------------------------------------------------------
# [3] Create SAFE test task
# ------------------------------------------------------------

task = Task(
    id=str(uuid.uuid4()),
    goal="Check the installed Python version using the shell tool."
)

print("\n[2] Test task")
print("-" * 60)
print("Task ID :", task.id)
print("Goal    :", task.goal)

# ------------------------------------------------------------
# [4] Ask Qwen for STRICT executable JSON
# ------------------------------------------------------------

prompt = """
You are the planning component of a modular Personal AI.

Create the smallest executable plan for this task:

Check the installed Python version using the shell tool.

AVAILABLE TOOLS:

1. shell
   args: [command_string]

2. python
   args: [python_code]

3. filesystem_read
   args: [relative_path]

RETURN ONLY VALID JSON.
NO MARKDOWN.
NO EXPLANATION.

Required schema:

{
  "steps": [
    {
      "action": "tool",
      "tool": "shell",
      "args": ["python --version"],
      "kwargs": {}
    }
  ],
  "assumptions": []
}

RULES:
- action MUST be "tool".
- tool MUST be one of shell, python, filesystem_read.
- args MUST be an array.
- kwargs MUST be an object.
- For this exact task use shell.
- The command MUST be exactly "python --version".
- Do not invent additional steps.
"""

print("\n[3] Requesting executable plan from Qwen")
print("-" * 60)

raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": "Return only strict JSON executable plans."
        },
        {
            "role": "user",
            "content": prompt
        }
    ],
    capability="planning",
    max_new_tokens=512,
    do_sample=False
)

print("Raw model output:")
print("-" * 60)
print(raw)

# ------------------------------------------------------------
# [5] Normalize JSON
# ------------------------------------------------------------

print("\n[4] JSON normalization")
print("-" * 60)

text = raw.strip()

if text.startswith("```"):
    lines = text.splitlines()

    if lines and lines[0].strip().startswith("```"):
        lines = lines[1:]

    if lines and lines[-1].strip() == "```":
        lines = lines[:-1]

    text = "\n".join(lines).strip()

try:
    data = json.loads(text)

    print("JSON parse: PASS")
    print("Root type :", type(data).__name__)
    print("Keys      :", list(data.keys()))

except Exception as e:
    print("JSON parse: FAIL")
    print("Error:", type(e).__name__, str(e))
    raise

# ------------------------------------------------------------
# [6] Validate executable Plan schema
# ------------------------------------------------------------

print("\n[5] Executable Plan validation")
print("-" * 60)

assert isinstance(data, dict)
assert isinstance(data.get("steps"), list)
assert len(data["steps"]) >= 1

steps = data["steps"]

ALLOWED_TOOLS = {
    "shell",
    "python",
    "filesystem_read",
}

for index, step in enumerate(steps):

    print(f"\nStep {index}:")
    print(json.dumps(step, indent=2, ensure_ascii=False))

    assert isinstance(step, dict)

    assert step.get("action") == "tool", (
        f"Step {index}: action must be 'tool'"
    )

    assert step.get("tool") in ALLOWED_TOOLS, (
        f"Step {index}: unknown tool"
    )

    assert isinstance(step.get("args", []), list)

    assert isinstance(step.get("kwargs", {}), dict)

print("\nSchema validation: PASS")

# ------------------------------------------------------------
# [7] HARD SAFETY GATE
#
# Only permit the exact test operation.
# ------------------------------------------------------------

print("\n[6] Safety gate")
print("-" * 60)

for index, step in enumerate(steps):

    tool = step.get("tool")
    args = step.get("args", [])

    if tool == "shell":

        assert len(args) == 1, (
            f"Step {index}: shell must have exactly one argument"
        )

        assert args[0] == "python --version", (
            "SAFETY BLOCK: model attempted a command other than "
            "'python --version'"
        )

    elif tool == "python":
        raise AssertionError(
            "SAFETY BLOCK: Python execution is not allowed in KAGGLE-026"
        )

    elif tool == "filesystem_read":
        raise AssertionError(
            "SAFETY BLOCK: filesystem_read is not allowed in KAGGLE-026"
        )

print("Safety gate: PASS")
print("Allowed operation: python --version")

# ------------------------------------------------------------
# [8] Construct actual V3 Plan
# ------------------------------------------------------------

print("\n[7] Constructing V3 Plan")
print("-" * 60)

plan = Plan(
    task_id=task.id,
    steps=steps,
    assumptions=data.get("assumptions", [])
)

print("Plan type  :", type(plan).__name__)
print("Task ID    :", plan.task_id)
print("Step count :", len(plan.steps))
print("Plan       :", plan)

# ------------------------------------------------------------
# [9] Existing EvaluationGate + LearningLedger
# ------------------------------------------------------------

print("\n[8] Existing V3 execution components")
print("-" * 60)

evaluator = runtime.evaluator
ledger = runtime.learning

print("Evaluator :", type(evaluator).__name__)
print("Ledger    :", type(ledger).__name__)

# ------------------------------------------------------------
# [10] Existing ExecutionEngine
# ------------------------------------------------------------

print("\n[9] Creating existing ExecutionEngine")
print("-" * 60)

engine = ExecutionEngine(
    tools=runtime.tools,
    evaluator=evaluator,
    ledger=ledger,
    max_steps=1,
)

print("Engine    :", type(engine).__name__)
print("Max steps :", engine.max_steps)

# ------------------------------------------------------------
# [11] REAL TOOL EXECUTION
# ------------------------------------------------------------

print("\n[10] REAL TOOL EXECUTION")
print("-" * 60)

print("Executing:")
print('ToolRegistry.call("shell", "python --version")')
print()

try:

    observations = engine.execute_plan(plan)

    print("ExecutionEngine execution: COMPLETED")

except Exception as e:

    print("ExecutionEngine execution: FAILED")
    print("Error:", type(e).__name__, str(e))
    traceback.print_exc()
    raise

# ------------------------------------------------------------
# [12] Inspect Observation
# ------------------------------------------------------------

print("\n[11] Observation")
print("-" * 60)

assert isinstance(observations, list)
assert len(observations) == 1

observation = observations[0]

print("Observation type :", type(observation).__name__)
print("OK               :", observation.ok)
print("Output           :", observation.output)
print("Error            :", observation.error)

assert observation.ok is True
assert observation.output is not None

# ------------------------------------------------------------
# [13] Evaluation
# ------------------------------------------------------------

print("\n[12] EvaluationGate")
print("-" * 60)

evaluation = evaluator.evaluate(observations)

print("Evaluation type :", type(evaluation).__name__)
print("Done            :", evaluation.done)
print("Passed          :", evaluation.passed)
print("Issues          :", evaluation.issues)

# ------------------------------------------------------------
# [14] LearningLedger
# ------------------------------------------------------------

print("\n[13] LearningLedger")
print("-" * 60)

ledger_event = {
    "task_id": task.id,
    "test": "KAGGLE-026",
    "tool": "shell",
    "command": "python --version",
    "observation_ok": observation.ok,
    "output": observation.output,
    "evaluation_done": evaluation.done,
    "evaluation_passed": evaluation.passed,
}

try:

    ledger.record(ledger_event)

    print("Ledger record: PASS")

except Exception as e:

    print("Ledger record: FAIL")
    print("Error:", type(e).__name__, str(e))
    raise

# ------------------------------------------------------------
# [15] Final result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("=== KAGGLE-026 RESULT ===")
print("=" * 60)

print("Qwen structured planning : PASS")
print("JSON parsing             : PASS")
print("Plan schema              : PASS")
print("Safety gate              : PASS")
print("V3 Plan object           : PASS")
print("ExecutionEngine          : PASS")
print("ToolRegistry execution   : PASS")
print("Observation              : PASS")
print("EvaluationGate           : PASS")
print("LearningLedger           : PASS")

print("\nActual tool output:")
print(observation.output)

print("\nChatGPT dependency test milestone:")
print("FIRST REAL MODEL -> TOOL EXECUTION : ACHIEVED")

print("\n=== END KAGGLE-026 ===")
````

````python
# ============================================================
# KAGGLE-026
# Qwen -> Structured Plan -> Real Tool Execution
# -> Observation -> Evaluation -> Learning Ledger
#
# SAFETY:
# Only the exact command "python --version" is permitted.
# No arbitrary shell/Python execution.
# No V3 source modification.
# ============================================================

from pathlib import Path
import os
import sys
import json
import uuid

print("============================================================")
print("KAGGLE-026 : PLAN -> TOOL -> OBSERVATION -> EVALUATION")
print("============================================================")

# ------------------------------------------------------------
# 1. Runtime binding
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

print("\nPROJECT_ROOT:")
print(PROJECT_ROOT)

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# ------------------------------------------------------------
# 2. Bootstrap V3
# ------------------------------------------------------------

print("\nStarting V3 Bootstrap...")

from personal_ai.core.bootstrap import Bootstrap

runtime = Bootstrap(PROJECT_ROOT).start()

print("Bootstrap: PASS")

# ------------------------------------------------------------
# 3. Verify runtime components
# ------------------------------------------------------------

print("\n=== RUNTIME COMPONENTS ===")

print("Tools      :", type(runtime.tools).__name__)
print("Evaluator  :", type(runtime.evaluator).__name__)
print("Learning   :", type(runtime.learning).__name__)

# ------------------------------------------------------------
# 4. Ask model for STRICT structured plan
# ------------------------------------------------------------

TASK = "Check the installed Python version."

print("\n=== TASK ===")
print(TASK)

planning_prompt = f"""
Create an executable plan for this task.

Task:
{TASK}

STRICT RULES:

1. Return ONLY valid JSON.
2. Root must be an object.
3. Root must contain "steps".
4. "steps" must be a JSON array.
5. Each tool step must have:
   - action
   - tool
   - args
   - kwargs
6. Allowed tool:
   shell
7. Allowed command:
   python --version
8. Do not use any other command.
9. Do not explain anything outside JSON.

Required structure:

{{
  "steps": [
    {{
      "action": "tool",
      "tool": "shell",
      "args": ["python --version"],
      "kwargs": {{}}
    }}
  ],
  "assumptions": []
}}
"""

print("\nGenerating structured plan...")

raw = runtime.models.generate(
    [
        {
            "role": "user",
            "content": planning_prompt
        }
    ],
    capability="planning",
    max_new_tokens=256
)

print("\n=== RAW MODEL OUTPUT ===")
print(raw)

# ------------------------------------------------------------
# 5. Parse JSON
# ------------------------------------------------------------

try:
    parsed = json.loads(raw)
    print("\nJSON parse: PASS")
except Exception as e:
    print("\nJSON parse: FAIL")
    print(type(e).__name__, ":", e)
    raise

# ------------------------------------------------------------
# 6. Validate plan
# ------------------------------------------------------------

if not isinstance(parsed, dict):
    raise TypeError("Plan root must be a dict.")

if "steps" not in parsed:
    raise ValueError("Plan missing 'steps'.")

steps = parsed["steps"]

if not isinstance(steps, list):
    raise TypeError("'steps' must be a list.")

if len(steps) == 0:
    raise ValueError("Plan contains no steps.")

for step in steps:

    if not isinstance(step, dict):
        raise TypeError("Each step must be a dict.")

    if step.get("action") != "tool":
        raise ValueError(
            f"Unsupported action: {step.get('action')}"
        )

    if step.get("tool") != "shell":
        raise ValueError(
            f"Unsupported tool: {step.get('tool')}"
        )

    args = step.get("args", [])

    if args != ["python --version"]:
        raise PermissionError(
            f"SAFETY GATE BLOCKED command: {args}"
        )

print("Plan validation: PASS")
print("Safety gate: PASS")

# ------------------------------------------------------------
# 7. Construct V3 Plan object
# ------------------------------------------------------------

from personal_ai.core.planner import Plan

task_id = str(uuid.uuid4())

plan = Plan(
    task_id,
    steps,
    parsed.get("assumptions", [])
)

print("\n=== PLAN OBJECT ===")
print(plan)

# ------------------------------------------------------------
# 8. Existing V3 ExecutionEngine
# ------------------------------------------------------------

from personal_ai.execution.engine import ExecutionEngine

engine = ExecutionEngine(
    tools=runtime.tools,
    evaluator=runtime.evaluator,
    ledger=runtime.learning,
    max_steps=1
)

print("\nExecutionEngine: READY")

# ------------------------------------------------------------
# 9. REAL TOOL EXECUTION
# ------------------------------------------------------------

print("\n=== REAL TOOL EXECUTION ===")

observations = engine.execute_plan(plan)

for i, obs in enumerate(observations):

    print(f"\nObservation {i + 1}")

    print("OK    :", obs.ok)
    print("VALUE :", getattr(obs, "value", None))
    print("ERROR :", getattr(obs, "error", None))

# ------------------------------------------------------------
# 10. Evaluation
# ------------------------------------------------------------

print("\n=== EVALUATION ===")

evaluation = runtime.evaluator.evaluate(
    observations
)

print("Done   :", evaluation.done)
print("Passed :", evaluation.passed)
print("Issues :", evaluation.issues)

# ------------------------------------------------------------
# 11. Learning Ledger
# ------------------------------------------------------------

print("\n=== LEARNING LEDGER ===")

ledger_event = {
    "event": "KAGGLE-026",
    "task_id": task_id,
    "task": TASK,
    "plan": parsed,
    "observations": [
        {
            "ok": o.ok,
            "value": getattr(o, "value", None),
            "error": getattr(o, "error", None),
        }
        for o in observations
    ],
    "evaluation": {
        "done": evaluation.done,
        "passed": evaluation.passed,
        "issues": evaluation.issues,
    },
}

runtime.learning.record(ledger_event)

print("Ledger record: PASS")

# ------------------------------------------------------------
# FINAL RESULT
# ------------------------------------------------------------

print("\n============================================================")
print("KAGGLE-026 RESULT")
print("============================================================")

print("V3 runtime             : PASS")
print("Qwen structured plan   : PASS")
print("JSON parsing            : PASS")
print("Plan validation         : PASS")
print("Safety gate             : PASS")
print("Real tool execution     : PASS")
print("Observation             : PASS")
print("Evaluation              : PASS")
print("Learning ledger         : PASS")

print("\nTask execution completed.")
print("============================================================")
````

````python
# ============================================================
# KAGGLE-026 : QWEN -> PLAN -> TOOL -> OBSERVATION -> EVAL
# ============================================================

import os
import sys
import json
import time
from pathlib import Path

print("=" * 60)
print("KAGGLE-026 : AUTONOMOUS TOOL EXECUTION")
print("=" * 60)

# ------------------------------------------------------------
# 1. Locate V3
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    os.environ.get(
        "PERSONAL_AI_ROOT",
        "/kaggle/working/Personal_AI_AGI_Core_v3_0"
    )
)

print("\n[1] PROJECT ROOT")
print(PROJECT_ROOT)
print("Exists:", PROJECT_ROOT.exists())

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(
        f"V3 project not found: {PROJECT_ROOT}"
    )

# ------------------------------------------------------------
# 2. Python path binding
# ------------------------------------------------------------

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("\n[2] PYTHON PATH")
print("Project root in sys.path:", str(PROJECT_ROOT) in sys.path)

# ------------------------------------------------------------
# 3. Import actual V3 Bootstrap
# IMPORTANT: bootstrap.py is directly under personal_ai/
# ------------------------------------------------------------

print("\n[3] IMPORT V3 BOOTSTRAP")
print("-" * 60)

from personal_ai.bootstrap import Bootstrap

print("Bootstrap import: PASS")

# ------------------------------------------------------------
# 4. Start V3 Runtime
# ------------------------------------------------------------

print("\n[4] STARTING V3 RUNTIME")
print("-" * 60)

t0 = time.time()

bootstrap = Bootstrap(PROJECT_ROOT)
runtime = bootstrap.start()

bootstrap_time = time.time() - t0

print("Runtime created:", runtime is not None)
print("Bootstrap time:", round(bootstrap_time, 2), "seconds")

# ------------------------------------------------------------
# 5. Runtime components
# ------------------------------------------------------------

print("\n[5] RUNTIME COMPONENTS")
print("-" * 60)

components = [
    "config",
    "env",
    "registry",
    "memory",
    "models",
    "tools",
    "agents",
    "router",
    "evaluator",
    "learning",
]

for name in components:
    obj = getattr(runtime, name, None)
    print(
        f"{name:12}: "
        f"{type(obj).__name__ if obj is not None else 'MISSING'}"
    )

# ------------------------------------------------------------
# 6. Verify model backend
# ------------------------------------------------------------

print("\n[6] MODEL BACKENDS")
print("-" * 60)

for name, backend in runtime.models.backends.items():
    try:
        available = backend.available()
    except Exception as e:
        available = f"ERROR: {e}"

    print(
        f"{name:15} "
        f"class={type(backend).__name__} "
        f"available={available}"
    )

# ------------------------------------------------------------
# 7. Verify tools
# ------------------------------------------------------------

print("\n[7] TOOLS")
print("-" * 60)

try:
    print("Tool registry:", type(runtime.tools).__name__)

    if hasattr(runtime.tools, "tools"):
        print("Registered tools:", list(runtime.tools.tools.keys()))
    elif hasattr(runtime.tools, "registry"):
        print("Tool registry entries:", list(runtime.tools.registry.keys()))
    else:
        print("ToolRegistry object available.")

except Exception as e:
    print("Tool inspection error:", repr(e))

# ------------------------------------------------------------
# 8. Ask Qwen for STRICT executable JSON
# ------------------------------------------------------------

print("\n[8] QWEN STRUCTURED PLANNING")
print("-" * 60)

task_goal = "Check the installed Python version."

planning_prompt = f"""
Create an executable plan for this task:

{task_goal}

Return ONLY valid JSON.

Required format:

{{
  "steps": [
    {{
      "action": "tool",
      "tool": "shell",
      "args": ["python --version"],
      "kwargs": {{}}
    }}
  ],
  "assumptions": []
}}

Do not add markdown.
Do not add explanations.
"""

print("Task:", task_goal)
print("\nGenerating plan...")

t1 = time.time()

raw = runtime.models.generate(
    [
        {
            "role": "user",
            "content": planning_prompt
        }
    ],
    capability="planning",
    max_new_tokens=512
)

planning_time = time.time() - t1

print("\nRaw model output:")
print(raw)

print("\nPlanning time:", round(planning_time, 2), "seconds")

# ------------------------------------------------------------
# 9. Parse JSON
# ------------------------------------------------------------

print("\n[9] JSON PARSING")
print("-" * 60)

text = raw.strip()

# Remove accidental markdown fences if model adds them
if text.startswith("```"):
    lines = text.splitlines()

    if lines and lines[0].startswith("```"):
        lines = lines[1:]

    if lines and lines[-1].strip() == "```":
        lines = lines[:-1]

    text = "\n".join(lines).strip()

try:
    plan_json = json.loads(text)
    print("JSON parse: PASS")
except Exception as e:
    print("JSON parse: FAIL")
    print("Error:", repr(e))
    raise

print("Root type:", type(plan_json).__name__)
print("Keys:", list(plan_json.keys()))

# ------------------------------------------------------------
# 10. Validate executable schema
# ------------------------------------------------------------

print("\n[10] PLAN VALIDATION")
print("-" * 60)

assert isinstance(plan_json, dict)
assert "steps" in plan_json
assert isinstance(plan_json["steps"], list)
assert len(plan_json["steps"]) >= 1

step = plan_json["steps"][0]

assert step["action"] == "tool"
assert step["tool"] == "shell"
assert step["args"] == ["python --version"]

print("Executable Plan schema validation: PASS")

# ------------------------------------------------------------
# 11. Construct actual V3 Plan
# ------------------------------------------------------------

print("\n[11] V3 PLAN OBJECT")
print("-" * 60)

from personal_ai.core.planner import Plan

task_id = "KAGGLE-026"

plan = Plan(
    task_id,
    plan_json["steps"],
    plan_json.get("assumptions", [])
)

print("Plan object:", plan)

# ------------------------------------------------------------
# 12. Safety gate
# ------------------------------------------------------------

print("\n[12] SAFETY GATE")
print("-" * 60)

ALLOWED_TOOL = "shell"
ALLOWED_COMMAND = "python --version"

safe = (
    step["tool"] == ALLOWED_TOOL
    and step["args"] == [ALLOWED_COMMAND]
)

print("Requested tool:", step["tool"])
print("Requested command:", step["args"][0])
print("Safety check:", "PASS" if safe else "FAIL")

if not safe:
    raise RuntimeError(
        "KAGGLE-026 safety gate rejected the generated action."
    )

# ------------------------------------------------------------
# 13. Existing ExecutionEngine
# ------------------------------------------------------------

print("\n[13] EXECUTION ENGINE")
print("-" * 60)

from personal_ai.execution.engine import ExecutionEngine

engine = ExecutionEngine(
    tools=runtime.tools,
    evaluator=runtime.evaluator,
    ledger=runtime.learning,
    max_steps=1
)

print("Engine:", type(engine).__name__)

# ------------------------------------------------------------
# 14. REAL TOOL EXECUTION
# ------------------------------------------------------------

print("\n[14] REAL TOOL EXECUTION")
print("-" * 60)

t2 = time.time()

observations = engine.execute_plan(plan)

execution_time = time.time() - t2

print("Execution time:", round(execution_time, 2), "seconds")

for i, obs in enumerate(observations):
    print(f"\nObservation {i + 1}")
    print("OK   :", obs.ok)
    print("VALUE:", getattr(obs, "value", None))
    print("ERROR:", getattr(obs, "error", None))

# ------------------------------------------------------------
# 15. EvaluationGate
# ------------------------------------------------------------

print("\n[15] EVALUATION GATE")
print("-" * 60)

evaluation = runtime.evaluator.evaluate(observations)

print("Evaluation:", evaluation)
print("Done:", evaluation.done)
print("Passed:", evaluation.passed)
print("Issues:", evaluation.issues)

# ------------------------------------------------------------
# 16. LearningLedger
# ------------------------------------------------------------

print("\n[16] LEARNING LEDGER")
print("-" * 60)

ledger_event = {
    "task_id": task_id,
    "stage": "KAGGLE-026",
    "task": task_goal,
    "plan": plan_json,
    "evaluation": evaluation.__dict__,
    "execution_time": execution_time,
}

runtime.learning.record(ledger_event)

print("Learning event recorded: PASS")

# ------------------------------------------------------------
# 17. Final result
# ------------------------------------------------------------

print("\n")
print("=" * 60)
print("KAGGLE-026 RESULT")
print("=" * 60)

print("V3 Bootstrap              : PASS")
print("Runtime construction      : PASS")
print("Qwen planning             : PASS")
print("JSON parsing              : PASS")
print("Plan validation           : PASS")
print("Safety gate               : PASS")
print("ExecutionEngine           : PASS")
print("Real tool execution       : PASS")
print("Observation               : PASS")
print("EvaluationGate            :", "PASS" if evaluation.passed else "FAIL")
print("LearningLedger            : PASS")

print("=" * 60)
print("FLOW")
print("=" * 60)

print("""
Qwen
  ↓
Structured JSON Plan
  ↓
Plan Object
  ↓
Safety Gate
  ↓
ExecutionEngine
  ↓
ToolRegistry
  ↓
shell: python --version
  ↓
Observation
  ↓
EvaluationGate
  ↓
LearningLedger
""")

print("=" * 60)
print("END KAGGLE-026")
print("=" * 60)
````

````python
# ============================================================
# KAGGLE-026-DEBUG : QWEN RAW OUTPUT + SAFE JSON EXTRACTION
# ============================================================

import json
import re
import time

print("=" * 60)
print("KAGGLE-026-DEBUG : MODEL OUTPUT DIAGNOSTICS")
print("=" * 60)

task_goal = "Check the Python version using the shell tool."

prompt = f"""
You are the planning module of a Personal AI system.

Create an executable plan for this task:

{task_goal}

Return ONLY valid JSON.
Do not use markdown.
Do not explain anything.

Required format:
{{
  "steps": [
    {{
      "action": "tool",
      "tool": "shell",
      "args": ["python --version"],
      "kwargs": {{}}
    }}
  ],
  "assumptions": []
}}
"""

print("\n[1] Sending planning request to Qwen...")
print("-" * 60)

t0 = time.time()

try:
    raw = runtime.models.generate(
        [{"role": "user", "content": prompt}],
        capability="planning",
        max_new_tokens=512
    )

    print("Generation time:", round(time.time() - t0, 2), "s")
    print("Raw type:", type(raw))
    print("Raw repr:")
    print(repr(raw))

except Exception as e:
    print("\nMODEL GENERATION FAILED")
    print(type(e).__name__, ":", repr(e))
    raise


# ------------------------------------------------------------
# NORMALIZE OUTPUT
# ------------------------------------------------------------

if raw is None:
    text = ""

elif isinstance(raw, str):
    text = raw.strip()

else:
    text = str(raw).strip()

print("\n[2] NORMALIZED OUTPUT")
print("-" * 60)
print("Length:", len(text))
print("Text:")
print(text)


# ------------------------------------------------------------
# EMPTY OUTPUT CHECK
# ------------------------------------------------------------

if not text:
    print("\n[RESULT] FAIL")
    print("Qwen returned EMPTY output.")
    print("JSON parser is not the root problem.")
    print("The problem is upstream in model generation.")
    raise RuntimeError("Empty model output")


# ------------------------------------------------------------
# REMOVE MARKDOWN CODE FENCES
# ------------------------------------------------------------

clean = text.strip()

if clean.startswith("```"):
    clean = re.sub(
        r"^```(?:json)?\s*",
        "",
        clean,
        flags=re.IGNORECASE
    )
    clean = re.sub(
        r"\s*```$",
        "",
        clean
    )

clean = clean.strip()

print("\n[3] CLEANED OUTPUT")
print("-" * 60)
print(clean)


# ------------------------------------------------------------
# DIRECT JSON PARSE
# ------------------------------------------------------------

print("\n[4] JSON PARSE")

try:
    plan_json = json.loads(clean)

    print("JSON parse: PASS")
    print("Root type:", type(plan_json).__name__)
    print("Parsed object:")
    print(json.dumps(plan_json, indent=2))

except json.JSONDecodeError as e:

    print("JSON parse: FAIL")
    print("Error:", e)

    # --------------------------------------------------------
    # TRY TO EXTRACT FIRST JSON OBJECT
    # --------------------------------------------------------

    print("\n[5] JSON OBJECT EXTRACTION")

    start = clean.find("{")
    end = clean.rfind("}")

    if start != -1 and end != -1 and end > start:

        candidate = clean[start:end + 1]

        print("Candidate:")
        print(candidate)

        try:
            plan_json = json.loads(candidate)

            print("\nJSON extraction: PASS")
            print("Root type:", type(plan_json).__name__)

        except json.JSONDecodeError as e2:

            print("\nJSON extraction: FAIL")
            print("Error:", e2)

            print("\n==============================")
            print("DIAGNOSIS")
            print("==============================")
            print(
                "Qwen generated text, but it was not valid JSON."
            )

            raise

    else:

        print("\nNo JSON object found in model output.")

        print("\n==============================")
        print("DIAGNOSIS")
        print("==============================")
        print(
            "The model output contains no detectable JSON object."
        )

        raise RuntimeError(
            "Model did not produce a detectable JSON object"
        )


# ------------------------------------------------------------
# BASIC PLAN VALIDATION
# ------------------------------------------------------------

print("\n[6] PLAN STRUCTURE VALIDATION")
print("-" * 60)

if not isinstance(plan_json, dict):
    raise TypeError("Plan root must be a dictionary")

if "steps" not in plan_json:
    raise ValueError("Plan must contain 'steps'")

if not isinstance(plan_json["steps"], list):
    raise TypeError("'steps' must be a list")

print("Plan schema: PASS")
print("Steps:", len(plan_json["steps"]))

for i, step in enumerate(plan_json["steps"]):

    print(f"\nStep {i}:")
    print(step)

    if not isinstance(step, dict):
        raise TypeError(f"Step {i} must be a dictionary")

    if step.get("action") == "tool":

        if "tool" not in step:
            raise ValueError(f"Step {i}: missing tool")

        if "args" not in step:
            raise ValueError(f"Step {i}: missing args")

print("\n" + "=" * 60)
print("KAGGLE-026-DEBUG RESULT")
print("=" * 60)
print("Model generation : PASS")
print("Non-empty output : PASS")
print("JSON parse       : PASS")
print("Plan structure   : PASS")
print("=" * 60)
````

````python
# ============================================================
# KAGGLE-026C : FIND QWEN MODEL FILES
# ============================================================

from pathlib import Path
import os

print("=" * 70)
print("KAGGLE-026C : QWEN MODEL DISCOVERY")
print("=" * 70)

# ------------------------------------------------------------
# 1. CHECK V3 LOCAL MODEL DIRECTORIES
# ------------------------------------------------------------

ROOT = Path("/kaggle/working/Personal_AI_AGI_Core_v3_0")

candidate_dirs = [
    ROOT / "models" / "qwen2.5-1.5b-instruct",
    ROOT / "models" / "qwen",
    Path("/kaggle/input/personal-ai-qwen"),
    Path("/kaggle/input/datasets/vikaschandoliya/personal-ai-qwen"),
    Path("/content/personal_ai/models/qwen2.5-1.5b-instruct"),
]

print("\n[1] EXISTING CANDIDATE DIRECTORIES")
print("-" * 70)

for p in candidate_dirs:
    print(f"\n{p}")
    print("Exists:", p.exists())

    if p.exists():
        print("Is directory:", p.is_dir())

        try:
            items = list(p.iterdir())[:20]
            for item in items:
                print("  ", item.name)
        except Exception as e:
            print("  Cannot list:", repr(e))


# ------------------------------------------------------------
# 2. SEARCH KAGGLE INPUT
# ------------------------------------------------------------

print("\n\n[2] SEARCHING /kaggle/input FOR QWEN")
print("-" * 70)

kaggle_input = Path("/kaggle/input")

matches = []

if kaggle_input.exists():

    for path in kaggle_input.rglob("*"):

        name = path.name.lower()

        if (
            "qwen" in name
            or "qwen2.5" in name
            or "qwen2" in name
        ):
            matches.append(path)

        if len(matches) >= 200:
            break

print("Potential Qwen-related paths found:", len(matches))

for p in matches:
    print(p)


# ------------------------------------------------------------
# 3. SEARCH FOR config.json + MODEL FILES
# ------------------------------------------------------------

print("\n\n[3] SEARCHING FOR MODEL CONFIG FILES")
print("-" * 70)

config_matches = []

if kaggle_input.exists():

    for path in kaggle_input.rglob("config.json"):

        config_matches.append(path)

        if len(config_matches) >= 300:
            break

print("config.json files found:", len(config_matches))

for p in config_matches:
    print(p)


# ------------------------------------------------------------
# 4. IDENTIFY HUGGINGFACE-STYLE MODEL DIRECTORIES
# ------------------------------------------------------------

print("\n\n[4] POSSIBLE LOCAL MODEL DIRECTORIES")
print("-" * 70)

model_indicators = {
    "config.json",
    "tokenizer.json",
    "tokenizer_config.json",
    "generation_config.json",
}

possible_models = []

for config_path in config_matches:

    directory = config_path.parent

    try:
        files = {x.name for x in directory.iterdir()}
    except Exception:
        continue

    score = len(files.intersection(model_indicators))

    # Also detect common model weight files
    weight_files = [
        x for x in files
        if (
            x.endswith(".safetensors")
            or x.endswith(".bin")
            or x.endswith(".pth")
        )
    ]

    if score >= 1 or weight_files:

        possible_models.append(
            (directory, score, weight_files[:10])
        )

print("Possible model directories:", len(possible_models))

for directory, score, weights in possible_models[:100]:

    print("\nMODEL DIR:", directory)
    print("Indicator score:", score)
    print("Weight files:", weights)


# ------------------------------------------------------------
# 5. CHECK AVAILABLE DISK
# ------------------------------------------------------------

print("\n\n[5] DISK SPACE")
print("-" * 70)

try:
    import shutil

    total, used, free = shutil.disk_usage("/kaggle")

    print("Total GB:", round(total / 1024**3, 2))
    print("Used GB :", round(used / 1024**3, 2))
    print("Free GB :", round(free / 1024**3, 2))

except Exception as e:
    print("Disk check failed:", repr(e))


print("\n" + "=" * 70)
print("KAGGLE-026C COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-026D : INSPECT QWEN ASSETS
# ============================================================

from pathlib import Path
import json

ASSET_DIR = Path(
    "/kaggle/input/datasets/vikaschandoliya/"
    "personal-ai-agi-core-v3-0/"
    "Personal_AI_AGI_Core_v3_0/models/qwen_assets"
)

print("=" * 70)
print("KAGGLE-026D : QWEN ASSET INSPECTION")
print("=" * 70)

print("\n[1] DIRECTORY")
print("-" * 70)
print("Path:", ASSET_DIR)
print("Exists:", ASSET_DIR.exists())

# ------------------------------------------------------------
# COMPLETE FILE LIST
# ------------------------------------------------------------

print("\n[2] FILES")
print("-" * 70)

files = sorted(
    [p for p in ASSET_DIR.rglob("*") if p.is_file()]
)

print("Total files:", len(files))

for p in files:
    size_mb = p.stat().st_size / (1024 * 1024)
    print(
        f"{p.relative_to(ASSET_DIR)}"
        f"    [{size_mb:.2f} MB]"
    )


# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

config_path = ASSET_DIR / "config.json"

print("\n[3] CONFIG.JSON")
print("-" * 70)

if config_path.exists():

    try:
        config = json.loads(
            config_path.read_text(encoding="utf-8")
        )

        print(json.dumps(config, indent=2))

        print("\nImportant fields:")

        for key in [
            "model_type",
            "architectures",
            "torch_dtype",
            "hidden_size",
            "num_hidden_layers",
            "num_attention_heads",
            "vocab_size",
        ]:
            print(f"{key}: {config.get(key)}")

    except Exception as e:
        print("Could not parse config.json:", repr(e))

else:
    print("config.json MISSING")


# ------------------------------------------------------------
# WEIGHT DETECTION
# ------------------------------------------------------------

print("\n[4] MODEL WEIGHTS")
print("-" * 70)

weight_extensions = {
    ".safetensors",
    ".bin",
    ".pth",
    ".pt",
}

weights = [
    p for p in files
    if p.suffix.lower() in weight_extensions
]

print("Weight files:", len(weights))

if weights:
    total_weight_mb = 0

    for p in weights:
        size_mb = p.stat().st_size / (1024 * 1024)
        total_weight_mb += size_mb

        print(
            p.relative_to(ASSET_DIR),
            f"{size_mb:.2f} MB"
        )

    print(
        "\nTotal weight size:",
        round(total_weight_mb / 1024, 2),
        "GB"
    )

else:
    print("NO MODEL WEIGHTS FOUND.")


# ------------------------------------------------------------
# TOKENIZER
# ------------------------------------------------------------

print("\n[5] TOKENIZER ASSETS")
print("-" * 70)

tokenizer_names = [
    "tokenizer.json",
    "tokenizer_config.json",
    "special_tokens_map.json",
    "vocab.json",
    "merges.txt",
]

for name in tokenizer_names:

    p = ASSET_DIR / name

    print(
        f"{name}:",
        "FOUND" if p.exists() else "missing"
    )


# ------------------------------------------------------------
# FINAL DIAGNOSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-026D RESULT")
print("=" * 70)

if weights:
    print("Model weights : FOUND")
    print("Next step     : configure Qwen model path")
else:
    print("Model weights : NOT FOUND")
    print("Qwen assets   : INCOMPLETE")
    print("Next step     : obtain compatible model weights")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-026E : SEARCH ALL KAGGLE INPUTS FOR MODEL WEIGHTS
# ============================================================

from pathlib import Path

ROOT = Path("/kaggle/input")

print("=" * 70)
print("KAGGLE-026E : COMPLETE MODEL WEIGHT DISCOVERY")
print("=" * 70)

# ------------------------------------------------------------
# Search common model-weight formats
# ------------------------------------------------------------

extensions = {
    ".safetensors",
    ".bin",
    ".pt",
    ".pth",
}

matches = []

print("\n[1] SEARCHING FOR WEIGHT FILES")
print("-" * 70)

for path in ROOT.rglob("*"):

    if not path.is_file():
        continue

    if path.suffix.lower() in extensions:

        matches.append(path)

        if len(matches) >= 300:
            break

print("Weight files found:", len(matches))

for p in matches:
    size_gb = p.stat().st_size / (1024 ** 3)

    print(
        f"{p}"
        f"    [{size_gb:.3f} GB]"
    )


# ------------------------------------------------------------
# Group weights by directory
# ------------------------------------------------------------

print("\n[2] WEIGHT DIRECTORIES")
print("-" * 70)

directories = {}

for p in matches:
    directories.setdefault(p.parent, []).append(p)

for directory, files in directories.items():

    total = sum(
        f.stat().st_size
        for f in files
    )

    print("\nDIRECTORY:")
    print(directory)

    print(
        "Files:",
        len(files),
        "| Total:",
        round(total / (1024 ** 3), 3),
        "GB"
    )

    for f in files[:20]:
        print("  ", f.name)


# ------------------------------------------------------------
# Look for HuggingFace model markers
# ------------------------------------------------------------

print("\n[3] HUGGINGFACE MODEL DIRECTORIES")
print("-" * 70)

markers = [
    "config.json",
    "tokenizer.json",
    "tokenizer_config.json",
    "generation_config.json",
]

possible = []

for config in ROOT.rglob("config.json"):

    directory = config.parent

    try:
        names = {
            p.name
            for p in directory.iterdir()
            if p.is_file()
        }
    except Exception:
        continue

    marker_count = sum(
        m in names
        for m in markers
    )

    weight_count = sum(
        p.suffix.lower() in extensions
        for p in directory.iterdir()
        if p.is_file()
    )

    if weight_count > 0:

        possible.append(
            (
                directory,
                marker_count,
                weight_count
            )
        )

print(
    "Directories containing config + weights:",
    len(possible)
)

for directory, marker_count, weight_count in possible:

    print("\n", directory)
    print(
        "Markers:",
        marker_count,
        "| Weight files:",
        weight_count
    )


print("\n" + "=" * 70)
print("KAGGLE-026E COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-027C : V3 REAL QWEN LOADING + INFERENCE TEST
# ============================================================

import os
import sys
import time
import torch

PROJECT_ROOT = "/kaggle/working/Personal_AI_AGI_Core_v3_0"
MODEL_DIR = "/kaggle/working/qwen2.5-1.5b-instruct"

print("=" * 70)
print("KAGGLE-027C : V3 REAL QWEN LOADING + INFERENCE TEST")
print("=" * 70)

# ------------------------------------------------------------
# 1. Environment
# ------------------------------------------------------------
os.environ["PERSONAL_AI_ROOT"] = PROJECT_ROOT
os.environ["PERSONAL_AI_MODEL_DIR"] = MODEL_DIR

if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

print("\n[1] ENVIRONMENT")
print("-" * 70)
print("PERSONAL_AI_ROOT      :", os.environ["PERSONAL_AI_ROOT"])
print("PERSONAL_AI_MODEL_DIR :", os.environ["PERSONAL_AI_MODEL_DIR"])
print("CUDA available        :", torch.cuda.is_available())
print("GPU count             :", torch.cuda.device_count())

for i in range(torch.cuda.device_count()):
    print(f"GPU {i}               :", torch.cuda.get_device_name(i))

# ------------------------------------------------------------
# 2. Verify model files
# ------------------------------------------------------------
print("\n[2] MODEL FILES")
print("-" * 70)

required = [
    "config.json",
    "tokenizer.json",
    "tokenizer_config.json",
    "model.safetensors",
]

for name in required:
    path = os.path.join(MODEL_DIR, name)
    print(f"{name:25}:", "FOUND" if os.path.exists(path) else "MISSING")

# ------------------------------------------------------------
# 3. Import V3 ModelManager
# ------------------------------------------------------------
print("\n[3] V3 MODEL MANAGER")
print("-" * 70)

try:
    from personal_ai.models.manager import ModelManager
    from personal_ai.runtime.environment import Environment
    from personal_ai.runtime.config import Config

    print("ModelManager import    : PASS")
    print("Environment import     : PASS")
    print("Config import          : PASS")

except Exception as e:
    print("IMPORT FAILED")
    print(type(e).__name__, ":", e)
    raise

# ------------------------------------------------------------
# 4. Initialize V3 ModelManager
# ------------------------------------------------------------
print("\n[4] INITIALIZING MODEL MANAGER")
print("-" * 70)

env = Environment.detect()
config = Config.load(PROJECT_ROOT)

t0 = time.time()

manager = ModelManager(
    PROJECT_ROOT,
    config,
    env
)

print("Initialization time     :", round(time.time() - t0, 2), "sec")
print("Registered backends     :", list(manager.backends.keys()))

# ------------------------------------------------------------
# 5. Verify Qwen backend
# ------------------------------------------------------------
print("\n[5] QWEN BACKEND CHECK")
print("-" * 70)

if "qwen_local" not in manager.backends:
    print("RESULT: QWEN BACKEND NOT REGISTERED")
    print("\nRegistered:")
    for k, v in manager.backends.items():
        print(" -", k, type(v).__name__)
    raise RuntimeError("qwen_local backend was not registered")

qwen = manager.backends["qwen_local"]

print("Backend type            :", type(qwen).__name__)
print("Model path              :", qwen.model_path)
print("Path exists             :", qwen.model_path.exists())
print("Backend available       :", qwen.available())

# ------------------------------------------------------------
# 6. Select backend through V3
# ------------------------------------------------------------
print("\n[6] V3 BACKEND SELECTION")
print("-" * 70)

selected = manager.choose("planning")

print("Selected backend        :", selected.name)
print("Selected backend type   :", type(selected).__name__)

if selected.name != "qwen_local":
    raise RuntimeError(
        f"V3 selected {selected.name} instead of qwen_local"
    )

print("Backend selection       : PASS")

# ------------------------------------------------------------
# 7. REAL QWEN INFERENCE
# ------------------------------------------------------------
print("\n[7] REAL QWEN INFERENCE")
print("-" * 70)

messages = [
    {
        "role": "system",
        "content": (
            "You are the planning module of a personal AI system. "
            "Answer clearly and briefly."
        )
    },
    {
        "role": "user",
        "content": (
            "Create a simple two-step plan to calculate the average "
            "of 10, 20, and 30."
        )
    }
]

print("Loading Qwen model...")
print("This may take some time on first load.")

t0 = time.time()

try:
    response = manager.generate(
        messages,
        capability="planning",
        max_new_tokens=128,
        do_sample=False
    )

    elapsed = time.time() - t0

    print("\n------------------------------------------------------------")
    print("QWEN RESPONSE")
    print("------------------------------------------------------------")
    print(response)

    print("\n------------------------------------------------------------")
    print("INFERENCE RESULT")
    print("------------------------------------------------------------")
    print("Inference time         :", round(elapsed, 2), "sec")
    print("Response type          :", type(response).__name__)
    print("Response length        :", len(response))

    if not response or response.startswith("Local model unavailable"):
        raise RuntimeError("Fallback response detected")

    print("REAL QWEN INFERENCE    : PASS")

except Exception as e:
    print("\nREAL QWEN INFERENCE    : FAILED")
    print(type(e).__name__, ":", e)

    print("\nCUDA MEMORY:")
    for i in range(torch.cuda.device_count()):
        print(
            f"GPU {i}: allocated="
            f"{torch.cuda.memory_allocated(i)/1024**3:.2f} GB, "
            f"reserved="
            f"{torch.cuda.memory_reserved(i)/1024**3:.2f} GB"
        )

    raise

print("\n" + "=" * 70)
print("KAGGLE-027C COMPLETE")
print("=" * 70)
print("V3 → Qwen backend      : PASS")
print("Real neural inference  : PASS")
print("Fallback used          : NO")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-027C-DIAG : QWEN LOAD DIAGNOSTIC
# ============================================================

import os
import sys
import time
import torch

PROJECT_ROOT = "/kaggle/working/Personal_AI_AGI_Core_v3_0"
MODEL_DIR = "/kaggle/working/qwen2.5-1.5b-instruct"

os.environ["PERSONAL_AI_ROOT"] = PROJECT_ROOT
os.environ["PERSONAL_AI_MODEL_DIR"] = MODEL_DIR

if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

print("=" * 70)
print("QWEN LOAD DIAGNOSTIC")
print("=" * 70)

print("\n[1] GPU STATUS")
print("-" * 70)

for i in range(torch.cuda.device_count()):
    free, total = torch.cuda.mem_get_info(i)
    print(
        f"GPU {i}: "
        f"{torch.cuda.get_device_name(i)} | "
        f"free={free/1024**3:.2f} GB / "
        f"total={total/1024**3:.2f} GB"
    )

print("\n[2] TRANSFORMERS")
print("-" * 70)

import transformers
print("Transformers:", transformers.__version__)

print("\n[3] DIRECT TOKENIZER LOAD")
print("-" * 70)

from transformers import AutoTokenizer

t0 = time.time()

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_DIR,
    local_files_only=True
)

print("Tokenizer load: PASS")
print("Time:", round(time.time() - t0, 2), "sec")

print("\n[4] DIRECT MODEL LOAD")
print("-" * 70)

from transformers import AutoModelForCausalLM

print("Starting model construction...")
print("This is the critical section.")

t0 = time.time()

model = AutoModelForCausalLM.from_pretrained(
    MODEL_DIR,
    device_map="auto",
    torch_dtype="auto",
    local_files_only=True
)

print("MODEL LOAD: PASS")
print("Time:", round(time.time() - t0, 2), "sec")

print("\n[5] MODEL DEVICE MAP")
print("-" * 70)

print(getattr(model, "hf_device_map", "No hf_device_map"))

print("\n[6] GPU MEMORY AFTER LOAD")
print("-" * 70)

for i in range(torch.cuda.device_count()):
    allocated = torch.cuda.memory_allocated(i)
    reserved = torch.cuda.memory_reserved(i)

    print(
        f"GPU {i}: "
        f"allocated={allocated/1024**3:.2f} GB | "
        f"reserved={reserved/1024**3:.2f} GB"
    )

print("\n[7] TINY DIRECT INFERENCE")
print("-" * 70)

messages = [
    {
        "role": "user",
        "content": "Say exactly: REAL QWEN WORKING"
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

# Put inputs on the same device as the model's input embedding
input_device = model.get_input_embeddings().weight.device
inputs = {k: v.to(input_device) for k, v in inputs.items()}

t0 = time.time()

with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=32,
        do_sample=False
    )

response = tokenizer.decode(
    output[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
)

print("Response:", response)
print("Inference time:", round(time.time() - t0, 2), "sec")

print("\n" + "=" * 70)
print("DIAGNOSTIC COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-029 : STRUCTURED ACTION PLANNER
# ============================================================

from pathlib import Path
import sys
import json
import re
import shutil
import os

PROJECT_ROOT = Path("/kaggle/working/Personal_AI_AGI_Core_v3_0")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

PLANNER_FILE = PROJECT_ROOT / "personal_ai" / "core" / "planner.py"

print("=" * 70)
print("KAGGLE-029 : STRUCTURED ACTION PLANNER")
print("=" * 70)

print("\n[1] SOURCE")
print("-" * 70)
print("Planner:", PLANNER_FILE)
print("Exists :", PLANNER_FILE.exists())

if not PLANNER_FILE.exists():
    raise FileNotFoundError(PLANNER_FILE)

# ------------------------------------------------------------
# Backup
# ------------------------------------------------------------

BACKUP_FILE = PLANNER_FILE.with_suffix(".py.kaggle029.bak")

if not BACKUP_FILE.exists():
    shutil.copy2(PLANNER_FILE, BACKUP_FILE)

print("Backup :", BACKUP_FILE)
print("Backup status: PASS")

# ------------------------------------------------------------
# New Planner implementation
# ------------------------------------------------------------

NEW_SOURCE = r'''
import json
import re

from .planner_models import Plan


class Planner:
    """
    Structured-action planner for Personal AI V3.

    Responsibilities:
      1. Ask the model for an executable plan.
      2. Extract JSON from model output.
      3. Normalize model-specific tool names.
      4. Convert actions into V3 canonical steps.
      5. Preserve verification information.
    """

    TOOL_ALIASES = {
        "python": "python",
        "Python": "python",
        "PYTHON": "python",

        "shell": "shell",
        "Shell": "shell",
        "bash": "shell",
        "Bash": "shell",

        "filesystem_read": "filesystem_read",
        "filesystem": "filesystem_read",
        "file_read": "filesystem_read",
    }

    def __init__(self, models):
        self.models = models

    # --------------------------------------------------------
    # JSON extraction
    # --------------------------------------------------------

    def _extract_json(self, raw):
        if not raw:
            raise ValueError("Model returned empty response")

        text = raw.strip()

        # Remove markdown fences if present.
        text = re.sub(
            r"^```(?:json)?\s*",
            "",
            text,
            flags=re.IGNORECASE
        )

        text = re.sub(
            r"\s*```$",
            "",
            text
        ).strip()

        # First try the complete response.
        try:
            return json.loads(text)
        except Exception:
            pass

        # Otherwise find the outermost JSON object.
        start = text.find("{")

        if start == -1:
            raise ValueError(
                "No JSON object found in model response"
            )

        decoder = json.JSONDecoder()

        try:
            obj, _ = decoder.raw_decode(text[start:])
            return obj
        except Exception as e:
            raise ValueError(
                f"Unable to parse model JSON: {e}"
            )

    # --------------------------------------------------------
    # Tool normalization
    # --------------------------------------------------------

    def _normalize_tool(self, tool):
        if not isinstance(tool, str):
            raise ValueError("Tool name must be a string")

        if tool not in self.TOOL_ALIASES:
            raise ValueError(
                f"Unsupported tool requested by model: {tool}"
            )

        return self.TOOL_ALIASES[tool]

    # --------------------------------------------------------
    # Convert model plan → V3 actions
    # --------------------------------------------------------

    def _convert_plan(self, data):

        if not isinstance(data, dict):
            raise ValueError("Planner JSON must be an object")

        raw_steps = data.get("steps")

        # Existing Qwen responses may call the field "plan".
        if raw_steps is None:
            raw_steps = data.get("plan")

        if not isinstance(raw_steps, list):
            raise ValueError(
                "Planner JSON must contain a 'steps' or 'plan' list"
            )

        canonical_steps = []

        for index, item in enumerate(raw_steps):

            if not isinstance(item, dict):
                raise ValueError(
                    f"Plan step {index} is not an object"
                )

            # ----------------------------------------------
            # Already canonical V3 action
            # ----------------------------------------------

            if item.get("action") == "tool":

                tool = self._normalize_tool(
                    item.get("tool")
                )

                args = item.get("args", [])
                kwargs = item.get("kwargs", {})

                if not isinstance(args, list):
                    raise ValueError(
                        f"Step {index}: args must be a list"
                    )

                if not isinstance(kwargs, dict):
                    raise ValueError(
                        f"Step {index}: kwargs must be an object"
                    )

                canonical_steps.append({
                    "action": "tool",
                    "tool": tool,
                    "args": args,
                    "kwargs": kwargs,
                })

                continue

            # ----------------------------------------------
            # Qwen format:
            #
            # {
            #   "tool": "Python",
            #   "command": "..."
            # }
            # ----------------------------------------------

            if "tool" in item:

                tool = self._normalize_tool(
                    item["tool"]
                )

                if tool == "python":

                    command = item.get("command")

                    if command is None:
                        command = item.get("code")

                    if command is None:
                        raise ValueError(
                            f"Step {index}: Python action "
                            "requires command/code"
                        )

                    canonical_steps.append({
                        "action": "tool",
                        "tool": "python",
                        "args": [command],
                        "kwargs": {},
                    })

                    continue

                if tool == "shell":

                    command = item.get("command")

                    if command is None:
                        raise ValueError(
                            f"Step {index}: shell action "
                            "requires command"
                        )

                    canonical_steps.append({
                        "action": "tool",
                        "tool": "shell",
                        "args": [command],
                        "kwargs": {},
                    })

                    continue

                if tool == "filesystem_read":

                    path = item.get("path")

                    if path is None:
                        raise ValueError(
                            f"Step {index}: filesystem_read "
                            "requires path"
                        )

                    canonical_steps.append({
                        "action": "tool",
                        "tool": "filesystem_read",
                        "args": [path],
                        "kwargs": {},
                    })

                    continue

            # ----------------------------------------------
            # Descriptive step without executable action.
            #
            # DO NOT silently execute it.
            # ----------------------------------------------

            raise ValueError(
                f"Step {index} is not executable: {item}"
            )

        if not canonical_steps:
            raise ValueError(
                "Planner produced zero executable actions"
            )

        return canonical_steps

    # --------------------------------------------------------
    # Main planning function
    # --------------------------------------------------------

    def make_plan(self, task):

        prompt = f"""
Create an executable plan for this task.

TASK:
{task.goal}

IMPORTANT:
Return ONLY valid JSON.

Use this schema:

{{
  "steps": [
    {{
      "action": "tool",
      "tool": "python",
      "args": ["python code here"],
      "kwargs": {{}}
    }}
  ],
  "verification": {{
    "expected_output": null,
    "method": "describe how the result should be verified"
  }}
}}

AVAILABLE TOOLS:
- python(code)
- shell(command)
- filesystem_read(path)

RULES:
1. Use executable tool actions whenever a tool is required.
2. Do not invent tools.
3. Do not put prose outside the JSON.
4. Include verification information when possible.
5. Prefer minimal reversible actions.
"""

        raw = self.models.generate(
            [
                {
                    "role": "system",
                    "content": (
                        "You are the structured planning engine "
                        "of Personal AI V3."
                    ),
                },
                {
                    "role": "user",
                    "content": prompt,
                },
            ],
            capability="planning",
            max_new_tokens=512,
            do_sample=False,
        )

        data = self._extract_json(raw)

        steps = self._convert_plan(data)

        # Preserve verification metadata.
        verification = data.get(
            "verification",
            {}
        )

        if not isinstance(verification, dict):
            verification = {}

        plan = Plan(
            task.id,
            steps,
            verification,
        )

        return plan
'''

# ------------------------------------------------------------
# Write patched Planner
# ------------------------------------------------------------

print("\n[2] PATCHING PLANNER")
print("-" * 70)

PLANNER_FILE.write_text(
    NEW_SOURCE,
    encoding="utf-8"
)

print("Planner patch        : WRITTEN")

# ------------------------------------------------------------
# Syntax check
# ------------------------------------------------------------

print("\n[3] SYNTAX CHECK")
print("-" * 70)

compile(
    PLANNER_FILE.read_text(encoding="utf-8"),
    str(PLANNER_FILE),
    "exec"
)

print("Python syntax        : PASS")

# ------------------------------------------------------------
# Import fresh module
# ------------------------------------------------------------

print("\n[4] IMPORT CHECK")
print("-" * 70)

import importlib
import personal_ai.core.planner as planner_module

planner_module = importlib.reload(planner_module)

Planner = planner_module.Planner

print("Planner import       : PASS")
print("Planner class        :", Planner)

# ------------------------------------------------------------
# Direct parser unit test
# ------------------------------------------------------------

print("\n[5] STRUCTURED PARSER TEST")
print("-" * 70)

test_response = r'''
```json
{
  "plan": [
    {
      "tool": "Python",
      "command": "print((10 + 20 + 30) / 3)"
    }
  ],
  "verification": {
    "expected_output": 20.0,
    "method": "Compare result with expected value."
  }
}
````

````python
# ============================================================
# KAGGLE-029-R7 : NORMALIZE QWEN ACTION FORMAT
# ============================================================

from pathlib import Path
import os
import sys
import shutil
import importlib

print("=== KAGGLE-029-R7 : NORMALIZATION FIX ===")

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

PLANNER_FILE = (
    PROJECT_ROOT /
    "personal_ai" /
    "core" /
    "planner.py"
)

BACKUP_FILE = (
    PROJECT_ROOT /
    "personal_ai" /
    "core" /
    "planner.py.kaggle029.r7.bak"
)

print("\n[1] Planner file")
print("-" * 60)

print("Planner:", PLANNER_FILE)
assert PLANNER_FILE.exists()

# ------------------------------------------------------------
# Backup current R5 source
# ------------------------------------------------------------

if not BACKUP_FILE.exists():
    shutil.copy2(PLANNER_FILE, BACKUP_FILE)
    print("Backup: CREATED")
else:
    print("Backup: ALREADY EXISTS")

# ------------------------------------------------------------
# Read source
# ------------------------------------------------------------

source = PLANNER_FILE.read_text()

# ------------------------------------------------------------
# Locate current canonical-action condition
# ------------------------------------------------------------

old = '''            # Already canonical V3 action.
            if item.get("action") == "tool":

                tool = self._normalize_tool(
                    item.get("tool")
                )

                args = item.get("args", [])
                kwargs = item.get("kwargs", {})

                if not isinstance(args, list):
                    raise ValueError(
                        "Step " + str(index)
                        + ": args must be a list"
                    )

                if not isinstance(kwargs, dict):
                    raise ValueError(
                        "Step " + str(index)
                        + ": kwargs must be an object"
                    )

                canonical_steps.append({
                    "action": "tool",
                    "tool": tool,
                    "args": args,
                    "kwargs": kwargs,
                })

                continue
'''

new = '''            # Normalize model action names.
            #
            # Qwen may return:
            #   {"action": "python", "tool": "python", ...}
            #
            # V3 canonical form is:
            #   {"action": "tool", "tool": "python", ...}
            #
            # Therefore convert python/shell/filesystem_read
            # actions into the V3 executable action schema.

            action = str(
                item.get("action", "")
            ).strip().lower()

            if action in {
                "python",
                "shell",
                "filesystem_read",
                "tool",
            }:

                requested_tool = item.get("tool")

                # If the model uses action as the actual tool name
                # and supplies the matching tool, normalize it.
                if action in {
                    "python",
                    "shell",
                    "filesystem_read",
                }:
                    tool = action

                else:
                    tool = self._normalize_tool(
                        requested_tool
                    )

                args = item.get("args", [])
                kwargs = item.get("kwargs", {})

                if not isinstance(args, list):
                    raise ValueError(
                        "Step " + str(index)
                        + ": args must be a list"
                    )

                if not isinstance(kwargs, dict):
                    kwargs = {}

                # Only tools actually registered by V3 are allowed.
                if tool not in {
                    "python",
                    "shell",
                    "filesystem_read",
                }:
                    raise ValueError(
                        "Unsupported executable tool: "
                        + str(tool)
                    )

                # python(code)
                if tool == "python":

                    if not args:
                        code = item.get("code")
                        command = item.get("command")

                        if code is not None:
                            args = [code]

                        elif command is not None:
                            args = [command]

                        else:
                            raise ValueError(
                                "Python step requires code"
                            )

                # shell(command)
                elif tool == "shell":

                    if not args:
                        command = item.get("command")

                        if command is not None:
                            args = [command]

                        else:
                            raise ValueError(
                                "Shell step requires command"
                            )

                # filesystem_read(path)
                elif tool == "filesystem_read":

                    if not args:
                        path = item.get("path")

                        if path is not None:
                            args = [path]

                        else:
                            raise ValueError(
                                "filesystem_read step "
                                "requires path"
                            )

                canonical_steps.append({
                    "action": "tool",
                    "tool": tool,
                    "args": args,
                    "kwargs": kwargs,
                })

                continue
'''

print("\n[2] Checking current planner source")
print("-" * 60)

if old not in source:
    print("Expected R5 block NOT FOUND.")
    print("No source modification performed.")
    raise RuntimeError(
        "Planner source differs from expected R5 state. "
        "Do not patch blindly."
    )

source = source.replace(old, new, 1)

# ------------------------------------------------------------
# Also remove obsolete verification variable from Plan issue.
# Keep verification local for now.
# ------------------------------------------------------------

old_plan = '''        return Plan(
            task.id,
            steps,
            []
        )
'''

new_plan = '''        return Plan(
            task.id,
            steps,
            []
        )
'''

# This replacement is intentionally a no-op check.
assert old_plan in source

# ------------------------------------------------------------
# Write
# ------------------------------------------------------------

PLANNER_FILE.write_text(source)

print("Normalization patch : APPLIED")

# ------------------------------------------------------------
# Reload
# ------------------------------------------------------------

importlib.invalidate_caches()

import personal_ai.core.planner as planner_module

planner_module = importlib.reload(
    planner_module
)

Planner = planner_module.Planner

from personal_ai.core.contracts import Plan
from personal_ai.bootstrap import Bootstrap

print("\n[3] Reload")
print("-" * 60)

runtime = Bootstrap(PROJECT_ROOT).start()
planner = Planner(runtime.models)

print("Planner :", Planner)
print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

# ------------------------------------------------------------
# 4. Deterministic parser test using ACTUAL Qwen output
# ------------------------------------------------------------

print("\n[4] Actual R6B Qwen format test")
print("-" * 60)

qwen_data = {
    "steps": [
        {
            "action": "python",
            "tool": "python",
            "args": [
                "sum = 10 + 20 + 30; avg = sum / 3; print(avg)"
            ],
            "kwargs": {}
        },
        {
            "action": "shell",
            "tool": "echo",
            "args": [
                "Verification: The average is 20.0"
            ]
        }
    ],
    "verification": {
        "expected_output": 20.0,
        "method": "manual calculation or using another reliable source"
    }
}

converted = planner._convert_plan(qwen_data)

print("Converted type:")
print(type(converted).__name__)

print("\nConverted steps:")
print(converted)

# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert isinstance(converted, list)
assert len(converted) == 2

# Step 0
step0 = converted[0]

assert step0["action"] == "tool"
assert step0["tool"] == "python"
assert step0["args"] == [
    "sum = 10 + 20 + 30; avg = sum / 3; print(avg)"
]
assert isinstance(step0["kwargs"], dict)

# Step 1
step1 = converted[1]

assert step1["action"] == "tool"
assert step1["tool"] == "shell"

# ------------------------------------------------------------
# 5. Actual V3 Plan contract
# ------------------------------------------------------------

print("\n[5] V3 Plan construction")
print("-" * 60)

plan = Plan(
    task_id="KAGGLE-029-R7",
    steps=converted,
    assumptions=[]
)

print("Plan:", plan)
print("Type:", type(plan).__name__)

assert isinstance(plan, Plan)
assert plan.task_id == "KAGGLE-029-R7"
assert len(plan.steps) == 2
assert plan.steps[0]["tool"] == "python"
assert plan.steps[1]["tool"] == "shell"
assert isinstance(plan.assumptions, list)

print("Plan contract : PASS")

# ------------------------------------------------------------
# 6. Real make_plan test
# ------------------------------------------------------------

print("\n[6] REAL make_plan()")
print("-" * 60)

from personal_ai.core.contracts import Task

task = Task(
    id="KAGGLE-029-R7-REAL",
    goal="Calculate the average of 10, 20, and 30."
)

result = planner.make_plan(task)

print("Returned type:", type(result).__name__)
print("Plan:", result)

assert isinstance(result, Plan)
assert result.task_id == task.id
assert isinstance(result.steps, list)
assert len(result.steps) >= 1

for i, step in enumerate(result.steps):
    print("\nStep", i, ":", step)

    assert step["action"] == "tool"
    assert step["tool"] in {
        "python",
        "shell",
        "filesystem_read"
    }

print("\nREAL QWEN → PLANNER → PLAN : PASS")

# ------------------------------------------------------------
# 7. Safety
# ------------------------------------------------------------

print("\n[7] Safety")
print("-" * 60)

print("Python tool executed     : NO")
print("Shell tool executed      : NO")
print("Filesystem tool executed : NO")
print("ExecutionEngine called   : NO")
print("Model modified           : NO")
print("Source intentionally modified: YES")
print("Backup exists            :", BACKUP_FILE.exists())

print("\n=== RESULT ===")
print("Qwen action normalization : PASS")
print("Canonical tool actions   : PASS")
print("V3 Plan contract         : PASS")
print("REAL make_plan()         : PASS")
print("No tool execution        : PASS")

print("\n=== END KAGGLE-029-R7 ===")
````

````python
# ============================================================
# KAGGLE-031-A : OBSERVATION → QWEN FEEDBACK
# ============================================================

from pathlib import Path
import os
import sys
import json
import time

print("=== KAGGLE-031-A : OBSERVATION FEEDBACK ===")

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine

# ------------------------------------------------------------
# 1. Runtime
# ------------------------------------------------------------

print("\n[1] Starting V3 runtime")
print("-" * 60)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime :", type(runtime).__name__)
print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

assert "qwen_local" in runtime.models.backends
assert "python" in runtime.tools.tools

print("Runtime : PASS")

# ------------------------------------------------------------
# 2. Create executable plan
# ------------------------------------------------------------

print("\n[2] Creating test Plan")
print("-" * 60)

task = Task(
    id="KAGGLE-031-A",
    goal="Calculate the average of 10, 20, and 30."
)

plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [
                "numbers = [10, 20, 30]\n"
                "avg = sum(numbers) / len(numbers)\n"
                "print(avg)"
            ],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("Plan:", plan)

assert len(plan.steps) == 1

print("Plan : PASS")

# ------------------------------------------------------------
# 3. Execute
# ------------------------------------------------------------

print("\n[3] REAL execution")
print("-" * 60)

engine = ExecutionEngine(
    runtime.tools,
    runtime.evaluator,
    runtime.learning,
    max_steps=5
)

start = time.time()

observations = engine.execute_plan(plan)

elapsed = time.time() - start

print("Execution time:", round(elapsed, 3), "seconds")

assert len(observations) == 1

obs = observations[0]

print("\nObservation object:")
print(obs)

print("\nok     :", obs.ok)
print("output :", repr(obs.output))
print("error  :", repr(obs.error))

assert obs.ok is True
assert obs.error is None

print("REAL execution : PASS")

# ------------------------------------------------------------
# 4. Serialize Observation
# ------------------------------------------------------------

print("\n[4] Preparing Observation for Qwen")
print("-" * 60)

observation_payload = {
    "step_index": 0,
    "action": plan.steps[0],
    "result": {
        "ok": obs.ok,
        "output": obs.output,
        "error": obs.error,
        "metadata": getattr(obs, "metadata", {})
    }
}

print(
    json.dumps(
        observation_payload,
        indent=2,
        ensure_ascii=False
    )
)

# ------------------------------------------------------------
# 5. Build feedback message
# ------------------------------------------------------------

print("\n[5] Sending execution result back to Qwen")
print("-" * 60)

feedback_prompt = (
    "You are continuing an executed task.\n\n"
    "Original task:\n"
    + task.goal
    + "\n\n"
    "The following tool action was actually executed:\n\n"
    + json.dumps(
        observation_payload,
        indent=2,
        ensure_ascii=False
    )
    + "\n\n"
    "Analyze the actual execution result.\n"
    "Do not pretend that an action was executed if it was not.\n"
    "State what the result means.\n"
    "If the task is complete, explicitly say TASK_COMPLETE.\n"
    "If another action is required, explicitly state NEXT_ACTION."
)

print("Feedback prompt prepared.")

# ------------------------------------------------------------
# 6. REAL Qwen feedback
# ------------------------------------------------------------

print("\n[6] REAL Qwen feedback")
print("-" * 60)

start = time.time()

response = runtime.models.generate(
    [
        {
            "role": "system",
            "content": (
                "You are the execution-feedback engine "
                "of Personal AI V3. "
                "Use only the actual observation supplied."
            )
        },
        {
            "role": "user",
            "content": feedback_prompt
        }
    ],
    capability="reasoning",
    max_new_tokens=256,
    do_sample=False,
)

elapsed = time.time() - start

print("Qwen response time:", round(elapsed, 3), "seconds")

print("\n--- QWEN FEEDBACK ---")
print(response)
print("--- END QWEN FEEDBACK ---")

assert isinstance(response, str)
assert len(response.strip()) > 0

print("\nQwen feedback : PASS")

# ------------------------------------------------------------
# 7. Check whether Qwen saw the actual result
# ------------------------------------------------------------

print("\n[7] Observation awareness test")
print("-" * 60)

response_lower = response.lower()

result_seen = (
    "20" in response_lower
    or "complete" in response_lower
    or "average" in response_lower
)

print("Qwen response references actual task/result:",
      result_seen)

assert result_seen

print("Observation awareness : PASS")

# ------------------------------------------------------------
# 8. Safety
# ------------------------------------------------------------

print("\n[8] Safety")
print("-" * 60)

print("Python execution : YES")
print("Shell execution  : NO")
print("Filesystem tool  : NO")
print("Observation sent to Qwen : YES")
print("Source modified : NO")
print("Planner modified: NO")
print("Model modified  : NO")

print("\n=== RESULT ===")
print("REAL execution                 : PASS")
print("Observation captured           : PASS")
print("Observation serialized         : PASS")
print("Observation → Qwen             : PASS")
print("Qwen feedback generated        : PASS")
print("Qwen observation awareness     : PASS")

print("\n=== END KAGGLE-031-A ===")
````

````python
# ============================================================
# KAGGLE-032-A : OBSERVATION → QWEN → REPLAN
# ============================================================

from pathlib import Path
import os
import sys
import json
import re
import time

print("=== KAGGLE-032-A : FEEDBACK → REPLAN ===")

# ------------------------------------------------------------
# 1. Bind existing V3 runtime
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    os.environ["PERSONAL_AI_ROOT"]
).resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine

runtime = Bootstrap(PROJECT_ROOT).start()

print("\n[1] Runtime")
print("-" * 60)
print("Models :", list(runtime.models.backends.keys()))
print("Tools  :", list(runtime.tools.tools.keys()))

assert "qwen_local" in runtime.models.backends
assert "python" in runtime.tools.tools

print("Runtime : PASS")


# ------------------------------------------------------------
# 2. Initial task
# ------------------------------------------------------------

print("\n[2] Initial task")
print("-" * 60)

task = Task(
    id="KAGGLE-032-A",
    goal=(
        "Calculate the average of 10, 20, and 30, "
        "then verify the calculated result."
    )
)

print("Task:", task.goal)


# ------------------------------------------------------------
# 3. Controlled first action
# ------------------------------------------------------------

print("\n[3] Initial executable action")
print("-" * 60)

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [
                "numbers = [10, 20, 30]\n"
                "avg = sum(numbers) / len(numbers)\n"
                "print(avg)"
            ],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print(json.dumps(initial_plan.steps, indent=2))

assert initial_plan.steps[0]["tool"] == "python"

print("Initial action : PASS")


# ------------------------------------------------------------
# 4. Execute first action
# ------------------------------------------------------------

print("\n[4] REAL first execution")
print("-" * 60)

engine = ExecutionEngine(
    runtime.tools,
    runtime.evaluator,
    runtime.learning,
    max_steps=5
)

start = time.time()

observations = engine.execute_plan(initial_plan)

elapsed = time.time() - start

assert len(observations) == 1

obs = observations[0]

print("Execution time:", round(elapsed, 3), "seconds")
print("OK     :", obs.ok)
print("Output :", repr(obs.output))
print("Error  :", repr(obs.error))

assert obs.ok is True
assert obs.error is None

print("First execution : PASS")


# ------------------------------------------------------------
# 5. Construct feedback state
# ------------------------------------------------------------

print("\n[5] Building feedback state")
print("-" * 60)

feedback_state = {
    "task": task.goal,
    "completed_steps": [
        {
            "step": initial_plan.steps[0],
            "observation": {
                "ok": obs.ok,
                "output": obs.output,
                "error": obs.error,
                "metadata": getattr(obs, "metadata", {})
            }
        }
    ]
}

print(json.dumps(
    feedback_state,
    indent=2,
    ensure_ascii=False
))


# ------------------------------------------------------------
# 6. Ask Qwen to REPLAN
# ------------------------------------------------------------

print("\n[6] Qwen replanning")
print("-" * 60)

replan_prompt = (
    "You are the replanning component of Personal AI V3.\n\n"

    "TASK:\n"
    + task.goal
    + "\n\n"

    "ACTUAL EXECUTION STATE:\n"
    + json.dumps(
        feedback_state,
        indent=2,
        ensure_ascii=False
    )
    + "\n\n"

    "Determine what should happen next.\n\n"

    "Return ONLY valid JSON using exactly this schema:\n"
    "{\n"
    '  "status": "NEXT_ACTION" or "TASK_COMPLETE",\n'
    '  "next_action": {\n'
    '    "action": "tool",\n'
    '    "tool": "python",\n'
    '    "args": ["python code"],\n'
    '    "kwargs": {}\n'
    "  }\n"
    "}\n\n"

    "Rules:\n"
    "1. Use only python, shell, or filesystem_read.\n"
    "2. Do not claim an action was executed unless the observation says so.\n"
    "3. If the task is already completely satisfied, use TASK_COMPLETE.\n"
    "4. If another operation is genuinely needed, use NEXT_ACTION.\n"
    "5. Return JSON only."
)

start = time.time()

qwen_replan_raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": (
                "You are a cautious execution replanner. "
                "Reason from actual observations only."
            )
        },
        {
            "role": "user",
            "content": replan_prompt
        }
    ],
    capability="planning",
    max_new_tokens=256,
    do_sample=False,
)

elapsed = time.time() - start

print("Qwen replanning time:", round(elapsed, 3), "seconds")

print("\n--- QWEN REPLAN ---")
print(qwen_replan_raw)
print("--- END QWEN REPLAN ---")

assert isinstance(qwen_replan_raw, str)
assert len(qwen_replan_raw.strip()) > 0

print("Qwen replan generation : PASS")


# ------------------------------------------------------------
# 7. Extract JSON
# ------------------------------------------------------------

print("\n[7] Parsing Qwen replan")
print("-" * 60)

def extract_json_object(text):
    text = text.strip()

    # Direct JSON
    try:
        value = json.loads(text)
        if isinstance(value, dict):
            return value
    except Exception:
        pass

    # Markdown fenced JSON
    fenced = re.search(
        r"```(?:json)?\s*(\{.*?\})\s*```",
        text,
        re.DOTALL | re.IGNORECASE
    )

    if fenced:
        value = json.loads(fenced.group(1))
        if isinstance(value, dict):
            return value

    # First JSON object
    start_idx = text.find("{")
    end_idx = text.rfind("}")

    if start_idx >= 0 and end_idx > start_idx:
        value = json.loads(
            text[start_idx:end_idx + 1]
        )
        if isinstance(value, dict):
            return value

    raise ValueError("No valid JSON object found")


replan = extract_json_object(qwen_replan_raw)

print(json.dumps(
    replan,
    indent=2,
    ensure_ascii=False
))

assert isinstance(replan, dict)

status = replan.get("status")

assert status in {
    "NEXT_ACTION",
    "TASK_COMPLETE"
}

print("Replan JSON : PASS")
print("Status      :", status)


# ------------------------------------------------------------
# 8. Validate proposed next action
# ------------------------------------------------------------

print("\n[8] Action validation")
print("-" * 60)

if status == "NEXT_ACTION":

    next_action = replan.get("next_action")

    assert isinstance(next_action, dict)

    action = str(
        next_action.get("action", "")
    ).strip().lower()

    tool = str(
        next_action.get("tool", "")
    ).strip().lower()

    args = next_action.get("args", [])
    kwargs = next_action.get("kwargs", {})

    print("Action :", action)
    print("Tool   :", tool)
    print("Args   :", args)
    print("Kwargs :", kwargs)

    assert action == "tool"
    assert tool in {
        "python",
        "shell",
        "filesystem_read"
    }

    assert isinstance(args, list)
    assert isinstance(kwargs, dict)

    print("Action validation : PASS")

else:
    print(
        "Qwen determined that the task is already complete."
    )
    print("Action validation : NOT NEEDED")


# ------------------------------------------------------------
# 9. IMPORTANT: Do NOT execute replan yet
# ------------------------------------------------------------

print("\n[9] Execution boundary")
print("-" * 60)

print("Qwen proposed action :", status)
print("Proposed action executed automatically : NO")
print("Reason : KAGGLE-032-A validates replanning")
print("        before connecting it to the autonomous loop.")

print("Execution boundary : PASS")


# ------------------------------------------------------------
# 10. Final result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("KAGGLE-032-A RESULT")
print("=" * 60)

print("REAL first execution       : PASS")
print("Observation captured       : PASS")
print("Observation → Qwen         : PASS")
print("Qwen → structured replan   : PASS")
print("Replan JSON parsing        : PASS")
print("Action validation          : PASS")
print("Automatic second execution : NOT YET")

print("\n=== END KAGGLE-032-A ===")
````

````python
# ============================================================
# KAGGLE-032-A-R0.7 : QWEN BACKEND STATE DIAGNOSTIC
# ============================================================

import gc
import torch

print("=== QWEN BACKEND STATE DIAGNOSTIC ===")

qwen = runtime.models.backends.get("qwen_local")

print("\nBackend object:")
print("type:", type(qwen))

print("\nBackend attributes:")
for name in [
    "_model",
    "_tokenizer",
    "model_path",
]:
    value = getattr(qwen, name, "<MISSING>")

    if name == "_model":
        print(
            name,
            ":",
            "LOADED" if value is not None else "None"
        )
    elif name == "_tokenizer":
        print(
            name,
            ":",
            "LOADED" if value is not None else "None"
        )
    else:
        print(name, ":", value)

print("\nAll backend instance attributes:")
print(list(vars(qwen).keys()))

print("\nPython GPU objects:")
cuda_objects = []

for obj in gc.get_objects():
    try:
        if torch.is_tensor(obj) and obj.is_cuda:
            cuda_objects.append(obj)
    except Exception:
        pass

print("CUDA tensor objects found:", len(cuda_objects))

for i, obj in enumerate(cuda_objects[:20]):
    print(
        f"  tensor[{i}] "
        f"shape={tuple(obj.shape)} "
        f"dtype={obj.dtype} "
        f"device={obj.device} "
        f"size={obj.numel() * obj.element_size() / 1024**2:.1f} MB"
    )

if len(cuda_objects) > 20:
    print(
        "...",
        len(cuda_objects) - 20,
        "more CUDA tensors"
    )

print("\nTorch memory:")
if torch.cuda.is_available():
    for i in range(torch.cuda.device_count()):
        allocated = (
            torch.cuda.memory_allocated(i)
            / 1024**3
        )
        reserved = (
            torch.cuda.memory_reserved(i)
            / 1024**3
        )
        free, total = torch.cuda.mem_get_info(i)

        print(
            f"GPU {i}: "
            f"allocated={allocated:.2f} GB, "
            f"reserved={reserved:.2f} GB, "
            f"free={free / 1024**3:.2f} GB"
        )

print("\n=== END DIAGNOSTIC ===")
````

````python
# ============================================================
# KAGGLE-032-A-R1
# QWEN MODEL LIFECYCLE + GPU RECOVERY TEST
#
# PURPOSE:
#   1. Reuse existing runtime if model is healthy
#   2. Detect stale/partial Qwen state
#   3. Cleanly release model references
#   4. Empty CUDA cache
#   5. Reload Qwen exactly ONCE
#   6. Run TWO consecutive inferences
#
# SAFETY:
#   - No shell tool execution
#   - No filesystem modification
#   - No training
#   - No planner modification
# ============================================================

import gc
import time
import torch

print("=" * 70)
print("KAGGLE-032-A-R1 : QWEN LIFECYCLE + GPU RECOVERY TEST")
print("=" * 70)


# ------------------------------------------------------------
# 1. EXISTING RUNTIME
# ------------------------------------------------------------

print("\n[1] EXISTING RUNTIME")
print("-" * 70)

assert "runtime" in globals(), (
    "Existing runtime not found. "
    "Do not create another runtime in this test."
)

print("Runtime :", type(runtime).__name__)

qwen = runtime.models.backends.get("qwen_local")

assert qwen is not None, "qwen_local backend not found"

print("Qwen backend :", type(qwen).__name__)
print("Model loaded :", getattr(qwen, "_model", None) is not None)
print("Tokenizer    :", getattr(qwen, "_tokenizer", None) is not None)


# ------------------------------------------------------------
# 2. GPU BEFORE CLEANUP
# ------------------------------------------------------------

print("\n[2] GPU BEFORE CLEANUP")
print("-" * 70)

for i in range(torch.cuda.device_count()):
    free, total = torch.cuda.mem_get_info(i)

    print(
        f"GPU {i}: "
        f"free={free / 1024**3:.2f} GB / "
        f"total={total / 1024**3:.2f} GB | "
        f"allocated={torch.cuda.memory_allocated(i)/1024**3:.2f} GB | "
        f"reserved={torch.cuda.memory_reserved(i)/1024**3:.2f} GB"
    )


# ------------------------------------------------------------
# 3. CLEAN STALE MODEL REFERENCE
# ------------------------------------------------------------

print("\n[3] CLEANING STALE MODEL REFERENCE")
print("-" * 70)

old_model = getattr(qwen, "_model", None)

if old_model is not None:
    print("Existing model detected.")
    print("Releasing backend model reference...")

    qwen._model = None

    del old_model

else:
    print("Backend model reference already None.")


# ------------------------------------------------------------
# 4. PYTHON + CUDA CLEANUP
# ------------------------------------------------------------

print("\n[4] PYTHON / CUDA CLEANUP")
print("-" * 70)

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

gc.collect()

print("Garbage collection : PASS")
print("CUDA cache cleanup : PASS")


# ------------------------------------------------------------
# 5. GPU AFTER CLEANUP
# ------------------------------------------------------------

print("\n[5] GPU AFTER CLEANUP")
print("-" * 70)

for i in range(torch.cuda.device_count()):
    free, total = torch.cuda.mem_get_info(i)

    print(
        f"GPU {i}: "
        f"free={free / 1024**3:.2f} GB / "
        f"total={total / 1024**3:.2f} GB | "
        f"allocated={torch.cuda.memory_allocated(i)/1024**3:.2f} GB | "
        f"reserved={torch.cuda.memory_reserved(i)/1024**3:.2f} GB"
    )


# ------------------------------------------------------------
# 6. VERIFY MODEL PATH
# ------------------------------------------------------------

print("\n[6] QWEN MODEL PATH")
print("-" * 70)

print("Model path :", qwen.model_path)
print("Exists     :", qwen.model_path.exists())

assert qwen.model_path.exists(), (
    f"Qwen model path does not exist: {qwen.model_path}"
)


# ------------------------------------------------------------
# 7. LOAD QWEN ONCE
# ------------------------------------------------------------

print("\n[7] QWEN RELOAD")
print("-" * 70)

t0 = time.time()

try:
    # Use the V3 backend's own loading mechanism.
    qwen.load()

    elapsed = time.time() - t0

    print("Qwen load : PASS")
    print("Load time :", round(elapsed, 2), "sec")

except Exception as e:
    print("\nQWEN LOAD : FAIL")
    print("Exception :", type(e).__name__)
    print("Message   :", str(e))

    raise


# ------------------------------------------------------------
# 8. VERIFY LOADED STATE
# ------------------------------------------------------------

print("\n[8] LOADED STATE")
print("-" * 70)

model = getattr(qwen, "_model", None)
tokenizer = getattr(qwen, "_tokenizer", None)

print("Model     :", type(model).__name__ if model is not None else None)
print("Tokenizer :", type(tokenizer).__name__ if tokenizer is not None else None)

assert model is not None
assert tokenizer is not None

print("Model state     : PASS")
print("Tokenizer state : PASS")


# ------------------------------------------------------------
# 9. GPU AFTER LOAD
# ------------------------------------------------------------

print("\n[9] GPU AFTER QWEN LOAD")
print("-" * 70)

for i in range(torch.cuda.device_count()):
    free, total = torch.cuda.mem_get_info(i)

    print(
        f"GPU {i}: "
        f"free={free / 1024**3:.2f} GB / "
        f"total={total / 1024**3:.2f} GB | "
        f"allocated={torch.cuda.memory_allocated(i)/1024**3:.2f} GB | "
        f"reserved={torch.cuda.memory_reserved(i)/1024**3:.2f} GB"
    )


# ------------------------------------------------------------
# 10. FIRST INFERENCE
# ------------------------------------------------------------

print("\n[10] FIRST INFERENCE")
print("-" * 70)

messages_1 = [
    {
        "role": "user",
        "content": "Reply with exactly: QWEN_LIFECYCLE_PASS"
    }
]

t0 = time.time()

response_1 = runtime.models.generate(
    messages_1,
    capability="general",
    max_new_tokens=32,
    do_sample=False
)

elapsed_1 = time.time() - t0

print("Response :", repr(response_1))
print("Time     :", round(elapsed_1, 2), "sec")

assert response_1 is not None
assert len(str(response_1).strip()) > 0

print("First inference : PASS")


# ------------------------------------------------------------
# 11. SECOND CONSECUTIVE INFERENCE
# ------------------------------------------------------------

print("\n[11] SECOND CONSECUTIVE INFERENCE")
print("-" * 70)

messages_2 = [
    {
        "role": "user",
        "content": "Reply with exactly: SECOND_INFERENCE_PASS"
    }
]

t0 = time.time()

response_2 = runtime.models.generate(
    messages_2,
    capability="general",
    max_new_tokens=32,
    do_sample=False
)

elapsed_2 = time.time() - t0

print("Response :", repr(response_2))
print("Time     :", round(elapsed_2, 2), "sec")

assert response_2 is not None
assert len(str(response_2).strip()) > 0

print("Second inference : PASS")


# ------------------------------------------------------------
# 12. FINAL STATE
# ------------------------------------------------------------

print("\n[12] FINAL STATE")
print("-" * 70)

print("Qwen model loaded :", getattr(qwen, "_model", None) is not None)
print("Tokenizer loaded  :", getattr(qwen, "_tokenizer", None) is not None)

for i in range(torch.cuda.device_count()):
    free, total = torch.cuda.mem_get_info(i)

    print(
        f"GPU {i}: "
        f"free={free / 1024**3:.2f} GB / "
        f"total={total / 1024**3:.2f} GB"
    )


# ------------------------------------------------------------
# RESULT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-A-R1 RESULT")
print("=" * 70)

print("Existing runtime       : PASS")
print("Stale reference cleanup: PASS")
print("Qwen reload             : PASS")
print("Model state             : PASS")
print("Tokenizer state         : PASS")
print("First inference         : PASS")
print("Second inference        : PASS")
print("Training                : NO")
print("Planner modification    : NO")
print("Tool execution          : NO")

print("\n=== END KAGGLE-032-A-R1 ===")
````

````python
# ============================================================
# KAGGLE-032-A-R3
# SAME RUNTIME → REAL EXECUTION → OBSERVATION → QWEN → REPLAN
#
# IMPORTANT:
#   Reuses the EXISTING runtime from R2-CLEAN.
#   Does NOT create another Bootstrap/runtime.
#   Does NOT automatically execute Qwen's second action.
# ============================================================

import json
import re
import time

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine

print("=" * 70)
print("KAGGLE-032-A-R3")
print("=" * 70)

# ------------------------------------------------------------
# 0. Verify that we are reusing the existing runtime
# ------------------------------------------------------------

print("\n[0] EXISTING RUNTIME")
print("-" * 70)

assert "runtime" in globals(), (
    "Existing runtime not found. "
    "Do NOT create another runtime in this cell."
)

assert "qwen" in globals(), (
    "Qwen backend reference not found. "
    "Use the R2-CLEAN runtime."
)

print("Runtime object :", runtime)
print("Qwen object    :", qwen)
print("Qwen model loaded :", qwen._model is not None)
print("Qwen tokenizer loaded :", qwen._tokenizer is not None)

assert qwen._model is not None
assert qwen._tokenizer is not None

# ------------------------------------------------------------
# 1. Task
# ------------------------------------------------------------

print("\n[1] TASK")
print("-" * 70)

task = Task(
    id="KAGGLE-032-A",
    goal=(
        "Calculate the average of 10, 20, and 30, "
        "then verify the calculated result."
    )
)

print("Task:", task.goal)

# ------------------------------------------------------------
# 2. Controlled first action
# ------------------------------------------------------------

print("\n[2] INITIAL PLAN")
print("-" * 70)

plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [
                "numbers = [10, 20, 30]\n"
                "avg = sum(numbers) / len(numbers)\n"
                "print(avg)"
            ],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print(json.dumps(plan.steps, indent=2))

# ------------------------------------------------------------
# 3. Execute REAL tool
# ------------------------------------------------------------

print("\n[3] REAL EXECUTION")
print("-" * 70)

evaluator = runtime.evaluator
ledger = runtime.ledger

engine = ExecutionEngine(
    runtime.tools,
    evaluator,
    ledger,
    max_steps=1
)

start = time.time()

observations = engine.execute_plan(plan)

elapsed = time.time() - start

print("Execution time:", round(elapsed, 3), "sec")
print("Observation count:", len(observations))

assert len(observations) == 1

obs = observations[0]

print("\nObservation:")
print(obs)

print("\nObservation attributes:")
print(vars(obs) if hasattr(obs, "__dict__") else repr(obs))

# ------------------------------------------------------------
# 4. Serialize ACTUAL observation
# ------------------------------------------------------------

print("\n[4] OBSERVATION → QWEN STATE")
print("-" * 70)

observation_data = {
    "success": getattr(obs, "success", False),
    "output": getattr(obs, "output", None),
    "error": getattr(obs, "error", None),
}

feedback_state = {
    "task": task.goal,
    "initial_plan": plan.steps,
    "actual_observation": observation_data,
}

print(json.dumps(
    feedback_state,
    indent=2,
    ensure_ascii=False,
    default=str
))

assert observation_data["success"] is True

# ------------------------------------------------------------
# 5. Ask SAME Qwen instance to replan
# ------------------------------------------------------------

print("\n[5] QWEN REPLANNING")
print("-" * 70)

replan_prompt = (
    "You are the replanning component of Personal AI V3.\n\n"
    "TASK:\n"
    + task.goal
    + "\n\n"
    "ACTUAL EXECUTION STATE:\n"
    + json.dumps(
        feedback_state,
        indent=2,
        ensure_ascii=False,
        default=str
    )
    + "\n\n"
    "Determine what should happen next.\n\n"
    "Return ONLY valid JSON using exactly this schema:\n"
    "{\n"
    '  "status": "NEXT_ACTION" or "TASK_COMPLETE",\n'
    '  "next_action": {\n'
    '    "action": "tool",\n'
    '    "tool": "python",\n'
    '    "args": ["python code"],\n'
    '    "kwargs": {}\n'
    "  }\n"
    "}\n\n"
    "Rules:\n"
    "1. Use only python, shell, or filesystem_read.\n"
    "2. Do not claim an action was executed unless the observation says so.\n"
    "3. If the task is already completely satisfied, use TASK_COMPLETE.\n"
    "4. If another operation is genuinely needed, use NEXT_ACTION.\n"
    "5. Return JSON only."
)

start = time.time()

qwen_replan_raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": (
                "You are a cautious execution replanner. "
                "Reason from actual observations only."
            )
        },
        {
            "role": "user",
            "content": replan_prompt
        }
    ],
    capability="planning",
    max_new_tokens=256,
    do_sample=False
)

elapsed = time.time() - start

print("Qwen replanning time:", round(elapsed, 3), "sec")

print("\n--- QWEN REPLAN ---")
print(qwen_replan_raw)
print("--- END QWEN REPLAN ---")

assert isinstance(qwen_replan_raw, str)
assert len(qwen_replan_raw.strip()) > 0

# ------------------------------------------------------------
# 6. Parse JSON
# ------------------------------------------------------------

print("\n[6] PARSING REPLAN")
print("-" * 70)

def extract_json_object(text):
    text = text.strip()

    # Direct JSON
    try:
        value = json.loads(text)
        if isinstance(value, dict):
            return value
    except Exception:
        pass

    # Markdown fenced JSON
    fenced = re.search(
        r"```(?:json)?\s*(\{.*?\})\s*```",
        text,
        re.DOTALL | re.IGNORECASE
    )

    if fenced:
        value = json.loads(fenced.group(1))
        if isinstance(value, dict):
            return value

    # First JSON object
    start_idx = text.find("{")
    end_idx = text.rfind("}")

    if start_idx >= 0 and end_idx > start_idx:
        value = json.loads(
            text[start_idx:end_idx + 1]
        )

        if isinstance(value, dict):
            return value

    raise ValueError("No valid JSON object found")

replan = extract_json_object(qwen_replan_raw)

print(json.dumps(
    replan,
    indent=2,
    ensure_ascii=False
))

assert isinstance(replan, dict)

status = replan.get("status")

assert status in {
    "NEXT_ACTION",
    "TASK_COMPLETE"
}

print("\nReplan JSON : PASS")
print("Status      :", status)

# ------------------------------------------------------------
# 7. Validate proposed action
# ------------------------------------------------------------

print("\n[7] ACTION VALIDATION")
print("-" * 70)

if status == "NEXT_ACTION":

    next_action = replan.get("next_action")

    assert isinstance(next_action, dict)

    action = str(
        next_action.get("action", "")
    ).strip().lower()

    tool = str(
        next_action.get("tool", "")
    ).strip().lower()

    args = next_action.get("args", [])
    kwargs = next_action.get("kwargs", {})

    print("Action :", action)
    print("Tool   :", tool)
    print("Args   :", args)
    print("Kwargs :", kwargs)

    assert action == "tool"

    assert tool in {
        "python",
        "shell",
        "filesystem_read"
    }

    assert isinstance(args, list)
    assert isinstance(kwargs, dict)

    print("Action validation : PASS")

else:
    print(
        "Qwen determined that the task is already complete."
    )
    print("Action validation : NOT NEEDED")

# ------------------------------------------------------------
# 8. HARD EXECUTION BOUNDARY
# ------------------------------------------------------------

print("\n[8] EXECUTION BOUNDARY")
print("-" * 70)

print("Qwen proposed :", status)
print("Automatic second execution : NO")
print("Reason : replan validation only")

# ------------------------------------------------------------
# 9. GPU STATE
# ------------------------------------------------------------

print("\n[9] GPU STATE")
print("-" * 70)

try:
    import torch

    for i in range(torch.cuda.device_count()):
        free, total = torch.cuda.mem_get_info(i)

        print(
            f"GPU {i}: "
            f"free={free / 1024**3:.2f} GB / "
            f"total={total / 1024**3:.2f} GB | "
            f"allocated={torch.cuda.memory_allocated(i)/1024**3:.2f} GB | "
            f"reserved={torch.cuda.memory_reserved(i)/1024**3:.2f} GB"
        )
except Exception as e:
    print("GPU diagnostic unavailable:", repr(e))

# ------------------------------------------------------------
# 10. FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-A-R3 RESULT")
print("=" * 70)

print("REAL first execution       : PASS")
print("Observation                : PASS")
print("Observation → Qwen        : PASS")
print("Qwen → structured replan  : PASS")
print("Replan JSON                : PASS")
print("Action validation          : PASS")
print("Second action execution    : NOT YET")

print("\n=== END KAGGLE-032-A-R3 ===")
````

````python
# ============================================================
# KAGGLE-032-B : REPLAN DECISION CONTRACT VALIDATOR
#
# Purpose:
#   Validate Qwen's decision before ANY second execution.
#
# Safety:
#   - No Bootstrap()
#   - No new model
#   - No source modification
#   - No second tool execution
# ============================================================

import json

print("=" * 70)
print("KAGGLE-032-B : DECISION CONTRACT VALIDATOR")
print("=" * 70)

# ------------------------------------------------------------
# 1. Confirm previous R3 result exists
# ------------------------------------------------------------

print("\n[1] PREVIOUS REPLAN")
print("-" * 70)

assert "replan" in globals()
assert isinstance(replan, dict)

print(json.dumps(
    replan,
    indent=2,
    ensure_ascii=False
))

status = replan.get("status")
next_action = replan.get("next_action")

print("\nStatus     :", status)
print("Next action:", next_action)

# ------------------------------------------------------------
# 2. Validate status
# ------------------------------------------------------------

print("\n[2] STATUS CONTRACT")
print("-" * 70)

assert status in {
    "NEXT_ACTION",
    "TASK_COMPLETE"
}

print("Status enum : PASS")

# ------------------------------------------------------------
# 3. Strict decision contract
# ------------------------------------------------------------

print("\n[3] STRICT DECISION CONTRACT")
print("-" * 70)

contract_valid = True
contract_issue = None

if status == "TASK_COMPLETE":

    # A completed task must not simultaneously request
    # another executable action.
    if next_action not in (None, {}):
        contract_valid = False
        contract_issue = (
            "TASK_COMPLETE contains a non-empty next_action"
        )

elif status == "NEXT_ACTION":

    if not isinstance(next_action, dict):
        contract_valid = False
        contract_issue = (
            "NEXT_ACTION requires a next_action object"
        )

    else:

        if next_action.get("action") != "tool":
            contract_valid = False
            contract_issue = (
                "next_action.action must be 'tool'"
            )

        elif next_action.get("tool") not in {
            "python",
            "shell",
            "filesystem_read"
        }:
            contract_valid = False
            contract_issue = (
                "Unsupported tool"
            )

        elif not isinstance(
            next_action.get("args", []),
            list
        ):
            contract_valid = False
            contract_issue = (
                "next_action.args must be a list"
            )

        elif not isinstance(
            next_action.get("kwargs", {}),
            dict
        ):
            contract_valid = False
            contract_issue = (
                "next_action.kwargs must be a dict"
            )

if contract_valid:

    print("Decision contract : PASS")

else:

    print("Decision contract : FAIL")
    print("Issue :", contract_issue)

# ------------------------------------------------------------
# 4. Important diagnostic
# ------------------------------------------------------------

print("\n[4] DIAGNOSTIC")
print("-" * 70)

if status == "TASK_COMPLETE" and next_action:
    print(
        "Qwen produced a contradictory decision:"
    )
    print(
        "TASK_COMPLETE + executable next_action"
    )
    print()
    print(
        "This is a MODEL OUTPUT CONTRACT ISSUE,"
    )
    print(
        "not an ExecutionEngine failure."
    )

elif status == "NEXT_ACTION":
    print(
        "Qwen requested another executable step."
    )

else:
    print(
        "Qwen returned a clean completion decision."
    )

# ------------------------------------------------------------
# 5. Safety boundary
# ------------------------------------------------------------

print("\n[5] EXECUTION SAFETY")
print("-" * 70)

print("Second action executed : NO")
print("Shell executed         : NO")
print("Filesystem accessed    : NO")
print("Model modified         : NO")
print("Source modified        : NO")

# ------------------------------------------------------------
# 6. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-B RESULT")
print("=" * 70)

print("Previous Qwen decision : PASS")
print(
    "Decision contract     :",
    "PASS" if contract_valid else "FAIL"
)
print("Second execution      : NOT PERFORMED")

if not contract_valid:
    print("\nNEXT REQUIRED STEP:")
    print("Fix/normalize contradictory Qwen decisions")
    print("before permitting autonomous execution.")

print("\n=== END KAGGLE-032-B ===")
````

````python
# ============================================================
# KAGGLE-032-C : DECISION NORMALIZER
#
# Goal:
#   Convert imperfect Qwen decisions into one canonical format.
#
# Canonical contract:
#
# TASK_COMPLETE
#   {
#       "status": "TASK_COMPLETE",
#       "next_action": None
#   }
#
# NEXT_ACTION
#   {
#       "status": "NEXT_ACTION",
#       "next_action": {
#           "action": "tool",
#           "tool": "...",
#           "args": [...],
#           "kwargs": {}
#       }
#   }
#
# Safety:
#   NO execution
#   NO Bootstrap
#   NO model loading
#   NO source modification
# ============================================================

import json

print("=" * 70)
print("KAGGLE-032-C : DECISION NORMALIZER")
print("=" * 70)

# ------------------------------------------------------------
# 1. Get previous Qwen decision
# ------------------------------------------------------------

print("\n[1] RAW QWEN DECISION")
print("-" * 70)

assert "replan" in globals()
assert isinstance(replan, dict)

print(json.dumps(
    replan,
    indent=2,
    ensure_ascii=False
))

# ------------------------------------------------------------
# 2. Normalizer
# ------------------------------------------------------------

print("\n[2] NORMALIZATION")
print("-" * 70)

def normalize_decision(decision):

    status = decision.get("status")
    next_action = decision.get("next_action")

    # --------------------------------------------------------
    # Case A:
    # Qwen says TASK_COMPLETE but also provides an action.
    #
    # For this milestone we interpret the explicit action as
    # evidence that another step is intended.
    # --------------------------------------------------------

    if status == "TASK_COMPLETE" and next_action:

        normalized_action = dict(next_action)

        # Qwen sometimes invents semantic action names such as
        # "verify_result". V3 execution requires:
        #
        # action == "tool"
        #
        if normalized_action.get("action") != "tool":

            normalized_action["action"] = "tool"

        if "kwargs" not in normalized_action:
            normalized_action["kwargs"] = {}

        return {
            "status": "NEXT_ACTION",
            "next_action": normalized_action,
            "normalization": {
                "reason": (
                    "TASK_COMPLETE contained an explicit "
                    "next_action; interpreted as pending action."
                ),
                "original_status": status
            }
        }

    # --------------------------------------------------------
    # Case B:
    # Clean TASK_COMPLETE
    # --------------------------------------------------------

    if status == "TASK_COMPLETE":

        return {
            "status": "TASK_COMPLETE",
            "next_action": None,
            "normalization": {
                "reason": "Clean completion decision."
            }
        }

    # --------------------------------------------------------
    # Case C:
    # NEXT_ACTION
    # --------------------------------------------------------

    if status == "NEXT_ACTION":

        if not isinstance(next_action, dict):
            raise ValueError(
                "NEXT_ACTION requires next_action object"
            )

        normalized_action = dict(next_action)

        if normalized_action.get("action") != "tool":
            normalized_action["action"] = "tool"

        if "kwargs" not in normalized_action:
            normalized_action["kwargs"] = {}

        return {
            "status": "NEXT_ACTION",
            "next_action": normalized_action,
            "normalization": {
                "reason": "Existing NEXT_ACTION normalized."
            }
        }

    # --------------------------------------------------------
    # Unknown status
    # --------------------------------------------------------

    raise ValueError(
        f"Unsupported decision status: {status!r}"
    )


normalized = normalize_decision(replan)

print(json.dumps(
    normalized,
    indent=2,
    ensure_ascii=False
))

# ------------------------------------------------------------
# 3. Validate normalized status
# ------------------------------------------------------------

print("\n[3] NORMALIZED STATUS")
print("-" * 70)

assert normalized["status"] in {
    "NEXT_ACTION",
    "TASK_COMPLETE"
}

print("Status :", normalized["status"])
print("Status contract : PASS")

# ------------------------------------------------------------
# 4. Validate normalized action
# ------------------------------------------------------------

print("\n[4] NORMALIZED ACTION")
print("-" * 70)

if normalized["status"] == "NEXT_ACTION":

    action = normalized["next_action"]

    print("Action :", action)

    assert isinstance(action, dict)
    assert action.get("action") == "tool"

    assert action.get("tool") in {
        "python",
        "shell",
        "filesystem_read"
    }

    assert isinstance(
        action.get("args", []),
        list
    )

    assert isinstance(
        action.get("kwargs", {}),
        dict
    )

    print("Executable action schema : PASS")

else:

    assert normalized["next_action"] is None

    print("No next action.")
    print("Completion schema : PASS")

# ------------------------------------------------------------
# 5. Check contradiction has been removed
# ------------------------------------------------------------

print("\n[5] CONTRADICTION CHECK")
print("-" * 70)

if normalized["status"] == "TASK_COMPLETE":

    assert normalized["next_action"] is None

else:

    assert normalized["status"] == "NEXT_ACTION"
    assert isinstance(
        normalized["next_action"],
        dict
    )

print("Contradiction removed : PASS")

# ------------------------------------------------------------
# 6. Safety boundary
# ------------------------------------------------------------

print("\n[6] SAFETY")
print("-" * 70)

print("Second action executed : NO")
print("Python executed        : NO")
print("Shell executed         : NO")
print("Filesystem accessed    : NO")
print("Model modified         : NO")
print("Source modified        : NO")

# ------------------------------------------------------------
# 7. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-C RESULT")
print("=" * 70)

print("Raw Qwen decision       : RECEIVED")
print("Decision normalization  : PASS")
print("Canonical status        :", normalized["status"])
print("Canonical action        : PASS")
print("Contradiction removal   : PASS")
print("Second execution        : NOT PERFORMED")

print("\n=== END KAGGLE-032-C ===")
````

````python
# ============================================================
# KAGGLE-032-D : CONTROLLED SECOND ACTION EXECUTION
#
# Goal:
#   Execute ONLY the normalized NEXT_ACTION from 032-C.
#
# Safety:
#   - Reuse existing runtime
#   - No Bootstrap
#   - No model reload
#   - No new model instance
#   - Exactly ONE additional tool execution
#   - Only python is permitted for this milestone
# ============================================================

import json

print("=" * 70)
print("KAGGLE-032-D : CONTROLLED SECOND ACTION EXECUTION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Verify normalized decision exists
# ------------------------------------------------------------

print("\n[1] NORMALIZED DECISION")
print("-" * 70)

assert "normalized" in globals()
assert isinstance(normalized, dict)

print(json.dumps(
    normalized,
    indent=2,
    ensure_ascii=False
))

assert normalized["status"] == "NEXT_ACTION"

action = normalized["next_action"]

assert isinstance(action, dict)
assert action["action"] == "tool"

# ------------------------------------------------------------
# 2. Strict execution allow-list
# ------------------------------------------------------------

print("\n[2] EXECUTION GATE")
print("-" * 70)

allowed_tools = {
    "python"
}

tool_name = action.get("tool")

print("Requested tool :", tool_name)

assert tool_name in allowed_tools, (
    f"Tool {tool_name!r} is not permitted "
    "for KAGGLE-032-D"
)

args = action.get("args", [])
kwargs = action.get("kwargs", {})

assert isinstance(args, list)
assert isinstance(kwargs, dict)

print("Execution gate : PASS")
print("Tool           :", tool_name)
print("Args           :", args)
print("Kwargs         :", kwargs)

# ------------------------------------------------------------
# 3. Verify existing runtime
# ------------------------------------------------------------

print("\n[3] RUNTIME SAFETY")
print("-" * 70)

assert "runtime" in globals()
assert runtime is not None

print("Existing runtime : PASS")
print("Runtime reused   : YES")
print("Bootstrap        : NOT CALLED")
print("Model reload     : NOT REQUESTED")

# ------------------------------------------------------------
# 4. Execute EXACTLY ONE action
# ------------------------------------------------------------

print("\n[4] SECOND ACTION EXECUTION")
print("-" * 70)

tool_registry = runtime.tools

print("Executing tool:", tool_name)

try:
    second_output = tool_registry.call(
        tool_name,
        *args,
        **kwargs
    )

    second_observation = {
        "ok": True,
        "output": second_output,
        "error": None,
        "metadata": {
            "execution_stage": 2,
            "tool": tool_name
        }
    }

except Exception as e:

    second_observation = {
        "ok": False,
        "output": None,
        "error": str(e),
        "metadata": {
            "execution_stage": 2,
            "tool": tool_name
        }
    }

print(json.dumps(
    second_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 5. Observation contract
# ------------------------------------------------------------

print("\n[5] OBSERVATION CONTRACT")
print("-" * 70)

assert "ok" in second_observation
assert "output" in second_observation
assert "error" in second_observation
assert "metadata" in second_observation

assert second_observation["metadata"]["execution_stage"] == 2

print("Observation schema : PASS")

# ------------------------------------------------------------
# 6. Validate expected verification result
# ------------------------------------------------------------

print("\n[6] RESULT VALIDATION")
print("-" * 70)

assert second_observation["ok"] is True

second_output = second_observation["output"]

print("Second execution output:")
print(second_output)

# ToolRegistry python() returns:
# {
#   "returncode": 0,
#   "stdout": "...",
#   "stderr": "..."
# }

assert isinstance(second_output, dict)
assert second_output.get("returncode") == 0

stdout = second_output.get("stdout", "").strip()

assert stdout == "20.0", (
    f"Expected verification output '20.0', got {stdout!r}"
)

print("Verification result : PASS")
print("Verified average    : 20.0")

# ------------------------------------------------------------
# 7. Preserve both observations
# ------------------------------------------------------------

print("\n[7] TWO-STAGE EXECUTION TRACE")
print("-" * 70)

assert "obs" in globals()

first_observation = {
    "ok": obs.ok,
    "output": obs.output,
    "error": obs.error,
    "metadata": obs.metadata
}

execution_trace = {
    "stage_1": first_observation,
    "stage_2": second_observation
}

print(json.dumps(
    execution_trace,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 8. Confirm stage separation
# ------------------------------------------------------------

print("\n[8] STAGE SEPARATION")
print("-" * 70)

assert execution_trace["stage_1"] is not execution_trace["stage_2"]

assert execution_trace["stage_1"]["metadata"] != \
       execution_trace["stage_2"]["metadata"]

print("Stage 1 : initial calculation")
print("Stage 2 : verification")
print("Stage separation : PASS")

# ------------------------------------------------------------
# 9. Safety audit
# ------------------------------------------------------------

print("\n[9] SAFETY AUDIT")
print("-" * 70)

print("Second action executed : YES")
print("Tool executed          :", tool_name)
print("Python executed        : YES")
print("Shell executed         : NO")
print("Filesystem accessed    : NO")
print("New model created      : NO")
print("Bootstrap called       : NO")
print("Source modified        : NO")

# ------------------------------------------------------------
# 10. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-D RESULT")
print("=" * 70)

print("Normalized decision    : PASS")
print("Execution gate         : PASS")
print("Second action          : PASS")
print("Second observation     : PASS")
print("Verification result    : PASS")
print("Two-stage trace        : PASS")

print("\nFIRST RESULT  : 20.0")
print("SECOND RESULT : 20.0")

print("\nKAGGLE-032-D STATUS : PASS")
print("=== END KAGGLE-032-D ===")
````

````python
# ============================================================
# KAGGLE-032-E : FINAL VERIFICATION DECISION
#
# Goal:
#   Feed Observation #2 back to Qwen and require an
#   unambiguous TASK_COMPLETE decision.
#
# Expected:
#
# {
#   "status": "TASK_COMPLETE",
#   "next_action": null
# }
#
# Safety:
#   - Existing runtime only
#   - No Bootstrap
#   - No model reload
#   - NO tool execution
# ============================================================

import json

print("=" * 70)
print("KAGGLE-032-E : FINAL VERIFICATION DECISION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Verify second observation
# ------------------------------------------------------------

print("\n[1] OBSERVATION #2")
print("-" * 70)

assert "second_observation" in globals()
assert isinstance(second_observation, dict)

print(json.dumps(
    second_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

assert second_observation["ok"] is True

# ------------------------------------------------------------
# 2. Build final verification state
# ------------------------------------------------------------

print("\n[2] VERIFICATION STATE")
print("-" * 70)

verification_state = {
    "task": "Calculate the average of 10, 20, and 30, then verify the calculated result.",
    "initial_observation": {
        "ok": first_observation["ok"],
        "output": first_observation["output"],
        "error": first_observation["error"]
    },
    "verification_observation": {
        "ok": second_observation["ok"],
        "output": second_observation["output"],
        "error": second_observation["error"]
    }
}

print(json.dumps(
    verification_state,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 3. Ask Qwen for FINAL decision
# ------------------------------------------------------------

print("\n[3] QWEN FINAL DECISION")
print("-" * 70)

final_prompt = f"""
You are the final verification controller.

Task:
Calculate the average of 10, 20, and 30, then verify the result.

Execution history:
{json.dumps(verification_state, ensure_ascii=False)}

The verification execution succeeded and produced:
20.0

Determine whether the task is now completely verified.

Return ONLY valid JSON.

STRICT CONTRACT:

If the task is complete:
{{
  "status": "TASK_COMPLETE",
  "next_action": null
}}

If verification is genuinely incomplete:
{{
  "status": "NEXT_ACTION",
  "next_action": {{
    "action": "tool",
    "tool": "python",
    "args": ["..."],
    "kwargs": {{}}
  }}
}}

IMPORTANT:
- Do NOT put next_action together with TASK_COMPLETE.
- For TASK_COMPLETE, next_action MUST be null.
- Do NOT invent another action if the task is already verified.
"""

final_raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": (
                "You are a strict autonomous verification controller. "
                "Return only the requested JSON contract."
            )
        },
        {
            "role": "user",
            "content": final_prompt
        }
    ],
    capability="planning",
    max_new_tokens=256,
    do_sample=False
)

print("Raw Qwen output:")
print(final_raw)

# ------------------------------------------------------------
# 4. Parse JSON
# ------------------------------------------------------------

print("\n[4] JSON PARSING")
print("-" * 70)

text = final_raw.strip()

# Handle possible markdown fences defensively
if text.startswith("```"):
    text = text.replace("```json", "").replace("```", "").strip()

final_decision = json.loads(text)

assert isinstance(final_decision, dict)

print(json.dumps(
    final_decision,
    indent=2,
    ensure_ascii=False
))

print("JSON parsing : PASS")

# ------------------------------------------------------------
# 5. Strict final decision contract
# ------------------------------------------------------------

print("\n[5] STRICT FINAL DECISION CONTRACT")
print("-" * 70)

status = final_decision.get("status")
next_action = final_decision.get("next_action")

assert status in {
    "TASK_COMPLETE",
    "NEXT_ACTION"
}

if status == "TASK_COMPLETE":

    assert next_action is None or next_action == {}, (
        "TASK_COMPLETE must have null/empty next_action"
    )

    print("Status       : TASK_COMPLETE")
    print("Next action  : None")
    print("Completion contract : PASS")

else:

    assert isinstance(next_action, dict)

    assert next_action.get("action") == "tool"

    assert next_action.get("tool") in {
        "python",
        "shell",
        "filesystem_read"
    }

    assert isinstance(
        next_action.get("args", []),
        list
    )

    assert isinstance(
        next_action.get("kwargs", {}),
        dict
    )

    print("Status       : NEXT_ACTION")
    print("Next action  :", next_action)
    print("Continuation contract : PASS")

# ------------------------------------------------------------
# 6. Verification result independently checked
# ------------------------------------------------------------

print("\n[6] INDEPENDENT VERIFICATION")
print("-" * 70)

verified_stdout = (
    second_observation["output"]
    .get("stdout", "")
    .strip()
)

assert verified_stdout == "20.0"

print("Expected : 20.0")
print("Observed : 20.0")
print("Independent verification : PASS")

# ------------------------------------------------------------
# 7. Safety boundary
# ------------------------------------------------------------

print("\n[7] SAFETY")
print("-" * 70)

print("Final decision requested : YES")
print("Additional tool executed : NO")
print("Python executed           : NO")
print("Shell executed            : NO")
print("Filesystem accessed       : NO")
print("Bootstrap called          : NO")
print("New model created         : NO")
print("Source modified           : NO")

# ------------------------------------------------------------
# 8. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-032-E RESULT")
print("=" * 70)

print("Observation #2            : PASS")
print("Qwen final decision       : RECEIVED")
print("JSON parsing              : PASS")
print("Decision contract         : PASS")
print("Independent verification  : PASS")
print("Additional execution      : NOT PERFORMED")

if final_decision["status"] == "TASK_COMPLETE":
    print("Final state               : TASK_COMPLETE")
    print("\nKAGGLE-032-E STATUS : PASS")
else:
    print("Final state               : NEXT_ACTION")
    print("\nKAGGLE-032-E STATUS : CONTINUATION REQUIRED")

print("=== END KAGGLE-032-E ===")
````

````python
# ============================================================
# KAGGLE-033-A : CONTROLLED FAILURE INJECTION
#
# Goal:
#   Prove that V3 can safely capture a deterministic tool
#   failure as an Observation.
#
# This milestone DOES NOT:
#   - ask Qwen to repair
#   - retry
#   - execute shell
#   - modify files
#   - modify source
#   - reload the model
#
# Failure used:
#   Python ZeroDivisionError
# ============================================================

import json

print("=" * 70)
print("KAGGLE-033-A : CONTROLLED FAILURE INJECTION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Verify existing runtime
# ------------------------------------------------------------

print("\n[1] RUNTIME")
print("-" * 70)

assert "runtime" in globals()
assert runtime is not None

print("Existing runtime : PASS")
print("Runtime reused   : YES")
print("Bootstrap        : NOT CALLED")
print("Model reload     : NOT REQUESTED")

# ------------------------------------------------------------
# 2. Construct deliberately failing action
# ------------------------------------------------------------

print("\n[2] FAILURE ACTION")
print("-" * 70)

failure_code = """
print(10 / 0)
""".strip()

failure_action = {
    "action": "tool",
    "tool": "python",
    "args": [failure_code],
    "kwargs": {}
}

print(json.dumps(
    failure_action,
    indent=2,
    ensure_ascii=False
))

assert failure_action["action"] == "tool"
assert failure_action["tool"] == "python"

print("Failure action contract : PASS")

# ------------------------------------------------------------
# 3. Execute exactly ONE failing action
# ------------------------------------------------------------

print("\n[3] FAILURE EXECUTION")
print("-" * 70)

try:

    failure_output = runtime.tools.call(
        failure_action["tool"],
        *failure_action["args"],
        **failure_action["kwargs"]
    )

    # If the tool itself returns a non-zero process result,
    # preserve that as an unsuccessful observation.
    returncode = failure_output.get("returncode")

    if returncode == 0:

        failure_observation = {
            "ok": True,
            "output": failure_output,
            "error": None,
            "metadata": {
                "execution_stage": "failure_injection"
            }
        }

    else:

        failure_observation = {
            "ok": False,
            "output": failure_output,
            "error": failure_output.get("stderr"),
            "metadata": {
                "execution_stage": "failure_injection"
            }
        }

except Exception as e:

    failure_observation = {
        "ok": False,
        "output": None,
        "error": str(e),
        "metadata": {
            "execution_stage": "failure_injection"
        }
    }

print(json.dumps(
    failure_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 4. Observation contract
# ------------------------------------------------------------

print("\n[4] OBSERVATION CONTRACT")
print("-" * 70)

assert isinstance(failure_observation, dict)

assert "ok" in failure_observation
assert "output" in failure_observation
assert "error" in failure_observation
assert "metadata" in failure_observation

print("Observation schema : PASS")

# ------------------------------------------------------------
# 5. Failure detection
# ------------------------------------------------------------

print("\n[5] FAILURE DETECTION")
print("-" * 70)

assert failure_observation["ok"] is False

print("ok    :", failure_observation["ok"])
print("error :", failure_observation["error"])

assert (
    "ZeroDivisionError"
    in str(failure_observation["error"])
    or (
        isinstance(failure_observation["output"], dict)
        and "ZeroDivisionError"
        in str(failure_observation["output"])
    )
)

print("Expected failure detected : PASS")

# ------------------------------------------------------------
# 6. Confirm no accidental success
# ------------------------------------------------------------

print("\n[6] FAILURE INTEGRITY")
print("-" * 70)

assert failure_observation["ok"] is False

if isinstance(failure_observation["output"], dict):

    rc = failure_observation["output"].get("returncode")

    if rc is not None:
        assert rc != 0

print("Execution classified as failure : PASS")

# ------------------------------------------------------------
# 7. Safety boundary
# ------------------------------------------------------------

print("\n[7] SAFETY AUDIT")
print("-" * 70)

print("Failure action executed : YES")
print("Number of executions    : 1")
print("Python executed         : YES")
print("Shell executed          : NO")
print("Filesystem accessed     : NO")
print("Network accessed        : NO")
print("Retry performed         : NO")
print("Qwen repair requested   : NO")
print("Model modified          : NO")
print("Bootstrap called        : NO")
print("Source modified         : NO")

# ------------------------------------------------------------
# 8. Save globally for next milestone
# ------------------------------------------------------------

failure_state = {
    "action": failure_action,
    "observation": failure_observation
}

print("\n[8] FAILURE STATE")
print("-" * 70)

print(json.dumps(
    failure_state,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 9. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-033-A RESULT")
print("=" * 70)

print("Controlled failure        : PASS")
print("Observation generation    : PASS")
print("Failure detection         : PASS")
print("Failure state preserved   : PASS")
print("Retry                     : NOT PERFORMED")
print("Repair                    : NOT PERFORMED")

print("\nKAGGLE-033-A STATUS : PASS")
print("=== END KAGGLE-033-A ===")
````

````python
# ============================================================
# KAGGLE-033-B : FAILURE DIAGNOSIS + REPAIR PROPOSAL
#
# Goal:
#   Feed the real failure Observation to Qwen.
#
# Qwen must:
#   1. diagnose the failure
#   2. identify the cause
#   3. propose a repair
#
# Safety:
#   - NO repair execution
#   - NO retry
#   - NO shell
#   - NO filesystem modification
#   - NO Bootstrap
#   - NO model reload
# ============================================================

import json

print("=" * 70)
print("KAGGLE-033-B : FAILURE DIAGNOSIS + REPAIR PROPOSAL")
print("=" * 70)

# ------------------------------------------------------------
# 1. Verify failure state
# ------------------------------------------------------------

print("\n[1] FAILURE OBSERVATION")
print("-" * 70)

assert "failure_state" in globals()
assert isinstance(failure_state, dict)

failure_action = failure_state["action"]
failure_observation = failure_state["observation"]

print(json.dumps(
    failure_state,
    indent=2,
    ensure_ascii=False,
    default=str
))

assert failure_observation["ok"] is False

print("Failure observation : PASS")

# ------------------------------------------------------------
# 2. Build diagnosis state
# ------------------------------------------------------------

print("\n[2] DIAGNOSIS INPUT")
print("-" * 70)

diagnosis_input = {
    "original_action": failure_action,
    "observation": failure_observation
}

print(json.dumps(
    diagnosis_input,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 3. Ask Qwen to diagnose and repair
# ------------------------------------------------------------

print("\n[3] QWEN DIAGNOSIS")
print("-" * 70)

diagnosis_prompt = f"""
You are an autonomous failure-diagnosis controller.

A tool execution failed.

Original action:
{json.dumps(failure_action, ensure_ascii=False)}

Observed result:
{json.dumps(failure_observation, ensure_ascii=False)}

Analyze the failure and propose ONE safe repair.

For this test:
- Only the python tool may be proposed.
- Do not propose shell.
- Do not propose filesystem access.
- Do not propose network access.
- The repaired action must be executable Python.
- The repaired action should address the actual failure.
- Do not merely repeat the failing action.

Return ONLY valid JSON using exactly this structure:

{{
  "status": "REPAIR_REQUIRED",
  "diagnosis": {{
    "failure_type": "...",
    "cause": "...",
    "explanation": "..."
  }},
  "repair_action": {{
    "action": "tool",
    "tool": "python",
    "args": ["..."],
    "kwargs": {{}}
  }}
}}

If and only if the failure cannot safely be repaired, return:

{{
  "status": "REPAIR_UNSAFE",
  "diagnosis": {{
    "failure_type": "...",
    "cause": "...",
    "explanation": "..."
  }},
  "repair_action": null
}}
"""

diagnosis_raw = runtime.models.generate(
    [
        {
            "role": "system",
            "content": (
                "You are a strict failure diagnosis and repair planner. "
                "Return only valid JSON."
            )
        },
        {
            "role": "user",
            "content": diagnosis_prompt
        }
    ],
    capability="planning",
    max_new_tokens=512,
    do_sample=False
)

print("Raw Qwen output:")
print(diagnosis_raw)

# ------------------------------------------------------------
# 4. Parse JSON
# ------------------------------------------------------------

print("\n[4] JSON PARSING")
print("-" * 70)

text = diagnosis_raw.strip()

if text.startswith("```"):
    text = (
        text
        .replace("```json", "")
        .replace("```", "")
        .strip()
    )

diagnosis = json.loads(text)

assert isinstance(diagnosis, dict)

print(json.dumps(
    diagnosis,
    indent=2,
    ensure_ascii=False
))

print("JSON parsing : PASS")

# ------------------------------------------------------------
# 5. Status contract
# ------------------------------------------------------------

print("\n[5] STATUS CONTRACT")
print("-" * 70)

status = diagnosis.get("status")

assert status in {
    "REPAIR_REQUIRED",
    "REPAIR_UNSAFE"
}

print("Status :", status)
print("Status contract : PASS")

# ------------------------------------------------------------
# 6. Diagnosis contract
# ------------------------------------------------------------

print("\n[6] DIAGNOSIS CONTRACT")
print("-" * 70)

diag = diagnosis.get("diagnosis")

assert isinstance(diag, dict)

assert isinstance(
    diag.get("failure_type"),
    str
)

assert isinstance(
    diag.get("cause"),
    str
)

assert isinstance(
    diag.get("explanation"),
    str
)

print("Failure type :", diag["failure_type"])
print("Cause        :", diag["cause"])
print("Explanation  :", diag["explanation"])

print("Diagnosis contract : PASS")

# ------------------------------------------------------------
# 7. Repair contract
# ------------------------------------------------------------

print("\n[7] REPAIR CONTRACT")
print("-" * 70)

repair_action = diagnosis.get("repair_action")

if status == "REPAIR_REQUIRED":

    assert isinstance(repair_action, dict)

    assert repair_action.get("action") == "tool"

    assert repair_action.get("tool") == "python"

    assert isinstance(
        repair_action.get("args"),
        list
    )

    assert len(repair_action["args"]) >= 1

    assert isinstance(
        repair_action.get("kwargs", {}),
        dict
    )

    repair_code = repair_action["args"][0]

    assert isinstance(repair_code, str)
    assert len(repair_code.strip()) > 0

    # Prevent the model from simply repeating the original
    # failing action.
    assert repair_code.strip() != \
        failure_action["args"][0].strip()

    print("Repair action :", repair_action)
    print("Repair contract : PASS")

else:

    assert repair_action is None

    print("Qwen declared repair unsafe.")
    print("Repair contract : PASS")

# ------------------------------------------------------------
# 8. Check diagnosis against known failure
# ------------------------------------------------------------

print("\n[8] FAILURE UNDERSTANDING")
print("-" * 70)

known_error = str(
    failure_observation.get("error", "")
)

assert (
    "ZeroDivisionError" in known_error
    or "division by zero" in known_error
)

# Qwen's diagnosis should meaningfully mention the
# division-by-zero problem.

diagnosis_text = (
    diag["failure_type"] + " " +
    diag["cause"] + " " +
    diag["explanation"]
).lower()

assert (
    "zero" in diagnosis_text
    or "division" in diagnosis_text
    or "divide" in diagnosis_text
)

print("Actual failure : ZeroDivisionError")
print("Qwen diagnosis : relevant")
print("Failure understanding : PASS")

# ------------------------------------------------------------
# 9. Safety boundary
# ------------------------------------------------------------

print("\n[9] SAFETY AUDIT")
print("-" * 70)

print("Failure analyzed       : YES")
print("Qwen diagnosis         : YES")
print("Repair proposed        :",
      "YES" if status == "REPAIR_REQUIRED" else "NO")
print("Repair executed        : NO")
print("Retry performed        : NO")
print("Shell executed         : NO")
print("Filesystem accessed    : NO")
print("Network accessed       : NO")
print("Model modified         : NO")
print("Bootstrap called       : NO")
print("Source modified        : NO")

# ------------------------------------------------------------
# 10. Save for next milestone
# ------------------------------------------------------------

failure_diagnosis_state = {
    "failure": failure_state,
    "diagnosis": diagnosis
}

print("\n[10] FAILURE DIAGNOSIS STATE")
print("-" * 70)

print(json.dumps(
    failure_diagnosis_state,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 11. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-033-B RESULT")
print("=" * 70)

print("Failure observation      : PASS")
print("Qwen diagnosis           : PASS")
print("Diagnosis contract       : PASS")
print("Repair proposal contract : PASS")
print("Failure understanding    : PASS")
print("Repair execution         : NOT PERFORMED")
print("Retry                    : NOT PERFORMED")

if status == "REPAIR_REQUIRED":
    print("\nQwen proposed a repair.")
else:
    print("\nQwen classified repair as unsafe.")

print("\nKAGGLE-033-B STATUS : PASS")
print("=== END KAGGLE-033-B ===")
````

````python
# ============================================================
# KAGGLE-033-D : CONTROLLED REPAIR EXECUTION + RECOVERY
#
# Goal:
#   Execute the safety-approved repair exactly once.
#   Capture the resulting Observation.
#   Verify recovery.
#
# Safety:
#   - Existing runtime only
#   - Python-only
#   - Exactly ONE repair execution
#   - No shell
#   - No filesystem
#   - No network
#   - No Bootstrap
#   - No model reload
#   - No automatic second retry
# ============================================================

import json

print("=" * 70)
print("KAGGLE-033-D : CONTROLLED REPAIR EXECUTION + RECOVERY")
print("=" * 70)

# ------------------------------------------------------------
# 1. Verify approved repair
# ------------------------------------------------------------

print("\n[1] APPROVED REPAIR")
print("-" * 70)

assert "approved_repair" in globals()
assert isinstance(approved_repair, dict)

assert approved_repair["status"] == "REPAIR_APPROVED"

repair = approved_repair["repair_action"]

assert isinstance(repair, dict)
assert repair["action"] == "tool"
assert repair["tool"] == "python"

repair_args = repair.get("args", [])
repair_kwargs = repair.get("kwargs", {})

assert isinstance(repair_args, list)
assert len(repair_args) >= 1
assert isinstance(repair_kwargs, dict)

print(json.dumps(
    approved_repair,
    indent=2,
    ensure_ascii=False
))

print("Approved repair : PASS")

# ------------------------------------------------------------
# 2. Re-check authorization immediately before execution
# ------------------------------------------------------------

print("\n[2] FINAL EXECUTION GATE")
print("-" * 70)

assert approved_repair["authorization"]["tool_allowed"] is True
assert approved_repair["authorization"]["schema_valid"] is True
assert approved_repair["authorization"]["different_from_failure"] is True
assert approved_repair["authorization"]["addresses_known_failure"] is True
assert approved_repair["authorization"]["static_safety_check"] is True

assert repair["tool"] == "python"

print("Tool allow-list          : PASS")
print("Schema validation        : PASS")
print("Repair difference       : PASS")
print("Failure-specific repair : PASS")
print("Static safety            : PASS")
print("FINAL EXECUTION GATE    : PASS")

# ------------------------------------------------------------
# 3. Verify existing runtime
# ------------------------------------------------------------

print("\n[3] RUNTIME SAFETY")
print("-" * 70)

assert "runtime" in globals()
assert runtime is not None

print("Existing runtime : PASS")
print("Runtime reused   : YES")
print("Bootstrap        : NOT CALLED")
print("Model reload     : NOT REQUESTED")

# ------------------------------------------------------------
# 4. Execute exactly ONE repair
# ------------------------------------------------------------

print("\n[4] REPAIR EXECUTION")
print("-" * 70)

print("Executing approved repair:")
print(repair_args[0])

try:

    repair_output = runtime.tools.call(
        repair["tool"],
        *repair_args,
        **repair_kwargs
    )

    returncode = repair_output.get("returncode")

    if returncode == 0:

        repair_observation = {
            "ok": True,
            "output": repair_output,
            "error": None,
            "metadata": {
                "execution_stage": "repair",
                "tool": repair["tool"]
            }
        }

    else:

        repair_observation = {
            "ok": False,
            "output": repair_output,
            "error": repair_output.get("stderr"),
            "metadata": {
                "execution_stage": "repair",
                "tool": repair["tool"]
            }
        }

except Exception as e:

    repair_observation = {
        "ok": False,
        "output": None,
        "error": str(e),
        "metadata": {
            "execution_stage": "repair",
            "tool": repair["tool"]
        }
    }

print(json.dumps(
    repair_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 5. Observation contract
# ------------------------------------------------------------

print("\n[5] REPAIR OBSERVATION CONTRACT")
print("-" * 70)

assert isinstance(repair_observation, dict)

assert "ok" in repair_observation
assert "output" in repair_observation
assert "error" in repair_observation
assert "metadata" in repair_observation

print("Observation schema : PASS")

# ------------------------------------------------------------
# 6. Recovery validation
# ------------------------------------------------------------

print("\n[6] RECOVERY VALIDATION")
print("-" * 70)

assert repair_observation["ok"] is True

output = repair_observation["output"]

assert isinstance(output, dict)

assert output.get("returncode") == 0

stdout = output.get("stdout", "").strip()

print("Return code :", output.get("returncode"))
print("Stdout      :", repr(stdout))
print("Stderr      :", repr(output.get("stderr", "")))

assert stdout == "5.0"

print("Repair result : 5.0")
print("Recovery validation : PASS")

# ------------------------------------------------------------
# 7. Compare failure → repair
# ------------------------------------------------------------

print("\n[7] FAILURE → REPAIR TRACE")
print("-" * 70)

failed_code = failure_state["action"]["args"][0]

print("FAILED:")
print(failed_code)

print("\nREPAIRED:")
print(repair_args[0])

print("\nFAILURE OBSERVATION:")
print(json.dumps(
    failure_state["observation"],
    indent=2,
    ensure_ascii=False
))

print("\nREPAIR OBSERVATION:")
print(json.dumps(
    repair_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

assert failure_state["observation"]["ok"] is False
assert repair_observation["ok"] is True

print("\nFailure → recovery transition : PASS")

# ------------------------------------------------------------
# 8. Build complete recovery trace
# ------------------------------------------------------------

print("\n[8] COMPLETE RECOVERY TRACE")
print("-" * 70)

recovery_trace = {
    "failure": {
        "action": failure_state["action"],
        "observation": failure_state["observation"]
    },
    "diagnosis": diagnosis,
    "approved_repair": approved_repair,
    "repair_observation": repair_observation
}

print(json.dumps(
    recovery_trace,
    indent=2,
    ensure_ascii=False,
    default=str
))

# ------------------------------------------------------------
# 9. Safety audit
# ------------------------------------------------------------

print("\n[9] SAFETY AUDIT")
print("-" * 70)

print("Original failure executed : YES")
print("Repair authorized         : YES")
print("Repair executed           : YES")
print("Repair execution count    : 1")
print("Additional retry          : NO")
print("Shell executed            : NO")
print("Filesystem accessed       : NO")
print("Network accessed          : NO")
print("Bootstrap called          : NO")
print("New model created         : NO")
print("Source modified           : NO")

# ------------------------------------------------------------
# 10. Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-033-D RESULT")
print("=" * 70)

print("Approved repair           : PASS")
print("Final execution gate      : PASS")
print("Repair execution          : PASS")
print("Repair observation        : PASS")
print("Recovery validation       : PASS")
print("Failure → recovery trace : PASS")
print("Additional retry          : NOT PERFORMED")

print("\nFAILED RESULT  : ZeroDivisionError")
print("REPAIRED RESULT: 5.0")

print("\nKAGGLE-033-D STATUS : PASS")
print("=== END KAGGLE-033-D ===")
````

````python
# ============================================================
# KAGGLE-033-C-REV2
# Controlled Repair Safety Gate
# FIX: nested diagnosis.repair_action
# ============================================================

import json

print("=" * 70)
print("KAGGLE-033-C-REV2 : REPAIR SAFETY GATE")
print("=" * 70)

# ------------------------------------------------------------
# [1] FAILURE DIAGNOSIS STATE
# ------------------------------------------------------------
print("\n[1] FAILURE DIAGNOSIS STATE")
print("-" * 70)

assert "failure_diagnosis_state" in globals(), (
    "failure_diagnosis_state is missing. "
    "Run KAGGLE-033-B first."
)

fd = failure_diagnosis_state

assert isinstance(fd, dict)

print(json.dumps(fd, indent=2, ensure_ascii=False, default=str))

# ------------------------------------------------------------
# [2] EXTRACT FAILURE
# ------------------------------------------------------------
print("\n[2] FAILURE EXTRACTION")
print("-" * 70)

failure = fd.get("failure")
assert isinstance(failure, dict)

failure_action = failure.get("action")
failure_observation = failure.get("observation")

assert isinstance(failure_action, dict)
assert isinstance(failure_observation, dict)

print("Failure action :")
print(json.dumps(
    failure_action,
    indent=2,
    ensure_ascii=False,
    default=str
))

print("\nFailure observation :")
print(json.dumps(
    failure_observation,
    indent=2,
    ensure_ascii=False,
    default=str
))

print("Failure extraction : PASS")

# ------------------------------------------------------------
# [3] EXTRACT DIAGNOSIS
# ------------------------------------------------------------
print("\n[3] DIAGNOSIS EXTRACTION")
print("-" * 70)

diagnosis_state = fd.get("diagnosis")

assert isinstance(diagnosis_state, dict)

status = diagnosis_state.get("status")
diagnosis = diagnosis_state.get("diagnosis")
repair = diagnosis_state.get("repair_action")

print("Diagnosis status :", status)

print("\nDiagnosis:")
print(json.dumps(
    diagnosis,
    indent=2,
    ensure_ascii=False,
    default=str
))

print("\nRepair action:")
print(json.dumps(
    repair,
    indent=2,
    ensure_ascii=False,
    default=str
))

assert status == "REPAIR_REQUIRED"
assert isinstance(diagnosis, dict)
assert isinstance(repair, dict)

print("\nDiagnosis extraction : PASS")

# ------------------------------------------------------------
# [4] REPAIR SCHEMA
# ------------------------------------------------------------
print("\n[4] REPAIR SCHEMA")
print("-" * 70)

action = repair.get("action")
tool = repair.get("tool")
args = repair.get("args")
kwargs = repair.get("kwargs", {})

assert action == "tool"
assert tool == "python"
assert isinstance(args, list)
assert len(args) >= 1
assert isinstance(kwargs, dict)

repair_code = str(args[0])

print("Action :", action)
print("Tool   :", tool)
print("Args   :", args)
print("Kwargs :", kwargs)

print("\nRepair schema : PASS")

# ------------------------------------------------------------
# [5] FAILURE DIFFERENCE CHECK
# ------------------------------------------------------------
print("\n[5] FAILURE DIFFERENCE CHECK")
print("-" * 70)

failed_args = failure_action.get("args", [])

assert isinstance(failed_args, list)

failed_code = (
    str(failed_args[0])
    if failed_args
    else ""
)

print("Failed code :")
print(failed_code)

print("\nRepair code :")
print(repair_code)

assert repair_code.strip() != failed_code.strip()

print("\nRepair differs from failed action : PASS")

# ------------------------------------------------------------
# [6] FAILURE-SPECIFIC CHECK
# ------------------------------------------------------------
print("\n[6] FAILURE-SPECIFIC CHECK")
print("-" * 70)

failure_text = json.dumps(
    failure_diagnosis_state,
    ensure_ascii=False,
    default=str
)

assert (
    "ZeroDivisionError" in failure_text
    or "division by zero" in failure_text.lower()
)

print("Known failure : ZeroDivisionError")
print("Failure-specific diagnosis : PASS")

# ------------------------------------------------------------
# [7] REPAIR CORRECTION CHECK
# ------------------------------------------------------------
print("\n[7] REPAIR CORRECTION CHECK")
print("-" * 70)

normalized_repair = (
    repair_code
    .replace(" ", "")
    .replace("\n", "")
)

assert "10/0" not in normalized_repair
assert "10/2" in normalized_repair

print("Original dangerous expression : 10 / 0")
print("Approved safe expression      : 10 / 2")
print("Correction check : PASS")

# ------------------------------------------------------------
# [8] STATIC SAFETY GATE
# ------------------------------------------------------------
print("\n[8] STATIC SAFETY GATE")
print("-" * 70)

forbidden_patterns = [
    "import os",
    "import subprocess",
    "import socket",
    "import requests",
    "open(",
    "__import__",
    "eval(",
    "exec(",
]

code_lower = repair_code.lower()

blocked = [
    pattern
    for pattern in forbidden_patterns
    if pattern.lower() in code_lower
]

assert not blocked, (
    f"Forbidden pattern detected: {blocked}"
)

print("Forbidden-pattern scan : PASS")

# ------------------------------------------------------------
# [9] AUTHORIZATION
# ------------------------------------------------------------
print("\n[9] AUTHORIZATION")
print("-" * 70)

authorization = {
    "schema_valid": True,
    "tool_allowed": True,
    "different_from_failure": True,
    "addresses_known_failure": True,
    "static_safety_check": True,
}

assert all(authorization.values())

print(json.dumps(
    authorization,
    indent=2
))

print("Repair authorization : PASS")

# ------------------------------------------------------------
# [10] CREATE APPROVED REPAIR
# ------------------------------------------------------------
print("\n[10] APPROVED REPAIR")
print("-" * 70)

approved_repair = {
    "status": "REPAIR_APPROVED",

    "repair_action": {
        "action": "tool",
        "tool": "python",
        "args": args,
        "kwargs": kwargs,
    },

    "authorization": authorization,

    "normalization": {
        "source": "KAGGLE-033-B",
        "source_path": "diagnosis.repair_action",
        "controlled_milestone": True,
    },
}

assert isinstance(approved_repair, dict)
assert approved_repair["status"] == "REPAIR_APPROVED"

approved_action = approved_repair["repair_action"]

assert approved_action["action"] == "tool"
assert approved_action["tool"] == "python"
assert isinstance(approved_action["args"], list)
assert isinstance(approved_action["kwargs"], dict)

print(json.dumps(
    approved_repair,
    indent=2,
    ensure_ascii=False,
    default=str
))

print("\napproved_repair variable : CREATED")
print("Execution performed      : NO")
print("Repair authorized        : YES")

print("\n" + "=" * 70)
print("KAGGLE-033-C-REV2 STATUS : PASS")
print("=" * 70)
````

````python
# ================================================================
# KAGGLE-033-D : CONTROLLED REPAIR EXECUTION
# ================================================================

print("=" * 70)
print("KAGGLE-033-D : CONTROLLED REPAIR EXECUTION")
print("=" * 70)

# ------------------------------------------------
# [1] APPROVED REPAIR CHECK
# ------------------------------------------------
assert "approved_repair" in globals(), \
    "approved_repair missing — run KAGGLE-033-C-REV2 first."

ar = approved_repair

assert ar["status"] == "REPAIR_APPROVED"

repair_action = ar["repair_action"]

assert repair_action["action"] == "tool"
assert repair_action["tool"] == "python"
assert isinstance(repair_action["args"], list)
assert isinstance(repair_action["kwargs"], dict)

repair_code = repair_action["args"][0]

print("\n[1] APPROVED REPAIR")
print("-" * 70)
print(repair_code)
print("Authorization : PASS")


# ------------------------------------------------
# [2] SAFETY RE-CHECK
# ------------------------------------------------
assert "10 / 0" not in repair_code.replace(" ", "")
assert "10 / 2" in repair_code.replace(" ", "")

forbidden = [
    "import os",
    "import subprocess",
    "import socket",
    "import requests",
    "open(",
    "__import__",
    "eval(",
    "exec("
]

for pattern in forbidden:
    assert pattern not in repair_code, \
        f"Forbidden pattern detected: {pattern}"

print("\n[2] SAFETY RE-CHECK")
print("-" * 70)
print("Forbidden-pattern scan : PASS")
print("Dangerous expression   : ABSENT")
print("Approved correction    : PRESENT")


# ------------------------------------------------
# [3] EXECUTE THROUGH EXISTING V3 RUNTIME
# ------------------------------------------------
assert "runtime" in globals(), \
    "Existing V3 runtime not found."

executor = runtime.execution

# Execute ONLY the authorized action.
repair_observations = executor.execute_plan(
    type("ApprovedRepairPlan", (), {
        "steps": [repair_action]
    })()
)

assert repair_observations, "No observation returned."

obs = repair_observations[0]

print("\n[3] REPAIR EXECUTION")
print("-" * 70)

if hasattr(obs, "__dict__"):
    print(obs.__dict__)
else:
    print(obs)


# ------------------------------------------------
# [4] NORMALIZE OBSERVATION
# ------------------------------------------------
if hasattr(obs, "ok"):
    repair_ok = bool(obs.ok)
    repair_output = getattr(obs, "output", None)
    repair_error = getattr(obs, "error", None)
else:
    repair_ok = bool(getattr(obs, "success", False))
    repair_output = getattr(obs, "output", None)
    repair_error = getattr(obs, "error", None)

assert repair_ok is True, \
    f"Approved repair execution failed: {repair_error}"

print("\nRepair execution : PASS")


# ------------------------------------------------
# [5] RESULT VALIDATION
# ------------------------------------------------
result_text = str(repair_output)

assert "5.0" in result_text, \
    f"Expected repaired result 5.0, got: {result_text}"

assert "10 / 0" not in result_text.replace(" ", "")

print("\n[4] RESULT VALIDATION")
print("-" * 70)
print("Expected result : 5.0")
print("Actual result   :", result_text.strip())
print("Result check    : PASS")


# ------------------------------------------------
# [6] CREATE REPAIR TRACE
# ------------------------------------------------
repair_execution_state = {
    "status": "REPAIR_EXECUTED",
    "failed_action": {
        "action": "tool",
        "tool": "python",
        "args": ["print(10 / 0)"],
        "kwargs": {}
    },
    "approved_repair": repair_action,
    "observation": {
        "ok": repair_ok,
        "output": repair_output,
        "error": repair_error
    },
    "verification": {
        "expected": "5.0",
        "actual": result_text.strip(),
        "success": True
    }
}

print("\n[5] REPAIR TRACE")
print("-" * 70)
print(repair_execution_state)


# ------------------------------------------------
# [7] FINAL SAFETY AUDIT
# ------------------------------------------------
print("\n[6] SAFETY AUDIT")
print("-" * 70)
print("New Bootstrap       : NO")
print("New Qwen load       : NO")
print("Shell execution     : NO")
print("Filesystem access   : NO")
print("Network access      : NO")
print("Source modification : NO")
print("Only approved tool  : YES")


# ------------------------------------------------
# FINAL
# ------------------------------------------------
print("\n" + "=" * 70)
print("KAGGLE-033-D STATUS : PASS")
print("=" * 70)
````

````python
import signal
import time
import torch
from transformers import AutoModelForCausalLM

MODEL_PATH = "/kaggle/working/qwen3-8b-awq"

def timeout_handler(signum, frame):
    raise TimeoutError("MODEL LOAD exceeded 120 seconds")

signal.signal(signal.SIGALRM, timeout_handler)
signal.alarm(120)

print("LOAD TEST START", flush=True)
print("Torch:", torch.__version__, flush=True)
print("CUDA:", torch.cuda.is_available(), flush=True)

start = time.time()

try:
    print("Calling from_pretrained()...", flush=True)

    model = AutoModelForCausalLM.from_pretrained(
        MODEL_PATH,
        local_files_only=True,
        device_map="auto",
        torch_dtype="auto",
    )

    signal.alarm(0)

    print("MODEL LOAD PASS", flush=True)
    print("Model:", type(model).__name__, flush=True)
    print("Time:", round(time.time() - start, 1), "seconds", flush=True)
    print("Device map:", getattr(model, "hf_device_map", None), flush=True)

except Exception as e:
    signal.alarm(0)

    print("MODEL LOAD FAILED", flush=True)
    print("Exception:", type(e).__name__, flush=True)
    print("Message:", str(e), flush=True)

    for i in range(torch.cuda.device_count()):
        print(
            f"GPU {i}: "
            f"{torch.cuda.memory_allocated(i)/1024**3:.2f} GB allocated / "
            f"{torch.cuda.memory_reserved(i)/1024**3:.2f} GB reserved",
            flush=True
        )
````

````python
import torch
import time
from transformers import AutoTokenizer

MODEL_PATH = "/kaggle/working/qwen3-8b-awq"

print("=== QWEN3 FIRST INFERENCE ===", flush=True)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_PATH,
    local_files_only=True
)

messages = [
    {
        "role": "user",
        "content": "Explain in 3 simple points what an AI agent is."
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

# Put inputs on the device containing the input embeddings
input_device = next(model.parameters()).device
inputs = {k: v.to(input_device) for k, v in inputs.items()}

print("Input device:", input_device, flush=True)
print("Input tokens:", inputs["input_ids"].shape[-1], flush=True)

start = time.time()

with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=128,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
    )

elapsed = time.time() - start

new_tokens = outputs[0][inputs["input_ids"].shape[-1]:]

answer = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print("\n=== MODEL RESPONSE ===")
print(answer)

print("\n=== INFERENCE METRICS ===")
print("Generation time:", round(elapsed, 2), "sec")
print("Generated tokens:", len(new_tokens))
print(
    "Approx tokens/sec:",
    round(len(new_tokens) / elapsed, 2)
)

print("\n=== QWEN3 INFERENCE TEST COMPLETE ===")
````

````python
# ============================================================
# KAGGLE-038
# QWEN3-8B-AWQ — FIRST REAL INFERENCE TEST
# Purpose:
#   Prompt → Tokenizer → Loaded Model → Generation → Decode
# ============================================================

import time
import torch

print("=" * 70)
print("KAGGLE-038 — FIRST REAL QWEN3 INFERENCE TEST")
print("=" * 70, flush=True)

# ------------------------------------------------------------
# STEP 1 — Verify existing objects
# ------------------------------------------------------------

assert "model" in globals(), "ERROR: model is not loaded in memory."
assert "tokenizer" in globals(), "ERROR: tokenizer is not loaded in memory."

print("\n[038.1] Model object: PASS")
print("[038.2] Tokenizer object: PASS")

# ------------------------------------------------------------
# STEP 2 — Test prompt
# ------------------------------------------------------------

messages = [
    {
        "role": "user",
        "content": (
            "Explain in exactly 3 simple points what an AI agent is."
        )
    }
]

print("\n[038.3] Prompt prepared: PASS")

# ------------------------------------------------------------
# STEP 3 — Apply Qwen chat template
# ------------------------------------------------------------

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

print("[038.4] Chat template: PASS")

# ------------------------------------------------------------
# STEP 4 — Tokenize
# ------------------------------------------------------------

inputs = tokenizer(
    text,
    return_tensors="pt"
)

# Important:
# Input must start on the same device as embedding layer.
input_device = model.get_input_embeddings().weight.device

inputs = {
    key: value.to(input_device)
    for key, value in inputs.items()
}

print("[038.5] Tokenization: PASS")
print("Input device:", input_device)
print("Input tokens:", inputs["input_ids"].shape[-1])

# ------------------------------------------------------------
# STEP 5 — Real generation
# ------------------------------------------------------------

print("\n[038.6] Starting generation...", flush=True)

start_time = time.time()

with torch.inference_mode():

    outputs = model.generate(
        **inputs,

        max_new_tokens=128,

        # Explicit sampling settings
        # Avoid conflict with Qwen generation_config defaults
        do_sample=True,
        temperature=0.6,
        top_p=0.95,
        top_k=20,

        pad_token_id=tokenizer.eos_token_id,
    )

elapsed = time.time() - start_time

print("[038.7] Generation completed: PASS")

# ------------------------------------------------------------
# STEP 6 — Extract only generated tokens
# ------------------------------------------------------------

input_length = inputs["input_ids"].shape[-1]

generated_tokens = outputs[0][input_length:]

answer = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True,
)

print("\n" + "=" * 70)
print("QWEN3 RESPONSE")
print("=" * 70)

print(answer)

# ------------------------------------------------------------
# STEP 7 — Metrics
# ------------------------------------------------------------

generated_count = len(generated_tokens)

tokens_per_second = (
    generated_count / elapsed
    if elapsed > 0
    else 0
)

print("\n" + "=" * 70)
print("KAGGLE-038 METRICS")
print("=" * 70)

print("Generation time:", round(elapsed, 2), "seconds")
print("Generated tokens:", generated_count)
print("Approx tokens/sec:", round(tokens_per_second, 2))

# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-038 STATUS : PASS")
print("Qwen3-8B-AWQ real inference is operational.")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-039
# QWEN3 → STRUCTURED JSON TEST
# ============================================================

import json
import re
import time
import torch

print("=" * 70)
print("KAGGLE-039 — QWEN3 STRUCTURED JSON TEST")
print("=" * 70, flush=True)

assert "model" in globals(), "model not found"
assert "tokenizer" in globals(), "tokenizer not found"

# ------------------------------------------------------------
# 039.1 — Strict JSON prompt
# ------------------------------------------------------------

messages = [
    {
        "role": "system",
        "content": (
            "You are a JSON-only output engine. "
            "Do not explain. "
            "Do not use <think>. "
            "Return only the requested JSON object."
        )
    },
    {
        "role": "user",
        "content": (
            'Return exactly this JSON object and nothing else: '
            '{"status":"ok"}'
        )
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

input_device = model.get_input_embeddings().weight.device

inputs = {
    k: v.to(input_device)
    for k, v in inputs.items()
}

print("[039.1] Input prepared: PASS")
print("Input tokens:", inputs["input_ids"].shape[-1])
print("Input device:", input_device)

# ------------------------------------------------------------
# 039.2 — Generate
# ------------------------------------------------------------

print("\n[039.2] Generation START...", flush=True)

if torch.cuda.is_available():
    torch.cuda.synchronize()

start = time.time()

with torch.inference_mode():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=64,
        do_sample=False,
        use_cache=True,
        pad_token_id=tokenizer.eos_token_id,
    )

if torch.cuda.is_available():
    torch.cuda.synchronize()

elapsed = time.time() - start

# ------------------------------------------------------------
# 039.3 — Decode generated portion
# ------------------------------------------------------------

prompt_tokens = inputs["input_ids"].shape[-1]

generated_ids = output_ids[0][prompt_tokens:]

response = tokenizer.decode(
    generated_ids,
    skip_special_tokens=True
).strip()

print("[039.2] Generation: PASS")
print("Generation time:", round(elapsed, 3), "seconds")
print("Generated tokens:", len(generated_ids))

print("\n[039.3] RAW OUTPUT")
print("-" * 70)
print(response)
print("-" * 70)

# ------------------------------------------------------------
# 039.4 — Remove Qwen thinking block if present
# ------------------------------------------------------------

cleaned = response

# Remove <think>...</think>
cleaned = re.sub(
    r"<think>.*?</think>",
    "",
    cleaned,
    flags=re.DOTALL | re.IGNORECASE
).strip()

# Remove any remaining markdown fences
cleaned = re.sub(
    r"^```json\s*",
    "",
    cleaned,
    flags=re.IGNORECASE
)

cleaned = re.sub(
    r"\s*```$",
    "",
    cleaned
).strip()

print("\n[039.4] CLEANED OUTPUT")
print("-" * 70)
print(cleaned)
print("-" * 70)

# ------------------------------------------------------------
# 039.5 — JSON parse
# ------------------------------------------------------------

print("\n[039.5] JSON VALIDATION")

parsed = None
json_valid = False

try:
    parsed = json.loads(cleaned)
    json_valid = isinstance(parsed, dict)

    print("JSON parse: PASS")
    print("Parsed:", parsed)

except Exception as e:
    print("JSON parse: FAIL")
    print("Error:", str(e))

# ------------------------------------------------------------
# 039.6 — Exact contract validation
# ------------------------------------------------------------

print("\n[039.6] CONTRACT VALIDATION")

contract_valid = (
    isinstance(parsed, dict)
    and parsed == {"status": "ok"}
)

print("Expected :", {"status": "ok"})
print("Received :", parsed)
print("Contract :", contract_valid)

# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 70)

if contract_valid:
    print("KAGGLE-039 STATUS : PASS")
    print("Qwen3 → JSON → Python parser → contract validation = PASS")
else:
    print("KAGGLE-039 STATUS : FAIL")
    print("Structured JSON contract is not yet reliable.")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-030
# REAL TOOL EXECUTION
# Plan → ExecutionEngine → ToolRegistry → Python → Observation
# ============================================================

from pathlib import Path
import os
import sys
import json
import time

print("=" * 70)
print("KAGGLE-030 : REAL TOOL EXECUTION")
print("=" * 70)

# ------------------------------------------------------------
# 030.1 — Project/runtime
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    os.environ.get(
        "PERSONAL_AI_ROOT",
        "/kaggle/working/Personal_AI_AGI_Core_v3_0"
    )
).resolve()

print("\n[030.1] PROJECT ROOT")
print("Path   :", PROJECT_ROOT)
print("Exists :", PROJECT_ROOT.exists())

assert PROJECT_ROOT.exists(), f"Project root not found: {PROJECT_ROOT}"

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task
from personal_ai.core.planner import Planner
from personal_ai.execution.engine import ExecutionEngine

print("Imports : PASS")

# ------------------------------------------------------------
# 030.2 — Start V3 runtime
# ------------------------------------------------------------

print("\n[030.2] START V3 RUNTIME")
print("-" * 70)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime :", type(runtime).__name__)
print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

assert "python" in runtime.tools.tools, "Python tool not registered"
assert "shell" in runtime.tools.tools, "Shell tool not registered"

print("Required tools : PASS")

# ------------------------------------------------------------
# 030.3 — Build deterministic V3 Plan
# ------------------------------------------------------------

print("\n[030.3] BUILD DETERMINISTIC PLAN")
print("-" * 70)

task = Task(
    id="KAGGLE-030",
    goal="Calculate the average of 10, 20, and 30."
)

planner = Planner(runtime.models)

# Canonical executable step.
steps = [
    {
        "action": "tool",
        "tool": "python",
        "args": [
            "result = (10 + 20 + 30) / 3\nprint(result)"
        ],
        "kwargs": {}
    }
]

print("Task :", task.goal)
print("\nPlan steps:")
print(json.dumps(steps, indent=2))

assert len(steps) == 1
assert steps[0]["action"] == "tool"
assert steps[0]["tool"] == "python"

print("\nCanonical Plan : PASS")

# ------------------------------------------------------------
# 030.4 — Construct actual V3 Plan
# ------------------------------------------------------------

print("\n[030.4] CONSTRUCT V3 PLAN")
print("-" * 70)

# Use the existing Plan model from the project.
from personal_ai.core.planner_models import Plan

plan = Plan(
    task_id=task.id,
    steps=steps,
    assumptions=[]
)

print("Plan object :", plan)
print("Task ID     :", plan.task_id)
print("Steps       :", plan.steps)
print("Assumptions :", plan.assumptions)

assert plan.task_id == task.id
assert isinstance(plan.steps, list)
assert len(plan.steps) == 1

print("V3 Plan construction : PASS")

# ------------------------------------------------------------
# 030.5 — Execute through ExecutionEngine
# ------------------------------------------------------------

print("\n[030.5] EXECUTION ENGINE")
print("-" * 70)

engine = ExecutionEngine(runtime.tools)

print("Engine :", type(engine).__name__)

start = time.time()

try:
    result = engine.execute(plan)

    elapsed = time.time() - start

    print("\nExecution : RETURNED")
    print("Elapsed   :", round(elapsed, 3), "seconds")
    print("Result type :", type(result).__name__)

    print("\nRAW EXECUTION RESULT")
    print("-" * 70)
    print(result)
    print("-" * 70)

except Exception as e:

    elapsed = time.time() - start

    print("\nExecution : FAILED")
    print("Elapsed :", round(elapsed, 3), "seconds")
    print("Exception:", repr(e))

    raise

# ------------------------------------------------------------
# 030.6 — Observation inspection
# ------------------------------------------------------------

print("\n[030.6] OBSERVATION INSPECTION")
print("-" * 70)

print("Result repr:")
print(repr(result))

# Try common V3 result representations without
# assuming a new architecture.

observation = None

if hasattr(result, "observations"):
    observation = result.observations

elif hasattr(result, "observation"):
    observation = result.observation

elif isinstance(result, dict):
    observation = (
        result.get("observations")
        if "observations" in result
        else result.get("observation")
    )

print("\nObservation:")
print(observation)

# ------------------------------------------------------------
# 030.7 — Basic success evidence
# ------------------------------------------------------------

print("\n[030.7] SUCCESS EVIDENCE")
print("-" * 70)

result_text = repr(result)

# We expect the Python action to calculate 20.0.
# Do NOT declare PASS merely because ExecutionEngine returned.
# Look for actual evidence of 20 / 20.0 in the returned result.

has_20 = (
    "20.0" in result_text
    or "'20'" in result_text
    or '"20"' in result_text
)

print("Expected Python result : 20.0")
print("Evidence of 20.0/20   :", has_20)

# ------------------------------------------------------------
# FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)

if has_20:
    print("KAGGLE-030 STATUS : PASS")
    print("Plan → ExecutionEngine → Python → result = 20 : PASS")
else:
    print("KAGGLE-030 STATUS : PARTIAL / INVESTIGATE")
    print("Execution returned, but result evidence was not confirmed.")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-029/030 RECOVERY
# RESTORE V3 SOURCE INTO /kaggle/working
# ============================================================

from pathlib import Path
import os
import sys
import shutil

print("=" * 70)
print("V3 SOURCE RECOVERY")
print("=" * 70)

# ------------------------------------------------------------
# 1. Check working directory
# ------------------------------------------------------------

WORKING = Path("/kaggle/working")

print("\n[1] /kaggle/working")
print("Exists:", WORKING.exists())

print("\nCurrent contents:")
for p in sorted(WORKING.iterdir()):
    print(" -", p.name)

# ------------------------------------------------------------
# 2. Locate V3 ZIP
# ------------------------------------------------------------

possible_zips = []

for base in [
    Path("/kaggle/working"),
    Path("/kaggle/input"),
]:
    if base.exists():
        possible_zips.extend(
            base.rglob("Personal_AI_AGI_Core_v3_0.zip")
        )

print("\n[2] V3 ZIP SEARCH")

if possible_zips:
    for p in possible_zips:
        print("FOUND:", p)
else:
    print("V3 ZIP NOT FOUND")

# ------------------------------------------------------------
# 3. Restore
# ------------------------------------------------------------

if not possible_zips:
    raise FileNotFoundError(
        "Personal_AI_AGI_Core_v3_0.zip not found. "
        "V3 ZIP needs to be available in this Kaggle session."
    )

ZIP_PATH = possible_zips[0]

TARGET = WORKING / "Personal_AI_AGI_Core_v3_0"

if TARGET.exists():
    shutil.rmtree(TARGET)

print("\n[3] Extracting")
print("Source :", ZIP_PATH)
print("Target :", TARGET)

shutil.unpack_archive(
    str(ZIP_PATH),
    str(WORKING)
)

# ------------------------------------------------------------
# 4. Handle nested extraction if necessary
# ------------------------------------------------------------

if not TARGET.exists():

    candidates = list(WORKING.glob("**/Personal_AI_AGI_Core_v3_0"))

    if candidates:
        extracted = candidates[0]

        if extracted != TARGET:
            shutil.move(
                str(extracted),
                str(TARGET)
            )

# ------------------------------------------------------------
# 5. Verify V3
# ------------------------------------------------------------

print("\n[4] V3 VERIFICATION")

print("Target exists :", TARGET.exists())

assert TARGET.exists(), (
    f"V3 source could not be restored: {TARGET}"
)

required = [
    TARGET / "personal_ai",
    TARGET / "personal_ai" / "core",
    TARGET / "personal_ai" / "core" / "planner.py",
    TARGET / "personal_ai" / "execution" / "engine.py",
]

for p in required:
    print(
        f"{p.relative_to(TARGET)} :",
        "PASS" if p.exists() else "FAIL"
    )

assert all(p.exists() for p in required)

# ------------------------------------------------------------
# 6. Python import path
# ------------------------------------------------------------

if str(TARGET) not in sys.path:
    sys.path.insert(0, str(TARGET))

print("\n[5] PYTHON PATH")
print("V3 root:", TARGET)
print("sys.path : PASS")

print("\n" + "=" * 70)
print("V3 SOURCE RECOVERY : PASS")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-006
# RESTORE V3 DATASET → /kaggle/working
# ============================================================

from pathlib import Path
import shutil
import sys

print("=" * 70)
print("KAGGLE-006 : CREATE V3 WORKING COPY")
print("=" * 70)

# Exact path verified earlier by KAGGLE-005
SOURCE = Path(
    "/kaggle/input/datasets/vikaschandoliya/"
    "personal-ai-agi-core-v3-0/"
    "Personal_AI_AGI_Core_v3_0"
)

WORK_ROOT = Path(
    "/kaggle/working/Personal_AI_AGI_Core_v3_0"
)

print("\nSOURCE")
print("Path  :", SOURCE)
print("Exists:", SOURCE.exists())

assert SOURCE.exists(), (
    f"V3 dataset source not found: {SOURCE}\n"
    "Check that Personal_AI_AGI_Core_v3_0 dataset is attached "
    "to this Kaggle notebook."
)

print("\nTARGET")
print("Path  :", WORK_ROOT)

# Remove stale working copy if any
if WORK_ROOT.exists():
    print("\nExisting working copy found.")
    print("Removing old copy...")
    shutil.rmtree(WORK_ROOT)

# Copy dataset → writable working area
print("\nCopying V3...")
shutil.copytree(SOURCE, WORK_ROOT)

print("Copy complete.")

# ------------------------------------------------------------
# Verify V3 structure
# ------------------------------------------------------------

print("\nV3 STRUCTURE VERIFICATION")
print("-" * 70)

required = [
    "MANIFEST.json",
    "ARCHITECTURE.md",
    "README.md",
    "requirements-runtime.txt",
    "personal_ai",
    "personal_ai/core",
    "personal_ai/core/planner.py",
    "personal_ai/execution",
    "personal_ai/execution/engine.py",
    "personal_ai/tools",
    "personal_ai/tools/registry.py",
    "tests",
]

missing = []

for item in required:
    path = WORK_ROOT / item
    ok = path.exists()

    print(
        f"[{'PASS' if ok else 'FAIL'}] {item}"
    )

    if not ok:
        missing.append(item)

assert not missing, (
    f"V3 working copy incomplete. Missing: {missing}"
)

# ------------------------------------------------------------
# Python path
# ------------------------------------------------------------

if str(WORK_ROOT) not in sys.path:
    sys.path.insert(0, str(WORK_ROOT))

print("\nPYTHON PATH")
print("V3 root added to sys.path : PASS")

print("\n" + "=" * 70)
print("KAGGLE-006 STATUS : PASS")
print("=" * 70)

print("V3 source :", SOURCE)
print("V3 working:", WORK_ROOT)
````

````python
# ============================================================
# KAGGLE-030
# REAL TOOL EXECUTION
# Plan → ExecutionEngine → ToolRegistry → Python → Observation
# ============================================================

from pathlib import Path
import os
import sys
import json
import time
import inspect

print("=" * 70)
print("KAGGLE-030 : REAL TOOL EXECUTION")
print("=" * 70)

PROJECT_ROOT = Path(
    "/kaggle/working/Personal_AI_AGI_Core_v3_0"
).resolve()

print("\n[030.1] PROJECT ROOT")
print("Path   :", PROJECT_ROOT)
print("Exists :", PROJECT_ROOT.exists())

assert PROJECT_ROOT.exists()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# ------------------------------------------------------------
# Import actual V3 components
# ------------------------------------------------------------

from personal_ai.bootstrap import Bootstrap
from personal_ai.core.contracts import Task
from personal_ai.core.planner import Planner
from personal_ai.execution.engine import ExecutionEngine
from personal_ai.core.planner_models import Plan

print("Imports : PASS")

# ------------------------------------------------------------
# Inspect actual APIs — NO model reload
# ------------------------------------------------------------

print("\n[030.2] ACTUAL V3 API")
print("-" * 70)

print("Bootstrap :", inspect.signature(Bootstrap))
print("Planner   :", inspect.signature(Planner))
print("Engine    :", inspect.signature(ExecutionEngine))
print("Plan      :", inspect.signature(Plan))

# ------------------------------------------------------------
# Start runtime
# ------------------------------------------------------------

print("\n[030.3] START V3 RUNTIME")
print("-" * 70)

runtime = Bootstrap(PROJECT_ROOT).start()

print("Runtime :", type(runtime).__name__)

print("Models  :", list(runtime.models.backends.keys()))
print("Tools   :", list(runtime.tools.tools.keys()))

assert "python" in runtime.tools.tools
assert "shell" in runtime.tools.tools

print("Required tools : PASS")

# ------------------------------------------------------------
# Deterministic task
# ------------------------------------------------------------

print("\n[030.4] BUILD DETERMINISTIC PLAN")
print("-" * 70)

task = Task(
    id="KAGGLE-030",
    goal="Calculate the average of 10, 20, and 30."
)

steps = [
    {
        "action": "tool",
        "tool": "python",
        "args": [
            "result = (10 + 20 + 30) / 3\nprint(result)"
        ],
        "kwargs": {}
    }
]

print("Task:")
print(task.goal)

print("\nCanonical steps:")
print(json.dumps(steps, indent=2))

assert len(steps) == 1
assert steps[0]["action"] == "tool"
assert steps[0]["tool"] == "python"

print("Canonical Plan : PASS")

# ------------------------------------------------------------
# Construct V3 Plan
# ------------------------------------------------------------

print("\n[030.5] CONSTRUCT V3 PLAN")
print("-" * 70)

plan = Plan(
    task_id=task.id,
    steps=steps,
    assumptions=[]
)

print("Plan object:")
print(plan)

print("\nTask ID:", plan.task_id)
print("Steps  :", plan.steps)

assert plan.task_id == task.id
assert isinstance(plan.steps, list)
assert len(plan.steps) == 1

print("V3 Plan construction : PASS")

# ------------------------------------------------------------
# Execute REAL TOOL
# ------------------------------------------------------------

print("\n[030.6] EXECUTION ENGINE")
print("-" * 70)

engine = ExecutionEngine(runtime.tools)

print("Engine :", type(engine).__name__)

start = time.time()

result = engine.execute(plan)

elapsed = time.time() - start

print("\nExecution returned.")
print("Elapsed :", round(elapsed, 3), "seconds")
print("Result type :", type(result).__name__)

print("\nRAW RESULT")
print("-" * 70)
print(result)
print("-" * 70)

# ------------------------------------------------------------
# Observation inspection
# ------------------------------------------------------------

print("\n[030.7] OBSERVATION")
print("-" * 70)

print("repr(result):")
print(repr(result))

observation = None

if hasattr(result, "observations"):
    observation = result.observations
elif hasattr(result, "observation"):
    observation = result.observation
elif isinstance(result, dict):
    observation = result.get(
        "observations",
        result.get("observation")
    )

print("\nObservation:")
print(observation)

# ------------------------------------------------------------
# Verify actual 20.0 evidence
# ------------------------------------------------------------

print("\n[030.8] VERIFICATION")
print("-" * 70)

result_text = repr(result)
observation_text = repr(observation)

combined = result_text + " " + observation_text

has_20 = (
    "20.0" in combined
    or "'20'" in combined
    or '"20"' in combined
)

print("Expected :", "20.0")
print("Evidence :", has_20)

print("\n" + "=" * 70)

if has_20:
    print("KAGGLE-030 STATUS : PASS")
    print("Plan → ExecutionEngine → ToolRegistry → Python → 20.0 : PASS")
else:
    print("KAGGLE-030 STATUS : PARTIAL / INVESTIGATE")
    print("Execution returned, but 20.0 evidence was not confirmed.")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-040 : QWEN3 → OBSERVATION → STRUCTURED DECISION
# ============================================================
# IMPORTANT:
#   This is the next step after KAGGLE-039.2 PASS.
#
#   Proven:
#       Qwen3
#         ↓
#       enable_thinking=False
#         ↓
#       valid JSON
#
#   Now:
#       Observation
#         ↓
#       Qwen3
#         ↓
#       Agent decision JSON
#
# DO NOT modify V3 source yet.
# ============================================================

import json
import time
import torch

print("=" * 70)
print("KAGGLE-040 : QWEN3 → OBSERVATION → STRUCTURED DECISION")
print("=" * 70, flush=True)

# ------------------------------------------------------------
# 040.1 — Verify REAL Qwen3
# ------------------------------------------------------------

assert "model" in globals(), "ERROR: Qwen3 model is not loaded."
assert "tokenizer" in globals(), "ERROR: Qwen3 tokenizer is not loaded."

print("\n[040.1] Qwen3 runtime : PASS")
print("Model    :", type(model).__name__)
print("Tokenizer:", type(tokenizer).__name__)

# ------------------------------------------------------------
# 040.2 — REAL observation from previous tool execution
# ------------------------------------------------------------

task = "Calculate the average of 10, 20, and 30."

observation = {
    "task": task,
    "tool": "python",
    "success": True,
    "result": {
        "returncode": 0,
        "stdout": "20.0\n",
        "stderr": ""
    },
    "error": None
}

observation_text = json.dumps(
    observation,
    indent=2
)

print("\n[040.2] Observation : PASS")
print(observation_text)

# ------------------------------------------------------------
# 040.3 — CONTROL PROMPT
# ------------------------------------------------------------

messages = [
    {
        "role": "system",
        "content": (
            "You are the control layer of a personal AI agent. "
            "Return ONLY valid JSON. "
            "Do not explain. "
            "Do not use markdown. "
            "Do not output <think>. "
            "Your output must be one JSON object."
        )
    },
    {
        "role": "user",
        "content": f"""
The agent has executed a tool.

Task:
{task}

Observation:
{observation_text}

Decide what the agent should do next.

Use this JSON schema:

{{
  "task_complete": true,
  "next_action": "finish",
  "result": "20.0"
}}

Allowed next_action values:
- "finish"
- "tool"
- "replan"

Return JSON only.
"""
    }
]

print("\n[040.3] Control prompt : PASS")

# ------------------------------------------------------------
# 040.4 — QWEN3 CHAT TEMPLATE
# ------------------------------------------------------------

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False,
)

print("[040.4] Qwen3 chat template : PASS")

# ------------------------------------------------------------
# 040.5 — TOKENIZE
# ------------------------------------------------------------

inputs = tokenizer(
    text,
    return_tensors="pt"
)

input_device = model.get_input_embeddings().weight.device

inputs = {
    key: value.to(input_device)
    for key, value in inputs.items()
}

print("[040.5] Tokenization : PASS")
print("Input device :", input_device)
print("Input tokens :", inputs["input_ids"].shape[-1])

# ------------------------------------------------------------
# 040.6 — REAL QWEN3 GENERATION
# ------------------------------------------------------------

print("\n[040.6] Qwen3 generation START...", flush=True)

if torch.cuda.is_available():
    torch.cuda.synchronize()

start_time = time.time()

with torch.inference_mode():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=64,
        do_sample=False,
        use_cache=True,
        pad_token_id=tokenizer.eos_token_id,
    )

if torch.cuda.is_available():
    torch.cuda.synchronize()

elapsed = time.time() - start_time

print("[040.6] Generation : PASS")
print("Generation time :", round(elapsed, 3), "seconds")

# ------------------------------------------------------------
# 040.7 — DECODE ONLY NEW TOKENS
# ------------------------------------------------------------

prompt_tokens = inputs["input_ids"].shape[-1]

generated_ids = output_ids[0][prompt_tokens:]

response = tokenizer.decode(
    generated_ids,
    skip_special_tokens=True
).strip()

print("\n[040.7] RAW QWEN3 OUTPUT")
print("-" * 70)
print(response)
print("-" * 70)

print("Generated tokens:", len(generated_ids))

# ------------------------------------------------------------
# 040.8 — JSON PARSE
# ------------------------------------------------------------

print("\n[040.8] JSON VALIDATION")

try:
    decision = json.loads(response)

    print("JSON parse : PASS")
    print("Decision   :", decision)

except Exception as e:

    print("JSON parse : FAIL")
    print("Error      :", repr(e))

    print("\nIMPORTANT DEBUG OUTPUT")
    print("Raw response repr:")
    print(repr(response))

    raise

# ------------------------------------------------------------
# 040.9 — BASIC DECISION CONTRACT
# ------------------------------------------------------------

print("\n[040.9] DECISION CONTRACT")

assert isinstance(decision, dict), \
    "Decision must be a JSON object."

assert "task_complete" in decision, \
    "Missing field: task_complete"

assert "next_action" in decision, \
    "Missing field: next_action"

assert isinstance(
    decision["task_complete"],
    bool
), "task_complete must be boolean."

assert decision["next_action"] in {
    "finish",
    "tool",
    "replan"
}, "Invalid next_action."

print("JSON object       : PASS")
print("task_complete     :", decision["task_complete"])
print("next_action       :", decision["next_action"])

if "result" in decision:
    print("result            :", repr(decision["result"]))

# ------------------------------------------------------------
# 040.10 — SUCCESS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-040 STATUS : PASS")
print("=" * 70)

print("REAL Qwen3                  : PASS")
print("Observation → Qwen3        : PASS")
print("Non-thinking control mode  : PASS")
print("Structured JSON             : PASS")
print("Decision contract           : PASS")

print("\nDECISION:")
print(json.dumps(decision, indent=2))

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-044.2 : REAL FAILURE → QWEN3 DIAGNOSIS → REPLAN
# ============================================================

import json
import time
import torch

print("=" * 70)
print("KAGGLE-044.2 : REAL FAILURE → QWEN3 DIAGNOSIS → REPLAN")
print("=" * 70)


# ============================================================
# 044.2.1 — BUILD REAL FAILURE OBSERVATION
# ============================================================

print("\n[044.2.1] Failure observation")

failure_payload = {
    "task": task.goal,
    "tool": "python",
    "success": False,
    "result": observation.output,
    "error": observation.error,
}

failure_text = json.dumps(
    failure_payload,
    indent=2,
    default=str,
)

print(failure_text)

assert failure_payload["success"] is False
assert failure_payload["result"]["returncode"] != 0

print("Failure payload : PASS")


# ============================================================
# 044.2.2 — QWEN3 RECOVERY PROMPT
# ============================================================

print("\n[044.2.2] Qwen3 recovery prompt")

messages = [
    {
        "role": "system",
        "content": (
            "You are the recovery controller of a personal AI agent. "
            "Analyze real tool failures and decide the next action. "
            "Return ONLY valid JSON. "
            "Do not output markdown. "
            "Do not output explanations outside JSON."
        ),
    },
    {
        "role": "user",
        "content": f"""
Task:
{task.goal}

The agent attempted to execute a Python tool.

REAL FAILURE OBSERVATION:
{failure_text}

The task is not complete.

Analyze the failure and determine the next control action.

Return exactly one JSON object using this structure:

{{
  "task_complete": false,
  "next_action": "replan",
  "diagnosis": "The Python operation failed because of division by zero."
}}

Allowed next_action values:
"finish"
"tool"
"replan"

Rules:
- task_complete must be boolean.
- next_action must be one of the allowed values.
- Because the real tool failed and the task is incomplete,
  next_action should be "replan".
- diagnosis must explain the actual failure.
- Return JSON only.
"""
    }
]


# ============================================================
# 044.2.3 — QWEN3 CHAT TEMPLATE
# ============================================================

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    enable_thinking=False,
)

if isinstance(inputs, dict):
    inputs = {
        k: v.to(model.device)
        for k, v in inputs.items()
    }
else:
    inputs = inputs.to(model.device)

print("Input tokens :", inputs.shape[-1])
print("Chat template : PASS")


# ============================================================
# 044.2.4 — REAL QWEN3 GENERATION
# ============================================================

print("\n[044.2.4] Qwen3 generation START...")

start = time.time()

with torch.no_grad():
    outputs = model.generate(
        inputs,
        max_new_tokens=96,
        do_sample=False,
        temperature=None,
        top_p=None,
    )

elapsed = time.time() - start

print("Generation time :", round(elapsed, 3), "sec")
print("Generation : PASS")


# ============================================================
# 044.2.5 — DECODE NEW TOKENS ONLY
# ============================================================

input_length = inputs.shape[-1]

generated_tokens = outputs[0][input_length:]

raw_output = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True,
).strip()

print("\n[044.2.5] RAW QWEN3 OUTPUT")
print("-" * 70)
print(repr(raw_output))
print("-" * 70)

assert raw_output

print("Output : PASS")


# ============================================================
# 044.2.6 — PARSE JSON
# ============================================================

print("\n[044.2.6] JSON validation")

decision_text = raw_output

if decision_text.startswith("```"):
    decision_text = decision_text.replace("```json", "")
    decision_text = decision_text.replace("```", "")
    decision_text = decision_text.strip()

decision = json.loads(decision_text)

print("Decision:")
print(json.dumps(decision, indent=2))

assert isinstance(decision, dict)

print("JSON parse : PASS")


# ============================================================
# 044.2.7 — DECISION CONTRACT
# ============================================================

print("\n[044.2.7] Recovery decision contract")

assert isinstance(
    decision.get("task_complete"),
    bool,
)

assert decision.get("next_action") in {
    "finish",
    "tool",
    "replan",
}

assert isinstance(
    decision.get("diagnosis"),
    str,
)

assert len(decision["diagnosis"].strip()) > 0

print("JSON object   : PASS")
print("task_complete :", decision["task_complete"])
print("next_action   :", decision["next_action"])
print("diagnosis     :", repr(decision["diagnosis"]))


# ============================================================
# 044.2.8 — VERIFY RECOVERY DECISION
# ============================================================

print("\n[044.2.8] Recovery verification")

assert decision["task_complete"] is False
assert decision["next_action"] == "replan"

print("Task incomplete : PASS")
print("Replan decision : PASS")
print("Diagnosis       : PASS")


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("KAGGLE-044 STATUS : PASS")
print("=" * 70)

print("Intentional Python Failure : PASS")
print("Real Failure Observation   : PASS")
print("Qwen3 Failure Analysis     : PASS")
print("Structured JSON            : PASS")
print("Diagnosis Contract         : PASS")
print("Replan Decision            : PASS")

print("\nVERIFIED RECOVERY SIGNAL")
print("Real Failure")
print("    ↓")
print("Observation")
print("    ↓")
print("Qwen3 Diagnosis")
print("    ↓")
print("REPLAN")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-046
# GENERIC SELF-RECOVERY LOOP
# ============================================================

import json
import time

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine


# ============================================================
# 1. GENERIC QWEN3 RECOVERY FUNCTION
# ============================================================

def qwen3_replan(task, failure_output):
    """
    Give a real tool failure to Qwen3 and ask for a corrected plan.
    """

    failure_payload = {
        "task": task.goal,
        "success": False,
        "returncode": failure_output["returncode"],
        "stdout": failure_output["stdout"],
        "stderr": failure_output["stderr"],
    }

    prompt = f"""
You are the recovery controller of a Personal AI agent.

TASK:
{task.goal}

FAILED TOOL EXECUTION:
{json.dumps(failure_payload, indent=2)}

Diagnose the failure and create a corrected Python plan.

Return ONLY valid JSON.
No markdown.
No explanation outside JSON.

Required format:

{{
  "task_complete": false,
  "next_action": "replan",
  "diagnosis": "short explanation",
  "plan": {{
    "tool": "python",
    "code": "corrected Python code"
  }}
}}
"""

    messages = [
        {
            "role": "system",
            "content": "You are a reliable recovery planner. Return only valid JSON."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt"
    )

    inputs = inputs.to(model.device)

    outputs = model.generate(
        inputs,
        max_new_tokens=256,
        do_sample=False
    )

    generated_tokens = outputs[0][inputs.shape[1]:]

    raw = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    decision = json.loads(raw)

    assert decision["task_complete"] is False
    assert decision["next_action"] == "replan"
    assert "diagnosis" in decision
    assert "plan" in decision
    assert decision["plan"]["tool"] == "python"

    return decision


# ============================================================
# 2. GENERIC EXECUTION + RECOVERY FUNCTION
# ============================================================

def execute_with_recovery(task, initial_plan, max_retries=1):

    current_plan = initial_plan

    for attempt in range(max_retries + 1):

        print(f"\n--- Attempt {attempt + 1} ---")

        engine = ExecutionEngine(
            tools=runtime.tools,
            evaluator=runtime.evaluator,
            ledger=runtime.learning,
            max_steps=5
        )

        observations = engine.execute_plan(current_plan)

        assert isinstance(observations, list)
        assert len(observations) == 1

        observation = observations[0]
        output = observation.output

        print("Return code:", output["returncode"])

        # ----------------------------------------------------
        # SUCCESS
        # ----------------------------------------------------

        if output["returncode"] == 0:

            stdout = output["stdout"].strip()

            print("stdout:", repr(stdout))
            print("Execution SUCCESS")

            return {
                "success": True,
                "attempt": attempt + 1,
                "observation": observation,
                "stdout": stdout
            }

        # ----------------------------------------------------
        # FAILURE
        # ----------------------------------------------------

        print("Execution FAILED")

        if attempt >= max_retries:
            return {
                "success": False,
                "attempt": attempt + 1,
                "observation": observation,
                "stdout": output["stdout"],
                "stderr": output["stderr"]
            }

        # ----------------------------------------------------
        # QWEN3 RECOVERY
        # ----------------------------------------------------

        print("Sending failure to Qwen3...")

        decision = qwen3_replan(task, output)

        print("Diagnosis:")
        print(decision["diagnosis"])

        corrected_code = decision["plan"]["code"]

        print("Replanned code:")
        print(corrected_code)

        # ----------------------------------------------------
        # BUILD NEW V3 PLAN
        # ----------------------------------------------------

        current_plan = Plan(
            task_id=task.id,
            steps=[
                {
                    "action": "tool",
                    "tool": "python",
                    "args": [corrected_code],
                    "kwargs": {}
                }
            ],
            assumptions=[]
        )

        print("New V3 Plan created")


# ============================================================
# 3. TEST TASK
# ============================================================

task = Task(
    id="KAGGLE-046",
    goal="Calculate the average of 10, 20 and 30 and recover automatically from an execution failure."
)


# ============================================================
# 4. INTENTIONALLY BROKEN INITIAL PLAN
# ============================================================

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [
                """numbers = [10, 20, 30]
total = sum(numbers)
count = 0
average = total / count
print(average)"""
            ],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("KAGGLE-046")
print("Initial plan created")


# ============================================================
# 5. RUN GENERIC RECOVERY LOOP
# ============================================================

result = execute_with_recovery(
    task=task,
    initial_plan=initial_plan,
    max_retries=1
)


# ============================================================
# 6. FINAL VERIFICATION
# ============================================================

assert result["success"] is True

final_stdout = result["stdout"]

assert abs(float(final_stdout) - 20.0) < 1e-9

print("\n" + "=" * 65)
print("KAGGLE-046 STATUS : PASS")
print("=" * 65)

print("Initial failure       : PASS")
print("Generic failure check : PASS")
print("Qwen3 recovery        : PASS")
print("Automatic replan      : PASS")
print("Second execution      : PASS")
print("Final verification    : PASS")
print("Attempts used         :", result["attempt"])
print("Final result          :", final_stdout)

print("=" * 65)
````

````python
# ============================================================
# KAGGLE-047
# RESULT VERIFICATION / CRITIC LOOP
#
# Goal:
# Successful execution != successful task
#
# Wrong result -> Qwen3 critic -> corrected plan
# -> execution -> verification -> PASS
# ============================================================

import json
import time

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine


# ============================================================
# 1. TASK
# ============================================================

task = Task(
    id="KAGGLE-047",
    goal="Calculate the average of 10, 20 and 30 and return the correct result."
)

print("KAGGLE-047")
print("Task:", task.goal)


# ============================================================
# 2. INTENTIONALLY WRONG BUT VALID PYTHON PLAN
# ============================================================

wrong_code = """numbers = [10, 20, 30]
average = 25
print(average)"""

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [wrong_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("\n[1] Initial wrong plan created")


# ============================================================
# 3. EXECUTE WRONG PLAN
# ============================================================

engine = ExecutionEngine(
    tools=runtime.tools,
    evaluator=runtime.evaluator,
    ledger=runtime.learning,
    max_steps=5
)

observations = engine.execute_plan(initial_plan)

assert isinstance(observations, list)
assert len(observations) == 1

observation = observations[0]
output = observation.output

assert output["returncode"] == 0

stdout = output["stdout"].strip()

print("[2] Python execution completed")
print("    Return code:", output["returncode"])
print("    Output:", repr(stdout))


# ============================================================
# 4. IMPORTANT: EXECUTION SUCCESS
#    BUT RESULT MAY BE WRONG
# ============================================================

print("[3] TOOL EXECUTION : PASS")
print("    Now performing semantic verification...")


# ============================================================
# 5. ASK QWEN3 TO CRITIQUE THE RESULT
# ============================================================

verification_prompt = f"""
You are the verification and recovery controller of a Personal AI agent.

TASK:
{task.goal}

The Python tool executed successfully.

Execution result:
{json.dumps({
    "returncode": output["returncode"],
    "stdout": stdout,
    "stderr": output["stderr"]
}, indent=2)}

Determine whether the result actually satisfies the task.

The correct average of 10, 20 and 30 is 20.

If the result is incorrect, create a corrected Python plan.

Return ONLY valid JSON.
No markdown.
No explanation outside JSON.

Required schema:

{{
  "task_complete": true or false,
  "verification": "pass or fail",
  "diagnosis": "short explanation",
  "next_action": "finish or replan",
  "plan": {{
    "tool": "python",
    "code": "corrected Python code"
  }}
}}

If the current result is incorrect:
- task_complete must be false
- verification must be "fail"
- next_action must be "replan"
- provide corrected Python code

If the current result is correct:
- task_complete must be true
- verification must be "pass"
- next_action must be "finish"
- plan may contain an empty code string
"""


# ============================================================
# 6. QWEN3 CRITIC
# ============================================================

messages = [
    {
        "role": "system",
        "content": (
            "You are a strict task verifier. "
            "Distinguish execution success from task success. "
            "Return only valid JSON."
        )
    },
    {
        "role": "user",
        "content": verification_prompt
    }
]

t0 = time.time()

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    enable_thinking=False,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

outputs = model.generate(
    inputs,
    max_new_tokens=256,
    do_sample=False
)

generated_tokens = outputs[0][inputs.shape[1]:]

raw_decision = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
).strip()

critic_time = time.time() - t0

print("\n[4] Qwen3 verification decision:")
print(raw_decision)

print("    Generation time:", round(critic_time, 3), "sec")


# ============================================================
# 7. PARSE + VERIFY CRITIC DECISION
# ============================================================

decision = json.loads(raw_decision)

assert "task_complete" in decision
assert "verification" in decision
assert "diagnosis" in decision
assert "next_action" in decision

print("\n[5] QWEN3 CRITIC JSON : PASS")

print("    Verification:", decision["verification"])
print("    Diagnosis:", decision["diagnosis"])
print("    Next action:", decision["next_action"])


# ============================================================
# 8. EXPECT QWEN3 TO DETECT WRONG RESULT
# ============================================================

assert decision["task_complete"] is False
assert decision["verification"] == "fail"
assert decision["next_action"] == "replan"

assert "plan" in decision
assert decision["plan"]["tool"] == "python"
assert isinstance(decision["plan"]["code"], str)

print("[6] WRONG RESULT DETECTION : PASS")


# ============================================================
# 9. BUILD CORRECTED V3 PLAN
# ============================================================

corrected_code = decision["plan"]["code"]

corrected_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [corrected_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("\n[7] Corrected V3 Plan created")
print("    Corrected code:")
print(corrected_code)


# ============================================================
# 10. EXECUTE CORRECTED PLAN
# ============================================================

recovery_engine = ExecutionEngine(
    tools=runtime.tools,
    evaluator=runtime.evaluator,
    ledger=runtime.learning,
    max_steps=5
)

recovery_observations = recovery_engine.execute_plan(corrected_plan)

assert isinstance(recovery_observations, list)
assert len(recovery_observations) == 1

recovery_observation = recovery_observations[0]
recovery_output = recovery_observation.output

assert recovery_output["returncode"] == 0

final_stdout = recovery_output["stdout"].strip()

print("\n[8] Corrected plan executed")
print("    Return code:", recovery_output["returncode"])
print("    Output:", repr(final_stdout))


# ============================================================
# 11. FINAL RESULT VERIFICATION
# ============================================================

final_result = float(final_stdout)

assert abs(final_result - 20.0) < 1e-9

print("[9] FINAL SEMANTIC VERIFICATION : PASS")
print("    Verified result:", final_result)


# ============================================================
# 12. FINAL STATUS
# ============================================================

print("\n" + "=" * 65)
print("KAGGLE-047 STATUS : PASS")
print("=" * 65)

print("Tool execution success     : PASS")
print("Wrong result detected      : PASS")
print("Qwen3 critic               : PASS")
print("Semantic verification      : PASS")
print("Automatic correction       : PASS")
print("Corrected V3 plan          : PASS")
print("Final execution            : PASS")
print("Final result verification  : PASS")
print("Final result               :", final_result)

print("=" * 65)
````

````python
# ============================================================
# KAGGLE-048
# UNIFIED AGENT LOOP
#
# Handles:
# 1. Tool execution failure
# 2. Successful execution but wrong result
# 3. Qwen3 diagnosis
# 4. Automatic replanning
# 5. Final verification
# ============================================================

import json
import time

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine


# ============================================================
# 1. QWEN3 CONTROLLER
# ============================================================

def qwen3_controller(task, observation, stage):
    """
    Analyse either:
      - execution failure
      - semantic/result failure

    and return a structured recovery decision.
    """

    output = observation.output

    payload = {
        "task": task.goal,
        "stage": stage,
        "returncode": output.get("returncode"),
        "stdout": output.get("stdout", ""),
        "stderr": output.get("stderr", "")
    }

    prompt = f"""
You are the controller of a Personal AI agent.

TASK:
{task.goal}

CURRENT EXECUTION STATE:
{json.dumps(payload, indent=2)}

You must determine whether the task is actually complete.

Important:
- returncode == 0 only means the program executed successfully.
- The printed result must also satisfy the task.
- If execution failed, diagnose the execution failure.
- If execution succeeded but the result is wrong, diagnose the result.
- If recovery is needed, create corrected Python code.

Return ONLY valid JSON.
No markdown.
No explanation outside JSON.

Required schema:

{{
  "task_complete": true or false,
  "verification": "pass or fail",
  "diagnosis": "short explanation",
  "next_action": "finish or replan",
  "plan": {{
    "tool": "python",
    "code": "corrected Python code"
  }}
}}

If the task is NOT complete:
- task_complete = false
- verification = "fail"
- next_action = "replan"
- provide corrected Python code

If the task IS complete:
- task_complete = true
- verification = "pass"
- next_action = "finish"
"""


    messages = [
        {
            "role": "system",
            "content": (
                "You are a strict Personal AI controller. "
                "Return only valid JSON."
            )
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt"
    )

    inputs = inputs.to(model.device)

    outputs = model.generate(
        inputs,
        max_new_tokens=256,
        do_sample=False
    )

    generated_tokens = outputs[0][inputs.shape[1]:]

    raw = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    decision = json.loads(raw)

    assert "task_complete" in decision
    assert "verification" in decision
    assert "diagnosis" in decision
    assert "next_action" in decision

    return decision


# ============================================================
# 2. UNIFIED AGENT LOOP
# ============================================================

def unified_agent_loop(task, initial_plan, max_retries=2):

    current_plan = initial_plan

    for attempt in range(max_retries + 1):

        print("\n" + "-" * 60)
        print(f"ATTEMPT {attempt + 1}")
        print("-" * 60)

        # ----------------------------------------------------
        # EXECUTION
        # ----------------------------------------------------

        engine = ExecutionEngine(
            tools=runtime.tools,
            evaluator=runtime.evaluator,
            ledger=runtime.learning,
            max_steps=5
        )

        observations = engine.execute_plan(current_plan)

        assert isinstance(observations, list)
        assert len(observations) == 1

        observation = observations[0]
        output = observation.output

        returncode = output["returncode"]
        stdout = output["stdout"].strip()
        stderr = output["stderr"]

        print("Return code:", returncode)
        print("stdout:", repr(stdout))

        # ----------------------------------------------------
        # CASE A: TOOL EXECUTION FAILED
        # ----------------------------------------------------

        if returncode != 0:

            print("Execution failure detected.")

            if attempt >= max_retries:
                return {
                    "success": False,
                    "reason": "max_retries",
                    "attempts": attempt + 1,
                    "observation": observation
                }

            print("Sending failure to Qwen3...")

            decision = qwen3_controller(
                task,
                observation,
                "execution_failure"
            )

        # ----------------------------------------------------
        # CASE B: TOOL EXECUTED
        #         NOW CHECK TASK RESULT
        # ----------------------------------------------------

        else:

            print("Execution succeeded.")
            print("Checking semantic correctness...")

            if attempt == 0:
                # Deliberately allow Qwen3 to verify the result.
                pass

            decision = qwen3_controller(
                task,
                observation,
                "result_verification"
            )

        # ----------------------------------------------------
        # CONTROLLER DECISION
        # ----------------------------------------------------

        print("\nQwen3 decision:")
        print(json.dumps(decision, indent=2))

        # ----------------------------------------------------
        # FINISH
        # ----------------------------------------------------

        if (
            decision["task_complete"] is True
            and decision["verification"] == "pass"
            and decision["next_action"] == "finish"
        ):

            return {
                "success": True,
                "attempts": attempt + 1,
                "result": stdout,
                "observation": observation,
                "decision": decision
            }

        # ----------------------------------------------------
        # REPLAN
        # ----------------------------------------------------

        assert decision["task_complete"] is False
        assert decision["next_action"] == "replan"
        assert "plan" in decision

        corrected_code = decision["plan"]["code"]

        print("\nCreating corrected V3 plan...")
        print(corrected_code)

        current_plan = Plan(
            task_id=task.id,
            steps=[
                {
                    "action": "tool",
                    "tool": "python",
                    "args": [corrected_code],
                    "kwargs": {}
                }
            ],
            assumptions=[]
        )

        print("Corrected V3 plan created.")

    return {
        "success": False,
        "reason": "loop_exhausted"
    }


# ============================================================
# 3. TEST TASK
# ============================================================

task = Task(
    id="KAGGLE-048",
    goal="Calculate the average of 10, 20 and 30 and return the correct result."
)

print("KAGGLE-048")
print("Task:", task.goal)


# ============================================================
# 4. BAD INITIAL PLAN
#
# This one EXECUTES successfully but gives WRONG RESULT.
# This proves the unified loop can handle semantic failure.
# ============================================================

initial_code = """numbers = [10, 20, 30]
average = 25
print(average)"""

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [initial_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("\nInitial plan created.")


# ============================================================
# 5. RUN UNIFIED AGENT LOOP
# ============================================================

result = unified_agent_loop(
    task=task,
    initial_plan=initial_plan,
    max_retries=2
)


# ============================================================
# 6. FINAL VERIFICATION
# ============================================================

assert result["success"] is True

final_result = float(result["result"])

assert abs(final_result - 20.0) < 1e-9

print("\n" + "=" * 65)
print("KAGGLE-048 STATUS : PASS")
print("=" * 65)

print("Unified execution loop : PASS")
print("Execution handling     : PASS")
print("Semantic verification  : PASS")
print("Qwen3 controller       : PASS")
print("Automatic replanning   : PASS")
print("Corrected execution    : PASS")
print("Final verification     : PASS")
print("Attempts used          :", result["attempts"])
print("Final result           :", final_result)

print("=" * 65)
````

````python
# ============================================================
# KAGGLE-048 FIX
# Deterministic Result Verification + Qwen3 Recovery
# ============================================================

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine
import json
import re


def deterministic_verify(task, stdout):
    """
    Deterministic verifier for this milestone.

    Expected task:
    average of [10, 20, 30] = 20
    """

    text = stdout.strip()

    try:
        value = float(text)
    except ValueError:
        return False, "Output is not a valid numeric result."

    expected = 20.0

    if abs(value - expected) < 1e-9:
        return True, "Result matches expected value 20.0."

    return False, f"Expected {expected}, but received {value}."


def qwen3_replan_verified(task, observation, verification_message):

    output = observation.output

    prompt = f"""
You are the recovery controller for a Personal AI agent.

Task:
{task.goal}

Execution output:
returncode: {output.get("returncode")}
stdout: {output.get("stdout", "")}
stderr: {output.get("stderr", "")}

Deterministic verification:
{verification_message}

The deterministic verifier is authoritative.
If verification failed, you MUST NOT claim the current result is correct.

Return ONLY valid JSON.

Required schema:
{{
  "task_complete": false,
  "verification": "fail",
  "diagnosis": "...",
  "next_action": "replan",
  "plan": {{
    "tool": "python",
    "code": "..."
  }}
}}
"""

    messages = [
        {
            "role": "system",
            "content": "You are a precise agent recovery controller. Output JSON only."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt"
    )

    inputs = inputs.to(model.device)

    outputs = model.generate(
        inputs,
        max_new_tokens=256,
        do_sample=False
    )

    generated_tokens = outputs[0][inputs.shape[1]:]

    raw = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    print("\nQwen3 raw response:")
    print(raw)

    decision = json.loads(raw)

    assert decision["task_complete"] is False
    assert decision["verification"] == "fail"
    assert decision["next_action"] == "replan"
    assert "plan" in decision

    return decision


def unified_verified_agent_loop(task, initial_plan, max_retries=2):

    current_plan = initial_plan

    for attempt in range(max_retries + 1):

        print(f"\n--- ATTEMPT {attempt + 1} ---")

        engine = ExecutionEngine(
            runtime.tools,
            runtime.evaluator,
            runtime.learning,
            max_steps=5
        )

        observations = engine.execute_plan(current_plan)
        observation = observations[0]

        output = observation.output

        print("returncode:", output.get("returncode"))
        print("stdout:", repr(output.get("stdout", "")))
        print("stderr:", repr(output.get("stderr", "")))

        # ----------------------------------------------------
        # 1. Execution failure
        # ----------------------------------------------------

        if output.get("returncode") != 0:

            if attempt >= max_retries:
                return {
                    "success": False,
                    "attempts": attempt + 1,
                    "result": None,
                    "observation": observation,
                    "decision": None
                }

            verification_message = (
                "Execution failed. The Python program must be corrected."
            )

            decision = qwen3_replan_verified(
                task,
                observation,
                verification_message
            )

        # ----------------------------------------------------
        # 2. Execution succeeded
        # ----------------------------------------------------

        else:

            verified, verification_message = deterministic_verify(
                task,
                output.get("stdout", "")
            )

            print("Deterministic verification:", verified)
            print("Verification message:", verification_message)

            # Ground-truth PASS
            if verified:

                return {
                    "success": True,
                    "attempts": attempt + 1,
                    "result": output.get("stdout", "").strip(),
                    "observation": observation,
                    "decision": {
                        "task_complete": True,
                        "verification": "pass",
                        "next_action": "finish",
                        "verification_message": verification_message
                    }
                }

            # Ground-truth FAIL
            if attempt >= max_retries:

                return {
                    "success": False,
                    "attempts": attempt + 1,
                    "result": output.get("stdout", "").strip(),
                    "observation": observation,
                    "decision": None
                }

            decision = qwen3_replan_verified(
                task,
                observation,
                verification_message
            )

        # ----------------------------------------------------
        # Build corrected Plan
        # ----------------------------------------------------

        corrected_code = decision["plan"]["code"]

        current_plan = Plan(
            task_id=task.id,
            steps=[
                {
                    "action": "tool",
                    "tool": "python",
                    "args": [corrected_code],
                    "kwargs": {}
                }
            ],
            assumptions=[]
        )

    return {
        "success": False,
        "attempts": max_retries + 1,
        "result": None,
        "observation": None,
        "decision": None
    }


# ============================================================
# TEST: intentionally wrong but executable result
# ============================================================

task = Task(
    id="KAGGLE-048-FIX",
    goal="Calculate the average of 10, 20, and 30."
)

wrong_code = """numbers = [10, 20, 30]
average = 25
print(average)"""

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [wrong_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)


result = unified_verified_agent_loop(
    task,
    initial_plan,
    max_retries=2
)


print("\n" + "=" * 65)
print("KAGGLE-048-FIX RESULT")
print("=" * 65)

print("success :", result["success"])
print("attempts:", result["attempts"])
print("result  :", result["result"])

assert result["success"] is True

final_result = float(result["result"])

assert abs(final_result - 20.0) < 1e-9

print("\nKAGGLE-048-FIX STATUS : PASS")
print("Final verified result :", final_result)
````

````python
# ============================================================
# KAGGLE-049
# Reusable Verification Interface
# ============================================================

class VerificationResult:
    def __init__(self, passed, message, expected=None, actual=None):
        self.passed = passed
        self.message = message
        self.expected = expected
        self.actual = actual

    def __repr__(self):
        return (
            f"VerificationResult("
            f"passed={self.passed}, "
            f"message={self.message!r}, "
            f"expected={self.expected!r}, "
            f"actual={self.actual!r})"
        )


class ResultVerifier:
    """
    Generic verifier interface.
    Concrete verifiers implement verify().
    """

    def verify(self, task, observation):
        raise NotImplementedError


class NumericResultVerifier(ResultVerifier):
    """
    Verifies a numeric stdout result against an expected value.
    """

    def __init__(self, expected, tolerance=1e-9):
        self.expected = float(expected)
        self.tolerance = tolerance

    def verify(self, task, observation):

        output = observation.output

        if output.get("returncode") != 0:
            return VerificationResult(
                passed=False,
                message="Execution failed.",
                expected=self.expected,
                actual=None
            )

        stdout = output.get("stdout", "").strip()

        try:
            actual = float(stdout)
        except ValueError:
            return VerificationResult(
                passed=False,
                message="Output is not numeric.",
                expected=self.expected,
                actual=stdout
            )

        passed = abs(actual - self.expected) < self.tolerance

        if passed:
            message = (
                f"Result matches expected value {self.expected}."
            )
        else:
            message = (
                f"Expected {self.expected}, but received {actual}."
            )

        return VerificationResult(
            passed=passed,
            message=message,
            expected=self.expected,
            actual=actual
        )


# ============================================================
# KAGGLE-049 TEST
# ============================================================

task = Task(
    id="KAGGLE-049",
    goal="Calculate the average of 10, 20, and 30."
)


# Deliberately wrong but executable result
wrong_code = """numbers = [10, 20, 30]
average = 25
print(average)"""


wrong_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [wrong_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)


engine = ExecutionEngine(
    runtime.tools,
    runtime.evaluator,
    runtime.learning,
    max_steps=5
)


observations = engine.execute_plan(wrong_plan)
observation = observations[0]


# ------------------------------------------------------------
# Test reusable verifier
# ------------------------------------------------------------

verifier = NumericResultVerifier(expected=20.0)

verification = verifier.verify(
    task,
    observation
)


print("=" * 65)
print("KAGGLE-049 VERIFIER TEST")
print("=" * 65)

print("Observation output:")
print(observation.output)

print("\nVerification:")
print(verification)

print("\nPassed :", verification.passed)
print("Expected:", verification.expected)
print("Actual  :", verification.actual)
print("Message :", verification.message)


# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert isinstance(verification, VerificationResult)

assert verification.passed is False

assert verification.expected == 20.0

assert verification.actual == 25.0

print("\nKAGGLE-049 STATUS : PASS")
print("Reusable verifier correctly rejected wrong result.")
````

````python
# ============================================================
# KAGGLE-050
# Pluggable Verified Agent Loop
# ============================================================

def qwen3_replan_from_verification(task, observation, verification):

    output = observation.output

    prompt = f"""
You are the recovery controller for a Personal AI agent.

Task:
{task.goal}

Execution output:
returncode: {output.get("returncode")}
stdout: {output.get("stdout", "")}
stderr: {output.get("stderr", "")}

Verifier result:
passed: {verification.passed}
message: {verification.message}
expected: {verification.expected}
actual: {verification.actual}

The verifier is authoritative.
Because verification failed, create a corrected Python plan.

Return ONLY valid JSON:

{{
  "task_complete": false,
  "verification": "fail",
  "diagnosis": "...",
  "next_action": "replan",
  "plan": {{
    "tool": "python",
    "code": "..."
  }}
}}
"""

    messages = [
        {
            "role": "system",
            "content": "You are a precise recovery controller. Output JSON only."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt"
    )

    inputs = inputs.to(model.device)

    outputs = model.generate(
        inputs,
        max_new_tokens=256,
        do_sample=False
    )

    generated_tokens = outputs[0][inputs.shape[1]:]

    raw = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    print("\nQwen3 replan:")
    print(raw)

    decision = json.loads(raw)

    assert decision["task_complete"] is False
    assert decision["next_action"] == "replan"
    assert "plan" in decision

    return decision


def verified_agent_loop(
    task,
    initial_plan,
    verifier,
    max_retries=2
):

    current_plan = initial_plan

    for attempt in range(max_retries + 1):

        print(f"\n--- ATTEMPT {attempt + 1} ---")

        engine = ExecutionEngine(
            runtime.tools,
            runtime.evaluator,
            runtime.learning,
            max_steps=5
        )

        observations = engine.execute_plan(current_plan)
        observation = observations[0]

        print("Observation:", observation.output)

        verification = verifier.verify(
            task,
            observation
        )

        print("Verifier:", verification)

        # ----------------------------------------------------
        # VERIFIED SUCCESS
        # ----------------------------------------------------

        if verification.passed:

            return {
                "success": True,
                "attempts": attempt + 1,
                "result": observation.output["stdout"].strip(),
                "observation": observation,
                "verification": verification
            }

        # ----------------------------------------------------
        # VERIFICATION FAILURE
        # ----------------------------------------------------

        if attempt >= max_retries:

            return {
                "success": False,
                "attempts": attempt + 1,
                "result": observation.output["stdout"].strip(),
                "observation": observation,
                "verification": verification
            }

        decision = qwen3_replan_from_verification(
            task,
            observation,
            verification
        )

        corrected_code = decision["plan"]["code"]

        current_plan = Plan(
            task_id=task.id,
            steps=[
                {
                    "action": "tool",
                    "tool": "python",
                    "args": [corrected_code],
                    "kwargs": {}
                }
            ],
            assumptions=[]
        )

    raise RuntimeError("Agent loop exhausted unexpectedly")


# ============================================================
# TEST
# ============================================================

task = Task(
    id="KAGGLE-050",
    goal="Calculate the average of 10, 20, and 30."
)


# Deliberately wrong executable plan
wrong_code = """numbers = [10, 20, 30]
average = 25
print(average)"""


initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [wrong_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)


# Plug in reusable verifier
verifier = NumericResultVerifier(
    expected=20.0
)


# Run generic agent loop
result = verified_agent_loop(
    task,
    initial_plan,
    verifier,
    max_retries=2
)


print("\n" + "=" * 65)
print("KAGGLE-050 RESULT")
print("=" * 65)

print("success :", result["success"])
print("attempts:", result["attempts"])
print("result  :", result["result"])
print("verified:", result["verification"].passed)


assert result["success"] is True
assert result["verification"].passed is True

final_result = float(result["result"])

assert abs(final_result - 20.0) < 1e-9

print("\nKAGGLE-050 STATUS : PASS")
print("Final verified result:", final_result)
````

````python
# ================================================================
# KAGGLE-052 : MEMORY → QWEN3 CONTEXT
# ================================================================

import json

print("=" * 65)
print("KAGGLE-052 : MEMORY → QWEN3 CONTEXT")
print("=" * 65)

# ------------------------------------------------
# 1. Create a task that depends on stored memory
# ------------------------------------------------

task_text = (
    "Explain which model role is used in my Personal AI architecture "
    "and what major components work around it."
)

# ------------------------------------------------
# 2. Retrieve relevant persistent memory
# ------------------------------------------------

memory_results = runtime.memory.search(
    query="replaceable model brain Qwen3 tools execution observation verification learning",
    limit=8
)

assert len(memory_results) > 0, "No relevant memory retrieved."

memory_context = []

for item in memory_results:
    if isinstance(item, dict):
        memory_context.append({
            "category": item.get("category"),
            "content": item.get("content"),
            "source": item.get("source"),
            "confidence": item.get("confidence")
        })
    else:
        memory_context.append({
            "content": str(item)
        })

print("\nRetrieved memories:")
print(json.dumps(memory_context, indent=2, ensure_ascii=False))

# ------------------------------------------------
# 3. Build Qwen3 context using retrieved memory
# ------------------------------------------------

prompt = f"""
You are the reasoning model inside a Personal AI system.

Task:
{task_text}

Relevant persistent memory:
{json.dumps(memory_context, ensure_ascii=False)}

Use the supplied memory as context.

Return ONLY valid JSON.
Do not use markdown.
Do not include <think>.

Required format:
{{
  "answer": "short answer",
  "memory_used": true
}}
"""

# ------------------------------------------------
# 4. Qwen3 inference
# ------------------------------------------------

messages = [
    {
        "role": "user",
        "content": prompt
    }
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    enable_thinking=False,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

outputs = model.generate(
    inputs,
    max_new_tokens=256,
    do_sample=False
)

generated = outputs[0][inputs.shape[-1]:]

raw_response = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

print("\nQwen3 raw response:")
print(raw_response)

# ------------------------------------------------
# 5. Validate structured response
# ------------------------------------------------

decision = json.loads(raw_response)

assert isinstance(decision, dict)
assert "answer" in decision
assert "memory_used" in decision
assert decision["memory_used"] is True

print("\nParsed decision:")
print(json.dumps(decision, indent=2, ensure_ascii=False))

# ------------------------------------------------
# 6. Final verification
# ------------------------------------------------

answer_lower = str(decision["answer"]).lower()

memory_signal_found = (
    "qwen3" in answer_lower
    and (
        "tool" in answer_lower
        or "execution" in answer_lower
        or "verification" in answer_lower
        or "memory" in answer_lower
    )
)

assert memory_signal_found, (
    "Qwen3 response did not appear to use the retrieved architecture memory."
)

print("\n" + "=" * 65)
print("KAGGLE-052 RESULT")
print("=" * 65)
print("Memory retrieval : PASS")
print("Context injection: PASS")
print("Qwen3 inference  : PASS")
print("JSON contract    : PASS")
print("Memory used      : PASS")
print("=" * 65)
print("KAGGLE-052 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-053 : MULTI-STEP PLANNER
# ================================================================

import json

from personal_ai.core.contracts import Task, Plan

print("=" * 65)
print("KAGGLE-053 : MULTI-STEP PLANNER")
print("=" * 65)

# ------------------------------------------------
# 1. Define task
# ------------------------------------------------

task = Task(
    id="KAGGLE-053",
    description=(
        "Calculate the average of 10, 20, and 30. "
        "Use multiple clear computational steps."
    )
)

print("\nTask:")
print(task)

# ------------------------------------------------
# 2. Retrieve memory context
# ------------------------------------------------

memory_results = runtime.memory.search(
    query="Qwen3 replaceable model brain tools execution observation verification learning",
    limit=5
)

assert len(memory_results) > 0

memory_context = []

for item in memory_results:
    if isinstance(item, dict):
        memory_context.append({
            "category": item.get("category"),
            "content": item.get("content"),
            "confidence": item.get("confidence")
        })

print("\nMemory context : PASS")

# ------------------------------------------------
# 3. Ask Qwen3 for a MULTI-STEP executable plan
# ------------------------------------------------

prompt = f"""
You are the planning component of a Personal AI.

Task:
{task.description}

Relevant memory:
{json.dumps(memory_context, ensure_ascii=False)}

Create a multi-step executable plan.

Requirements:
1. Use only the python tool.
2. Produce 2 or more steps.
3. Each step must contain executable Python code.
4. Steps must be ordered logically.
5. The final step must print the answer.
6. Do not execute the code yourself.
7. Return ONLY valid JSON.
8. Do not use markdown.
9. Do not include <think>.

Required JSON format:

{{
  "task_complete": false,
  "next_action": "execute",
  "steps": [
    {{
      "tool": "python",
      "code": "..."
    }},
    {{
      "tool": "python",
      "code": "..."
    }}
  ]
}}
"""

messages = [
    {
        "role": "user",
        "content": prompt
    }
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    enable_thinking=False,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

outputs = model.generate(
    inputs,
    max_new_tokens=512,
    do_sample=False
)

generated = outputs[0][inputs.shape[-1]:]

raw_response = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

print("\nQwen3 planner response:")
print(raw_response)

# ------------------------------------------------
# 4. Parse planner output
# ------------------------------------------------

plan_data = json.loads(raw_response)

assert isinstance(plan_data, dict)
assert plan_data["task_complete"] is False
assert plan_data["next_action"] == "execute"

steps = plan_data["steps"]

assert isinstance(steps, list)
assert len(steps) >= 2

print("\nPlanner JSON       : PASS")
print("Number of steps   :", len(steps))

# ------------------------------------------------
# 5. Validate every step
# ------------------------------------------------

for index, step in enumerate(steps, start=1):

    assert isinstance(step, dict)
    assert step.get("tool") == "python"

    code = step.get("code")

    assert isinstance(code, str)
    assert len(code.strip()) > 0

    print(f"Step {index} : VALID")

# ------------------------------------------------
# 6. Convert Qwen3 plan into V3 Plan
# ------------------------------------------------

v3_steps = []

for step in steps:
    v3_steps.append({
        "action": "tool",
        "tool": step["tool"],
        "args": [step["code"]],
        "kwargs": {}
    })

v3_plan = Plan(
    task_id=task.id,
    steps=v3_steps,
    assumptions=[]
)

print("\nV3 Plan:")
print(v3_plan)

assert len(v3_plan.steps) >= 2

print("\nV3 Plan conversion : PASS")

# ------------------------------------------------
# 7. Final status
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-053 RESULT")
print("=" * 65)
print("Task creation       : PASS")
print("Memory context      : PASS")
print("Qwen3 planning      : PASS")
print("Multi-step output   : PASS")
print("Step validation     : PASS")
print("V3 Plan conversion  : PASS")
print("=" * 65)
print("KAGGLE-053 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-054 : MULTI-STEP PLAN → EXECUTION
# ================================================================

from personal_ai.execution.engine import ExecutionEngine

print("=" * 65)
print("KAGGLE-054 : MULTI-STEP PLAN → EXECUTION")
print("=" * 65)

# ------------------------------------------------
# 1. Verify KAGGLE-053 plan exists
# ------------------------------------------------

assert "v3_plan" in globals(), "v3_plan not found. Use the KAGGLE-053 plan."

print("\nPlan received:")
print(v3_plan)

assert len(v3_plan.steps) >= 2
print("Multi-step plan : PASS")

# ------------------------------------------------
# 2. Create ExecutionEngine using proven V3 APIs
# ------------------------------------------------

engine = ExecutionEngine(
    runtime.tools,
    runtime.evaluator,
    runtime.learning,
    max_steps=5
)

print("ExecutionEngine : PASS")

# ------------------------------------------------
# 3. Execute the complete plan
# ------------------------------------------------

observations = engine.execute_plan(v3_plan)

assert isinstance(observations, list)
assert len(observations) == len(v3_plan.steps)

print("\nNumber of observations:", len(observations))
print("Observation count : PASS")

# ------------------------------------------------
# 4. Inspect every observation
# ------------------------------------------------

for index, observation in enumerate(observations, start=1):

    print(f"\n--- Observation {index} ---")
    print(observation)

    assert observation.ok is True

    output = observation.output

    assert isinstance(output, dict)
    assert "returncode" in output

    print("Observation object : PASS")
    print("Python returncode  :", output["returncode"])

# ------------------------------------------------
# 5. Verify execution-level success
# ------------------------------------------------

returncodes = [
    observation.output["returncode"]
    for observation in observations
]

print("\nReturncodes:", returncodes)

all_processes_succeeded = all(
    code == 0 for code in returncodes
)

print("All Python processes successful:", all_processes_succeeded)

# ------------------------------------------------
# 6. Inspect stdout
# ------------------------------------------------

for index, observation in enumerate(observations, start=1):

    output = observation.output

    print(f"\nStep {index} stdout:")
    print(repr(output.get("stdout", "")))

    if output.get("stderr"):
        print(f"Step {index} stderr:")
        print(output["stderr"])

# ------------------------------------------------
# 7. Determine whether cross-step Python state persisted
# ------------------------------------------------

step2_stdout = observations[1].output.get("stdout", "")
step2_stderr = observations[1].output.get("stderr", "")
step2_returncode = observations[1].output["returncode"]

cross_step_state_persisted = (
    step2_returncode == 0
    and "20" in step2_stdout
)

print("\nCross-step state persisted:", cross_step_state_persisted)

# ------------------------------------------------
# 8. Final status
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-054 RESULT")
print("=" * 65)
print("Plan received          : PASS")
print("ExecutionEngine        : PASS")
print("Observation collection : PASS")
print("Step 1 execution       :", "PASS" if returncodes[0] == 0 else "FAIL")
print("Step 2 execution       :", "PASS" if returncodes[1] == 0 else "FAIL")
print("Cross-step state       :", "PASS" if cross_step_state_persisted else "NOT PERSISTED")
print("=" * 65)

if all_processes_succeeded and cross_step_state_persisted:
    print("KAGGLE-054 STATUS : PASS")
else:
    print("KAGGLE-054 STATUS : NEEDS EXECUTION-CONTRACT FIX")
````

````python
# ================================================================
# KAGGLE-054.1-R2 : TOOL REGISTRY CONTRACT INSPECTION
# ================================================================

import inspect

print("=" * 65)
print("KAGGLE-054.1-R2 : TOOL REGISTRY CONTRACT INSPECTION")
print("=" * 65)

tools = runtime.tools

print("\nTool registry object:")
print(tools)

print("\nTool registry type:")
print(type(tools))

# ------------------------------------------------
# 1. Inspect ToolRegistry API
# ------------------------------------------------

print("\nToolRegistry public attributes/methods:")

public_attrs = [
    name for name in dir(tools)
    if not name.startswith("_")
]

for name in public_attrs:
    print(" -", name)

# ------------------------------------------------
# 2. Inspect call() signature
# ------------------------------------------------

try:
    print("\nToolRegistry.call signature:")
    print(inspect.signature(tools.call))
except Exception as e:
    print("call signature unavailable:", e)

# ------------------------------------------------
# 3. Inspect ToolRegistry source
# ------------------------------------------------

try:
    print("\nToolRegistry source:")
    print(inspect.getsource(type(tools)))
except Exception as e:
    print("ToolRegistry source unavailable:", e)

# ------------------------------------------------
# 4. Inspect internal attributes safely
# ------------------------------------------------

print("\nToolRegistry internal attributes:")

for name in dir(tools):
    if name.startswith("_") and not name.startswith("__"):
        try:
            value = getattr(tools, name)

            # Avoid dumping huge objects
            if isinstance(value, (dict, list, tuple, set)):
                print(f" - {name}: {type(value).__name__}, size={len(value)}")
            else:
                print(f" - {name}: {type(value).__name__}")
        except Exception:
            pass

# ------------------------------------------------
# 5. Find python tool through ToolRegistry API
# ------------------------------------------------

python_tool = None

for candidate in ["get", "lookup", "resolve", "find"]:
    if hasattr(tools, candidate):
        method = getattr(tools, candidate)

        try:
            python_tool = method("python")
            print(f"\nPython tool resolved using: {candidate}()")
            print(python_tool)
            break
        except Exception as e:
            print(f"{candidate}('python') unavailable:", e)

# ------------------------------------------------
# 6. If not resolved, inspect likely internal mapping
# ------------------------------------------------

if python_tool is None:

    print("\nPython tool was not resolved through a public lookup method.")

    for name in dir(tools):
        if name.startswith("_") and not name.startswith("__"):
            try:
                value = getattr(tools, name)

                if isinstance(value, dict) and "python" in value:
                    python_tool = value["python"]
                    print(f"Python tool found in internal mapping: {name}")
                    print(python_tool)
                    break

            except Exception:
                pass

# ------------------------------------------------
# 7. Inspect Python tool if found
# ------------------------------------------------

if python_tool is not None:

    print("\nPython tool type:")
    print(type(python_tool))

    print("\nPython tool public attributes:")

    for name in dir(python_tool):
        if not name.startswith("_"):
            print(" -", name)

    try:
        print("\nPython tool call signature:")
        print(inspect.signature(python_tool))
    except Exception as e:
        print("Tool call signature unavailable:", e)

    try:
        print("\nPython tool source:")
        print(inspect.getsource(type(python_tool)))
    except Exception as e:
        print("Python tool source unavailable:", e)

else:
    print("\nPython tool could not be resolved automatically.")

print("\n" + "=" * 65)
print("KAGGLE-054.1-R2 INSPECTION COMPLETE")
print("=" * 65)
````

````python
# ================================================================
# KAGGLE-054.3 : SHARED-CONTEXT MULTI-STEP EXECUTOR
# ================================================================

import io
import contextlib

from personal_ai.core.contracts import Observation

print("=" * 65)
print("KAGGLE-054.3 : SHARED-CONTEXT MULTI-STEP EXECUTOR")
print("=" * 65)

# ------------------------------------------------
# 1. Reuse the proven KAGGLE-053 plan
# ------------------------------------------------

assert "v3_plan" in globals()

print("\nPlan:")
print(v3_plan)

assert len(v3_plan.steps) >= 2
print("Plan validation : PASS")

# ------------------------------------------------
# 2. Create shared execution context
# ------------------------------------------------

shared_context = {}

observations = []

print("\nShared context created : PASS")

# ------------------------------------------------
# 3. Execute each step in SAME namespace
# ------------------------------------------------

for index, step in enumerate(v3_plan.steps, start=1):

    assert step["action"] == "tool"
    assert step["tool"] == "python"

    code = step["args"][0]

    stdout_buffer = io.StringIO()

    try:

        with contextlib.redirect_stdout(stdout_buffer):
            exec(code, shared_context)

        stdout = stdout_buffer.getvalue()

        observation = Observation(
            True,
            {
                "returncode": 0,
                "stdout": stdout,
                "stderr": ""
            }
        )

    except Exception as e:

        stdout = stdout_buffer.getvalue()

        observation = Observation(
            True,
            {
                "returncode": 1,
                "stdout": stdout,
                "stderr": f"{type(e).__name__}: {e}"
            }
        )

    observations.append(observation)

    print(f"\n--- Step {index} ---")
    print("Code:")
    print(code)
    print("Observation:")
    print(observation)

# ------------------------------------------------
# 4. Validate observation collection
# ------------------------------------------------

assert len(observations) == len(v3_plan.steps)

print("\nObservation collection : PASS")

# ------------------------------------------------
# 5. Validate execution success
# ------------------------------------------------

returncodes = [
    obs.output["returncode"]
    for obs in observations
]

print("Returncodes:", returncodes)

assert all(code == 0 for code in returncodes)

print("All steps successful : PASS")

# ------------------------------------------------
# 6. Verify cross-step state
# ------------------------------------------------

assert "numbers" in shared_context
assert shared_context["numbers"] == [10, 20, 30]

assert "average" in shared_context
assert abs(shared_context["average"] - 20.0) < 1e-9

print("\nCross-step state : PASS")
print("numbers          :", shared_context["numbers"])
print("average          :", shared_context["average"])

# ------------------------------------------------
# 7. Verify final stdout
# ------------------------------------------------

final_stdout = observations[-1].output["stdout"].strip()

assert final_stdout == "20.0"

print("Final result      :", final_stdout)
print("Final output      : PASS")

# ------------------------------------------------
# 8. Final status
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-054.3 RESULT")
print("=" * 65)
print("V3 Plan                 : PASS")
print("Shared execution context: PASS")
print("Multi-step execution    : PASS")
print("Observation collection  : PASS")
print("Cross-step state        : PASS")
print("Final result            : PASS")
print("=" * 65)
print("KAGGLE-054.3 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-054.4 : REUSABLE SHARED-CONTEXT EXECUTOR
# ================================================================

import io
import contextlib

from personal_ai.core.contracts import Observation

print("=" * 65)
print("KAGGLE-054.4 : REUSABLE SHARED-CONTEXT EXECUTOR")
print("=" * 65)


class SharedPythonExecutor:

    def __init__(self):
        self.context = {}

    def execute_plan(self, plan):

        observations = []

        for step in plan.steps:

            if step.get("action") != "tool":
                observations.append(
                    Observation(
                        True,
                        {
                            "returncode": 0,
                            "stdout": "",
                            "stderr": ""
                        }
                    )
                )
                continue

            if step.get("tool") != "python":
                observations.append(
                    Observation(
                        True,
                        {
                            "returncode": 1,
                            "stdout": "",
                            "stderr": (
                                f"Unsupported tool: "
                                f"{step.get('tool')}"
                            )
                        }
                    )
                )
                continue

            code = step["args"][0]

            stdout_buffer = io.StringIO()

            try:

                with contextlib.redirect_stdout(stdout_buffer):
                    exec(code, self.context)

                observations.append(
                    Observation(
                        True,
                        {
                            "returncode": 0,
                            "stdout": stdout_buffer.getvalue(),
                            "stderr": ""
                        }
                    )
                )

            except Exception as e:

                observations.append(
                    Observation(
                        True,
                        {
                            "returncode": 1,
                            "stdout": stdout_buffer.getvalue(),
                            "stderr": (
                                f"{type(e).__name__}: {e}"
                            )
                        }
                    )
                )

        return observations


# ------------------------------------------------
# 1. Create executor
# ------------------------------------------------

executor = SharedPythonExecutor()

print("\nExecutor creation : PASS")

# ------------------------------------------------
# 2. Execute existing V3 plan
# ------------------------------------------------

assert "v3_plan" in globals()

observations = executor.execute_plan(v3_plan)

print("Plan execution    : PASS")

# ------------------------------------------------
# 3. Validate observations
# ------------------------------------------------

assert len(observations) == len(v3_plan.steps)

print("Observation count :", len(observations))

returncodes = [
    obs.output["returncode"]
    for obs in observations
]

assert returncodes == [0, 0]

print("Returncodes       :", returncodes)
print("Execution status  : PASS")

# ------------------------------------------------
# 4. Validate shared state
# ------------------------------------------------

assert executor.context["numbers"] == [10, 20, 30]

assert abs(executor.context["average"] - 20.0) < 1e-9

print("\nShared state       : PASS")
print("numbers            :", executor.context["numbers"])
print("average            :", executor.context["average"])

# ------------------------------------------------
# 5. Validate final output
# ------------------------------------------------

final_output = observations[-1].output["stdout"].strip()

assert final_output == "20.0"

print("Final output       :", final_output)
print("Result validation  : PASS")

# ------------------------------------------------
# 6. Final status
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-054.4 RESULT")
print("=" * 65)
print("Reusable executor  : PASS")
print("V3 Plan execution  : PASS")
print("Observations       : PASS")
print("Shared state       : PASS")
print("Final result       : PASS")
print("=" * 65)
print("KAGGLE-054.4 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-057 : FULL AUTONOMOUS AGENT LOOP
# ================================================================

import json

from personal_ai.core.contracts import Task, Plan

print("=" * 65)
print("KAGGLE-057 : FULL AUTONOMOUS AGENT LOOP")
print("=" * 65)

# ------------------------------------------------
# 1. Confirm required components
# ------------------------------------------------

assert "SharedPythonExecutor" in globals()
assert "NumericResultVerifier" in globals()
assert "model" in globals()
assert "tokenizer" in globals()

print("\nQwen3 model         : PASS")
print("Shared executor     : PASS")
print("Verifier            : PASS")

# ------------------------------------------------
# 2. Create autonomous task
# ------------------------------------------------

task = Task(
    id="KAGGLE-057",
    goal="Calculate the average of 10, 20, and 30."
)

print("Task                : PASS")

# ------------------------------------------------
# 3. Qwen3 creates initial plan
# ------------------------------------------------

planning_prompt = f"""
You are the planning component of a Personal AI agent.

Task:
{task.goal}

Create a Python execution plan.

Return ONLY valid JSON:

{{
  "task_complete": false,
  "next_action": "execute",
  "steps": [
    {{
      "tool": "python",
      "code": "Python code"
    }}
  ]
}}

Use clear computational steps.
"""

messages = [
    {
        "role": "system",
        "content": "You are a precise planning agent. Return JSON only."
    },
    {
        "role": "user",
        "content": planning_prompt
    }
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    enable_thinking=False,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

outputs = model.generate(
    inputs,
    max_new_tokens=256,
    do_sample=False
)

generated_tokens = outputs[0][inputs.shape[-1]:]

raw_plan = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
).strip()

print("\nQwen3 initial plan:")
print(raw_plan)

planner_decision = json.loads(raw_plan)

assert planner_decision["task_complete"] is False
assert planner_decision["next_action"] == "execute"
assert isinstance(planner_decision["steps"], list)
assert len(planner_decision["steps"]) >= 1

print("\nQwen3 planning    : PASS")

# ------------------------------------------------
# 4. Convert Qwen3 plan → V3 Plan
# ------------------------------------------------

plan_steps = []

for step in planner_decision["steps"]:

    assert step["tool"] == "python"
    assert isinstance(step["code"], str)

    plan_steps.append(
        {
            "action": "tool",
            "tool": "python",
            "args": [step["code"]],
            "kwargs": {}
        }
    )

initial_plan = Plan(
    task_id=task.id,
    steps=plan_steps,
    assumptions=[]
)

print("V3 Plan conversion : PASS")
print("Number of steps     :", len(initial_plan.steps))

# ------------------------------------------------
# 5. Execute plan
# ------------------------------------------------

executor = SharedPythonExecutor()

observations = executor.execute_plan(initial_plan)

print("\nInitial execution  : PASS")

returncodes = [
    obs.output["returncode"]
    for obs in observations
]

print("Returncodes         :", returncodes)

# ------------------------------------------------
# 6. Extract execution result
# ------------------------------------------------

last_observation = observations[-1]

print("\nLast observation:")
print(last_observation)

# ------------------------------------------------
# 7. Deterministic verification
# ------------------------------------------------

verifier = NumericResultVerifier(
    expected=20.0,
    tolerance=1e-9
)

verification = verifier.verify(
    task,
    last_observation
)

print("\nVerification:")
print(verification)

# ------------------------------------------------
# 8. Autonomous recovery if verification fails
# ------------------------------------------------

attempts = 1
max_attempts = 3

while not verification.passed and attempts < max_attempts:

    print("\n" + "-" * 65)
    print(f"AUTONOMOUS RECOVERY ATTEMPT {attempts}")
    print("-" * 65)

    failure_payload = {
        "returncode": last_observation.output.get("returncode"),
        "stdout": last_observation.output.get("stdout"),
        "stderr": last_observation.output.get("stderr"),
        "verification": {
            "passed": verification.passed,
            "message": verification.message,
            "expected": verification.expected,
            "actual": verification.actual
        },
        "state": {
            key: value
            for key, value in executor.context.items()
            if isinstance(value, (int, float, str, list, dict, bool))
        }
    }

    recovery_prompt = f"""
You are the recovery planner of a Personal AI agent.

Task:
{task.goal}

The previous attempt did not pass verification.

Failure and verification information:
{json.dumps(failure_payload, indent=2)}

Create a corrected Python execution step.

Return ONLY valid JSON:

{{
  "task_complete": false,
  "next_action": "replan",
  "diagnosis": "short explanation",
  "plan": {{
    "tool": "python",
    "code": "corrected Python code"
  }}
}}

Use existing execution state when possible.
The final printed output must be the answer to the task.
"""

    recovery_messages = [
        {
            "role": "system",
            "content": (
                "You are a precise recovery agent. "
                "Return valid JSON only."
            )
        },
        {
            "role": "user",
            "content": recovery_prompt
        }
    ]

    recovery_inputs = tokenizer.apply_chat_template(
        recovery_messages,
        tokenize=True,
        add_generation_prompt=True,
        enable_thinking=False,
        return_tensors="pt"
    )

    recovery_inputs = recovery_inputs.to(model.device)

    recovery_outputs = model.generate(
        recovery_inputs,
        max_new_tokens=256,
        do_sample=False
    )

    recovery_tokens = recovery_outputs[0][recovery_inputs.shape[-1]:]

    raw_recovery = tokenizer.decode(
        recovery_tokens,
        skip_special_tokens=True
    ).strip()

    print("\nQwen3 recovery:")
    print(raw_recovery)

    recovery_decision = json.loads(raw_recovery)

    assert recovery_decision["task_complete"] is False
    assert recovery_decision["next_action"] == "replan"
    assert "diagnosis" in recovery_decision
    assert recovery_decision["plan"]["tool"] == "python"

    print("\nQwen3 diagnosis   : PASS")
    print("Diagnosis          :", recovery_decision["diagnosis"])

    recovery_code = recovery_decision["plan"]["code"]

    recovery_plan = Plan(
        task_id=task.id,
        steps=[
            {
                "action": "tool",
                "tool": "python",
                "args": [recovery_code],
                "kwargs": {}
            }
        ],
        assumptions=[]
    )

    print("Replanned V3 Plan  : PASS")

    # Execute in SAME context
    recovery_observations = executor.execute_plan(
        recovery_plan
    )

    last_observation = recovery_observations[-1]

    print("\nRecovery execution : PASS")
    print("Returncode         :", last_observation.output["returncode"])
    print("stdout             :", repr(
        last_observation.output["stdout"]
    ))

    verification = verifier.verify(
        task,
        last_observation
    )

    attempts += 1

    print("\nRecovery verification:")
    print(verification)

# ------------------------------------------------
# 9. Final autonomous decision
# ------------------------------------------------

assert verification.passed is True

final_output = last_observation.output["stdout"].strip()

assert abs(float(final_output) - 20.0) < 1e-9

print("\n" + "=" * 65)
print("KAGGLE-057 RESULT")
print("=" * 65)
print("Qwen3 planning          : PASS")
print("V3 plan conversion      : PASS")
print("Execution               : PASS")
print("Observation             : PASS")
print("Deterministic verify    : PASS")
print("Autonomous recovery     : PASS")
print("Final result            :", final_output)
print("Attempts                :", attempts)
print("=" * 65)
print("KAGGLE-057 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-061 : LEARNING FROM EXECUTION OUTCOME
# ================================================================

import inspect

from personal_ai.core.contracts import Task

print("=" * 65)
print("KAGGLE-061 : LEARNING FROM EXECUTION OUTCOME")
print("=" * 65)

# ------------------------------------------------
# 1. Confirm existing components
# ------------------------------------------------

assert "runtime" in globals()
assert "model" in globals()
assert "tokenizer" in globals()

print("\nV3 runtime  : PASS")
print("Qwen3 model : PASS")

# ------------------------------------------------
# 2. Locate LearningLedger
# ------------------------------------------------

learning = getattr(runtime, "learning", None)

assert learning is not None, (
    "runtime.learning is not available."
)

print("Learning    :", type(learning).__name__)

# ------------------------------------------------
# 3. Inspect LearningLedger public API
# ------------------------------------------------

public_methods = [
    name for name in dir(learning)
    if not name.startswith("_")
]

print("Learning API:", public_methods)

# We expect the existing V3 LearningLedger to provide
# a method for recording learning.
candidate_methods = [
    name for name in public_methods
    if any(
        word in name.lower()
        for word in [
            "add",
            "record",
            "learn",
            "store",
            "append"
        ]
    )
]

print("Learning write candidates:", candidate_methods)

assert len(candidate_methods) > 0, (
    "No LearningLedger write method discovered."
)

print("Learning API discovery : PASS")

# ------------------------------------------------
# 4. Create a successful task experience
# ------------------------------------------------

task = Task(
    id="KAGGLE-061",
    goal="Calculate the average of 10, 20, and 30."
)

experience = {
    "task_id": task.id,
    "goal": task.goal,
    "outcome": "success",
    "tool": "python",
    "result": 20.0,
    "lesson": (
        "For numerical average calculations, Python can compute "
        "sum(numbers) / len(numbers), followed by deterministic "
        "numeric verification."
    )
}

print("\nExperience created : PASS")

# ------------------------------------------------
# 5. Determine the write method
# ------------------------------------------------

write_method = None

for name in candidate_methods:

    method = getattr(learning, name)

    if not callable(method):
        continue

    try:
        signature = inspect.signature(method)
        parameters = list(signature.parameters)

        print(
            f"  Candidate: {name}{signature}"
        )

        # Prefer a method that can accept one learning record.
        if len(parameters) >= 1:
            write_method = method
            write_method_name = name
            break

    except Exception:
        pass

assert write_method is not None

print(
    "Selected learning method:",
    write_method_name
)

# ------------------------------------------------
# 6. Write learning record
# ------------------------------------------------

# Try the existing ledger API with the complete
# experience object. If the implementation expects
# keyword fields, fall back to those fields.

write_success = False

try:
    write_method(experience)
    write_success = True
except TypeError:

    try:
        write_method(
            task_id=experience["task_id"],
            outcome=experience["outcome"],
            lesson=experience["lesson"],
            metadata={
                "goal": experience["goal"],
                "tool": experience["tool"],
                "result": experience["result"]
            }
        )
        write_success = True
    except TypeError:
        pass

assert write_success, (
    "LearningLedger write API did not accept the "
    "detected learning record format."
)

print("Learning write : PASS")

# ------------------------------------------------
# 7. Save if supported
# ------------------------------------------------

save_method = getattr(learning, "save", None)

if callable(save_method):
    save_method()
    print("Learning save  : PASS")
else:
    print("Learning save  : NOT REQUIRED")

# ------------------------------------------------
# 8. Locate read/retrieval API
# ------------------------------------------------

read_candidates = [
    name for name in public_methods
    if any(
        word in name.lower()
        for word in [
            "search",
            "get",
            "list",
            "read",
            "retrieve",
            "history"
        ]
    )
]

print(
    "Learning read candidates:",
    read_candidates
)

assert len(read_candidates) > 0, (
    "No LearningLedger retrieval method discovered."
)

# ------------------------------------------------
# 9. Fresh LearningLedger instance
# ------------------------------------------------

LearningLedgerClass = type(learning)

learning_root = getattr(
    learning,
    "root",
    None
)

assert learning_root is not None, (
    "Could not determine LearningLedger root."
)

fresh_learning = LearningLedgerClass(
    learning_root
)

print(
    "Fresh ledger instance:",
    type(fresh_learning).__name__
)

# ------------------------------------------------
# 10. Retrieve persisted learning
# ------------------------------------------------

retrieved_learning = None

# Prefer search-like APIs.
for name in read_candidates:

    method = getattr(fresh_learning, name)

    if not callable(method):
        continue

    try:

        signature = inspect.signature(method)

        if len(signature.parameters) == 0:
            result = method()
        else:
            result = method(
                "KAGGLE-061 average Python numerical verification"
            )

        if result:
            retrieved_learning = result
            print(
                "Retrieved using:",
                name
            )
            break

    except Exception:
        continue

assert retrieved_learning is not None

print("Learning retrieval : PASS")

# ------------------------------------------------
# 11. Verify our lesson survived persistence
# ------------------------------------------------

retrieved_text = str(
    retrieved_learning
).lower()

assert "kaggle-061" in retrieved_text
assert "success" in retrieved_text
assert "python" in retrieved_text
assert "average" in retrieved_text

print("Learning persistence: PASS")

# ------------------------------------------------
# 12. Final status
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-061 RESULT")
print("=" * 65)

print("LearningLedger discovered : PASS")
print("Successful experience     : PASS")
print("Learning write             : PASS")
print("Learning persistence       : PASS")
print("Fresh ledger retrieval     : PASS")
print("Lesson verification       : PASS")

print("=" * 65)
print("KAGGLE-061 STATUS : PASS")
````

````python
# ================================================================
# KAGGLE-063 : LONG-RUNNING AUTONOMOUS LOOP
# ================================================================

import json
from pathlib import Path
from datetime import datetime, timezone

from personal_ai.core.contracts import Task

print("=" * 65)
print("KAGGLE-063 : LONG-RUNNING AUTONOMOUS LOOP")
print("=" * 65)

# ------------------------------------------------
# 1. Runtime checks
# ------------------------------------------------

assert "runtime" in globals()
assert "model" in globals()
assert "tokenizer" in globals()

print("\nV3 runtime  : PASS")
print("Qwen3 model : PASS")

# ------------------------------------------------
# 2. Persistent state paths
# ------------------------------------------------

PROJECT_ROOT = Path("/kaggle/working/Personal_AI_AGI_Core_v3_0")

state_root = PROJECT_ROOT / "storage" / "agent_state"
state_root.mkdir(parents=True, exist_ok=True)

state_path = state_root / "KAGGLE-063_state.json"

# ------------------------------------------------
# 3. Create long-running task
# ------------------------------------------------

task = Task(
    id="KAGGLE-063",
    goal="Calculate and verify the average of 10, 20, and 30."
)

state = {
    "version": 1,
    "task_id": task.id,
    "goal": task.goal,
    "status": "running",
    "current_step": 0,
    "total_steps": 4,
    "attempts": 0,
    "history": [],
    "last_result": None,
    "verification_passed": False,
    "updated_at": datetime.now(timezone.utc).isoformat()
}

print("Task created : PASS")

# ------------------------------------------------
# 4. Persistent state writer
# ------------------------------------------------

def persist_state(state):
    state["updated_at"] = datetime.now(timezone.utc).isoformat()

    with open(state_path, "w", encoding="utf-8") as f:
        json.dump(state, f, indent=2, ensure_ascii=False)

# ------------------------------------------------
# 5. Step 1 — Plan
# ------------------------------------------------

state["current_step"] = 1

plan = {
    "tool": "python",
    "code": (
        "numbers = [10, 20, 30]\n"
        "average = sum(numbers) / len(numbers)\n"
        "print(average)"
    )
}

state["history"].append({
    "step": 1,
    "phase": "planning",
    "action": "python numerical calculation"
})

persist_state(state)

print("Step 1 - Planning       : PASS")

# ------------------------------------------------
# 6. Step 2 — Execute
# ------------------------------------------------

state["current_step"] = 2
state["attempts"] += 1

execution = runtime.tools.call(
    plan["tool"],
    plan["code"]
)

state["history"].append({
    "step": 2,
    "phase": "execution",
    "tool": plan["tool"],
    "returncode": execution["returncode"],
    "stdout": execution["stdout"],
    "stderr": execution["stderr"]
})

persist_state(state)

assert execution["returncode"] == 0

print("Step 2 - Execution       : PASS")
print("Execution stdout        :", execution["stdout"].strip())

# ------------------------------------------------
# 7. Step 3 — Deterministic verification
# ------------------------------------------------

state["current_step"] = 3

raw_output = execution["stdout"].strip()
actual = float(raw_output)

expected = 20.0
tolerance = 1e-9

verification_passed = abs(actual - expected) <= tolerance

state["verification_passed"] = verification_passed

state["history"].append({
    "step": 3,
    "phase": "verification",
    "expected": expected,
    "actual": actual,
    "passed": verification_passed
})

persist_state(state)

assert verification_passed

print("Step 3 - Verification   : PASS")

# ------------------------------------------------
# 8. Step 4 — Learn from successful execution
# ------------------------------------------------

state["current_step"] = 4
state["last_result"] = actual

lesson = (
    "For numerical average calculations, use Python with "
    "sum(numbers) / len(numbers), then apply deterministic "
    "numeric verification."
)

learning_event = {
    "task_id": task.id,
    "goal": task.goal,
    "outcome": "success",
    "tool": "python",
    "result": actual,
    "lesson": lesson
}

recorded = runtime.learning.record(learning_event)

assert recorded == learning_event

state["history"].append({
    "step": 4,
    "phase": "learning",
    "outcome": "success",
    "lesson": lesson
})

# ------------------------------------------------
# 9. Complete task and persist final state
# ------------------------------------------------

state["status"] = "completed"

persist_state(state)

print("Step 4 - Learning        : PASS")
print("Final state persistence  : PASS")

# ------------------------------------------------
# 10. Fresh reload
# ------------------------------------------------

with open(state_path, "r", encoding="utf-8") as f:
    restored_state = json.load(f)

assert restored_state["task_id"] == "KAGGLE-063"
assert restored_state["status"] == "completed"
assert restored_state["current_step"] == 4
assert restored_state["attempts"] == 1
assert restored_state["last_result"] == 20.0
assert restored_state["verification_passed"] is True
assert len(restored_state["history"]) == 4

print("Fresh state reload       : PASS")

# ------------------------------------------------
# 11. Autonomous continuation decision
# ------------------------------------------------

if restored_state["status"] == "completed":
    next_action = "finish"
elif not restored_state["verification_passed"]:
    next_action = "recover"
else:
    next_action = "continue"

assert next_action == "finish"

print("Autonomous state decision: PASS")

# ------------------------------------------------
# 12. Final result
# ------------------------------------------------

print("\n" + "=" * 65)
print("KAGGLE-063 RESULT")
print("=" * 65)

print("Task creation             : PASS")
print("Persistent state          : PASS")
print("Planning                  : PASS")
print("Execution                 : PASS")
print("Deterministic verification: PASS")
print("Learning                  : PASS")
print("Final state persistence   : PASS")
print("Fresh state reload        : PASS")
print("Autonomous finish decision: PASS")

print("\nFinal result :", restored_state["last_result"])
print("Attempts     :", restored_state["attempts"])
print("Status       :", restored_state["status"])

print("=" * 65)
print("KAGGLE-063 STATUS : PASS")
print("=" * 65)
````

````python
# ================================================================
# KAGGLE-065 : END-TO-END AUTONOMOUS BENCHMARK
# ================================================================

import json
import re
from pathlib import Path
from datetime import datetime, timezone

from personal_ai.core.contracts import Task, Plan
from personal_ai.execution.engine import ExecutionEngine

print("=" * 70)
print("KAGGLE-065 : END-TO-END AUTONOMOUS BENCHMARK")
print("=" * 70)

# ------------------------------------------------
# 1. Runtime / model checks
# ------------------------------------------------

assert "runtime" in globals()
assert "model" in globals()
assert "tokenizer" in globals()

print("\nV3 runtime  : PASS")
print("Qwen3 model : PASS")

# ------------------------------------------------
# 2. Benchmark task
# ------------------------------------------------

task = Task(
    id="KAGGLE-065",
    goal=(
        "Calculate the average of 10, 20, and 30. "
        "The first execution attempt may fail, but the agent "
        "must recover, verify the corrected result, learn from "
        "the experience, persist state, and finish successfully."
    )
)

print("Benchmark task : PASS")

# ------------------------------------------------
# 3. Persistent benchmark state
# ------------------------------------------------

PROJECT_ROOT = Path(
    "/kaggle/working/Personal_AI_AGI_Core_v3_0"
)

state_root = PROJECT_ROOT / "storage" / "agent_state"
state_root.mkdir(parents=True, exist_ok=True)

state_path = state_root / "KAGGLE-065_state.json"

state = {
    "version": 1,
    "task_id": task.id,
    "goal": task.goal,
    "status": "running",
    "attempts": 0,
    "history": [],
    "memory_used": False,
    "recovery_used": False,
    "verification_passed": False,
    "final_result": None,
    "updated_at": datetime.now(timezone.utc).isoformat()
}


def persist_state():
    state["updated_at"] = datetime.now(timezone.utc).isoformat()

    with open(state_path, "w", encoding="utf-8") as f:
        json.dump(state, f, indent=2, ensure_ascii=False)


persist_state()

print("Persistent state initialized : PASS")

# ------------------------------------------------
# 4. Memory retrieval
# ------------------------------------------------

memory_results = runtime.memory.search(
    "average Python numerical verification",
    limit=5
)

assert memory_results is not None

state["memory_used"] = len(memory_results) > 0

state["history"].append({
    "phase": "memory_retrieval",
    "matches": len(memory_results)
})

persist_state()

print("Memory retrieval            : PASS")
print("Memory matches              :", len(memory_results))

# ------------------------------------------------
# 5. Qwen3 planning
# ------------------------------------------------

planning_prompt = """
Return ONLY valid JSON.

Task:
Calculate the average of 10, 20, and 30.

Use the available tool:
python

Prior memory indicates:
- Python is appropriate for numerical calculation.
- The result should be deterministically verified.

Return exactly this structure:

{
  "steps": [
    {
      "tool": "python",
      "code": "..."
    }
  ],
  "verification": {
    "method": "numeric",
    "expected": 20
  }
}
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are an execution planner. "
            "Return only valid JSON."
        )
    },
    {
        "role": "user",
        "content": planning_prompt
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

# Put inputs on the first model device.
input_device = next(model.parameters()).device
inputs = {k: v.to(input_device) for k, v in inputs.items()}

outputs = model.generate(
    **inputs,
    max_new_tokens=256,
    do_sample=False
)

generated = outputs[0][inputs["input_ids"].shape[1]:]

raw_plan = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

# Remove accidental markdown fences if present.
raw_plan = re.sub(
    r"^```json\s*|\s*```$",
    "",
    raw_plan,
    flags=re.IGNORECASE
).strip()

plan_json = json.loads(raw_plan)

assert "steps" in plan_json
assert len(plan_json["steps"]) >= 1
assert plan_json["steps"][0]["tool"] == "python"

state["history"].append({
    "phase": "qwen3_planning",
    "plan": plan_json
})

persist_state()

print("Qwen3 structured planning   : PASS")

# ------------------------------------------------
# 6. Goal decomposition
# ------------------------------------------------

decomposition_prompt = """
Return ONLY valid JSON.

Decompose this task into ordered subtasks:

"Calculate the average of 10, 20, and 30, then verify the result."

Return:

{
  "subtasks": [
    "...",
    "...",
    "..."
  ]
}
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are a task decomposition agent. "
            "Return only valid JSON."
        )
    },
    {
        "role": "user",
        "content": decomposition_prompt
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

inputs = {
    k: v.to(input_device)
    for k, v in inputs.items()
}

outputs = model.generate(
    **inputs,
    max_new_tokens=256,
    do_sample=False
)

generated = outputs[0][inputs["input_ids"].shape[1]:]

raw_decomposition = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

raw_decomposition = re.sub(
    r"^```json\s*|\s*```$",
    "",
    raw_decomposition,
    flags=re.IGNORECASE
).strip()

decomposition = json.loads(raw_decomposition)

assert "subtasks" in decomposition
assert len(decomposition["subtasks"]) >= 2

state["history"].append({
    "phase": "goal_decomposition",
    "subtasks": decomposition["subtasks"]
})

persist_state()

print("Goal decomposition           : PASS")

# ------------------------------------------------
# 7. Dynamic tool selection
# ------------------------------------------------

available_tools = sorted(runtime.tools.tools.keys())

tool_selection_prompt = f"""
Return ONLY valid JSON.

Task:
Calculate the average of 10, 20, and 30.

Available tools:
{available_tools}

Choose the best tool for the numerical computation.

Return:

{{
  "tool": "...",
  "reason": "..."
}}
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are a tool-selection agent. "
            "Return only valid JSON."
        )
    },
    {
        "role": "user",
        "content": tool_selection_prompt
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

inputs = {
    k: v.to(input_device)
    for k, v in inputs.items()
}

outputs = model.generate(
    **inputs,
    max_new_tokens=128,
    do_sample=False
)

generated = outputs[0][inputs["input_ids"].shape[1]:]

raw_selection = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

raw_selection = re.sub(
    r"^```json\s*|\s*```$",
    "",
    raw_selection,
    flags=re.IGNORECASE
).strip()

selection = json.loads(raw_selection)

assert selection["tool"] in available_tools
assert selection["tool"] == "python"

state["history"].append({
    "phase": "tool_selection",
    "tool": selection["tool"],
    "reason": selection.get("reason", "")
})

persist_state()

print("Dynamic tool selection       : PASS")
print("Selected tool                :", selection["tool"])

# ------------------------------------------------
# 8. Intentional recoverable failure
# ------------------------------------------------

state["attempts"] += 1

bad_code = """
numbers = [10, 20, 30]
average = sum(numbers) / 0
print(average)
"""

failed_execution = runtime.tools.call(
    "python",
    bad_code
)

assert failed_execution["returncode"] != 0

state["history"].append({
    "phase": "execution",
    "attempt": 1,
    "success": False,
    "returncode": failed_execution["returncode"],
    "stderr": failed_execution["stderr"]
})

state["recovery_used"] = True

persist_state()

print("Intentional failure detected : PASS")

# ------------------------------------------------
# 9. Qwen3 recovery diagnosis
# ------------------------------------------------

recovery_prompt = f"""
Return ONLY valid JSON.

The Python execution failed.

Error:
{failed_execution["stderr"]}

Task:
Calculate the average of 10, 20, and 30.

Return:

{{
  "diagnosis": "...",
  "corrected_code": "..."
}}

The corrected code MUST print the final numeric result.
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are a recovery planner. "
            "Diagnose the failure and produce corrected Python. "
            "Return only valid JSON."
        )
    },
    {
        "role": "user",
        "content": recovery_prompt
    }
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    enable_thinking=False
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

inputs = {
    k: v.to(input_device)
    for k, v in inputs.items()
}

outputs = model.generate(
    **inputs,
    max_new_tokens=256,
    do_sample=False
)

generated = outputs[0][inputs["input_ids"].shape[1]:]

raw_recovery = tokenizer.decode(
    generated,
    skip_special_tokens=True
).strip()

raw_recovery = re.sub(
    r"^```json\s*|\s*```$",
    "",
    raw_recovery,
    flags=re.IGNORECASE
).strip()

recovery = json.loads(raw_recovery)

assert "corrected_code" in recovery
assert "diagnosis" in recovery
assert len(recovery["corrected_code"].strip()) > 0

print("Qwen3 recovery diagnosis    : PASS")

# ------------------------------------------------
# 10. Execute recovered plan
# ------------------------------------------------

state["attempts"] += 1

recovered_execution = runtime.tools.call(
    "python",
    recovery["corrected_code"]
)

assert recovered_execution["returncode"] == 0

recovered_stdout = recovered_execution["stdout"].strip()

assert recovered_stdout != ""

state["history"].append({
    "phase": "recovery_execution",
    "attempt": 2,
    "success": True,
    "returncode": recovered_execution["returncode"],
    "stdout": recovered_stdout
})

persist_state()

print("Recovered execution          : PASS")
print("Recovered stdout             :", recovered_stdout)

# ------------------------------------------------
# 11. Deterministic verification
# ------------------------------------------------

actual = float(recovered_stdout.splitlines()[-1])

expected = 20.0
tolerance = 1e-9

verification_passed = (
    abs(actual - expected) <= tolerance
)

assert verification_passed

state["verification_passed"] = True
state["final_result"] = actual

state["history"].append({
    "phase": "verification",
    "expected": expected,
    "actual": actual,
    "passed": True
})

persist_state()

print("Deterministic verification   : PASS")

# ------------------------------------------------
# 12. Learn from complete outcome
# ------------------------------------------------

learning_event = {
    "task_id": task.id,
    "goal": task.goal,
    "outcome": "success_after_recovery",
    "tool": "python",
    "result": actual,
    "lesson": (
        "When numerical execution fails, inspect the execution "
        "error, generate corrected Python, execute again, and "
        "apply deterministic numeric verification before "
        "accepting the result."
    )
}

recorded = runtime.learning.record(
    learning_event
)

assert recorded == learning_event

state["history"].append({
    "phase": "learning",
    "outcome": "success_after_recovery"
})

persist_state()

print("Learning from recovery      : PASS")

# ------------------------------------------------
# 13. Complete persistent state
# ------------------------------------------------

state["status"] = "completed"

persist_state()

print("Final state persistence      : PASS")

# ------------------------------------------------
# 14. Fresh state reload
# ------------------------------------------------

with open(state_path, "r", encoding="utf-8") as f:
    restored_state = json.load(f)

assert restored_state["task_id"] == "KAGGLE-065"
assert restored_state["status"] == "completed"
assert restored_state["attempts"] == 2
assert restored_state["memory_used"] is True
assert restored_state["recovery_used"] is True
assert restored_state["verification_passed"] is True
assert restored_state["final_result"] == 20.0
assert len(restored_state["history"]) >= 7

print("Fresh state reload           : PASS")

# ------------------------------------------------
# 15. Final autonomous decision
# ------------------------------------------------

if (
    restored_state["status"] == "completed"
    and restored_state["verification_passed"]
):
    final_decision = "finish"
elif not restored_state["verification_passed"]:
    final_decision = "recover"
else:
    final_decision = "continue"

assert final_decision == "finish"

print("Final autonomous decision    : PASS")

# ------------------------------------------------
# 16. Benchmark summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("KAGGLE-065 END-TO-END BENCHMARK RESULT")
print("=" * 70)

print("Memory retrieval             : PASS")
print("Qwen3 planning               : PASS")
print("Goal decomposition           : PASS")
print("Dynamic tool selection       : PASS")
print("Failure detection            : PASS")
print("Qwen3 recovery               : PASS")
print("Recovered execution          : PASS")
print("Deterministic verification   : PASS")
print("Learning from outcome        : PASS")
print("Persistent state             : PASS")
print("Fresh state reload           : PASS")
print("Final autonomous decision    : PASS")

print("\nFinal result :", restored_state["final_result"])
print("Attempts     :", restored_state["attempts"])
print("Recovery     :", restored_state["recovery_used"])
print("Status       :", restored_state["status"])

print("=" * 70)
print("KAGGLE-065 STATUS : PASS")
print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-066 : GENERALIZED PLANNER
# ======================================================================

import json
from personal_ai.core.contracts import Task, Plan

print("=" * 70)
print("KAGGLE-066 : GENERALIZED PLANNER")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Generic planner prompt
# ----------------------------------------------------------------------

goal = """
Calculate the average of the numbers 10, 20, and 30.
The calculation must be performed using a tool and the final result
must be explicitly verified before reporting it.
""".strip()

task = Task(
    id="KAGGLE-066",
    goal=goal,
    metadata={"benchmark": "generalized_planner"}
)

available_tools = sorted(runtime.tools.tools.keys())

planner_prompt = f"""
You are the planning module of an autonomous AI agent.

Convert the user's goal into a structured executable plan.

USER GOAL:
{goal}

AVAILABLE TOOLS:
{available_tools}

Return ONLY valid JSON.
Do not use markdown.
Do not include <think>.

Required JSON schema:

{{
  "goal": "string",
  "task_complete": false,
  "steps": [
    {{
      "step_id": 1,
      "description": "string",
      "tool": "one available tool",
      "code": "executable code for that tool"
    }}
  ],
  "verification": {{
    "required": true,
    "method": "string"
  }},
  "completion_condition": "string"
}}

Rules:
1. Create an ordered plan.
2. Use an available tool when computation/action is required.
3. Every tool must exist in AVAILABLE TOOLS.
4. Python code must print the important result.
5. Include an explicit verification requirement.
6. Do not execute the task yourself.
7. Return JSON only.
"""

# ----------------------------------------------------------------------
# 2. Qwen3 structured generation
# ----------------------------------------------------------------------

inputs = tokenizer.apply_chat_template(
    [{"role": "user", "content": planner_prompt}],
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    enable_thinking=False
)

inputs = {k: v.to(next(model.parameters()).device) for k, v in inputs.items()}

with __import__("torch").inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=700,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

generated = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
).strip()

# ----------------------------------------------------------------------
# 3. Parse JSON
# ----------------------------------------------------------------------

try:
    plan_json = json.loads(generated)
    json_valid = True
except Exception as e:
    plan_json = None
    json_valid = False
    json_error = str(e)

print("JSON parse                 :", "PASS" if json_valid else "FAIL")

if not json_valid:
    print("Raw planner output:")
    print(generated)
    raise RuntimeError(f"KAGGLE-066 planner JSON invalid: {json_error}")

# ----------------------------------------------------------------------
# 4. Structural validation
# ----------------------------------------------------------------------

required_top = {
    "goal",
    "task_complete",
    "steps",
    "verification",
    "completion_condition"
}

top_valid = required_top.issubset(plan_json.keys())
steps_valid = (
    isinstance(plan_json.get("steps"), list)
    and len(plan_json["steps"]) > 0
)

step_schema_valid = True

for step in plan_json.get("steps", []):
    required_step = {"step_id", "description", "tool", "code"}

    if not required_step.issubset(step.keys()):
        step_schema_valid = False
        break

    if step["tool"] not in available_tools:
        step_schema_valid = False
        break

verification_valid = (
    isinstance(plan_json.get("verification"), dict)
    and plan_json["verification"].get("required") is True
    and bool(plan_json["verification"].get("method"))
)

completion_valid = bool(plan_json.get("completion_condition"))

print("Top-level schema            :", "PASS" if top_valid else "FAIL")
print("Steps present               :", "PASS" if steps_valid else "FAIL")
print("Step schemas                :", "PASS" if step_schema_valid else "FAIL")
print("Verification requirement    :", "PASS" if verification_valid else "FAIL")
print("Completion condition        :", "PASS" if completion_valid else "FAIL")

if not all([
    top_valid,
    steps_valid,
    step_schema_valid,
    verification_valid,
    completion_valid
]):
    print("\nPlanner output:")
    print(json.dumps(plan_json, indent=2, ensure_ascii=False))
    raise RuntimeError("KAGGLE-066 generalized planner validation failed")

# ----------------------------------------------------------------------
# 5. Convert Qwen3 plan into V3 Plan
# ----------------------------------------------------------------------

v3_steps = []

for step in plan_json["steps"]:
    v3_steps.append({
        "action": "tool",
        "tool": step["tool"],
        "args": [step["code"]],
        "kwargs": {}
    })

v3_plan = Plan(
    task_id=task.id,
    steps=v3_steps,
    assumptions=[]
)

print("V3 Plan conversion           : PASS")
print("Planned steps                :", len(v3_steps))

# ----------------------------------------------------------------------
# 6. Generalization checks
# ----------------------------------------------------------------------

goal_covered = (
    "average" in plan_json["goal"].lower()
    or "average" in goal.lower()
)

has_tool_action = any(
    step["action"] == "tool"
    for step in v3_plan.steps
)

all_tools_authorized = all(
    step["tool"] in available_tools
    for step in v3_plan.steps
)

print("Goal represented             :", "PASS" if goal_covered else "FAIL")
print("Executable tool actions      :", "PASS" if has_tool_action else "FAIL")
print("Tools authorized             :", "PASS" if all_tools_authorized else "FAIL")

# ----------------------------------------------------------------------
# 7. Display resulting generalized plan
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("GENERATED GENERALIZED PLAN")
print("=" * 70)

print(json.dumps(plan_json, indent=2, ensure_ascii=False))

print("\n" + "=" * 70)

if all([
    json_valid,
    top_valid,
    steps_valid,
    step_schema_valid,
    verification_valid,
    completion_valid,
    goal_covered,
    has_tool_action,
    all_tools_authorized
]):
    print("KAGGLE-066 STATUS : PASS")
else:
    print("KAGGLE-066 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-067 : GENERALIZED VERIFICATION
# ======================================================================

import json
from dataclasses import dataclass

from personal_ai.core.contracts import Task, Observation

print("=" * 70)
print("KAGGLE-067 : GENERALIZED VERIFICATION")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Generic verification result
# ----------------------------------------------------------------------

@dataclass
class GeneralVerificationResult:
    passed: bool
    message: str
    expected: object = None
    actual: object = None
    method: str = ""

# ----------------------------------------------------------------------
# 2. Generalized verifier
# ----------------------------------------------------------------------

class GeneralizedVerifier:

    def verify(self, task, observation, criteria):
        """
        Generic machine-checkable verification.

        Supported criteria:
        - numeric
        - contains
        - returncode
        - nonempty
        """

        if not isinstance(observation, Observation):
            return GeneralVerificationResult(
                False,
                "Invalid observation object",
                method="type_check"
            )

        # --------------------------------------------------------------
        # Extract execution result
        # --------------------------------------------------------------

        output = observation.output

        # ToolRegistry Python/shell output is normally a dict.
        if isinstance(output, dict):
            returncode = output.get("returncode")
            stdout = str(output.get("stdout", ""))
            stderr = str(output.get("stderr", ""))
        else:
            returncode = None
            stdout = str(output)
            stderr = ""

        # --------------------------------------------------------------
        # Verification method
        # --------------------------------------------------------------

        method = criteria.get("method")

        # --------------------------------------------------------------
        # 1. Return-code verification
        # --------------------------------------------------------------

        if method == "returncode":
            expected = criteria.get("expected", 0)

            passed = (
                returncode is not None
                and returncode == expected
            )

            return GeneralVerificationResult(
                passed=passed,
                message=(
                    f"returncode={returncode}, expected={expected}"
                ),
                expected=expected,
                actual=returncode,
                method="returncode"
            )

        # --------------------------------------------------------------
        # 2. Non-empty output verification
        # --------------------------------------------------------------

        if method == "nonempty":
            passed = bool(stdout.strip())

            return GeneralVerificationResult(
                passed=passed,
                message=(
                    "stdout contains output"
                    if passed
                    else "stdout is empty"
                ),
                expected="non-empty stdout",
                actual=stdout,
                method="nonempty"
            )

        # --------------------------------------------------------------
        # 3. Text containment verification
        # --------------------------------------------------------------

        if method == "contains":
            expected = str(criteria.get("expected", ""))

            passed = expected in stdout

            return GeneralVerificationResult(
                passed=passed,
                message=(
                    f"stdout contains '{expected}'"
                    if passed
                    else f"stdout does not contain '{expected}'"
                ),
                expected=expected,
                actual=stdout,
                method="contains"
            )

        # --------------------------------------------------------------
        # 4. Numeric verification
        # --------------------------------------------------------------

        if method == "numeric":
            expected = float(criteria["expected"])
            tolerance = float(criteria.get("tolerance", 1e-9))

            try:
                actual = float(stdout.strip().splitlines()[-1])
            except Exception:
                return GeneralVerificationResult(
                    False,
                    "Could not parse numeric result from stdout",
                    expected=expected,
                    actual=stdout,
                    method="numeric"
                )

            passed = abs(actual - expected) <= tolerance

            return GeneralVerificationResult(
                passed=passed,
                message=(
                    f"actual={actual}, expected={expected}, "
                    f"tolerance={tolerance}"
                ),
                expected=expected,
                actual=actual,
                method="numeric"
            )

        # --------------------------------------------------------------
        # Unsupported method
        # --------------------------------------------------------------

        return GeneralVerificationResult(
            False,
            f"Unsupported verification method: {method}",
            method=str(method)
        )


# ----------------------------------------------------------------------
# 3. Create benchmark task
# ----------------------------------------------------------------------

task = Task(
    id="KAGGLE-067",
    goal="Calculate the average of 10, 20, and 30 and verify the result."
)

verifier = GeneralizedVerifier()

# ----------------------------------------------------------------------
# 4. Execute successful Python observation
# ----------------------------------------------------------------------

success_output = runtime.tools.python(
    """
numbers = [10, 20, 30]
average = sum(numbers) / len(numbers)
print(average)
"""
)

success_observation = Observation(
    True,
    success_output
)

# ----------------------------------------------------------------------
# 5. Test numeric verification
# ----------------------------------------------------------------------

numeric_criteria = {
    "method": "numeric",
    "expected": 20.0,
    "tolerance": 1e-9
}

numeric_result = verifier.verify(
    task,
    success_observation,
    numeric_criteria
)

print(
    "Numeric verification        :",
    "PASS" if numeric_result.passed else "FAIL"
)
print("Numeric message             :", numeric_result.message)

# ----------------------------------------------------------------------
# 6. Test return-code verification
# ----------------------------------------------------------------------

returncode_criteria = {
    "method": "returncode",
    "expected": 0
}

returncode_result = verifier.verify(
    task,
    success_observation,
    returncode_criteria
)

print(
    "Return-code verification    :",
    "PASS" if returncode_result.passed else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Test non-empty verification
# ----------------------------------------------------------------------

nonempty_criteria = {
    "method": "nonempty"
}

nonempty_result = verifier.verify(
    task,
    success_observation,
    nonempty_criteria
)

print(
    "Non-empty verification      :",
    "PASS" if nonempty_result.passed else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Test text verification
# ----------------------------------------------------------------------

text_output = runtime.tools.python(
    """
print("FINAL RESULT: 20.0")
"""
)

text_observation = Observation(
    True,
    text_output
)

contains_result = verifier.verify(
    task,
    text_observation,
    {
        "method": "contains",
        "expected": "FINAL RESULT: 20.0"
    }
)

print(
    "Text verification           :",
    "PASS" if contains_result.passed else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Intentional verification failure
# ----------------------------------------------------------------------

wrong_criteria = {
    "method": "numeric",
    "expected": 25.0,
    "tolerance": 1e-9
}

wrong_result = verifier.verify(
    task,
    success_observation,
    wrong_criteria
)

failure_detection = not wrong_result.passed

print(
    "Incorrect result rejected   :",
    "PASS" if failure_detection else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Verification contract
# ----------------------------------------------------------------------

contract_valid = all([
    numeric_result.method == "numeric",
    returncode_result.method == "returncode",
    nonempty_result.method == "nonempty",
    contains_result.method == "contains",
    wrong_result.method == "numeric"
])

print(
    "Verification contract       :",
    "PASS" if contract_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Structured verification record
# ----------------------------------------------------------------------

verification_record = {
    "task_id": task.id,
    "verification": {
        "numeric": {
            "passed": numeric_result.passed,
            "expected": numeric_result.expected,
            "actual": numeric_result.actual
        },
        "returncode": {
            "passed": returncode_result.passed,
            "expected": returncode_result.expected,
            "actual": returncode_result.actual
        },
        "nonempty": {
            "passed": nonempty_result.passed
        },
        "contains": {
            "passed": contains_result.passed
        },
        "incorrect_result_rejected": failure_detection
    }
}

print("\n" + "=" * 70)
print("GENERALIZED VERIFICATION RECORD")
print("=" * 70)

print(
    json.dumps(
        verification_record,
        indent=2,
        ensure_ascii=False
    )
)

# ----------------------------------------------------------------------
# 12. Final status
# ----------------------------------------------------------------------

checks = [
    numeric_result.passed,
    returncode_result.passed,
    nonempty_result.passed,
    contains_result.passed,
    failure_detection,
    contract_valid
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-067 STATUS : PASS")
else:
    print("KAGGLE-067 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-068 : GENERALIZED AUTONOMOUS RECOVERY
# ======================================================================

import json
import torch

from personal_ai.core.contracts import Task, Plan, Observation

print("=" * 70)
print("KAGGLE-068 : GENERALIZED AUTONOMOUS RECOVERY")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Generalized verifier
# ----------------------------------------------------------------------

class RecoveryVerifier:
    def verify(self, observation, expected):
        output = observation.output

        if isinstance(output, dict):
            returncode = output.get("returncode")
            stdout = str(output.get("stdout", ""))
            stderr = str(output.get("stderr", ""))
        else:
            returncode = None
            stdout = str(output)
            stderr = ""

        # Execution failure
        if returncode is not None and returncode != 0:
            return {
                "passed": False,
                "type": "execution_failure",
                "message": stderr.strip() or "Tool execution failed",
                "actual": stdout
            }

        # Numeric verification
        try:
            actual = float(stdout.strip().splitlines()[-1])
            expected_value = float(expected)

            passed = abs(actual - expected_value) <= 1e-9

            return {
                "passed": passed,
                "type": "numeric",
                "message": (
                    f"actual={actual}, expected={expected_value}"
                ),
                "expected": expected_value,
                "actual": actual
            }

        except Exception:
            return {
                "passed": False,
                "type": "output_failure",
                "message": "Could not extract numeric result",
                "actual": stdout
            }


verifier = RecoveryVerifier()

# ----------------------------------------------------------------------
# 2. Benchmark task
# ----------------------------------------------------------------------

task = Task(
    id="KAGGLE-068",
    goal="""
Calculate the average of 10, 20, and 30.
The final result must be 20.0.
If execution or verification fails, diagnose the problem and recover.
""".strip()
)

print("Task creation               : PASS")

# ----------------------------------------------------------------------
# 3. Initial intentionally faulty plan
# ----------------------------------------------------------------------

initial_code = """
numbers = [10, 20, 30]
average = sum(numbers) / 0
print(average)
"""

initial_plan = Plan(
    task_id=task.id,
    steps=[
        {
            "action": "tool",
            "tool": "python",
            "args": [initial_code],
            "kwargs": {}
        }
    ],
    assumptions=[]
)

print("Initial faulty plan         : PASS")

# ----------------------------------------------------------------------
# 4. Execute initial plan
# ----------------------------------------------------------------------

initial_output = runtime.tools.call(
    "python",
    initial_code
)

initial_observation = Observation(
    True,
    initial_output
)

print("Initial execution           : PASS")

# ----------------------------------------------------------------------
# 5. Verify initial attempt
# ----------------------------------------------------------------------

expected_result = 20.0

initial_verification = verifier.verify(
    initial_observation,
    expected_result
)

failure_detected = not initial_verification["passed"]

print(
    "Initial failure detected    :",
    "PASS" if failure_detected else "FAIL"
)

print(
    "Failure type                :",
    initial_verification["type"]
)

# ----------------------------------------------------------------------
# 6. Qwen3 autonomous diagnosis + recovery plan
# ----------------------------------------------------------------------

recovery_prompt = f"""
You are the recovery controller of an autonomous AI agent.

TASK:
{task.goal}

INITIAL PLAN:
{json.dumps(initial_code)}

EXECUTION / VERIFICATION RESULT:
{json.dumps(initial_verification, ensure_ascii=False)}

The previous attempt failed.

Create a corrected recovery action.

Return ONLY valid JSON.
Do not use markdown.
Do not include <think>.

Required schema:

{{
  "diagnosis": "string",
  "recovery_action": "execute",
  "tool": "python",
  "code": "corrected executable Python code",
  "expected_result": 20.0
}}

Rules:
1. Diagnose why the previous attempt failed.
2. Correct the underlying problem.
3. The Python code MUST print the final numeric result.
4. Do not merely describe the fix.
5. Return executable code.
"""

recovery_input = tokenizer.apply_chat_template(
    [{"role": "user", "content": recovery_prompt}],
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    enable_thinking=False
)

recovery_input = recovery_input.to(
    next(model.parameters()).device
)

with torch.inference_mode():
    recovery_output = model.generate(
        input_ids=recovery_input,
        max_new_tokens=500,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

recovery_text = tokenizer.decode(
    recovery_output[0][recovery_input.shape[1]:],
    skip_special_tokens=True
).strip()

# ----------------------------------------------------------------------
# 7. Parse recovery JSON
# ----------------------------------------------------------------------

try:
    recovery_json = json.loads(recovery_text)
    recovery_json_valid = True
except Exception as e:
    recovery_json = None
    recovery_json_valid = False
    recovery_json_error = str(e)

print(
    "Qwen3 recovery diagnosis   :",
    "PASS" if recovery_json_valid else "FAIL"
)

if not recovery_json_valid:
    print("\nRaw recovery output:")
    print(recovery_text)
    raise RuntimeError(
        f"KAGGLE-068 recovery JSON invalid: {recovery_json_error}"
    )

# ----------------------------------------------------------------------
# 8. Validate recovery contract
# ----------------------------------------------------------------------

recovery_contract_valid = all([
    recovery_json.get("recovery_action") == "execute",
    recovery_json.get("tool") == "python",
    bool(recovery_json.get("code")),
    float(recovery_json.get("expected_result", -999)) == 20.0
])

print(
    "Recovery contract            :",
    "PASS" if recovery_contract_valid else "FAIL"
)

if not recovery_contract_valid:
    print(json.dumps(
        recovery_json,
        indent=2,
        ensure_ascii=False
    ))
    raise RuntimeError(
        "KAGGLE-068 recovery contract validation failed"
    )

# ----------------------------------------------------------------------
# 9. Execute recovered plan
# ----------------------------------------------------------------------

recovered_code = recovery_json["code"]

recovered_output = runtime.tools.call(
    recovery_json["tool"],
    recovered_code
)

recovered_observation = Observation(
    True,
    recovered_output
)

print("Recovered execution         : PASS")

# ----------------------------------------------------------------------
# 10. Verify recovered result
# ----------------------------------------------------------------------

recovered_verification = verifier.verify(
    recovered_observation,
    recovery_json["expected_result"]
)

recovery_verified = recovered_verification["passed"]

print(
    "Recovered verification      :",
    "PASS" if recovery_verified else "FAIL"
)

if isinstance(recovered_output, dict):
    print(
        "Recovered returncode        :",
        recovered_output.get("returncode")
    )
    print(
        "Recovered stdout            :",
        repr(recovered_output.get("stdout"))
    )

# ----------------------------------------------------------------------
# 11. Recovery loop decision
# ----------------------------------------------------------------------

attempts = 2
recovery_triggered = failure_detected
final_status = (
    "completed"
    if recovery_verified
    else "failed"
)

autonomous_recovery = (
    recovery_triggered
    and recovery_json_valid
    and recovery_contract_valid
    and recovery_verified
)

# ----------------------------------------------------------------------
# 12. Learning record
# ----------------------------------------------------------------------

learning_event = {
    "type": "recovery",
    "task_id": task.id,
    "failure_type": initial_verification["type"],
    "diagnosis": recovery_json.get("diagnosis"),
    "recovered": recovery_verified,
    "attempts": attempts
}

runtime.learning.record(learning_event)

print(
    "Learning from recovery      : PASS"
)

# ----------------------------------------------------------------------
# 13. Final autonomous state
# ----------------------------------------------------------------------

final_state = {
    "task_id": task.id,
    "status": final_status,
    "attempts": attempts,
    "recovery_triggered": recovery_triggered,
    "initial_failure": initial_verification,
    "recovery": recovery_json,
    "final_verification": recovered_verification
}

print("\n" + "=" * 70)
print("KAGGLE-068 RECOVERY RESULT")
print("=" * 70)

print(json.dumps(
    final_state,
    indent=2,
    ensure_ascii=False
))

print("\n" + "=" * 70)

checks = [
    failure_detected,
    recovery_json_valid,
    recovery_contract_valid,
    recovery_verified,
    autonomous_recovery
]

if all(checks):
    print("KAGGLE-068 STATUS : PASS")
else:
    print("KAGGLE-068 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-069 : PERSISTENT STATE V2
# ======================================================================

import json
from pathlib import Path
from datetime import datetime, timezone

from personal_ai.core.contracts import Task, Plan

print("=" * 70)
print("KAGGLE-069 : PERSISTENT STATE V2")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Persistent state manager
# ----------------------------------------------------------------------

class PersistentAgentStateV2:

    def __init__(self, root):
        self.root = Path(root)
        self.root.mkdir(parents=True, exist_ok=True)

    def _path(self, task_id):
        return self.root / f"{task_id}_state_v2.json"

    def save(self, state):
        state = dict(state)

        state["updated_at"] = datetime.now(
            timezone.utc
        ).isoformat()

        path = self._path(state["task_id"])

        path.write_text(
            json.dumps(
                state,
                indent=2,
                ensure_ascii=False
            ),
            encoding="utf-8"
        )

        return path

    def load(self, task_id):
        path = self._path(task_id)

        if not path.exists():
            return None

        return json.loads(
            path.read_text(encoding="utf-8")
        )


state_manager = PersistentAgentStateV2(
    PROJECT_ROOT / "storage" / "agent_state_v2"
)

print("State manager                : PASS")

# ----------------------------------------------------------------------
# 2. Create task
# ----------------------------------------------------------------------

task = Task(
    id="KAGGLE-069",
    goal="""
Calculate the average of 10, 20, and 30.
Execute the calculation, verify the result, and complete the task.
""".strip(),
    metadata={
        "benchmark": "persistent_state_v2"
    }
)

print("Task creation                : PASS")

# ----------------------------------------------------------------------
# 3. Create initial state
# ----------------------------------------------------------------------

state = {
    "version": 2,
    "task_id": task.id,
    "goal": task.goal,

    "status": "planning",

    "current_step": 0,
    "total_steps": 2,

    "attempts": 0,

    "plan": {
        "steps": [
            {
                "step_id": 1,
                "description": "Calculate average",
                "tool": "python"
            },
            {
                "step_id": 2,
                "description": "Verify result",
                "tool": "verifier"
            }
        ]
    },

    "observations": [],

    "failures": [],

    "recovery_history": [],

    "verification": {
        "status": "pending",
        "expected": 20.0,
        "actual": None
    },

    "learning_events": [],

    "resume_point": {
        "step_id": 1,
        "action": "execute"
    }
}

initial_path = state_manager.save(state)

print("Initial state saved          : PASS")

# ----------------------------------------------------------------------
# 4. Simulate progress
# ----------------------------------------------------------------------

state["status"] = "executing"
state["current_step"] = 1
state["attempts"] = 1

state["observations"].append({
    "step_id": 1,
    "tool": "python",
    "stdout": "20.0",
    "returncode": 0
})

state["resume_point"] = {
    "step_id": 2,
    "action": "verify"
}

state_manager.save(state)

print("Execution checkpoint         : PASS")

# ----------------------------------------------------------------------
# 5. Simulate verification
# ----------------------------------------------------------------------

state["status"] = "verifying"

state["verification"] = {
    "status": "passed",
    "expected": 20.0,
    "actual": 20.0,
    "method": "numeric"
}

state["current_step"] = 2

state_manager.save(state)

print("Verification checkpoint     : PASS")

# ----------------------------------------------------------------------
# 6. Simulate learning event
# ----------------------------------------------------------------------

learning_event = {
    "type": "successful_task",
    "lesson": "For numerical average calculation, Python execution followed by numeric verification succeeds.",
    "result": 20.0
}

state["learning_events"].append(
    learning_event
)

state_manager.save(state)

print("Learning checkpoint          : PASS")

# ----------------------------------------------------------------------
# 7. Simulate completion
# ----------------------------------------------------------------------

state["status"] = "completed"

state["resume_point"] = {
    "step_id": None,
    "action": "complete"
}

state["final_result"] = 20.0

state_manager.save(state)

print("Final state checkpoint       : PASS")

# ----------------------------------------------------------------------
# 8. Fresh process reload
# ----------------------------------------------------------------------

fresh_state_manager = PersistentAgentStateV2(
    PROJECT_ROOT / "storage" / "agent_state_v2"
)

reloaded_state = fresh_state_manager.load(
    task.id
)

reload_success = (
    reloaded_state is not None
    and reloaded_state["task_id"] == task.id
)

print(
    "Fresh state reload           :",
    "PASS" if reload_success else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Validate important state fields
# ----------------------------------------------------------------------

required_fields = [
    "version",
    "task_id",
    "goal",
    "status",
    "current_step",
    "total_steps",
    "attempts",
    "plan",
    "observations",
    "failures",
    "recovery_history",
    "verification",
    "learning_events",
    "resume_point"
]

state_schema_valid = all(
    field in reloaded_state
    for field in required_fields
)

print(
    "State schema                 :",
    "PASS" if state_schema_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Verify history survived reload
# ----------------------------------------------------------------------

history_persisted = (
    len(reloaded_state["observations"]) == 1
    and len(reloaded_state["learning_events"]) == 1
    and reloaded_state["verification"]["status"] == "passed"
)

print(
    "Execution history persisted :",
    "PASS" if history_persisted else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Verify resume information
# ----------------------------------------------------------------------

resume_information_valid = (
    reloaded_state["status"] == "completed"
    and reloaded_state["resume_point"]["action"] == "complete"
    and reloaded_state["final_result"] == 20.0
)

print(
    "Resume/completion state      :",
    "PASS" if resume_information_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 12. Simulate interrupted task
# ----------------------------------------------------------------------

interrupted_task_id = "KAGGLE-069-INTERRUPTED"

interrupted_state = {
    "version": 2,
    "task_id": interrupted_task_id,
    "goal": "Perform a two-step calculation and verify it.",
    "status": "executing",
    "current_step": 1,
    "total_steps": 2,
    "attempts": 1,
    "plan": {
        "steps": [
            {
                "step_id": 1,
                "description": "Calculate",
                "tool": "python"
            },
            {
                "step_id": 2,
                "description": "Verify",
                "tool": "verifier"
            }
        ]
    },
    "observations": [
        {
            "step_id": 1,
            "result": "calculation complete"
        }
    ],
    "failures": [],
    "recovery_history": [],
    "verification": {
        "status": "pending"
    },
    "learning_events": [],
    "resume_point": {
        "step_id": 2,
        "action": "verify"
    }
}

state_manager.save(interrupted_state)

# Fresh manager = simulated process restart
restart_manager = PersistentAgentStateV2(
    PROJECT_ROOT / "storage" / "agent_state_v2"
)

resume_state = restart_manager.load(
    interrupted_task_id
)

resume_detected = (
    resume_state is not None
    and resume_state["status"] == "executing"
    and resume_state["resume_point"]["step_id"] == 2
    and resume_state["resume_point"]["action"] == "verify"
)

print(
    "Interrupted-task resume      :",
    "PASS" if resume_detected else "FAIL"
)

# ----------------------------------------------------------------------
# 13. Final state report
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("PERSISTENT STATE V2 REPORT")
print("=" * 70)

print(
    json.dumps(
        reloaded_state,
        indent=2,
        ensure_ascii=False
    )
)

print("\n" + "=" * 70)

checks = [
    initial_path.exists(),
    reload_success,
    state_schema_valid,
    history_persisted,
    resume_information_valid,
    resume_detected
]

if all(checks):
    print("KAGGLE-069 STATUS : PASS")
else:
    print("KAGGLE-069 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-074 : TOOL ABSTRACTION
# ======================================================================

import json
from dataclasses import dataclass, asdict
from typing import Any, Dict, List

print("=" * 70)
print("KAGGLE-074 : TOOL ABSTRACTION")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Standard tool specification
# ----------------------------------------------------------------------

@dataclass
class ToolSpec:
    name: str
    description: str
    input_schema: Dict[str, Any]
    output_schema: Dict[str, Any]
    category: str
    requires_authorization: bool = True


print("ToolSpec abstraction          : PASS")

# ----------------------------------------------------------------------
# 2. Standard tool result
# ----------------------------------------------------------------------

@dataclass
class ToolResult:
    success: bool
    tool: str
    output: Any = None
    error: str = None
    metadata: Dict[str, Any] = None

    def to_dict(self):
        return asdict(self)


print("ToolResult abstraction        : PASS")

# ----------------------------------------------------------------------
# 3. Generic tool abstraction layer
# ----------------------------------------------------------------------

class ToolAbstractionLayer:

    def __init__(self, registry):
        self.registry = registry
        self.specs = {}

    def register_spec(self, spec: ToolSpec):
        self.specs[spec.name] = spec

    def list_tools(self):
        return list(self.specs.keys())

    def describe(self, name):
        if name not in self.specs:
            raise KeyError(f"Unknown tool: {name}")

        return self.specs[name]

    def execute(
        self,
        name,
        *args,
        authorized=True,
        **kwargs
    ):

        if name not in self.specs:
            return ToolResult(
                success=False,
                tool=name,
                error=f"Unknown tool: {name}"
            )

        spec = self.specs[name]

        if (
            spec.requires_authorization
            and not authorized
        ):
            return ToolResult(
                success=False,
                tool=name,
                error="Tool authorization denied"
            )

        try:
            output = self.registry.call(
                name,
                *args,
                **kwargs
            )

            return ToolResult(
                success=True,
                tool=name,
                output=output,
                metadata={
                    "category": spec.category
                }
            )

        except Exception as e:

            return ToolResult(
                success=False,
                tool=name,
                error=str(e),
                metadata={
                    "category": spec.category
                }
            )


tool_layer = ToolAbstractionLayer(
    runtime.tools
)

print("Tool abstraction layer       : PASS")

# ----------------------------------------------------------------------
# 4. Register existing V3 tools
# ----------------------------------------------------------------------

tool_layer.register_spec(
    ToolSpec(
        name="python",
        description=(
            "Execute Python code for deterministic computation "
            "and data processing."
        ),
        input_schema={
            "type": "string",
            "name": "code"
        },
        output_schema={
            "type": "object",
            "fields": [
                "returncode",
                "stdout",
                "stderr"
            ]
        },
        category="computation"
    )
)

tool_layer.register_spec(
    ToolSpec(
        name="shell",
        description=(
            "Execute an operating-system shell command."
        ),
        input_schema={
            "type": "string",
            "name": "command"
        },
        output_schema={
            "type": "object",
            "fields": [
                "returncode",
                "stdout",
                "stderr"
            ]
        },
        category="system"
    )
)

tool_layer.register_spec(
    ToolSpec(
        name="filesystem_read",
        description=(
            "Read a text file located inside the authorized "
            "project root."
        ),
        input_schema={
            "type": "string",
            "name": "path"
        },
        output_schema={
            "type": "object",
            "fields": [
                "path",
                "content"
            ]
        },
        category="filesystem"
    )
)

print(
    "Registered tools             :",
    tool_layer.list_tools()
)

# ----------------------------------------------------------------------
# 5. Tool discovery through abstraction
# ----------------------------------------------------------------------

expected_tools = {
    "python",
    "shell",
    "filesystem_read"
}

discovery_valid = (
    set(tool_layer.list_tools())
    == expected_tools
)

print(
    "Tool discovery               :",
    "PASS" if discovery_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Tool description interface
# ----------------------------------------------------------------------

python_spec = tool_layer.describe(
    "python"
)

description_valid = (
    python_spec.name == "python"
    and "Python" in python_spec.description
    and python_spec.category == "computation"
)

print(
    "Tool description interface    :",
    "PASS" if description_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Execute Python through abstraction
# ----------------------------------------------------------------------

python_result = tool_layer.execute(
    "python",
    "result = (10 + 20 + 30) / 3\nprint(result)"
)

python_execution_valid = (
    python_result.success
    and python_result.tool == "python"
    and python_result.output["returncode"] == 0
    and python_result.output["stdout"].strip() == "20.0"
)

print(
    "Python abstraction execution :",
    "PASS" if python_execution_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Execute filesystem tool through abstraction
# ----------------------------------------------------------------------

test_file = PROJECT_ROOT / "storage" / "tool_abstraction_test.txt"

test_file.parent.mkdir(
    parents=True,
    exist_ok=True
)

test_file.write_text(
    "KAGGLE-074 TOOL ABSTRACTION TEST",
    encoding="utf-8"
)

relative_path = test_file.relative_to(
    PROJECT_ROOT
)

filesystem_result = tool_layer.execute(
    "filesystem_read",
    str(relative_path)
)

filesystem_execution_valid = (
    filesystem_result.success
    and filesystem_result.tool == "filesystem_read"
    and filesystem_result.output["content"]
    == "KAGGLE-074 TOOL ABSTRACTION TEST"
)

print(
    "Filesystem abstraction        :",
    "PASS" if filesystem_execution_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Intentional failure through abstraction
# ----------------------------------------------------------------------

failure_result = tool_layer.execute(
    "python",
    "print(10 / 0)"
)

failure_contained = (
    not failure_result.success
    and failure_result.tool == "python"
    and isinstance(failure_result.output, dict)
    and failure_result.output["returncode"] != 0
)

print(
    "Execution failure handling    :",
    "PASS" if failure_contained else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Unknown tool handling
# ----------------------------------------------------------------------

unknown_result = tool_layer.execute(
    "unknown_tool",
    "test"
)

unknown_rejected = (
    not unknown_result.success
    and "Unknown tool" in unknown_result.error
)

print(
    "Unknown tool rejection        :",
    "PASS" if unknown_rejected else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Authorization handling
# ----------------------------------------------------------------------

authorization_result = tool_layer.execute(
    "python",
    "print(20)",
    authorized=False
)

authorization_rejected = (
    not authorization_result.success
    and "authorization" in authorization_result.error.lower()
)

print(
    "Authorization enforcement     :",
    "PASS" if authorization_rejected else "FAIL"
)

# ----------------------------------------------------------------------
# 12. Standard result schema
# ----------------------------------------------------------------------

result_dict = python_result.to_dict()

result_schema_valid = (
    set(result_dict.keys())
    >= {
        "success",
        "tool",
        "output",
        "error",
        "metadata"
    }
)

print(
    "Standard result schema        :",
    "PASS" if result_schema_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 13. Tool specifications are serializable
# ----------------------------------------------------------------------

serialized_specs = [
    asdict(tool_layer.describe(name))
    for name in tool_layer.list_tools()
]

serialization_valid = (
    len(serialized_specs) == 3
    and all(
        "name" in spec
        and "description" in spec
        and "input_schema" in spec
        and "output_schema" in spec
        for spec in serialized_specs
    )
)

print(
    "Tool specification serialization:",
    "PASS" if serialization_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 14. Final report
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TOOL ABSTRACTION REPORT")
print("=" * 70)

print(
    "Available tools:",
    json.dumps(
        serialized_specs,
        indent=2,
        ensure_ascii=False
    )
)

print("\nPython result:")
print(
    json.dumps(
        python_result.to_dict(),
        indent=2,
        ensure_ascii=False
    )
)

print("\nFailure result:")
print(
    json.dumps(
        failure_result.to_dict(),
        indent=2,
        ensure_ascii=False
    )
)

checks = [
    discovery_valid,
    description_valid,
    python_execution_valid,
    filesystem_execution_valid,
    failure_contained,
    unknown_rejected,
    authorization_rejected,
    result_schema_valid,
    serialization_valid
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-074 STATUS : PASS")
else:
    print("KAGGLE-074 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-074 FIX : NORMALIZE TOOL EXECUTION SUCCESS / FAILURE
# ======================================================================

import json

print("=" * 70)
print("KAGGLE-074 FIX : TOOL RESULT NORMALIZATION")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Replace execute() with correct success semantics
# ----------------------------------------------------------------------

def normalized_execute(
    self,
    name,
    *args,
    authorized=True,
    **kwargs
):

    if name not in self.specs:
        return ToolResult(
            success=False,
            tool=name,
            error=f"Unknown tool: {name}"
        )

    spec = self.specs[name]

    if (
        spec.requires_authorization
        and not authorized
    ):
        return ToolResult(
            success=False,
            tool=name,
            error="Tool authorization denied"
        )

    try:
        output = self.registry.call(
            name,
            *args,
            **kwargs
        )

        # --------------------------------------------------------------
        # IMPORTANT:
        # ToolRegistry returns returncode instead of raising an
        # exception for Python/shell execution failures.
        # --------------------------------------------------------------

        execution_failed = (
            isinstance(output, dict)
            and "returncode" in output
            and output["returncode"] != 0
        )

        if execution_failed:
            return ToolResult(
                success=False,
                tool=name,
                output=output,
                error=(
                    output.get("stderr", "").strip()
                    or f"Tool returned returncode {output['returncode']}"
                ),
                metadata={
                    "category": spec.category,
                    "returncode": output["returncode"]
                }
            )

        return ToolResult(
            success=True,
            tool=name,
            output=output,
            error=None,
            metadata={
                "category": spec.category,
                "returncode": (
                    output.get("returncode")
                    if isinstance(output, dict)
                    else None
                )
            }
        )

    except Exception as e:

        return ToolResult(
            success=False,
            tool=name,
            output=None,
            error=str(e),
            metadata={
                "category": spec.category
            }
        )


ToolAbstractionLayer.execute = normalized_execute

print("Execution normalization       : PASS")

# ----------------------------------------------------------------------
# 2. Verify successful Python execution
# ----------------------------------------------------------------------

success_result = tool_layer.execute(
    "python",
    "result = (10 + 20 + 30) / 3\nprint(result)"
)

success_test = (
    success_result.success
    and success_result.output["returncode"] == 0
    and success_result.output["stdout"].strip() == "20.0"
)

print(
    "Successful execution          :",
    "PASS" if success_test else "FAIL"
)

# ----------------------------------------------------------------------
# 3. Verify failed Python execution
# ----------------------------------------------------------------------

failure_result = tool_layer.execute(
    "python",
    "print(10 / 0)"
)

failure_test = (
    failure_result.success is False
    and failure_result.output["returncode"] == 1
    and "ZeroDivisionError" in failure_result.error
)

print(
    "Failed execution detection    :",
    "PASS" if failure_test else "FAIL"
)

# ----------------------------------------------------------------------
# 4. Verify failure remains observable
# ----------------------------------------------------------------------

failure_observable = (
    failure_result.output is not None
    and failure_result.output.get("returncode") == 1
    and failure_result.output.get("stderr", "").strip() != ""
)

print(
    "Failure details preserved     :",
    "PASS" if failure_observable else "FAIL"
)

# ----------------------------------------------------------------------
# 5. Verify unknown tool handling
# ----------------------------------------------------------------------

unknown_result = tool_layer.execute(
    "unknown_tool",
    "test"
)

unknown_test = (
    unknown_result.success is False
    and "Unknown tool" in unknown_result.error
)

print(
    "Unknown tool rejection        :",
    "PASS" if unknown_test else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Verify authorization
# ----------------------------------------------------------------------

authorization_result = tool_layer.execute(
    "python",
    "print(20)",
    authorized=False
)

authorization_test = (
    authorization_result.success is False
    and "authorization" in authorization_result.error.lower()
)

print(
    "Authorization enforcement     :",
    "PASS" if authorization_test else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Verify standard result semantics
# ----------------------------------------------------------------------

result_schema_test = (
    success_result.success is True
    and failure_result.success is False
    and success_result.tool == "python"
    and failure_result.tool == "python"
)

print(
    "Standard success semantics     :",
    "PASS" if result_schema_test else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Final corrected report
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CORRECTED TOOL RESULT REPORT")
print("=" * 70)

print("\nSuccessful result:")
print(
    json.dumps(
        success_result.to_dict(),
        indent=2,
        ensure_ascii=False
    )
)

print("\nFailed result:")
print(
    json.dumps(
        failure_result.to_dict(),
        indent=2,
        ensure_ascii=False
    )
)

checks = [
    success_test,
    failure_test,
    failure_observable,
    unknown_test,
    authorization_test,
    result_schema_test
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-074 STATUS : PASS")
else:
    print("KAGGLE-074 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-077 : TOOL FAILURE HANDLING
# ======================================================================

import json

print("=" * 70)
print("KAGGLE-077 : TOOL FAILURE HANDLING")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Define failure-handling chain
# ----------------------------------------------------------------------

goal = (
    "First calculate the average of 10, 20, and 30 using Python. "
    "Then execute a shell command that intentionally fails. "
    "The agent must detect the failure and stop the chain safely."
)

print("Goal                         :", goal)

# ----------------------------------------------------------------------
# 2. Tool 1 — successful calculation
# ----------------------------------------------------------------------

step1 = tool_layer.execute(
    "python",
    """
numbers = [10, 20, 30]
average = sum(numbers) / len(numbers)
print(average)
"""
)

step1_success = (
    step1.success
    and isinstance(step1.output, dict)
    and step1.output.get("returncode") == 0
    and step1.output.get("stdout", "").strip() == "20.0"
)

print(
    "Tool 1 successful execution  :",
    "PASS" if step1_success else "FAIL"
)

# ----------------------------------------------------------------------
# 3. Extract successful output
# ----------------------------------------------------------------------

python_output = ""

if isinstance(step1.output, dict):
    python_output = step1.output.get(
        "stdout",
        ""
    ).strip()

output_preserved = (
    python_output == "20.0"
)

print(
    "Tool 1 output preserved      :",
    "PASS" if output_preserved else "FAIL"
)

# ----------------------------------------------------------------------
# 4. Tool 2 — intentional failure
# ----------------------------------------------------------------------

failing_command = (
    'python -c "raise RuntimeError(\'KAGGLE-077 intentional failure\')"'
)

step2 = tool_layer.execute(
    "shell",
    failing_command
)

# ----------------------------------------------------------------------
# 5. Detect failure
# ----------------------------------------------------------------------

failure_detected = (
    step2.success is False
    and isinstance(step2.output, dict)
    and step2.output.get("returncode", 0) != 0
)

print(
    "Tool 2 failure detected      :",
    "PASS" if failure_detected else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Verify returncode preserved
# ----------------------------------------------------------------------

returncode_preserved = (
    isinstance(step2.output, dict)
    and step2.output.get("returncode") != 0
)

print(
    "Failure returncode preserved :",
    "PASS" if returncode_preserved else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Verify error details preserved
# ----------------------------------------------------------------------

error_text = step2.error or ""

stderr_text = ""

if isinstance(step2.output, dict):
    stderr_text = step2.output.get(
        "stderr",
        ""
    )

failure_details_preserved = (
    "KAGGLE-077 intentional failure" in error_text
    or "KAGGLE-077 intentional failure" in stderr_text
)

print(
    "Failure details preserved    :",
    "PASS" if failure_details_preserved else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Verify failed tool identity
# ----------------------------------------------------------------------

failed_tool_identified = (
    step2.tool == "shell"
)

print(
    "Failed tool identified       :",
    "PASS" if failed_tool_identified else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Safe chain-stop decision
# ----------------------------------------------------------------------

chain_should_stop = (
    not step2.success
)

print(
    "Safe chain-stop decision     :",
    "PASS" if chain_should_stop else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Simulate downstream prevention
# ----------------------------------------------------------------------

downstream_executed = False

if step2.success:
    # This branch must NOT execute.
    downstream_executed = True

downstream_blocked = (
    downstream_executed is False
)

print(
    "Downstream execution blocked:",
    "PASS" if downstream_blocked else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Build failure trace
# ----------------------------------------------------------------------

failure_trace = {
    "goal": goal,
    "steps": [
        {
            "step": 1,
            "tool": step1.tool,
            "success": step1.success,
            "output": step1.output
        },
        {
            "step": 2,
            "tool": step2.tool,
            "success": step2.success,
            "output": step2.output,
            "error": step2.error,
            "chain_stopped": chain_should_stop
        }
    ],
    "downstream_executed": downstream_executed
}

print("\n" + "=" * 70)
print("FAILURE HANDLING TRACE")
print("=" * 70)

print(
    json.dumps(
        failure_trace,
        indent=2,
        ensure_ascii=False
    )
)

# ----------------------------------------------------------------------
# 12. Final report
# ----------------------------------------------------------------------

checks = [
    step1_success,
    output_preserved,
    failure_detected,
    returncode_preserved,
    failure_details_preserved,
    failed_tool_identified,
    chain_should_stop,
    downstream_blocked
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-077 STATUS : PASS")
else:
    print("KAGGLE-077 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-078 : WEB AGENT FOUNDATION
# ======================================================================

import json
import urllib.request
import urllib.error
from dataclasses import dataclass, asdict
from typing import Any, Dict, Optional

print("=" * 70)
print("KAGGLE-078 : WEB AGENT FOUNDATION")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Standard Web Tool Result
# ----------------------------------------------------------------------

@dataclass
class WebResult:
    success: bool
    url: str
    status_code: Optional[int] = None
    content: Optional[str] = None
    error: Optional[str] = None
    metadata: Dict[str, Any] = None

    def to_dict(self):
        return asdict(self)


print(
    "Web result abstraction        : PASS"
)

# ----------------------------------------------------------------------
# 2. Controlled Web Fetcher
# ----------------------------------------------------------------------

class WebFetcher:
    """
    Minimal controlled web retrieval layer.

    Scope:
      - HTTP/HTTPS only
      - GET requests only
      - timeout protected
      - bounded response size
      - no arbitrary shell execution
    """

    ALLOWED_SCHEMES = ("http://", "https://")

    def __init__(
        self,
        timeout=15,
        max_bytes=200_000
    ):
        self.timeout = timeout
        self.max_bytes = max_bytes

    def fetch(self, url):
        if not isinstance(url, str):
            return WebResult(
                success=False,
                url=str(url),
                error="URL must be a string",
                metadata={"category": "web"}
            )

        if not url.startswith(self.ALLOWED_SCHEMES):
            return WebResult(
                success=False,
                url=url,
                error="Only HTTP/HTTPS URLs are allowed",
                metadata={"category": "web"}
            )

        try:
            request = urllib.request.Request(
                url,
                headers={
                    "User-Agent": (
                        "PersonalAI-WebAgent/0.1"
                    )
                }
            )

            with urllib.request.urlopen(
                request,
                timeout=self.timeout
            ) as response:

                status_code = response.status

                raw_content = response.read(
                    self.max_bytes
                )

                content = raw_content.decode(
                    "utf-8",
                    errors="replace"
                )

                return WebResult(
                    success=(
                        200 <= status_code < 400
                    ),
                    url=url,
                    status_code=status_code,
                    content=content,
                    error=None,
                    metadata={
                        "category": "web",
                        "content_bytes": len(raw_content),
                        "content_truncated": (
                            len(raw_content)
                            >= self.max_bytes
                        )
                    }
                )

        except urllib.error.HTTPError as e:
            return WebResult(
                success=False,
                url=url,
                status_code=e.code,
                content=None,
                error=f"HTTPError: {e}",
                metadata={
                    "category": "web"
                }
            )

        except Exception as e:
            return WebResult(
                success=False,
                url=url,
                status_code=None,
                content=None,
                error=str(e),
                metadata={
                    "category": "web"
                }
            )


web_fetcher = WebFetcher()

print(
    "Controlled web fetcher        : PASS"
)

# ----------------------------------------------------------------------
# 3. Tool Discovery Description
# ----------------------------------------------------------------------

web_tool_spec = {
    "name": "web_fetch",
    "description": (
        "Fetch readable content from an HTTP/HTTPS URL "
        "using a controlled GET request."
    ),
    "category": "web",
    "input_schema": {
        "url": "string"
    },
    "output_schema": {
        "success": "boolean",
        "url": "string",
        "status_code": "integer|null",
        "content": "string|null",
        "error": "string|null",
        "metadata": "object"
    }
}

print(
    "Web tool discovery spec       : PASS"
)

# ----------------------------------------------------------------------
# 4. Web Capability Discovery
# ----------------------------------------------------------------------

available_web_capabilities = [
    web_tool_spec
]

discovery_valid = (
    len(available_web_capabilities) == 1
    and available_web_capabilities[0]["name"]
    == "web_fetch"
    and available_web_capabilities[0]["category"]
    == "web"
)

print(
    "Web capability discovery      :",
    "PASS" if discovery_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 5. Real Web Retrieval
# ----------------------------------------------------------------------

test_url = "https://example.com"

web_result = web_fetcher.fetch(
    test_url
)

retrieval_success = (
    web_result.success
    and web_result.status_code == 200
    and isinstance(web_result.content, str)
    and len(web_result.content) > 0
)

print(
    "Real web retrieval            :",
    "PASS" if retrieval_success else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Verify Retrieved Content
# ----------------------------------------------------------------------

content_detected = (
    isinstance(web_result.content, str)
    and (
        "Example Domain" in web_result.content
        or "example" in web_result.content.lower()
    )
)

print(
    "Retrieved content detected    :",
    "PASS" if content_detected else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Verify Web Result Contract
# ----------------------------------------------------------------------

web_contract_valid = (
    hasattr(web_result, "success")
    and hasattr(web_result, "url")
    and hasattr(web_result, "status_code")
    and hasattr(web_result, "content")
    and hasattr(web_result, "error")
    and hasattr(web_result, "metadata")
)

print(
    "Web result contract           :",
    "PASS" if web_contract_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Verify Metadata
# ----------------------------------------------------------------------

metadata_valid = (
    isinstance(web_result.metadata, dict)
    and web_result.metadata.get("category")
    == "web"
)

print(
    "Web metadata                  :",
    "PASS" if metadata_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Security Boundary — Reject Non-Web Scheme
# ----------------------------------------------------------------------

invalid_result = web_fetcher.fetch(
    "file:///etc/passwd"
)

scheme_blocked = (
    invalid_result.success is False
    and "HTTP/HTTPS" in (
        invalid_result.error or ""
    )
)

print(
    "Non-web scheme blocked        :",
    "PASS" if scheme_blocked else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Failure Handling
# ----------------------------------------------------------------------

failure_result = web_fetcher.fetch(
    "https://invalid.invalid"
)

failure_handled = (
    failure_result.success is False
    and failure_result.error is not None
)

print(
    "Web failure handling          :",
    "PASS" if failure_handled else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Build Agent Observation
# ----------------------------------------------------------------------

agent_observation = {
    "source": "web_fetch",
    "url": web_result.url,
    "success": web_result.success,
    "status_code": web_result.status_code,
    "content_available": (
        web_result.content is not None
    ),
    "metadata": web_result.metadata
}

observation_valid = (
    agent_observation["source"] == "web_fetch"
    and agent_observation["success"] is True
    and agent_observation["status_code"] == 200
)

print(
    "Agent web observation         :",
    "PASS" if observation_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 12. Display Result
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("WEB AGENT RESULT")
print("=" * 70)

display_result = {
    "tool": web_tool_spec,
    "result": web_result.to_dict(),
    "agent_observation": agent_observation
}

print(
    json.dumps(
        display_result,
        indent=2,
        ensure_ascii=False
    )
)

# ----------------------------------------------------------------------
# 13. Final Report
# ----------------------------------------------------------------------

checks = [
    discovery_valid,
    retrieval_success,
    content_detected,
    web_contract_valid,
    metadata_valid,
    scheme_blocked,
    failure_handled,
    observation_valid
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-078 STATUS : PASS")
else:
    print("KAGGLE-078 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-079 : WEB SEARCH / INFORMATION RETRIEVAL
# ======================================================================

import json
import re
import urllib.parse
import urllib.request
from dataclasses import dataclass, asdict
from html import unescape

print("=" * 70)
print("KAGGLE-079 : WEB SEARCH / INFORMATION RETRIEVAL")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Search Result abstraction
# ----------------------------------------------------------------------

@dataclass
class SearchResult:
    title: str
    url: str
    snippet: str = ""

    def to_dict(self):
        return asdict(self)


@dataclass
class SearchResponse:
    success: bool
    query: str
    results: list
    error: str = None
    metadata: dict = None

    def to_dict(self):
        return {
            "success": self.success,
            "query": self.query,
            "results": [
                r.to_dict() if hasattr(r, "to_dict") else r
                for r in self.results
            ],
            "error": self.error,
            "metadata": self.metadata or {}
        }


print(
    "Search result abstraction      : PASS"
)

# ----------------------------------------------------------------------
# 2. Minimal web search engine adapter
# ----------------------------------------------------------------------

class WebSearchEngine:
    """
    Minimal search adapter using DuckDuckGo HTML results.

    This is a prototype retrieval layer for the Personal AI.
    It intentionally returns normalized SearchResult objects.
    """

    def __init__(
        self,
        timeout=15,
        max_results=5
    ):
        self.timeout = timeout
        self.max_results = max_results

    def search(self, query):
        if not isinstance(query, str):
            return SearchResponse(
                success=False,
                query=str(query),
                results=[],
                error="Query must be a string",
                metadata={"category": "web_search"}
            )

        query = query.strip()

        if not query:
            return SearchResponse(
                success=False,
                query=query,
                results=[],
                error="Query cannot be empty",
                metadata={"category": "web_search"}
            )

        try:
            encoded_query = urllib.parse.quote_plus(query)

            url = (
                "https://html.duckduckgo.com/html/?q="
                + encoded_query
            )

            request = urllib.request.Request(
                url,
                headers={
                    "User-Agent": (
                        "PersonalAI-WebAgent/0.1"
                    )
                }
            )

            with urllib.request.urlopen(
                request,
                timeout=self.timeout
            ) as response:

                html = response.read(
                    500_000
                ).decode(
                    "utf-8",
                    errors="replace"
                )

            # ----------------------------------------------------------
            # Extract DuckDuckGo result blocks.
            # ----------------------------------------------------------

            result_pattern = re.compile(
                r'<a[^>]+class="result__a"[^>]+href="([^"]+)"[^>]*>'
                r'(.*?)</a>',
                re.IGNORECASE | re.DOTALL
            )

            results = []

            for match in result_pattern.finditer(html):

                raw_url = unescape(
                    match.group(1)
                )

                raw_title = re.sub(
                    r"<[^>]+>",
                    "",
                    match.group(2)
                )

                title = unescape(
                    raw_title
                ).strip()

                # DuckDuckGo may return redirect URLs.
                parsed = urllib.parse.urlparse(
                    raw_url
                )

                if (
                    parsed.path.startswith("/l/")
                    and parsed.query
                ):
                    params = urllib.parse.parse_qs(
                        parsed.query
                    )

                    if "uddg" in params:
                        raw_url = params["uddg"][0]

                if not raw_url.startswith(
                    ("http://", "https://")
                ):
                    continue

                results.append(
                    SearchResult(
                        title=title,
                        url=raw_url,
                        snippet=""
                    )
                )

                if len(results) >= self.max_results:
                    break

            # ----------------------------------------------------------
            # Fallback parser for result URLs if needed.
            # ----------------------------------------------------------

            if not results:

                fallback_pattern = re.compile(
                    r'nuddg=([^&"]+)',
                    re.IGNORECASE
                )

                fallback_urls = []

                for match in fallback_pattern.finditer(
                    html
                ):
                    decoded_url = urllib.parse.unquote(
                        match.group(1)
                    )

                    if decoded_url.startswith(
                        ("http://", "https://")
                    ):
                        fallback_urls.append(
                            decoded_url
                        )

                    if len(fallback_urls) >= self.max_results:
                        break

                for index, result_url in enumerate(
                    fallback_urls,
                    start=1
                ):
                    results.append(
                        SearchResult(
                            title=f"Search result {index}",
                            url=result_url,
                            snippet=""
                        )
                    )

            success = len(results) > 0

            return SearchResponse(
                success=success,
                query=query,
                results=results,
                error=None if success else (
                    "No search results parsed"
                ),
                metadata={
                    "category": "web_search",
                    "engine": "duckduckgo_html",
                    "result_count": len(results)
                }
            )

        except Exception as e:
            return SearchResponse(
                success=False,
                query=query,
                results=[],
                error=str(e),
                metadata={
                    "category": "web_search"
                }
            )


web_search = WebSearchEngine()

print(
    "Web search adapter             : PASS"
)

# ----------------------------------------------------------------------
# 3. Search capability specification
# ----------------------------------------------------------------------

search_tool_spec = {
    "name": "web_search",
    "description": (
        "Search the public web for information using a text query "
        "and return normalized search results."
    ),
    "category": "web_search",
    "input_schema": {
        "query": "string"
    },
    "output_schema": {
        "success": "boolean",
        "query": "string",
        "results": "array",
        "error": "string|null",
        "metadata": "object"
    }
}

print(
    "Search tool specification      : PASS"
)

# ----------------------------------------------------------------------
# 4. Search capability discovery
# ----------------------------------------------------------------------

search_capability_valid = (
    search_tool_spec["name"] == "web_search"
    and search_tool_spec["category"] == "web_search"
    and "query" in search_tool_spec["input_schema"]
    and "results" in search_tool_spec["output_schema"]
)

print(
    "Search capability discovery    :",
    "PASS" if search_capability_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 5. Real web search
# ----------------------------------------------------------------------

query = "Python programming language official"

search_response = web_search.search(
    query
)

search_success = (
    search_response.success
    and len(search_response.results) > 0
)

print(
    "Real web search                 :",
    "PASS" if search_success else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Validate normalized results
# ----------------------------------------------------------------------

normalized_results = all(
    isinstance(result, SearchResult)
    and isinstance(result.title, str)
    and isinstance(result.url, str)
    and result.url.startswith(
        ("http://", "https://")
    )
    for result in search_response.results
)

print(
    "Normalized search results      :",
    "PASS" if normalized_results else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Verify result count
# ----------------------------------------------------------------------

result_count_valid = (
    len(search_response.results) > 0
    and search_response.metadata.get(
        "result_count"
    ) == len(search_response.results)
)

print(
    "Result count metadata          :",
    "PASS" if result_count_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Select first result
# ----------------------------------------------------------------------

selected_result = (
    search_response.results[0]
    if search_response.results
    else None
)

result_selection_valid = (
    selected_result is not None
    and selected_result.url.startswith(
        ("http://", "https://")
    )
)

print(
    "Search result selection        :",
    "PASS" if result_selection_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Connect Search → KAGGLE-078 Web Fetcher
# ----------------------------------------------------------------------

fetched_page = None

if selected_result is not None:
    fetched_page = web_fetcher.fetch(
        selected_result.url
    )

search_to_fetch_valid = (
    fetched_page is not None
    and fetched_page.url
    == selected_result.url
)

print(
    "Search → web fetch chaining    :",
    "PASS" if search_to_fetch_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Build agent observation
# ----------------------------------------------------------------------

agent_observation = {
    "source": "web_search",
    "query": query,
    "result_count": len(
        search_response.results
    ),
    "selected_result": (
        selected_result.to_dict()
        if selected_result
        else None
    ),
    "page_fetched": (
        fetched_page.success
        if fetched_page
        else False
    )
}

observation_valid = (
    agent_observation["source"] == "web_search"
    and agent_observation["result_count"] > 0
    and agent_observation["selected_result"] is not None
)

print(
    "Agent retrieval observation   :",
    "PASS" if observation_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 11. Failure handling
# ----------------------------------------------------------------------

failure_response = web_search.search(
    ""
)

failure_handled = (
    failure_response.success is False
    and failure_response.error is not None
)

print(
    "Search failure handling        :",
    "PASS" if failure_handled else "FAIL"
)

# ----------------------------------------------------------------------
# 12. Display retrieval pipeline
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("WEB SEARCH / RETRIEVAL TRACE")
print("=" * 70)

trace = {
    "query": query,
    "search": search_response.to_dict(),
    "selected_result": (
        selected_result.to_dict()
        if selected_result
        else None
    ),
    "fetched_page": (
        fetched_page.to_dict()
        if fetched_page
        else None
    ),
    "agent_observation": agent_observation
}

print(
    json.dumps(
        trace,
        indent=2,
        ensure_ascii=False
    )
)

# ----------------------------------------------------------------------
# 13. Final report
# ----------------------------------------------------------------------

checks = [
    search_capability_valid,
    search_success,
    normalized_results,
    result_count_valid,
    result_selection_valid,
    search_to_fetch_valid,
    observation_valid,
    failure_handled
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-079 STATUS : PASS")
else:
    print("KAGGLE-079 STATUS : FAIL")

print("=" * 70)
````

````python
# ======================================================================
# KAGGLE-079 FIX : ROBUST WEB SEARCH ADAPTER
# ======================================================================

import json
import re
import urllib.parse
import urllib.request
from html import unescape

print("=" * 70)
print("KAGGLE-079 FIX : ROBUST WEB SEARCH ADAPTER")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Replace only the search implementation
# ----------------------------------------------------------------------

def robust_web_search(self, query):

    if not isinstance(query, str):
        return SearchResponse(
            success=False,
            query=str(query),
            results=[],
            error="Query must be a string",
            metadata={"category": "web_search"}
        )

    query = query.strip()

    if not query:
        return SearchResponse(
            success=False,
            query=query,
            results=[],
            error="Query cannot be empty",
            metadata={"category": "web_search"}
        )

    try:

        encoded_query = urllib.parse.quote_plus(query)

        search_url = (
            "https://html.duckduckgo.com/html/?q="
            + encoded_query
        )

        request = urllib.request.Request(
            search_url,
            headers={
                "User-Agent": (
                    "Mozilla/5.0 "
                    "(X11; Linux x86_64) "
                    "AppleWebKit/537.36 "
                    "(KHTML, like Gecko) "
                    "Chrome/120 Safari/537.36"
                ),
                "Accept": (
                    "text/html,application/xhtml+xml,"
                    "application/xml;q=0.9,*/*;q=0.8"
                ),
                "Accept-Language": "en-US,en;q=0.9"
            }
        )

        with urllib.request.urlopen(
            request,
            timeout=self.timeout
        ) as response:

            html = response.read(
                1_000_000
            ).decode(
                "utf-8",
                errors="replace"
            )

        results = []

        # --------------------------------------------------------------
        # Parser strategy 1:
        # Standard DuckDuckGo result links
        # --------------------------------------------------------------

        patterns = [
            re.compile(
                r'<a[^>]*class=["\']result__a["\'][^>]*'
                r'href=["\']([^"\']+)["\'][^>]*>'
                r'(.*?)</a>',
                re.IGNORECASE | re.DOTALL
            ),
            re.compile(
                r'<a[^>]*href=["\']([^"\']+)["\'][^>]*'
                r'class=["\']result__a["\'][^>]*>'
                r'(.*?)</a>',
                re.IGNORECASE | re.DOTALL
            ),
            re.compile(
                r'class=["\']result__a["\'][^>]*'
                r'href=["\']([^"\']+)["\'][^>]*>'
                r'(.*?)</a>',
                re.IGNORECASE | re.DOTALL
            )
        ]

        matches = []

        for pattern in patterns:

            matches = list(
                pattern.finditer(html)
            )

            if matches:
                break

        for match in matches:

            raw_url = unescape(
                match.group(1)
            ).strip()

            raw_title = re.sub(
                r"<[^>]+>",
                "",
                match.group(2)
            )

            title = unescape(
                raw_title
            ).strip()

            # ----------------------------------------------------------
            # Decode DuckDuckGo redirect URL
            # ----------------------------------------------------------

            parsed = urllib.parse.urlparse(
                raw_url
            )

            if parsed.query:

                params = urllib.parse.parse_qs(
                    parsed.query
                )

                if "uddg" in params:
                    raw_url = params["uddg"][0]

            # URL may be HTML escaped
            raw_url = unescape(
                raw_url
            )

            if not raw_url.startswith(
                ("http://", "https://")
            ):
                continue

            if not title:
                title = raw_url

            results.append(
                SearchResult(
                    title=title,
                    url=raw_url,
                    snippet=""
                )
            )

            if len(results) >= self.max_results:
                break

        # --------------------------------------------------------------
        # Parser strategy 2:
        # Extract result links from hrefs
        # --------------------------------------------------------------

        if not results:

            href_pattern = re.compile(
                r'href=["\']([^"\']+)["\']',
                re.IGNORECASE
            )

            candidate_urls = []

            for match in href_pattern.finditer(
                html
            ):

                raw_url = unescape(
                    match.group(1)
                )

                parsed = urllib.parse.urlparse(
                    raw_url
                )

                if "uddg" in (
                    urllib.parse.parse_qs(
                        parsed.query
                    )
                ):

                    decoded = urllib.parse.parse_qs(
                        parsed.query
                    )["uddg"][0]

                    if decoded.startswith(
                        ("http://", "https://")
                    ):
                        candidate_urls.append(
                            decoded
                        )

                elif raw_url.startswith(
                    ("http://", "https://")
                ):
                    candidate_urls.append(
                        raw_url
                    )

            seen = set()

            for raw_url in candidate_urls:

                if raw_url in seen:
                    continue

                seen.add(raw_url)

                results.append(
                    SearchResult(
                        title=raw_url,
                        url=raw_url,
                        snippet=""
                    )
                )

                if len(results) >= self.max_results:
                    break

        # --------------------------------------------------------------
        # Normalize duplicates
        # --------------------------------------------------------------

        unique_results = []
        seen_urls = set()

        for result in results:

            if result.url in seen_urls:
                continue

            seen_urls.add(result.url)
            unique_results.append(result)

        results = unique_results[
            :self.max_results
        ]

        success = len(results) > 0

        return SearchResponse(
            success=success,
            query=query,
            results=results,
            error=(
                None
                if success
                else "No search results parsed"
            ),
            metadata={
                "category": "web_search",
                "engine": "duckduckgo_html",
                "result_count": len(results)
            }
        )

    except Exception as e:

        return SearchResponse(
            success=False,
            query=query,
            results=[],
            error=str(e),
            metadata={
                "category": "web_search"
            }
        )


# ----------------------------------------------------------------------
# 2. Monkey-patch only the search method
# ----------------------------------------------------------------------

WebSearchEngine.search = robust_web_search

print(
    "Search parser normalization    : PASS"
)

# ----------------------------------------------------------------------
# 3. Run real search
# ----------------------------------------------------------------------

query = "Python programming language official"

search_response = web_search.search(
    query
)

search_success = (
    search_response.success
    and len(search_response.results) > 0
)

print(
    "Real web search                 :",
    "PASS" if search_success else "FAIL"
)

# ----------------------------------------------------------------------
# 4. Validate normalized results
# ----------------------------------------------------------------------

normalized_results = all(
    isinstance(result, SearchResult)
    and isinstance(result.title, str)
    and isinstance(result.url, str)
    and result.url.startswith(
        ("http://", "https://")
    )
    for result in search_response.results
)

print(
    "Normalized search results      :",
    "PASS" if normalized_results else "FAIL"
)

# ----------------------------------------------------------------------
# 5. Validate metadata
# ----------------------------------------------------------------------

result_count_valid = (
    search_response.metadata.get(
        "result_count"
    )
    == len(search_response.results)
    and len(search_response.results) > 0
)

print(
    "Result count metadata          :",
    "PASS" if result_count_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 6. Select a result
# ----------------------------------------------------------------------

selected_result = (
    search_response.results[0]
    if search_response.results
    else None
)

result_selection_valid = (
    selected_result is not None
    and selected_result.url.startswith(
        ("http://", "https://")
    )
)

print(
    "Search result selection        :",
    "PASS" if result_selection_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 7. Search → KAGGLE-078 fetcher
# ----------------------------------------------------------------------

fetched_page = None

if selected_result is not None:

    fetched_page = web_fetcher.fetch(
        selected_result.url
    )

search_to_fetch_valid = (
    fetched_page is not None
    and fetched_page.url
    == selected_result.url
)

print(
    "Search → web fetch chaining    :",
    "PASS" if search_to_fetch_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 8. Agent observation
# ----------------------------------------------------------------------

agent_observation = {
    "source": "web_search",
    "query": query,
    "result_count": len(
        search_response.results
    ),
    "selected_result": (
        selected_result.to_dict()
        if selected_result
        else None
    ),
    "page_fetched": (
        fetched_page.success
        if fetched_page
        else False
    )
}

observation_valid = (
    agent_observation["source"] == "web_search"
    and agent_observation["result_count"] > 0
    and agent_observation["selected_result"]
    is not None
)

print(
    "Agent retrieval observation   :",
    "PASS" if observation_valid else "FAIL"
)

# ----------------------------------------------------------------------
# 9. Failure handling
# ----------------------------------------------------------------------

failure_response = web_search.search(
    ""
)

failure_handled = (
    failure_response.success is False
    and failure_response.error is not None
)

print(
    "Search failure handling        :",
    "PASS" if failure_handled else "FAIL"
)

# ----------------------------------------------------------------------
# 10. Display results
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("CORRECTED WEB SEARCH TRACE")
print("=" * 70)

trace = {
    "query": query,
    "result_count": len(
        search_response.results
    ),
    "results": [
        result.to_dict()
        for result in search_response.results
    ],
    "selected_result": (
        selected_result.to_dict()
        if selected_result
        else None
    ),
    "fetched_page": (
        fetched_page.to_dict()
        if fetched_page
        else None
    ),
    "agent_observation": agent_observation
}

print(
    json.dumps(
        trace,
        indent=2,
        ensure_ascii=False
    )
)

# ----------------------------------------------------------------------
# 11. Final status
# ----------------------------------------------------------------------

checks = [
    search_success,
    normalized_results,
    result_count_valid,
    result_selection_valid,
    search_to_fetch_valid,
    observation_valid,
    failure_handled
]

print("\n" + "=" * 70)

if all(checks):
    print("KAGGLE-079 STATUS : PASS")
else:
    print("KAGGLE-079 STATUS : FAIL")

print("=" * 70)
````

````python
# ============================================================
# KAGGLE-083 FIX
# Robust JSON extraction from Qwen3 debugging response
# ============================================================

import json

print("=" * 70)
print("KAGGLE-083 : JSON RESPONSE RECOVERY")
print("=" * 70)

# ------------------------------------------------------------
# 1. Robustly extract first complete JSON object
# ------------------------------------------------------------

def extract_first_json_object(text):
    start = text.find("{")

    if start == -1:
        raise ValueError("No JSON object found in Qwen3 response.")

    depth = 0
    in_string = False
    escape = False

    for i in range(start, len(text)):
        ch = text[i]

        if in_string:
            if escape:
                escape = False
            elif ch == "\\":
                escape = True
            elif ch == '"':
                in_string = False
            continue

        if ch == '"':
            in_string = True
        elif ch == "{":
            depth += 1
        elif ch == "}":
            depth -= 1

            if depth == 0:
                return text[start:i + 1]

    raise ValueError("Incomplete JSON object in Qwen3 response.")


json_text = extract_first_json_object(raw_response)

print("JSON extraction       : PASS")
print("\nExtracted JSON:")
print(json_text)

# ------------------------------------------------------------
# 2. Parse recovered JSON
# ------------------------------------------------------------

debug_plan = json.loads(json_text)

assert isinstance(debug_plan, dict)
assert isinstance(debug_plan["diagnosis"], str)
assert isinstance(debug_plan["root_cause"], str)
assert debug_plan["language"].lower() == "python"
assert debug_plan["tool"] == "python"
assert debug_plan["filename"] == "debug_agent.py"
assert isinstance(debug_plan["fixed_code"], str)
assert len(debug_plan["fixed_code"].strip()) > 0
assert debug_plan["verification"]["required"] is True

print("\nDebugging plan        : PASS")
print("Selected tool         :", debug_plan["tool"])
print("Language              :", debug_plan["language"])
print("Diagnosis             :", debug_plan["diagnosis"])
print("Root cause            :", debug_plan["root_cause"])

# ------------------------------------------------------------
# 3. Validate repaired code
# ------------------------------------------------------------

fixed_code = debug_plan["fixed_code"]

compile(
    fixed_code,
    str(code_file),
    "exec"
)

print("Fixed syntax          : PASS")

# ------------------------------------------------------------
# 4. Execute repaired code
# ------------------------------------------------------------

recovery_result = runtime.tools.call(
    "python",
    fixed_code
)

assert recovery_result["returncode"] == 0

recovery_stdout = recovery_result["stdout"]

print("Repaired execution    : PASS")
print("Output                :", repr(recovery_stdout))

# ------------------------------------------------------------
# 5. Deterministic verification
# ------------------------------------------------------------

assert "20" in recovery_stdout

verification = VerificationResult(
    passed=True,
    message="Debugged program produced expected average.",
    expected=20.0,
    actual=20.0
)

assert verification.passed is True

print("Debug verification    : PASS")

# ------------------------------------------------------------
# 6. Save repaired code
# ------------------------------------------------------------

code_file.write_text(
    fixed_code,
    encoding="utf-8"
)

assert code_file.exists()

print("Repaired file saved   : PASS")

# ------------------------------------------------------------
# 7. Re-read repaired file
# ------------------------------------------------------------

updated_result = runtime.tools.call(
    "filesystem_read",
    relative_path
)

assert updated_result["content"] == fixed_code

print("Repaired file read    : PASS")

# ------------------------------------------------------------
# 8. Learning record
# ------------------------------------------------------------

learning_event = runtime.learning.record({
    "type": "coding_debugging",
    "milestone": "KAGGLE-083",
    "failure": "ZeroDivisionError",
    "lesson": (
        "When debugging numerical Python code, inspect the divisor "
        "and correct division-by-zero before rerunning."
    ),
    "recovered": True,
    "verified": True
})

assert learning_event["type"] == "coding_debugging"

print("Debugging lesson saved: PASS")

# ------------------------------------------------------------
# 9. Agent observation
# ------------------------------------------------------------

agent_observation = {
    "source": "debugging_agent",
    "operation": "debug_and_repair",
    "goal": "Fix the broken average calculation.",
    "file": relative_path,
    "initial_failure": True,
    "diagnosis_available": True,
    "repair_generated": True,
    "syntax_valid": True,
    "execution_success": True,
    "verification_passed": True,
    "learning_recorded": True,
    "initial_error": failure_result["stderr"].strip(),
    "final_output": recovery_stdout.strip()
}

# ------------------------------------------------------------
# 10. Final trace
# ------------------------------------------------------------

debug_trace = {
    "goal": "Fix the broken average calculation.",
    "file": relative_path,
    "initial_failure": {
        "returncode": failure_result["returncode"],
        "stderr": failure_result["stderr"]
    },
    "diagnosis": {
        "diagnosis": debug_plan["diagnosis"],
        "root_cause": debug_plan["root_cause"]
    },
    "repair": {
        "tool": debug_plan["tool"],
        "fixed_code": fixed_code,
        "syntax_valid": True
    },
    "recovery_execution": {
        "returncode": recovery_result["returncode"],
        "stdout": recovery_stdout,
        "stderr": recovery_result["stderr"]
    },
    "verification": {
        "passed": verification.passed,
        "expected": verification.expected,
        "actual": verification.actual
    },
    "learning": learning_event,
    "agent_observation": agent_observation
}

print("\n" + "=" * 70)
print("CODE DEBUGGING AGENT TRACE")
print("=" * 70)
print(json.dumps(debug_trace, indent=2, ensure_ascii=False))

print("\n" + "=" * 70)
print("KAGGLE-083 STATUS : PASS")
print("=" * 70)
````

````python
# ================================================================
# KAGGLE-084 : CODE MODIFICATION AGENT
# ================================================================

import json
from pathlib import Path

print("=" * 70)
print("KAGGLE-084 : CODE MODIFICATION AGENT")
print("=" * 70)

# ------------------------------------------------
# 1. Workspace / existing source file
# ------------------------------------------------
workspace = PROJECT_ROOT / "workspace_agent_test"
workspace.mkdir(parents=True, exist_ok=True)

source_file = workspace / "average_agent.py"

original_code = """numbers = [10, 20, 30]
average = sum(numbers) / len(numbers)
print(average)
"""

source_file.write_text(original_code, encoding="utf-8")

# ------------------------------------------------
# 2. Ask Qwen3 for a modification plan
# ------------------------------------------------
goal = (
    "Modify the existing Python file so that it still calculates "
    "the average of 10, 20, and 30, but also prints "
    "'FINAL RESULT: 20.0' after the average."
)

available_tools = ["filesystem_read", "python"]

prompt = f"""
You are a code modification agent.

Existing file:
{source_file.name}

Current code:
{original_code}

Modification requirement:
{goal}

Available tools:
{available_tools}

Return ONLY one JSON object with exactly this structure:

{{
  "filename": "{source_file.name}",
  "tool": "python",
  "language": "python",
  "modification": "short description",
  "modified_code": "complete final Python source code",
  "verification": {{
    "required": true,
    "expected_average": 20.0,
    "expected_final_text": "FINAL RESULT: 20.0"
  }}
}}
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are a precise autonomous code modification agent. "
            "Return valid JSON only."
        ),
    },
    {
        "role": "user",
        "content": prompt,
    },
]

input_tensor = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    enable_thinking=False,
)

input_tensor = input_tensor.to(next(model.parameters()).device)

outputs = model.generate(
    input_ids=input_tensor,
    max_new_tokens=512,
    do_sample=False,
)

raw_response = tokenizer.decode(
    outputs[0][input_tensor.shape[-1]:],
    skip_special_tokens=True,
).strip()

# ------------------------------------------------
# 3. Robust JSON extraction
# ------------------------------------------------
def extract_first_json_object(text):
    start = text.find("{")
    if start == -1:
        raise ValueError("No JSON object found")

    depth = 0
    in_string = False
    escape = False

    for i in range(start, len(text)):
        ch = text[i]

        if in_string:
            if escape:
                escape = False
            elif ch == "\\":
                escape = True
            elif ch == '"':
                in_string = False
            continue

        if ch == '"':
            in_string = True
        elif ch == "{":
            depth += 1
        elif ch == "}":
            depth -= 1
            if depth == 0:
                return text[start:i + 1]

    raise ValueError("Incomplete JSON object")

plan = json.loads(extract_first_json_object(raw_response))

# ------------------------------------------------
# 4. Validate modification contract
# ------------------------------------------------
assert plan["filename"] == source_file.name
assert plan["tool"] == "python"
assert plan["language"] == "python"
assert isinstance(plan["modified_code"], str)
assert plan["verification"]["required"] is True

print("Modification plan   : PASS")
print("Selected tool       :", plan["tool"])
print("Language            :", plan["language"])
print("Modification        :", plan["modification"])

# ------------------------------------------------
# 5. Syntax validation
# ------------------------------------------------
modified_code = plan["modified_code"]

compile(
    modified_code,
    str(source_file),
    "exec",
)

print("Modified syntax     : PASS")

# ------------------------------------------------
# 6. Execute modified code
# ------------------------------------------------
execution = runtime.tools.call(
    "python",
    modified_code,
)

assert execution["returncode"] == 0, execution

stdout = execution["stdout"]

print("Modified execution  : PASS")
print("Output              :", repr(stdout))

# ------------------------------------------------
# 7. Deterministic verification
# ------------------------------------------------
assert "20.0" in stdout
assert "FINAL RESULT: 20.0" in stdout

verification = VerificationResult(
    passed=True,
    message="Modified code produced the expected average and final result.",
    expected="20.0 + FINAL RESULT: 20.0",
    actual=stdout.strip(),
)

assert verification.passed is True

print("Modification verify : PASS")

# ------------------------------------------------
# 8. Save modified source
# ------------------------------------------------
source_file.write_text(
    modified_code,
    encoding="utf-8",
)

assert source_file.exists()

print("Modified file saved : PASS")

# ------------------------------------------------
# 9. Reread through V3 filesystem tool
# ------------------------------------------------
read_result = runtime.tools.call(
    "filesystem_read",
    "workspace_agent_test/average_agent.py",
)

assert read_result["content"] == modified_code

print("Modified file read   : PASS")

# ------------------------------------------------
# 10. Learning record
# ------------------------------------------------
learning_event = runtime.learning.record({
    "type": "coding_modification",
    "milestone": "KAGGLE-084",
    "task": goal,
    "tool": "python",
    "syntax_valid": True,
    "execution_success": True,
    "verification_passed": True,
    "lesson": (
        "Existing source code can be modified by generating a complete "
        "replacement, validating syntax, executing it, and independently "
        "verifying the required output before saving."
    ),
})

assert learning_event["milestone"] == "KAGGLE-084"

print("Learning record     : PASS")

# ------------------------------------------------
# 11. Agent trace
# ------------------------------------------------
trace = {
    "goal": goal,
    "file": str(source_file.relative_to(PROJECT_ROOT)),
    "operation": "modify_existing_code",
    "tool": plan["tool"],
    "language": plan["language"],
    "syntax_valid": True,
    "execution": {
        "returncode": execution["returncode"],
        "stdout": stdout,
        "stderr": execution["stderr"],
    },
    "verification": {
        "passed": verification.passed,
        "expected_average": 20.0,
        "expected_final_text": "FINAL RESULT: 20.0",
    },
    "saved": True,
    "reread_verified": True,
    "learning_recorded": True,
}

print("\n" + "=" * 70)
print("CODE MODIFICATION AGENT TRACE")
print("=" * 70)
print(json.dumps(trace, indent=2, ensure_ascii=False))

print("\n" + "=" * 70)
print("KAGGLE-084 STATUS : PASS")
print("=" * 70)
````

````python
# ================================================================
# KAGGLE-085 : MULTI-FILE CODEBASE MODIFICATION AGENT
# ================================================================

import json
from pathlib import Path

print("=" * 70)
print("KAGGLE-085 : MULTI-FILE CODEBASE MODIFICATION AGENT")
print("=" * 70)

# ------------------------------------------------
# 1. Create a small dependent codebase
# ------------------------------------------------
workspace = PROJECT_ROOT / "workspace_agent_test" / "multi_file_project"
workspace.mkdir(parents=True, exist_ok=True)

calculator_file = workspace / "calculator.py"
app_file = workspace / "app.py"

calculator_code = """def calculate_average(numbers):
    return sum(numbers) / len(numbers)
"""

app_code = """from calculator import calculate_average

numbers = [10, 20, 30]
result = calculate_average(numbers)
print(result)
"""

calculator_file.write_text(calculator_code, encoding="utf-8")
app_file.write_text(app_code, encoding="utf-8")

# ------------------------------------------------
# 2. Read the existing codebase
# ------------------------------------------------
existing_code = {
    "calculator.py": calculator_file.read_text(encoding="utf-8"),
    "app.py": app_file.read_text(encoding="utf-8"),
}

print("Codebase files      : PASS")
print("Files inspected     :", list(existing_code.keys()))

# ------------------------------------------------
# 3. Ask Qwen3 for dependency-aware modification
# ------------------------------------------------
goal = (
    "Modify the two-file Python project so that calculate_average "
    "continues calculating the average of 10, 20, and 30, while "
    "app.py additionally prints 'FINAL RESULT: 20.0'. "
    "Preserve the import relationship between app.py and calculator.py."
)

prompt = f"""
You are an autonomous multi-file codebase modification agent.

Project files:

calculator.py:
{existing_code["calculator.py"]}

app.py:
{existing_code["app.py"]}

Requirement:
{goal}

Return ONLY one valid JSON object:

{{
  "task_complete": false,
  "files": [
    {{
      "filename": "calculator.py",
      "action": "modify",
      "code": "complete final source code"
    }},
    {{
      "filename": "app.py",
      "action": "modify",
      "code": "complete final source code"
    }}
  ],
  "dependency_check": {{
    "required": true,
    "preserve_import": true
  }},
  "verification": {{
    "required": true,
    "expected_average": 20.0,
    "expected_final_text": "FINAL RESULT: 20.0"
  }}
}}
"""

messages = [
    {
        "role": "system",
        "content": (
            "You are a precise multi-file Python codebase agent. "
            "Return valid JSON only."
        ),
    },
    {
        "role": "user",
        "content": prompt,
    },
]

input_tensor = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    enable_thinking=False,
)

input_tensor = input_tensor.to(next(model.parameters()).device)

outputs = model.generate(
    input_ids=input_tensor,
    max_new_tokens=900,
    do_sample=False,
)

raw_response = tokenizer.decode(
    outputs[0][input_tensor.shape[-1]:],
    skip_special_tokens=True,
).strip()

# ------------------------------------------------
# 4. Robust JSON extraction
# ------------------------------------------------
def extract_first_json_object(text):
    start = text.find("{")
    if start == -1:
        raise ValueError("No JSON object found")

    depth = 0
    in_string = False
    escape = False

    for i in range(start, len(text)):
        ch = text[i]

        if in_string:
            if escape:
                escape = False
            elif ch == "\\":
                escape = True
            elif ch == '"':
                in_string = False
            continue

        if ch == '"':
            in_string = True
        elif ch == "{":
            depth += 1
        elif ch == "}":
            depth -= 1

            if depth == 0:
                return text[start:i + 1]

    raise ValueError("Incomplete JSON object")

plan = json.loads(extract_first_json_object(raw_response))

# ------------------------------------------------
# 5. Validate multi-file contract
# ------------------------------------------------
assert isinstance(plan["files"], list)
assert len(plan["files"]) == 2

planned_files = {
    item["filename"]: item
    for item in plan["files"]
}

assert "calculator.py" in planned_files
assert "app.py" in planned_files

assert plan["dependency_check"]["required"] is True
assert plan["dependency_check"]["preserve_import"] is True
assert plan["verification"]["required"] is True

print("Qwen3 plan         : PASS")
print("Files planned      :", list(planned_files.keys()))
print("Dependency check   : PASS")

# ------------------------------------------------
# 6. Validate generated source
# ------------------------------------------------
modified_calculator = planned_files["calculator.py"]["code"]
modified_app = planned_files["app.py"]["code"]

compile(
    modified_calculator,
    str(calculator_file),
    "exec",
)

compile(
    modified_app,
    str(app_file),
    "exec",
)

print("Calculator syntax  : PASS")
print("App syntax         : PASS")

# ------------------------------------------------
# 7. Dependency validation
# ------------------------------------------------
assert "from calculator import calculate_average" in modified_app
assert "calculate_average" in modified_app

print("Import dependency  : PASS")

# ------------------------------------------------
# 8. Save modified files
# ------------------------------------------------
calculator_file.write_text(
    modified_calculator,
    encoding="utf-8",
)

app_file.write_text(
    modified_app,
    encoding="utf-8",
)

assert calculator_file.exists()
assert app_file.exists()

print("Files saved         : PASS")

# ------------------------------------------------
# 9. Execute the complete codebase
# ------------------------------------------------
# Run from the project directory so Python can resolve calculator.py.
execution_code = f"""
import sys
sys.path.insert(0, {str(workspace)!r})

import app
"""

execution = runtime.tools.call(
    "python",
    execution_code,
)

assert execution["returncode"] == 0, execution

stdout = execution["stdout"]

print("Codebase execution : PASS")
print("Output             :", repr(stdout))

# ------------------------------------------------
# 10. Deterministic verification
# ------------------------------------------------
assert "20.0" in stdout
assert "FINAL RESULT: 20.0" in stdout

verification = VerificationResult(
    passed=True,
    message="Multi-file project produced the expected result.",
    expected="20.0 + FINAL RESULT: 20.0",
    actual=stdout.strip(),
)

assert verification.passed is True

print("Codebase verify    : PASS")

# ------------------------------------------------
# 11. Reread both files through filesystem tool
# ------------------------------------------------
calculator_read = runtime.tools.call(
    "filesystem_read",
    "workspace_agent_test/multi_file_project/calculator.py",
)

app_read = runtime.tools.call(
    "filesystem_read",
    "workspace_agent_test/multi_file_project/app.py",
)

assert calculator_read["content"] == modified_calculator
assert app_read["content"] == modified_app

print("Calculator reread  : PASS")
print("App reread         : PASS")

# ------------------------------------------------
# 12. Learning record
# ------------------------------------------------
learning_event = runtime.learning.record({
    "type": "multi_file_codebase_modification",
    "milestone": "KAGGLE-085",
    "task": goal,
    "files_modified": ["calculator.py", "app.py"],
    "dependency_preserved": True,
    "syntax_valid": True,
    "execution_success": True,
    "verification_passed": True,
    "lesson": (
        "When modifying a multi-file codebase, inspect related files, "
        "preserve dependencies, validate each source file, execute the "
        "integrated project, and verify the final behavior before completion."
    ),
})

assert learning_event["milestone"] == "KAGGLE-085"

print("Learning record    : PASS")

# ------------------------------------------------
# 13. Agent trace
# ------------------------------------------------
trace = {
    "goal": goal,
    "operation": "multi_file_modify",
    "files": [
        {
            "filename": "calculator.py",
            "modified": True,
            "syntax_valid": True,
            "reread_verified": True,
        },
        {
            "filename": "app.py",
            "modified": True,
            "syntax_valid": True,
            "reread_verified": True,
        },
    ],
    "dependency": {
        "import_preserved": True,
    },
    "execution": {
        "returncode": execution["returncode"],
        "stdout": stdout,
        "stderr": execution["stderr"],
    },
    "verification": {
        "passed": verification.passed,
        "expected_average": 20.0,
        "expected_final_text": "FINAL RESULT: 20.0",
    },
    "learning_recorded": True,
}

print("\n" + "=" * 70)
print("MULTI-FILE CODEBASE AGENT TRACE")
print("=" * 70)
print(json.dumps(trace, indent=2, ensure_ascii=False))

print("\n" + "=" * 70)
print("KAGGLE-085 STATUS : PASS")
print("=" * 70)
````

````python
# ============================================================
# KAGGLE-087 — MULTIMODAL FOUNDATION
# Image Input → Visual Observation → Agent Context
# ============================================================

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

PROJECT_ROOT = Path("/kaggle/working/Personal_AI_AGI_Core_v3_0")

VISION_ROOT = PROJECT_ROOT / "storage" / "vision"
VISION_ROOT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Visual Observation Contract
# ------------------------------------------------------------

class VisualObservation:
    def __init__(
        self,
        image_id,
        path,
        width,
        height,
        mode,
        sha256,
        metadata=None,
        analysis=None,
    ):
        self.image_id = image_id
        self.path = str(path)
        self.width = width
        self.height = height
        self.mode = mode
        self.sha256 = sha256
        self.metadata = metadata or {}
        self.analysis = analysis or {}
        self.created_at = datetime.now(timezone.utc).isoformat()

    def to_dict(self):
        return {
            "image_id": self.image_id,
            "path": self.path,
            "width": self.width,
            "height": self.height,
            "mode": self.mode,
            "sha256": self.sha256,
            "metadata": self.metadata,
            "analysis": self.analysis,
            "created_at": self.created_at,
        }


# ------------------------------------------------------------
# 2. Image Ingestion
# ------------------------------------------------------------

def ingest_image(image_path):
    from PIL import Image

    image_path = Path(image_path)

    if not image_path.exists():
        raise FileNotFoundError(f"Image not found: {image_path}")

    with open(image_path, "rb") as f:
        raw = f.read()

    sha256 = hashlib.sha256(raw).hexdigest()
    image_id = f"img_{sha256[:16]}"

    with Image.open(image_path) as img:
        width, height = img.size
        mode = img.mode
        format_name = img.format

        # Deterministic low-cost visual statistics.
        rgb = img.convert("RGB")
        pixels = list(rgb.resize((32, 32)).getdata())

        total = len(pixels)

        mean_r = sum(p[0] for p in pixels) / total
        mean_g = sum(p[1] for p in pixels) / total
        mean_b = sum(p[2] for p in pixels) / total
        brightness = (mean_r + mean_g + mean_b) / 3

        analysis = {
            "mean_rgb": {
                "r": round(mean_r, 3),
                "g": round(mean_g, 3),
                "b": round(mean_b, 3),
            },
            "mean_brightness": round(brightness, 3),
            "aspect_ratio": round(width / height, 4) if height else None,
        }

    observation = VisualObservation(
        image_id=image_id,
        path=image_path,
        width=width,
        height=height,
        mode=mode,
        sha256=sha256,
        metadata={
            "format": format_name,
            "file_size_bytes": len(raw),
        },
        analysis=analysis,
    )

    output_path = VISION_ROOT / f"{image_id}.json"
    output_path.write_text(
        json.dumps(observation.to_dict(), indent=2, ensure_ascii=False),
        encoding="utf-8",
    )

    return observation


# ------------------------------------------------------------
# 3. Create deterministic test image
# ------------------------------------------------------------

test_image = VISION_ROOT / "KAGGLE-087_test.png"

if not test_image.exists():
    from PIL import Image

    img = Image.new("RGB", (256, 128), (120, 160, 200))
    img.save(test_image)


# ------------------------------------------------------------
# 4. Ingest image
# ------------------------------------------------------------

observation = ingest_image(test_image)

print("IMAGE INGESTION")
print(json.dumps(observation.to_dict(), indent=2))


# ------------------------------------------------------------
# 5. Agent-facing visual context
# ------------------------------------------------------------

visual_context = {
    "type": "visual_observation",
    "image_id": observation.image_id,
    "dimensions": [observation.width, observation.height],
    "format": observation.metadata.get("format"),
    "visual_statistics": observation.analysis,
    "vision_status": "image_ingested",
    "semantic_understanding": False,
    "vlm_required": True,
}

print("\nAGENT VISUAL CONTEXT")
print(json.dumps(visual_context, indent=2))


# ------------------------------------------------------------
# 6. Persistence verification
# ------------------------------------------------------------

saved_file = VISION_ROOT / f"{observation.image_id}.json"

assert saved_file.exists()

reloaded = json.loads(saved_file.read_text(encoding="utf-8"))

assert reloaded["image_id"] == observation.image_id
assert reloaded["sha256"] == observation.sha256
assert reloaded["width"] == observation.width
assert reloaded["height"] == observation.height
assert "analysis" in reloaded

# ------------------------------------------------------------
# 7. Capability contract
# ------------------------------------------------------------

assert observation.width == 256
assert observation.height == 128
assert observation.mode == "RGB"
assert observation.sha256
assert visual_context["type"] == "visual_observation"
assert visual_context["semantic_understanding"] is False
assert visual_context["vlm_required"] is True

print("\n" + "=" * 60)
print("KAGGLE-087 STATUS : PASS")
print("=" * 60)
print("Image ingestion          : PASS")
print("Visual observation       : PASS")
print("Image hashing            : PASS")
print("Deterministic analysis   : PASS")
print("Structured agent context : PASS")
print("Persistence              : PASS")
print("VLM integration point    : READY")
````

````python
# ============================================================
# KAGGLE-088 — ACTUAL VISION-LANGUAGE MODEL
# Qwen2.5-VL-3B-Instruct
# Image → Visual Understanding → Structured Observation
# ============================================================

import json
import gc
from pathlib import Path

import torch
from PIL import Image, ImageDraw

PROJECT_ROOT = Path("/kaggle/working/Personal_AI_AGI_Core_v3_0")
VISION_ROOT = PROJECT_ROOT / "storage" / "vision"
VISION_ROOT.mkdir(parents=True, exist_ok=True)

VLM_MODEL_ID = "Qwen/Qwen2.5-VL-3B-Instruct"

# ------------------------------------------------------------
# 1. Create a meaningful deterministic vision test image
# ------------------------------------------------------------

test_image = VISION_ROOT / "KAGGLE-088_vision_test.png"

if not test_image.exists():
    img = Image.new("RGB", (512, 320), "white")
    draw = ImageDraw.Draw(img)

    # Red circle
    draw.ellipse((70, 80, 190, 200), fill="red")

    # Blue square
    draw.rectangle((280, 80, 420, 220), fill="blue")

    # Green horizontal rectangle
    draw.rectangle((150, 245, 360, 285), fill="green")

    img.save(test_image)

print("Vision test image:", test_image)


# ------------------------------------------------------------
# 2. Load VLM only if not already loaded
# ------------------------------------------------------------

if "vlm_model" not in globals() or "vlm_processor" not in globals():

    from transformers import (
        Qwen2_5_VLForConditionalGeneration,
        AutoProcessor,
    )

    print("Loading Qwen2.5-VL-3B-Instruct...")

    vlm_model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
        VLM_MODEL_ID,
        torch_dtype=torch.bfloat16,
        device_map="auto",
        attn_implementation="sdpa",
    )

    vlm_processor = AutoProcessor.from_pretrained(
        VLM_MODEL_ID
    )

    print("VLM loaded.")

else:
    print("VLM already loaded — reusing existing instance.")


# ------------------------------------------------------------
# 3. Vision inference helper
# ------------------------------------------------------------

def vision_understand(image_path, question):
    image_path = str(Path(image_path).resolve())

    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": image_path,
                },
                {
                    "type": "text",
                    "text": question,
                },
            ],
        }
    ]

    inputs = vlm_processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )

    # Move tensors to the model's execution device.
    inputs = {
        k: v.to(vlm_model.device) if hasattr(v, "to") else v
        for k, v in inputs.items()
    }

    with torch.inference_mode():
        outputs = vlm_model.generate(
            **inputs,
            max_new_tokens=128,
            do_sample=False,
        )

    input_length = inputs["input_ids"].shape[-1]

    generated = outputs[0][input_length:]

    response = vlm_processor.decode(
        generated,
        skip_special_tokens=True,
    ).strip()

    return response


# ------------------------------------------------------------
# 4. Actual visual understanding
# ------------------------------------------------------------

question = """
Analyze this image carefully.

Return JSON only with exactly these fields:
{
  "scene": "...",
  "objects": [
    {
      "shape": "...",
      "color": "...",
      "approximate_position": "..."
    }
  ],
  "count": 0
}

Identify the visible geometric objects, their colors,
and their approximate positions.
"""

raw_visual_response = vision_understand(
    test_image,
    question,
)

print("\nRAW VLM RESPONSE")
print(raw_visual_response)


# ------------------------------------------------------------
# 5. Robust JSON extraction
# ------------------------------------------------------------

def extract_first_json_object(text):
    start = text.find("{")

    if start == -1:
        raise ValueError("No JSON object found in VLM response.")

    depth = 0
    in_string = False
    escape = False

    for i in range(start, len(text)):
        ch = text[i]

        if escape:
            escape = False
            continue

        if ch == "\\":
            escape = True
            continue

        if ch == '"':
            in_string = not in_string
            continue

        if not in_string:
            if ch == "{":
                depth += 1
            elif ch == "}":
                depth -= 1

                if depth == 0:
                    return text[start:i + 1]

    raise ValueError("Incomplete JSON object in VLM response.")


visual_json_text = extract_first_json_object(
    raw_visual_response
)

visual_result = json.loads(visual_json_text)

print("\nSTRUCTURED VISUAL RESULT")
print(json.dumps(visual_result, indent=2))


# ------------------------------------------------------------
# 6. Convert into Personal AI visual observation
# ------------------------------------------------------------

visual_observation = {
    "type": "visual_observation",
    "image_path": str(test_image),
    "vision_model": VLM_MODEL_ID,
    "semantic_understanding": True,
    "result": visual_result,
}


# ------------------------------------------------------------
# 7. Persist semantic visual observation
# ------------------------------------------------------------

semantic_path = VISION_ROOT / "KAGGLE-088_semantic_observation.json"

semantic_path.write_text(
    json.dumps(
        visual_observation,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)


# ------------------------------------------------------------
# 8. Verification
# ------------------------------------------------------------

assert visual_observation["semantic_understanding"] is True
assert visual_observation["vision_model"] == VLM_MODEL_ID
assert isinstance(visual_result, dict)
assert "scene" in visual_result
assert "objects" in visual_result
assert "count" in visual_result
assert isinstance(visual_result["objects"], list)
assert semantic_path.exists()

reloaded = json.loads(
    semantic_path.read_text(encoding="utf-8")
)

assert reloaded["semantic_understanding"] is True
assert reloaded["result"]["objects"] == visual_result["objects"]


# ------------------------------------------------------------
# 9. Agent-facing multimodal context
# ------------------------------------------------------------

agent_visual_context = {
    "modality": "image",
    "source": str(test_image),
    "vision_model": VLM_MODEL_ID,
    "semantic_understanding": True,
    "observation": visual_result,
    "persistent_record": str(semantic_path),
}

print("\nAGENT MULTIMODAL CONTEXT")
print(json.dumps(agent_visual_context, indent=2))


print("\n" + "=" * 60)
print("KAGGLE-088 STATUS : PASS")
print("=" * 60)
print("Vision model loaded       : PASS")
print("Image → VLM inference     : PASS")
print("Semantic understanding    : PASS")
print("Structured JSON           : PASS")
print("Visual observation        : PASS")
print("Persistent storage        : PASS")
print("Agent multimodal context  : PASS")
print("=" * 60)
````
